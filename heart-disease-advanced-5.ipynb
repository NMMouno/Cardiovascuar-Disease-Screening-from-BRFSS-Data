```python
# ============================================================
# Module 0 — Setup, Config, Reproducibility, Paths
# ============================================================

import os
import gc
import re
import json
import math
import time
import random
import warnings
from pathlib import Path
from dataclasses import dataclass, asdict

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
    log_loss,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    matthews_corrcoef,
    precision_recall_curve,
    roc_curve
)
from sklearn.preprocessing import RobustScaler, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.calibration import calibration_curve

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

import matplotlib.pyplot as plt
import joblib

warnings.filterwarnings("ignore")

# Optional libraries
try:
    from xgboost import XGBClassifier
    HAS_XGB = True
except Exception:
    HAS_XGB = False

try:
    from lightgbm import LGBMClassifier
    HAS_LGBM = True
except Exception:
    HAS_LGBM = False

try:
    from catboost import CatBoostClassifier
    HAS_CATBOOST = True
except Exception:
    HAS_CATBOOST = False

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader


# ----------------------------
# Reproducibility
# ----------------------------
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


seed_everything(42)


# ----------------------------
# Central config
# ----------------------------
CFG = {
    "experiment_name": "heart_disease_scratch_v1",
    "seed": 42,
    "target_col": "HeartDiseaseorAttack",   # change if needed
    "id_col": None,                         # set if dataset has an ID column
    "test_size": 0.20,
    "n_splits": 5,
    "n_repeats": 1,
    "device": "cuda" if torch.cuda.is_available() else "cpu",
    "num_workers": 2,
    "batch_size": 512,
    "max_epochs": 10,
    "patience": 3,
    "lr": 1e-3,
    "weight_decay": 1e-5,
    "attention_embed_dim": 64,
    "attention_heads": 4,
    "attention_blocks": 2,
    "attention_dropout": 0.10,
    "mlp_hidden_dims": [256, 128],
    "mlp_dropout": 0.20,
    "positive_class": 1,
    "feature_engineering": True,
    "scale_dense_for_dl": True,
    "save_models": True,
    "verbose": 1,
}

print(json.dumps(CFG, indent=2))


# ----------------------------
# Paths
# ----------------------------
BASE_DIR = Path("/kaggle/working") / CFG["experiment_name"]
OUT_DIR = BASE_DIR / "outputs"
TABLE_DIR = OUT_DIR / "tables"
FIG_DIR = OUT_DIR / "figures"
MODEL_DIR = OUT_DIR / "models"
OOF_DIR = OUT_DIR / "oof"
PRED_DIR = OUT_DIR / "preds"
LOG_DIR = OUT_DIR / "logs"

for d in [BASE_DIR, OUT_DIR, TABLE_DIR, FIG_DIR, MODEL_DIR, OOF_DIR, PRED_DIR, LOG_DIR]:
    d.mkdir(parents=True, exist_ok=True)

with open(LOG_DIR / "config.json", "w") as f:
    json.dump(CFG, f, indent=2)

print(f"Working directory: {BASE_DIR}")
print(f"Device: {CFG['device']}")

# ============================================================
# Module 1 — Data Loading and Audit
# ============================================================

# Example Kaggle input path:
# DATA_PATH = "/kaggle/input/heart-disease-health-indicators-dataset/heart_disease_health_indicators_BRFSS2015.csv"

DATA_PATH = "/kaggle/input/datasets/alexteboul/heart-disease-health-indicators-dataset/heart_disease_health_indicators_BRFSS2015.csv"
TARGET_COL = CFG["target_col"]

df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
display(df.head())

assert TARGET_COL in df.columns, f"Target column '{TARGET_COL}' not found."


# ----------------------------
# Basic audit
# ----------------------------
def basic_data_audit(df: pd.DataFrame, target_col: str) -> pd.DataFrame:
    rows = []
    for c in df.columns:
        s = df[c]
        rows.append({
            "column": c,
            "dtype": str(s.dtype),
            "n_unique": s.nunique(dropna=False),
            "n_missing": s.isna().sum(),
            "missing_pct": s.isna().mean() * 100.0,
            "min": s.min() if pd.api.types.is_numeric_dtype(s) else None,
            "max": s.max() if pd.api.types.is_numeric_dtype(s) else None,
            "mean": s.mean() if pd.api.types.is_numeric_dtype(s) else None,
            "std": s.std() if pd.api.types.is_numeric_dtype(s) else None,
            "is_target": c == target_col
        })
    return pd.DataFrame(rows)


audit_df = basic_data_audit(df, TARGET_COL)
audit_df.to_csv(TABLE_DIR / "data_audit_summary.csv", index=False)
display(audit_df.head(20))


# ----------------------------
# Target audit
# ----------------------------
target_counts = df[TARGET_COL].value_counts(dropna=False).sort_index()
target_pct = df[TARGET_COL].value_counts(normalize=True, dropna=False).sort_index()

target_summary = pd.DataFrame({
    "class": target_counts.index,
    "count": target_counts.values,
    "pct": target_pct.values
})
target_summary.to_csv(TABLE_DIR / "target_distribution.csv", index=False)
display(target_summary)

print(f"Positive prevalence: {df[TARGET_COL].mean():.6f}")


# ----------------------------
# Duplicates and missingness
# ----------------------------
n_duplicates = df.duplicated().sum()
print("Duplicated rows:", n_duplicates)

missing_summary = (
    df.isna()
      .sum()
      .reset_index()
      .rename(columns={"index": "column", 0: "n_missing"})
)
missing_summary["missing_pct"] = missing_summary["n_missing"] / len(df) * 100
missing_summary.to_csv(TABLE_DIR / "missing_summary.csv", index=False)
display(missing_summary.sort_values("n_missing", ascending=False).head(20))


# ----------------------------
# Numeric summary
# ----------------------------
numeric_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c]) and c != TARGET_COL]
desc = df[numeric_cols].describe().T
desc.to_csv(TABLE_DIR / "numeric_describe.csv")
display(desc.head(20))


# ----------------------------
# Simple plots
# ----------------------------
plt.figure(figsize=(6, 4))
plt.bar(target_summary["class"].astype(str), target_summary["count"])
plt.title("Target Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(FIG_DIR / "target_distribution.png", dpi=300, bbox_inches="tight")
plt.show()

# ============================================================
# Module 2 — Leak-Safe Split Design
# ============================================================

# Optional ID drop
drop_cols = []
if CFG["id_col"] is not None and CFG["id_col"] in df.columns:
    drop_cols.append(CFG["id_col"])

X_full = df.drop(columns=drop_cols + [TARGET_COL]).copy()
y_full = df[TARGET_COL].astype(int).copy()

print("X_full shape:", X_full.shape)
print("y_full mean:", y_full.mean())


# ----------------------------
# Train-dev / holdout split
# ----------------------------
X_dev, X_test, y_dev, y_test = train_test_split(
    X_full,
    y_full,
    test_size=CFG["test_size"],
    stratify=y_full,
    random_state=CFG["seed"]
)

print("Development shape:", X_dev.shape, y_dev.mean())
print("Holdout shape:", X_test.shape, y_test.mean())


split_summary = pd.DataFrame([
    {"split": "full", "n": len(y_full), "pos_rate": y_full.mean()},
    {"split": "dev", "n": len(y_dev), "pos_rate": y_dev.mean()},
    {"split": "test", "n": len(y_test), "pos_rate": y_test.mean()},
])
split_summary.to_csv(TABLE_DIR / "split_summary.csv", index=False)
display(split_summary)


# ----------------------------
# Repeated stratified folds on dev
# ----------------------------
rskf = RepeatedStratifiedKFold(
    n_splits=CFG["n_splits"],
    n_repeats=CFG["n_repeats"],
    random_state=CFG["seed"]
)

fold_records = []
fold_indices = []

for fold_id, (tr_idx, va_idx) in enumerate(rskf.split(X_dev, y_dev)):
    fold_indices.append((tr_idx, va_idx))
    fold_records.append({
        "fold_id": fold_id,
        "train_n": len(tr_idx),
        "valid_n": len(va_idx),
        "train_pos_rate": y_dev.iloc[tr_idx].mean(),
        "valid_pos_rate": y_dev.iloc[va_idx].mean(),
    })

fold_df = pd.DataFrame(fold_records)
fold_df.to_csv(TABLE_DIR / "cv_fold_summary.csv", index=False)
joblib.dump(fold_indices, OUT_DIR / "fold_indices.pkl")

display(fold_df.head())
print("Total folds:", len(fold_indices))

# ============================================================
# Module 3 — Preprocessing Backbone
# ============================================================

ALL_FEATURES = X_dev.columns.tolist()

numeric_features = [c for c in ALL_FEATURES if pd.api.types.is_numeric_dtype(X_dev[c])]
categorical_features = [c for c in ALL_FEATURES if c not in numeric_features]

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))


# ----------------------------
# Tree preprocessor
# ----------------------------
def build_tree_preprocessor(numeric_cols, categorical_cols):
    numeric_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median"))
    ])
    
    if len(categorical_cols) == 0:
        preprocessor = ColumnTransformer([
            ("num", numeric_pipe, numeric_cols)
        ], remainder="drop")
    else:
        categorical_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent"))
        ])
        preprocessor = ColumnTransformer([
            ("num", numeric_pipe, numeric_cols),
            ("cat", categorical_pipe, categorical_cols)
        ], remainder="drop")
    return preprocessor


# ----------------------------
# Dense / DL preprocessor
# ----------------------------
def build_dense_preprocessor(numeric_cols, categorical_cols, scaler="robust"):
    if scaler == "standard":
        scaler_obj = StandardScaler()
    else:
        scaler_obj = RobustScaler()

    numeric_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", scaler_obj)
    ])

    if len(categorical_cols) == 0:
        preprocessor = ColumnTransformer([
            ("num", numeric_pipe, numeric_cols)
        ], remainder="drop")
    else:
        categorical_pipe = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent"))
        ])
        preprocessor = ColumnTransformer([
            ("num", numeric_pipe, numeric_cols),
            ("cat", categorical_pipe, categorical_cols)
        ], remainder="drop")
    return preprocessor


tree_preprocessor = build_tree_preprocessor(numeric_features, categorical_features)
dense_preprocessor = build_dense_preprocessor(numeric_features, categorical_features, scaler="robust")


# ----------------------------
# Utility: fit-transform for a fold
# ----------------------------
def fit_transform_fold(preprocessor, X_train, X_valid, X_test=None):
    X_train_t = preprocessor.fit_transform(X_train)
    X_valid_t = preprocessor.transform(X_valid)
    X_test_t = preprocessor.transform(X_test) if X_test is not None else None

    # Convert sparse to dense if needed
    if hasattr(X_train_t, "toarray"):
        X_train_t = X_train_t.toarray()
    if hasattr(X_valid_t, "toarray"):
        X_valid_t = X_valid_t.toarray()
    if X_test_t is not None and hasattr(X_test_t, "toarray"):
        X_test_t = X_test_t.toarray()

    return X_train_t.astype(np.float32), X_valid_t.astype(np.float32), (
        X_test_t.astype(np.float32) if X_test_t is not None else None
    )

# ============================================================
# Module 4 — Clinically Grounded Feature Engineering
# ============================================================

def safe_has(cols, name):
    return name in cols

def add_feature_engineering(df_in: pd.DataFrame) -> pd.DataFrame:
    df = df_in.copy()
    cols = df.columns.tolist()

    # ------------------------
    # Example interaction terms
    # Adjust names if your dataset differs
    # ------------------------
    if safe_has(cols, "Age") and safe_has(cols, "BMI"):
        df["Age_x_BMI"] = df["Age"] * df["BMI"]

    if safe_has(cols, "Age") and safe_has(cols, "Smoker"):
        df["Age_x_Smoker"] = df["Age"] * df["Smoker"]

    if safe_has(cols, "Age") and safe_has(cols, "Diabetes"):
        df["Age_x_Diabetes"] = df["Age"] * df["Diabetes"]

    if safe_has(cols, "PhysActivity") and safe_has(cols, "BMI"):
        df["PhysActivity_x_BMI"] = df["PhysActivity"] * df["BMI"]

    if safe_has(cols, "GenHlth") and safe_has(cols, "PhysHlth"):
        df["GenHlth_x_PhysHlth"] = df["GenHlth"] * df["PhysHlth"]

    if safe_has(cols, "MentHlth") and safe_has(cols, "SleepTime"):
        df["MentHlth_x_SleepTime"] = df["MentHlth"] * df["SleepTime"]

    if safe_has(cols, "Stroke") and safe_has(cols, "Age"):
        df["Stroke_x_Age"] = df["Stroke"] * df["Age"]

    # ------------------------
    # Simple burden scores
    # ------------------------
    risk_cols_binary = [
        c for c in [
            "HighBP", "HighChol", "Smoker", "Stroke", "Diabetes",
            "DiffWalk", "Sex"
        ] if safe_has(cols, c)
    ]

    if len(risk_cols_binary) > 0:
        df["RiskFactorCount"] = df[risk_cols_binary].sum(axis=1)

    lifestyle_cols = [c for c in ["Smoker", "PhysActivity"] if safe_has(cols, c)]
    if len(lifestyle_cols) > 0:
        df["LifestyleScore"] = df[lifestyle_cols].sum(axis=1)

    # ------------------------
    # Optional bins
    # ------------------------
    if safe_has(cols, "BMI"):
        df["BMI_ge_30"] = (df["BMI"] >= 30).astype(int)
        df["BMI_ge_25"] = (df["BMI"] >= 25).astype(int)

    if safe_has(cols, "Age"):
        df["Age_ge_60"] = (df["Age"] >= 60).astype(int)
        df["Age_ge_50"] = (df["Age"] >= 50).astype(int)

    return df


if CFG["feature_engineering"]:
    X_dev_fe = add_feature_engineering(X_dev)
    X_test_fe = add_feature_engineering(X_test)
else:
    X_dev_fe = X_dev.copy()
    X_test_fe = X_test.copy()

print("Before FE:", X_dev.shape[1])
print("After FE:", X_dev_fe.shape[1])

feature_inventory = pd.DataFrame({
    "feature": X_dev_fe.columns,
    "dtype": [str(X_dev_fe[c].dtype) for c in X_dev_fe.columns]
})
feature_inventory.to_csv(TABLE_DIR / "feature_inventory_after_fe.csv", index=False)
display(feature_inventory.head(30))

# ============================================================
# Module 5A — Metrics and Utility Functions
# ============================================================

def find_best_f1_threshold(y_true, y_prob):
    precisions, recalls, thresholds = precision_recall_curve(y_true, y_prob)
    thresholds = np.append(thresholds, 1.0)
    f1s = 2 * precisions * recalls / (precisions + recalls + 1e-12)
    best_idx = np.nanargmax(f1s)
    return float(thresholds[best_idx]), float(f1s[best_idx])


def expected_calibration_error(y_true, y_prob, n_bins=15):
    y_true = np.asarray(y_true)
    y_prob = np.asarray(y_prob)

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.digitize(y_prob, bins) - 1
    bin_ids = np.clip(bin_ids, 0, n_bins - 1)

    ece = 0.0
    for b in range(n_bins):
        mask = bin_ids == b
        if mask.sum() == 0:
            continue
        acc = y_true[mask].mean()
        conf = y_prob[mask].mean()
        ece += (mask.mean()) * abs(acc - conf)
    return float(ece)


def evaluate_predictions(y_true, y_prob, threshold=0.5, prefix=""):
    y_true = np.asarray(y_true).astype(int)
    y_prob = np.asarray(y_prob).astype(float)
    y_pred = (y_prob >= threshold).astype(int)

    out = {
        f"{prefix}PR_AUC": average_precision_score(y_true, y_prob),
        f"{prefix}ROC_AUC": roc_auc_score(y_true, y_prob),
        f"{prefix}Brier": brier_score_loss(y_true, y_prob),
        f"{prefix}NLL": log_loss(y_true, np.vstack([1-y_prob, y_prob]).T, labels=[0,1]),
        f"{prefix}ECE": expected_calibration_error(y_true, y_prob, n_bins=15),
        f"{prefix}Precision": precision_score(y_true, y_pred, zero_division=0),
        f"{prefix}Recall": recall_score(y_true, y_pred, zero_division=0),
        f"{prefix}F1": f1_score(y_true, y_pred, zero_division=0),
        f"{prefix}Accuracy": accuracy_score(y_true, y_pred),
        f"{prefix}MCC": matthews_corrcoef(y_true, y_pred),
        f"{prefix}Threshold": threshold,
        f"{prefix}PosRatePred": y_pred.mean(),
        f"{prefix}MeanProb": y_prob.mean(),
        f"{prefix}N": len(y_true)
    }
    return out

# ============================================================
# Module 5B — Classical Model Builders
# ============================================================

def get_class_weight_scale(y):
    pos = np.sum(y == 1)
    neg = np.sum(y == 0)
    return max(1.0, neg / max(pos, 1))


def build_lr():
    return LogisticRegression(
        C=1.0,
        max_iter=2000,
        class_weight="balanced",
        solver="liblinear",
        random_state=CFG["seed"]
    )


def build_rf():
    return RandomForestClassifier(
        n_estimators=400,
        max_depth=None,
        min_samples_leaf=2,
        n_jobs=-1,
        class_weight="balanced_subsample",
        random_state=CFG["seed"]
    )


def build_xgb(y_train):
    assert HAS_XGB, "XGBoost is not installed."
    scale_pos_weight = get_class_weight_scale(y_train)
    return XGBClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.03,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.0,
        reg_lambda=1.0,
        objective="binary:logistic",
        eval_metric="logloss",
        scale_pos_weight=scale_pos_weight,
        tree_method="hist",
        random_state=CFG["seed"]
    )


def build_lgbm(y_train):
    assert HAS_LGBM, "LightGBM is not installed."
    return LGBMClassifier(
        n_estimators=500,
        learning_rate=0.03,
        num_leaves=63,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="binary",
        class_weight="balanced",
        random_state=CFG["seed"]
    )


def build_catboost(y_train):
    assert HAS_CATBOOST, "CatBoost is not installed."
    return CatBoostClassifier(
        iterations=500,
        depth=6,
        learning_rate=0.03,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        verbose=False,
        random_seed=CFG["seed"]
    )
```

```python
# ============================================================
# Module 5C — Torch Dataset + MLP Baseline
# ============================================================

class TabularDataset(Dataset):
    def __init__(self, X, y=None):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = None if y is None else torch.tensor(np.asarray(y), dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        if self.y is None:
            return self.X[idx]
        return self.X[idx], self.y[idx]


class MLPClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dims=(256, 128), dropout=0.2):
        super().__init__()
        dims = [input_dim] + list(hidden_dims)
        layers = []
        for i in range(len(dims) - 1):
            layers.append(nn.Linear(dims[i], dims[i+1]))
            layers.append(nn.BatchNorm1d(dims[i+1]))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout))
        layers.append(nn.Linear(dims[-1], 1))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x).squeeze(1)

# ============================================================
# Module 5D — Our Own Attention-Based DL Model (FTAN)
# ============================================================

class FeatureTokenizer(nn.Module):
    def __init__(self, num_features, embed_dim):
        super().__init__()
        self.num_features = num_features
        self.embed_dim = embed_dim

        # Project each scalar feature value into embedding space
        self.value_weight = nn.Parameter(torch.randn(num_features, embed_dim) * 0.02)
        self.value_bias = nn.Parameter(torch.zeros(num_features, embed_dim))

        # Learn feature identity embeddings
        self.feature_embedding = nn.Parameter(torch.randn(num_features, embed_dim) * 0.02)

    def forward(self, x):
        # x: [B, F]
        # token_j = x_j * W_j + b_j + e_j
        x_exp = x.unsqueeze(-1)  # [B, F, 1]
        tokens = x_exp * self.value_weight.unsqueeze(0) + self.value_bias.unsqueeze(0)
        tokens = tokens + self.feature_embedding.unsqueeze(0)
        return tokens  # [B, F, D]


class TransformerBlock(nn.Module):
    def __init__(self, embed_dim, num_heads, dropout=0.1, ff_mult=4):
        super().__init__()
        self.norm1 = nn.LayerNorm(embed_dim)
        self.attn = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )
        self.dropout1 = nn.Dropout(dropout)

        self.norm2 = nn.LayerNorm(embed_dim)
        self.ffn = nn.Sequential(
            nn.Linear(embed_dim, ff_mult * embed_dim),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(ff_mult * embed_dim, embed_dim),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        x_norm = self.norm1(x)
        attn_out, _ = self.attn(x_norm, x_norm, x_norm, need_weights=False)
        x = x + self.dropout1(attn_out)

        x_norm = self.norm2(x)
        x = x + self.ffn(x_norm)
        return x


class FTANClassifier(nn.Module):
    """
    Feature-Token Attention Network
    Our own attention-based DL model for tabular binary classification.
    """
    def __init__(self, num_features, embed_dim=64, num_heads=4, num_blocks=2, dropout=0.1):
        super().__init__()
        self.tokenizer = FeatureTokenizer(num_features=num_features, embed_dim=embed_dim)
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, dropout=dropout)
            for _ in range(num_blocks)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        self.head = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, 1)
        )

    def forward(self, x):
        # x: [B, F]
        tok = self.tokenizer(x)   # [B, F, D]
        for blk in self.blocks:
            tok = blk(tok)
        tok = self.norm(tok)

        pooled = tok.mean(dim=1)  # mean pooling across feature tokens
        logits = self.head(pooled).squeeze(1)
        return logits

# ============================================================
# Module 5E — Torch Training Utilities
# ============================================================

class FocalLossBinary(nn.Module):
    def __init__(self, alpha=0.25, gamma=2.0, reduction="mean"):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction

    def forward(self, logits, targets):
        bce = F.binary_cross_entropy_with_logits(logits, targets, reduction="none")
        probs = torch.sigmoid(logits)
        pt = targets * probs + (1 - targets) * (1 - probs)
        alpha_t = targets * self.alpha + (1 - targets) * (1 - self.alpha)
        loss = alpha_t * ((1 - pt) ** self.gamma) * bce
        if self.reduction == "mean":
            return loss.mean()
        return loss.sum()


def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0.0

    for Xb, yb in loader:
        Xb = Xb.to(device)
        yb = yb.to(device)

        optimizer.zero_grad()
        logits = model(Xb)
        loss = criterion(logits, yb)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * len(Xb)

    return total_loss / len(loader.dataset)


@torch.no_grad()
def predict_proba_torch(model, loader, device):
    model.eval()
    preds = []
    for batch in loader:
        if isinstance(batch, (list, tuple)):
            Xb = batch[0]
        else:
            Xb = batch
        Xb = Xb.to(device)
        logits = model(Xb)
        probs = torch.sigmoid(logits).detach().cpu().numpy()
        preds.append(probs)
    return np.concatenate(preds)


@torch.no_grad()
def evaluate_loss(model, loader, criterion, device):
    model.eval()
    total_loss = 0.0
    for Xb, yb in loader:
        Xb = Xb.to(device)
        yb = yb.to(device)
        logits = model(Xb)
        loss = criterion(logits, yb)
        total_loss += loss.item() * len(Xb)
    return total_loss / len(loader.dataset)


def fit_torch_model(
    model,
    X_train, y_train,
    X_valid, y_valid,
    device,
    batch_size=512,
    max_epochs=10,
    lr=1e-3,
    weight_decay=1e-5,
    patience=4,
    use_focal=False
):
    train_ds = TabularDataset(X_train, y_train)
    valid_ds = TabularDataset(X_valid, y_valid)

    train_loader = DataLoader(
        train_ds, batch_size=batch_size, shuffle=True,
        num_workers=CFG["num_workers"], pin_memory=True
    )
    valid_loader = DataLoader(
        valid_ds, batch_size=batch_size, shuffle=False,
        num_workers=CFG["num_workers"], pin_memory=True
    )

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if use_focal:
        criterion = FocalLossBinary(alpha=0.25, gamma=2.0)
    else:
        pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        tr_loss = train_one_epoch(model, train_loader, optimizer, criterion, device)
        va_loss = evaluate_loss(model, valid_loader, criterion, device)
        va_prob = predict_proba_torch(model, valid_loader, device)
        va_pr = average_precision_score(y_valid, va_prob)

        history.append({
            "epoch": epoch + 1,
            "train_loss": tr_loss,
            "valid_loss": va_loss,
            "valid_pr_auc": va_pr
        })

        if CFG["verbose"]:
            print(
                f"Epoch {epoch+1:02d} | "
                f"train_loss={tr_loss:.5f} | "
                f"valid_loss={va_loss:.5f} | "
                f"valid_pr_auc={va_pr:.5f}"
            )

        if va_pr > best_pr:
            best_pr = va_pr
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    history_df = pd.DataFrame(history)
    return model, history_df

# ============================================================
# Module 5F — Unified CV Runner
# ============================================================

def run_cv_classical(model_name, X_dev_df, y_dev, X_test_df, preproc_type="tree"):
    print(f"\n{'='*70}")
    print(f"Running CV for model: {model_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        X_va = X_dev_df.iloc[va_idx].copy()
        y_tr = y_dev.iloc[tr_idx].values
        y_va = y_dev.iloc[va_idx].values

        if preproc_type == "tree":
            preproc = build_tree_preprocessor(
                numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
                categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])]
            )
        else:
            preproc = build_dense_preprocessor(
                numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
                categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])],
                scaler="robust"
            )

        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)

        if model_name == "lr":
            model = build_lr()
        elif model_name == "rf":
            model = build_rf()
        elif model_name == "xgb":
            model = build_xgb(y_tr)
        elif model_name == "lgbm":
            model = build_lgbm(y_tr)
        elif model_name == "catboost":
            model = build_catboost(y_tr)
        else:
            raise ValueError(f"Unknown model_name: {model_name}")

        model.fit(X_tr_t, y_tr)

        va_prob = model.predict_proba(X_va_t)[:, 1]
        te_prob = model.predict_proba(X_te_t)[:, 1]

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        fold_metrics = evaluate_predictions(y_va, va_prob, threshold=0.5)
        fold_metrics["model"] = model_name
        fold_metrics["fold_id"] = fold_id
        fold_rows.append(fold_metrics)

        print(
            f"Fold {fold_id:02d} | "
            f"PR_AUC={fold_metrics['PR_AUC']:.5f} | "
            f"ROC_AUC={fold_metrics['ROC_AUC']:.5f}"
        )

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    best_thr, best_f1 = find_best_f1_threshold(y_dev.values, oof_pred)
    summary_oof = evaluate_predictions(y_dev.values, oof_pred, threshold=best_thr, prefix="OOF_")
    summary_test = evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_")

    final_row = {"model": model_name, **summary_oof, **summary_test}
    return cv_df, pd.DataFrame([final_row]), oof_pred, test_pred

# ============================================================
# Module 5G — CV Runner for MLP and FTAN
# ============================================================

def run_cv_torch(model_name, X_dev_df, y_dev, X_test_df):
    print(f"\n{'='*70}")
    print(f"Running CV for torch model: {model_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        X_va = X_dev_df.iloc[va_idx].copy()
        y_tr = y_dev.iloc[tr_idx].values.astype(np.float32)
        y_va = y_dev.iloc[va_idx].values.astype(np.float32)

        preproc = build_dense_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])],
            scaler="robust"
        )

        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)
        input_dim = X_tr_t.shape[1]

        if model_name == "mlp":
            model = MLPClassifier(
                input_dim=input_dim,
                hidden_dims=CFG["mlp_hidden_dims"],
                dropout=CFG["mlp_dropout"]
            )
            use_focal = False

        elif model_name == "ftan":
            model = FTANClassifier(
                num_features=input_dim,
                embed_dim=CFG["attention_embed_dim"],
                num_heads=CFG["attention_heads"],
                num_blocks=CFG["attention_blocks"],
                dropout=CFG["attention_dropout"]
            )
            use_focal = False

        else:
            raise ValueError(f"Unknown torch model_name: {model_name}")

        model, hist_df = fit_torch_model(
            model=model,
            X_train=X_tr_t, y_train=y_tr,
            X_valid=X_va_t, y_valid=y_va,
            device=CFG["device"],
            batch_size=CFG["batch_size"],
            max_epochs=CFG["max_epochs"],
            lr=CFG["lr"],
            weight_decay=CFG["weight_decay"],
            patience=CFG["patience"],
            use_focal=use_focal
        )

        valid_loader = DataLoader(
            TabularDataset(X_va_t, y_va),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        test_loader = DataLoader(
            TabularDataset(X_te_t, None),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        va_prob = predict_proba_torch(model, valid_loader, CFG["device"])
        te_prob = predict_proba_torch(model, test_loader, CFG["device"])

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        fold_metrics = evaluate_predictions(y_va, va_prob, threshold=0.5)
        fold_metrics["model"] = model_name
        fold_metrics["fold_id"] = fold_id
        fold_rows.append(fold_metrics)

        print(
            f"Fold {fold_id:02d} | "
            f"PR_AUC={fold_metrics['PR_AUC']:.5f} | "
            f"ROC_AUC={fold_metrics['ROC_AUC']:.5f}"
        )

        if CFG["save_models"]:
            torch.save(
                model.state_dict(),
                MODEL_DIR / f"{model_name}_fold{fold_id}.pt"
            )
            hist_df.to_csv(LOG_DIR / f"{model_name}_fold{fold_id}_history.csv", index=False)

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    best_thr, best_f1 = find_best_f1_threshold(y_dev.values, oof_pred)
    summary_oof = evaluate_predictions(y_dev.values, oof_pred, threshold=best_thr, prefix="OOF_")
    summary_test = evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_")

    final_row = {"model": model_name, **summary_oof, **summary_test}
    return cv_df, pd.DataFrame([final_row]), oof_pred, test_pred
```

```python
# ============================================================
# Module 5H — Execute Baselines
# ============================================================

X_dev_run = X_dev_fe.copy()
X_test_run = X_test_fe.copy()

all_cv_rows = []
all_summary_rows = []
all_oof_preds = {}
all_test_preds = {}

classical_models = ["lr", "rf"]
if HAS_XGB:
    classical_models.append("xgb")
if HAS_LGBM:
    classical_models.append("lgbm")
if HAS_CATBOOST:
    classical_models.append("catboost")

torch_models = ["mlp", "ftan"]   # ftan is our own attention-based model

# Classical models
for m in classical_models:
    preproc_type = "dense" if m == "lr" else "tree"
    cv_df, summary_df, oof_pred, test_pred = run_cv_classical(
        model_name=m,
        X_dev_df=X_dev_run,
        y_dev=y_dev,
        X_test_df=X_test_run,
        preproc_type=preproc_type
    )

    cv_df.to_csv(TABLE_DIR / f"{m}_cv_fold_metrics.csv", index=False)
    summary_df.to_csv(TABLE_DIR / f"{m}_summary.csv", index=False)

    all_cv_rows.append(cv_df)
    all_summary_rows.append(summary_df)
    all_oof_preds[m] = oof_pred
    all_test_preds[m] = test_pred

# Torch models
for m in torch_models:
    cv_df, summary_df, oof_pred, test_pred = run_cv_torch(
        model_name=m,
        X_dev_df=X_dev_run,
        y_dev=y_dev,
        X_test_df=X_test_run
    )

    cv_df.to_csv(TABLE_DIR / f"{m}_cv_fold_metrics.csv", index=False)
    summary_df.to_csv(TABLE_DIR / f"{m}_summary.csv", index=False)

    all_cv_rows.append(cv_df)
    all_summary_rows.append(summary_df)
    all_oof_preds[m] = oof_pred
    all_test_preds[m] = test_pred


# ----------------------------
# Save combined results
# ----------------------------
cv_results = pd.concat(all_cv_rows, axis=0, ignore_index=True)
summary_results = pd.concat(all_summary_rows, axis=0, ignore_index=True)

cv_results.to_csv(TABLE_DIR / "baseline_cv_results_all.csv", index=False)
summary_results.to_csv(TABLE_DIR / "baseline_summary_results_all.csv", index=False)

display(summary_results.sort_values("OOF_PR_AUC", ascending=False))

# Save OOF / test predictions
oof_pred_df = pd.DataFrame({"y_true": y_dev.values})
test_pred_df = pd.DataFrame({"y_true": y_test.values})

for k, v in all_oof_preds.items():
    oof_pred_df[k] = v
for k, v in all_test_preds.items():
    test_pred_df[k] = v

oof_pred_df.to_csv(OOF_DIR / "all_model_oof_predictions.csv", index=False)
test_pred_df.to_csv(PRED_DIR / "all_model_test_predictions.csv", index=False)

display(oof_pred_df.head())
display(test_pred_df.head())
```

```python
# ============================================================
# Module 6A — Utility Functions for Imbalance Experiments
# ============================================================

def summarize_cv_and_test(y_dev_true, oof_pred, y_test_true, test_pred, model_name, variant_name):
    best_thr, best_f1 = find_best_f1_threshold(y_dev_true, oof_pred)

    oof_metrics = evaluate_predictions(
        y_true=y_dev_true,
        y_prob=oof_pred,
        threshold=best_thr,
        prefix="OOF_"
    )
    test_metrics = evaluate_predictions(
        y_true=y_test_true,
        y_prob=test_pred,
        threshold=best_thr,
        prefix="TEST_"
    )

    row = {
        "model": model_name,
        "variant": variant_name,
        **oof_metrics,
        **test_metrics
    }
    return pd.DataFrame([row])


def build_catboost_custom(class_weights=None):
    assert HAS_CATBOOST, "CatBoost is not installed."
    params = {
        "iterations": 500,
        "depth": 6,
        "learning_rate": 0.03,
        "loss_function": "Logloss",
        "eval_metric": "AUC",
        "verbose": False,
        "random_seed": CFG["seed"]
    }

    if class_weights is not None:
        params["class_weights"] = class_weights
    else:
        params["auto_class_weights"] = "Balanced"

    return CatBoostClassifier(**params)

# ============================================================
# Module 6B — Torch Training with Selectable Imbalance Loss
# ============================================================

def fit_torch_model_v2(
    model,
    X_train, y_train,
    X_valid, y_valid,
    device,
    batch_size=512,
    max_epochs=40,
    lr=1e-3,
    weight_decay=1e-5,
    patience=6,
    loss_name="weighted_bce",   # ["weighted_bce", "focal"]
    focal_alpha=0.25,
    focal_gamma=2.0
):
    train_ds = TabularDataset(X_train, y_train)
    valid_ds = TabularDataset(X_valid, y_valid)

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )
    valid_loader = DataLoader(
        valid_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=lr,
        weight_decay=weight_decay
    )

    if loss_name == "weighted_bce":
        pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    elif loss_name == "focal":
        criterion = FocalLossBinary(alpha=focal_alpha, gamma=focal_gamma)
    else:
        raise ValueError(f"Unsupported loss_name: {loss_name}")

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        tr_loss = train_one_epoch(model, train_loader, optimizer, criterion, device)
        va_loss = evaluate_loss(model, valid_loader, criterion, device)
        va_prob = predict_proba_torch(model, valid_loader, device)
        va_pr = average_precision_score(y_valid, va_prob)

        history.append({
            "epoch": epoch + 1,
            "train_loss": tr_loss,
            "valid_loss": va_loss,
            "valid_pr_auc": va_pr,
            "loss_name": loss_name
        })

        if CFG["verbose"]:
            print(
                f"Epoch {epoch+1:02d} | "
                f"loss={loss_name} | "
                f"train_loss={tr_loss:.5f} | "
                f"valid_loss={va_loss:.5f} | "
                f"valid_pr_auc={va_pr:.5f}"
            )

        if va_pr > best_pr:
            best_pr = va_pr
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    history_df = pd.DataFrame(history)
    return model, history_df

# ============================================================
# Module 6C — CV Runner for CatBoost Imbalance Variants
# ============================================================

def run_cv_catboost_variant(
    X_dev_df,
    y_dev,
    X_test_df,
    variant_name="cb_balanced",
    class_weights=None
):
    print(f"\n{'='*70}")
    print(f"Running CatBoost variant: {variant_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        X_va = X_dev_df.iloc[va_idx].copy()
        y_tr = y_dev.iloc[tr_idx].values
        y_va = y_dev.iloc[va_idx].values

        preproc = build_tree_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])]
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)

        model = build_catboost_custom(class_weights=class_weights)
        model.fit(X_tr_t, y_tr)

        va_prob = model.predict_proba(X_va_t)[:, 1]
        te_prob = model.predict_proba(X_te_t)[:, 1]

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        fold_metrics = evaluate_predictions(y_va, va_prob, threshold=0.5)
        fold_metrics["model"] = "catboost"
        fold_metrics["variant"] = variant_name
        fold_metrics["fold_id"] = fold_id
        fold_rows.append(fold_metrics)

        print(
            f"Fold {fold_id:02d} | "
            f"PR_AUC={fold_metrics['PR_AUC']:.5f} | "
            f"ROC_AUC={fold_metrics['ROC_AUC']:.5f}"
        )

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    summary_df = summarize_cv_and_test(
        y_dev_true=y_dev.values,
        oof_pred=oof_pred,
        y_test_true=y_test.values,
        test_pred=test_pred,
        model_name="catboost",
        variant_name=variant_name
    )

    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 6D — CV Runner for MLP / FTAN Imbalance Variants
# ============================================================

def run_cv_torch_variant(
    model_name,
    X_dev_df,
    y_dev,
    X_test_df,
    variant_name,
    loss_name="weighted_bce",
    focal_alpha=0.25,
    focal_gamma=2.0
):
    print(f"\n{'='*70}")
    print(f"Running torch variant: {model_name} | {variant_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        X_va = X_dev_df.iloc[va_idx].copy()
        y_tr = y_dev.iloc[tr_idx].values.astype(np.float32)
        y_va = y_dev.iloc[va_idx].values.astype(np.float32)

        preproc = build_dense_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])],
            scaler="robust"
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)

        input_dim = X_tr_t.shape[1]

        if model_name == "mlp":
            model = MLPClassifier(
                input_dim=input_dim,
                hidden_dims=CFG["mlp_hidden_dims"],
                dropout=CFG["mlp_dropout"]
            )
        elif model_name == "ftan":
            model = FTANClassifier(
                num_features=input_dim,
                embed_dim=CFG["attention_embed_dim"],
                num_heads=CFG["attention_heads"],
                num_blocks=CFG["attention_blocks"],
                dropout=CFG["attention_dropout"]
            )
        else:
            raise ValueError(f"Unsupported model_name: {model_name}")

        model, hist_df = fit_torch_model_v2(
            model=model,
            X_train=X_tr_t,
            y_train=y_tr,
            X_valid=X_va_t,
            y_valid=y_va,
            device=CFG["device"],
            batch_size=CFG["batch_size"],
            max_epochs=CFG["max_epochs"],
            lr=CFG["lr"],
            weight_decay=CFG["weight_decay"],
            patience=CFG["patience"],
            loss_name=loss_name,
            focal_alpha=focal_alpha,
            focal_gamma=focal_gamma
        )

        valid_loader = DataLoader(
            TabularDataset(X_va_t, y_va),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        test_loader = DataLoader(
            TabularDataset(X_te_t, None),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        va_prob = predict_proba_torch(model, valid_loader, CFG["device"])
        te_prob = predict_proba_torch(model, test_loader, CFG["device"])

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        fold_metrics = evaluate_predictions(y_va, va_prob, threshold=0.5)
        fold_metrics["model"] = model_name
        fold_metrics["variant"] = variant_name
        fold_metrics["fold_id"] = fold_id
        fold_rows.append(fold_metrics)

        print(
            f"Fold {fold_id:02d} | "
            f"PR_AUC={fold_metrics['PR_AUC']:.5f} | "
            f"ROC_AUC={fold_metrics['ROC_AUC']:.5f}"
        )

        if CFG["save_models"]:
            torch.save(model.state_dict(), MODEL_DIR / f"{model_name}_{variant_name}_fold{fold_id}.pt")
            hist_df.to_csv(LOG_DIR / f"{model_name}_{variant_name}_fold{fold_id}_history.csv", index=False)

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    summary_df = summarize_cv_and_test(
        y_dev_true=y_dev.values,
        oof_pred=oof_pred,
        y_test_true=y_test.values,
        test_pred=test_pred,
        model_name=model_name,
        variant_name=variant_name
    )

    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 6E — Execute Module 6 Imbalance Experiments
# ============================================================

X_dev_run = X_dev_fe.copy()
X_test_run = X_test_fe.copy()

module6_cv_rows = []
module6_summary_rows = []
module6_oof_preds = {"y_true": y_dev.values}
module6_test_preds = {"y_true": y_test.values}


# ----------------------------
# CatBoost variants
# ----------------------------
if HAS_CATBOOST:
    cat_variants = [
        {
            "variant_name": "cb_balanced_auto",
            "class_weights": None
        },
        {
            "variant_name": "cb_heavy_pos",
            # [weight_for_class_0, weight_for_class_1]
            "class_weights": [1.0, 3.0]
        },
    ]

    for cfg_cb in cat_variants:
        cv_df, summary_df, oof_pred, test_pred = run_cv_catboost_variant(
            X_dev_df=X_dev_run,
            y_dev=y_dev,
            X_test_df=X_test_run,
            variant_name=cfg_cb["variant_name"],
            class_weights=cfg_cb["class_weights"]
        )

        module6_cv_rows.append(cv_df)
        module6_summary_rows.append(summary_df)
        module6_oof_preds[cfg_cb["variant_name"]] = oof_pred
        module6_test_preds[cfg_cb["variant_name"]] = test_pred


# ----------------------------
# MLP variants
# ----------------------------
mlp_variants = [
    {
        "variant_name": "mlp_weighted_bce",
        "loss_name": "weighted_bce",
        "focal_alpha": 0.25,
        "focal_gamma": 2.0
    },
    {
        "variant_name": "mlp_focal",
        "loss_name": "focal",
        "focal_alpha": 0.25,
        "focal_gamma": 2.0
    },
]

for cfg_mlp in mlp_variants:
    cv_df, summary_df, oof_pred, test_pred = run_cv_torch_variant(
        model_name="mlp",
        X_dev_df=X_dev_run,
        y_dev=y_dev,
        X_test_df=X_test_run,
        variant_name=cfg_mlp["variant_name"],
        loss_name=cfg_mlp["loss_name"],
        focal_alpha=cfg_mlp["focal_alpha"],
        focal_gamma=cfg_mlp["focal_gamma"]
    )

    module6_cv_rows.append(cv_df)
    module6_summary_rows.append(summary_df)
    module6_oof_preds[cfg_mlp["variant_name"]] = oof_pred
    module6_test_preds[cfg_mlp["variant_name"]] = test_pred


# ----------------------------
# FTAN variants
# ----------------------------
ftan_variants = [
    {
        "variant_name": "ftan_weighted_bce",
        "loss_name": "weighted_bce",
        "focal_alpha": 0.25,
        "focal_gamma": 2.0
    },
    {
        "variant_name": "ftan_focal",
        "loss_name": "focal",
        "focal_alpha": 0.25,
        "focal_gamma": 2.0
    },
]

for cfg_ftan in ftan_variants:
    cv_df, summary_df, oof_pred, test_pred = run_cv_torch_variant(
        model_name="ftan",
        X_dev_df=X_dev_run,
        y_dev=y_dev,
        X_test_df=X_test_run,
        variant_name=cfg_ftan["variant_name"],
        loss_name=cfg_ftan["loss_name"],
        focal_alpha=cfg_ftan["focal_alpha"],
        focal_gamma=cfg_ftan["focal_gamma"]
    )

    module6_cv_rows.append(cv_df)
    module6_summary_rows.append(summary_df)
    module6_oof_preds[cfg_ftan["variant_name"]] = oof_pred
    module6_test_preds[cfg_ftan["variant_name"]] = test_pred


# ----------------------------
# Save outputs
# ----------------------------
module6_cv_results = pd.concat(module6_cv_rows, axis=0, ignore_index=True)
module6_summary_results = pd.concat(module6_summary_rows, axis=0, ignore_index=True)

module6_cv_results.to_csv(TABLE_DIR / "module6_cv_fold_metrics.csv", index=False)
module6_summary_results.to_csv(TABLE_DIR / "module6_summary_results.csv", index=False)

module6_oof_df = pd.DataFrame(module6_oof_preds)
module6_test_df = pd.DataFrame(module6_test_preds)

module6_oof_df.to_csv(OOF_DIR / "module6_oof_predictions.csv", index=False)
module6_test_df.to_csv(PRED_DIR / "module6_test_predictions.csv", index=False)

display(module6_summary_results.sort_values("OOF_PR_AUC", ascending=False))

# ============================================================
# Module 6F — Compact Comparison Table
# ============================================================

keep_cols = [
    "model", "variant",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC", "OOF_Threshold",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC", "TEST_Threshold"
]

comparison_df = module6_summary_results[keep_cols].copy()
comparison_df = comparison_df.sort_values(["OOF_PR_AUC", "TEST_PR_AUC"], ascending=False)

comparison_df.to_csv(TABLE_DIR / "module6_comparison_compact.csv", index=False)
display(comparison_df)
```

```python
# ============================================================
# Module 7A — Load Predictions from Module 6
# ============================================================

oof_df = pd.read_csv(OOF_DIR / "module6_oof_predictions.csv")
test_df = pd.read_csv(PRED_DIR / "module6_test_predictions.csv")

print("OOF shape:", oof_df.shape)
print("TEST shape:", test_df.shape)

display(oof_df.head())
display(test_df.head())


# ----------------------------
# Candidate selection
# ----------------------------
# Based on Module 6 results

CANDIDATES = [
    "cb_heavy_pos",
    "cb_balanced_auto",
    "ftan_focal",
    "mlp_focal"   # optional
]

# Filter only available columns
CANDIDATES = [c for c in CANDIDATES if c in oof_df.columns]

print("Selected candidates:", CANDIDATES)

# ============================================================
# Module 7B — Correlation Analysis
# ============================================================

import seaborn as sns

# ----------------------------
# OOF correlation
# ----------------------------
corr_oof = oof_df[CANDIDATES].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr_oof, annot=True, fmt=".4f", cmap="coolwarm")
plt.title("OOF Prediction Correlation")
plt.tight_layout()
plt.savefig(FIG_DIR / "corr_oof.png", dpi=300)
plt.show()

corr_oof.to_csv(TABLE_DIR / "corr_oof.csv")


# ----------------------------
# TEST correlation
# ----------------------------
corr_test = test_df[CANDIDATES].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr_test, annot=True, fmt=".4f", cmap="coolwarm")
plt.title("TEST Prediction Correlation")
plt.tight_layout()
plt.savefig(FIG_DIR / "corr_test.png", dpi=300)
plt.show()

corr_test.to_csv(TABLE_DIR / "corr_test.csv")

# ============================================================
# Module 7C — Disagreement Analysis
# ============================================================

def compute_disagreement(df, candidates):
    preds = df[candidates].values
    std = np.std(preds, axis=1)
    max_diff = np.max(preds, axis=1) - np.min(preds, axis=1)

    return pd.DataFrame({
        "std": std,
        "max_diff": max_diff
    })


dis_oof = compute_disagreement(oof_df, CANDIDATES)
dis_test = compute_disagreement(test_df, CANDIDATES)

oof_df["disagreement_std"] = dis_oof["std"]
oof_df["disagreement_maxdiff"] = dis_oof["max_diff"]

test_df["disagreement_std"] = dis_test["std"]
test_df["disagreement_maxdiff"] = dis_test["max_diff"]


# ----------------------------
# Summary
# ----------------------------
print("\nOOF Disagreement Summary")
display(oof_df[["disagreement_std", "disagreement_maxdiff"]].describe())

print("\nTEST Disagreement Summary")
display(test_df[["disagreement_std", "disagreement_maxdiff"]].describe())


# ----------------------------
# Plot distributions
# ----------------------------
plt.figure(figsize=(6,4))
sns.histplot(oof_df["disagreement_std"], bins=50)
plt.title("OOF Disagreement STD Distribution")
plt.tight_layout()
plt.savefig(FIG_DIR / "disagreement_oof.png", dpi=300)
plt.show()

# ============================================================
# Module 7D — Error Overlap Analysis
# ============================================================

y_true = oof_df["y_true"].values

def get_binary_preds(prob, thr):
    return (prob >= thr).astype(int)

# Use each model's optimal threshold (recompute from OOF)
thresholds = {}
for m in CANDIDATES:
    thr, _ = find_best_f1_threshold(y_true, oof_df[m].values)
    thresholds[m] = thr

print("Thresholds:", thresholds)


# ----------------------------
# Build error masks
# ----------------------------
errors = {}

for m in CANDIDATES:
    y_pred = get_binary_preds(oof_df[m].values, thresholds[m])
    errors[m] = (y_pred != y_true)


# ----------------------------
# Pairwise error overlap
# ----------------------------
rows = []

for m1 in CANDIDATES:
    for m2 in CANDIDATES:
        if m1 == m2:
            continue
        
        e1 = errors[m1]
        e2 = errors[m2]

        both_wrong = np.sum(e1 & e2)
        m1_wrong_m2_correct = np.sum(e1 & (~e2))
        m2_wrong_m1_correct = np.sum(e2 & (~e1))

        rows.append({
            "model_1": m1,
            "model_2": m2,
            "both_wrong": both_wrong,
            "m1_wrong_m2_correct": m1_wrong_m2_correct,
            "m2_wrong_m1_correct": m2_wrong_m1_correct
        })

error_overlap_df = pd.DataFrame(rows)
error_overlap_df.to_csv(TABLE_DIR / "error_overlap.csv", index=False)

display(error_overlap_df.head(20))

# ============================================================
# Module 7E — Targeted Analysis: FTAN vs CatBoost
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"

y_pred_main = get_binary_preds(oof_df[main_model].values, thresholds[main_model])
y_pred_alt = get_binary_preds(oof_df[alt_model].values, thresholds[alt_model])

main_wrong = (y_pred_main != y_true)
alt_correct = (y_pred_alt == y_true)

improvement_region = main_wrong & alt_correct

print(f"\nCases where {main_model} is wrong but {alt_model} is correct:")
print("Count:", improvement_region.sum())
print("Percentage:", improvement_region.mean())


# ----------------------------
# Hard region stats
# ----------------------------
hard_df = oof_df[improvement_region].copy()

print("\nHard region summary:")
display(hard_df.describe())


# ----------------------------
# Compare probabilities
# ----------------------------
plt.figure(figsize=(6,4))
plt.scatter(
    oof_df[main_model],
    oof_df[alt_model],
    alpha=0.3
)
plt.xlabel(main_model)
plt.ylabel(alt_model)
plt.title("Prediction Scatter: CatBoost vs FTAN")
plt.tight_layout()
plt.savefig(FIG_DIR / "scatter_cb_vs_ftan.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Module 8A — Prepare Hardness Signals
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"

y_true = oof_df["y_true"].values

# thresholds from Module 7
thr_main = thresholds[main_model]
thr_alt  = thresholds[alt_model]

# ----------------------------
# Basic signals
# ----------------------------
oof_df["cb_prob"] = oof_df[main_model]
oof_df["ftan_prob"] = oof_df[alt_model]

oof_df["cb_conf"] = np.abs(oof_df["cb_prob"] - thr_main)
oof_df["ftan_conf"] = np.abs(oof_df["ftan_prob"] - thr_alt)

oof_df["mean_prob"] = oof_df[CANDIDATES].mean(axis=1)

# already computed:
# disagreement_std
# disagreement_maxdiff

# ============================================================
# Module 8B — Hard Region Definitions
# ============================================================

# 1. Disagreement-based (top X%)
dis_thr = oof_df["disagreement_std"].quantile(0.80)

oof_df["hard_disagreement"] = oof_df["disagreement_std"] >= dis_thr


# 2. Uncertainty band (around CatBoost threshold)
margin = 0.05  # can tune
oof_df["hard_uncertainty"] = (
    (oof_df["cb_prob"] >= thr_main - margin) &
    (oof_df["cb_prob"] <= thr_main + margin)
)


# 3. Low-confidence CatBoost
conf_thr = oof_df["cb_conf"].quantile(0.20)
oof_df["hard_lowconf"] = oof_df["cb_conf"] <= conf_thr


# 4. Error-driven (ideal but only for OOF analysis)
y_pred_main = (oof_df["cb_prob"] >= thr_main).astype(int)
oof_df["hard_error"] = (y_pred_main != y_true)


# 5. Combined hard region (recommended)
oof_df["hard_combined"] = (
    oof_df["hard_disagreement"] |
    oof_df["hard_uncertainty"]
)

# ============================================================
# Module 8C — Evaluate Hard Regions
# ============================================================

def evaluate_region(df, mask, name):
    subset = df[mask]
    
    size = len(subset)
    pct = size / len(df)
    
    # error rate of CatBoost
    y_pred = (subset["cb_prob"] >= thr_main).astype(int)
    err_rate = (y_pred != subset["y_true"]).mean()
    
    # FTAN improvement
    y_pred_alt = (subset["ftan_prob"] >= thr_alt).astype(int)
    improvement = np.mean(
        (y_pred != subset["y_true"]) &
        (y_pred_alt == subset["y_true"])
    )
    
    return {
        "region": name,
        "size": size,
        "pct": pct,
        "cb_error_rate": err_rate,
        "ftan_fix_rate": improvement
    }


regions = [
    "hard_disagreement",
    "hard_uncertainty",
    "hard_lowconf",
    "hard_combined"
]

results = []

for r in regions:
    res = evaluate_region(oof_df, oof_df[r], r)
    results.append(res)

hardness_df = pd.DataFrame(results)
display(hardness_df)

hardness_df.to_csv(TABLE_DIR / "hardness_region_eval.csv", index=False)

# ============================================================
# Module 8D — Visualization
# ============================================================

plt.figure(figsize=(6,4))
sns.histplot(oof_df["cb_prob"], bins=50, label="All", color="blue", alpha=0.5)
sns.histplot(oof_df[oof_df["hard_combined"]]["cb_prob"], bins=50, 
             label="Hard Region", color="red", alpha=0.5)
plt.legend()
plt.title("Probability Distribution: Hard vs All")
plt.tight_layout()
plt.savefig(FIG_DIR / "hard_region_distribution.png", dpi=300)
plt.show()

# ============================================================
# Module 8E — Save Hard Mask
# ============================================================

oof_df[["y_true", "hard_combined"]].to_csv(
    TABLE_DIR / "hard_region_mask_oof.csv", index=False
)

test_df["hard_combined"] = (
    (test_df["disagreement_std"] >= dis_thr) |
    (
        (test_df["cb_heavy_pos"] >= thr_main - margin) &
        (test_df["cb_heavy_pos"] <= thr_main + margin)
    )
)

test_df[["hard_combined"]].to_csv(
    TABLE_DIR / "hard_region_mask_test.csv", index=False
)
```

```python
# ============================================================
# FIX — Recreate hard-region masks for OOF and TEST
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"

HARD_COL = "hard_uncertainty"

# thresholds from Module 7
thr_main = thresholds[main_model]
thr_alt = thresholds[alt_model]

# same margin used in Module 8
margin = 0.05

# ----------------------------
# OOF hard mask
# ----------------------------
oof_df["hard_uncertainty"] = (
    (oof_df[main_model] >= thr_main - margin) &
    (oof_df[main_model] <= thr_main + margin)
)

# ----------------------------
# TEST hard mask
# ----------------------------
test_df["hard_uncertainty"] = (
    (test_df[main_model] >= thr_main - margin) &
    (test_df[main_model] <= thr_main + margin)
)

print("OOF hard_uncertainty count:", oof_df["hard_uncertainty"].sum())
print("TEST hard_uncertainty count:", test_df["hard_uncertainty"].sum())
print("OOF hard_uncertainty pct:", oof_df["hard_uncertainty"].mean())
print("TEST hard_uncertainty pct:", test_df["hard_uncertainty"].mean())

# ============================================================
# Module 9A — Build Hard-Region Meta Features
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"
aux_model = "mlp_focal" if "mlp_focal" in oof_df.columns else None

# Hard region definition from Module 8
HARD_COL = "hard_uncertainty"

# Rebuild dev/test feature tables aligned with oof/test order
X_dev_meta_source = X_dev_fe.reset_index(drop=True).copy()
X_test_meta_source = X_test_fe.reset_index(drop=True).copy()

assert len(X_dev_meta_source) == len(oof_df)
assert len(X_test_meta_source) == len(test_df)

def build_meta_features(base_df, feature_df, use_aux_model=True):
    out = pd.DataFrame(index=base_df.index)

    # prediction-level signals
    out["cb_prob"] = base_df[main_model].values
    out["ftan_prob"] = base_df[alt_model].values
    out["prob_diff"] = base_df[main_model].values - base_df[alt_model].values
    out["prob_mean"] = (base_df[main_model].values + base_df[alt_model].values) / 2.0
    out["prob_max"] = np.maximum(base_df[main_model].values, base_df[alt_model].values)
    out["prob_min"] = np.minimum(base_df[main_model].values, base_df[alt_model].values)

    if use_aux_model and aux_model is not None:
        out["mlp_prob"] = base_df[aux_model].values
        out["cb_mlp_diff"] = base_df[main_model].values - base_df[aux_model].values
        out["ftan_mlp_diff"] = base_df[alt_model].values - base_df[aux_model].values

    # uncertainty / disagreement features
    out["disagreement_std"] = base_df["disagreement_std"].values
    out["disagreement_maxdiff"] = base_df["disagreement_maxdiff"].values

    out["cb_conf"] = np.abs(base_df[main_model].values - thresholds[main_model])
    out["ftan_conf"] = np.abs(base_df[alt_model].values - thresholds[alt_model])

    # optional original features
    feat = feature_df.reset_index(drop=True).copy()
    feat.columns = [f"orig__{c}" for c in feat.columns]

    out = pd.concat([out.reset_index(drop=True), feat.reset_index(drop=True)], axis=1)
    return out


meta_dev_all = build_meta_features(oof_df, X_dev_meta_source, use_aux_model=True)
meta_test_all = build_meta_features(test_df, X_test_meta_source, use_aux_model=True)

print(meta_dev_all.shape, meta_test_all.shape)
display(meta_dev_all.head())

# ============================================================
# Module 9B — Select Hard Region Rows
# ============================================================

hard_mask_dev = oof_df[HARD_COL].values.astype(bool)
hard_mask_test = test_df[HARD_COL].values.astype(bool)

X_meta_dev_hard = meta_dev_all.loc[hard_mask_dev].reset_index(drop=True).copy()
y_meta_dev_hard = oof_df.loc[hard_mask_dev, "y_true"].reset_index(drop=True).astype(int).copy()

X_meta_test_hard = meta_test_all.loc[hard_mask_test].reset_index(drop=True).copy()
y_meta_test_hard = test_df.loc[hard_mask_test, "y_true"].reset_index(drop=True).astype(int).copy()

print("Hard dev shape:", X_meta_dev_hard.shape, "pos_rate:", y_meta_dev_hard.mean())
print("Hard test shape:", X_meta_test_hard.shape, "pos_rate:", y_meta_test_hard.mean())

# ============================================================
# Module 9C — Meta-Corrector CV on Hard Region
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold

def build_meta_catboost():
    assert HAS_CATBOOST, "CatBoost is required for Module 9 meta-corrector."
    return CatBoostClassifier(
        iterations=300,
        depth=4,
        learning_rate=0.03,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        verbose=False,
        random_seed=CFG["seed"]
    )

meta_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=CFG["seed"]
)

meta_oof_hard = np.zeros(len(X_meta_dev_hard), dtype=float)
meta_test_hard_folds = []
meta_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(meta_cv.split(X_meta_dev_hard, y_meta_dev_hard)):
    X_tr = X_meta_dev_hard.iloc[tr_idx].copy()
    y_tr = y_meta_dev_hard.iloc[tr_idx].values
    X_va = X_meta_dev_hard.iloc[va_idx].copy()
    y_va = y_meta_dev_hard.iloc[va_idx].values

    model = build_meta_catboost()
    model.fit(X_tr, y_tr)

    va_prob = model.predict_proba(X_va)[:, 1]
    te_prob = model.predict_proba(X_meta_test_hard)[:, 1] if len(X_meta_test_hard) > 0 else np.array([])

    meta_oof_hard[va_idx] = va_prob
    if len(X_meta_test_hard) > 0:
        meta_test_hard_folds.append(te_prob)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "meta_catboost"
    row["fold_id"] = fold_id
    meta_rows.append(row)

    print(
        f"Meta fold {fold_id:02d} | "
        f"PR_AUC={row['PR_AUC']:.5f} | "
        f"ROC_AUC={row['ROC_AUC']:.5f} | "
        f"F1={row['F1']:.5f}"
    )

meta_cv_df = pd.DataFrame(meta_rows)
meta_cv_df.to_csv(TABLE_DIR / "module9_meta_cv_metrics.csv", index=False)

if len(meta_test_hard_folds) > 0:
    meta_test_hard_pred = np.mean(np.vstack(meta_test_hard_folds), axis=0)
else:
    meta_test_hard_pred = np.array([])

display(meta_cv_df.describe())

# ============================================================
# Module 9D — Hard-Region Comparison
# ============================================================

cb_hard_dev_prob = oof_df.loc[hard_mask_dev, main_model].reset_index(drop=True).values
ftan_hard_dev_prob = oof_df.loc[hard_mask_dev, alt_model].reset_index(drop=True).values

hard_dev_compare = pd.DataFrame([
    {"model": "cb_heavy_pos_hard", **evaluate_predictions(y_meta_dev_hard.values, cb_hard_dev_prob, threshold=thresholds[main_model])},
    {"model": "ftan_focal_hard", **evaluate_predictions(y_meta_dev_hard.values, ftan_hard_dev_prob, threshold=thresholds[alt_model])},
    {"model": "meta_catboost_hard", **evaluate_predictions(y_meta_dev_hard.values, meta_oof_hard, threshold=0.5)},
])

hard_dev_compare.to_csv(TABLE_DIR / "module9_hard_region_compare_dev.csv", index=False)
display(hard_dev_compare)

# ============================================================
# Module 9E — Two-Stage Final Prediction
# ============================================================

# Full dev prediction
final_dev_prob = oof_df[main_model].values.copy()
final_dev_prob[hard_mask_dev] = meta_oof_hard

# Full test prediction
final_test_prob = test_df[main_model].values.copy()
if len(meta_test_hard_pred) > 0:
    final_test_prob[hard_mask_test] = meta_test_hard_pred

# Save
two_stage_dev_df = pd.DataFrame({
    "y_true": oof_df["y_true"].values,
    "cb_heavy_pos": oof_df[main_model].values,
    "two_stage_prob": final_dev_prob,
    "hard_region": hard_mask_dev.astype(int)
})

two_stage_test_df = pd.DataFrame({
    "y_true": test_df["y_true"].values,
    "cb_heavy_pos": test_df[main_model].values,
    "two_stage_prob": final_test_prob,
    "hard_region": hard_mask_test.astype(int)
})

two_stage_dev_df.to_csv(OOF_DIR / "module9_two_stage_dev_predictions.csv", index=False)
two_stage_test_df.to_csv(PRED_DIR / "module9_two_stage_test_predictions.csv", index=False)

display(two_stage_dev_df.head())
display(two_stage_test_df.head())
```

```python
# ============================================================
# Module 9F — Evaluate Full Two-Stage System
# ============================================================

cb_dev_prob = oof_df[main_model].values
cb_test_prob = test_df[main_model].values

cb_thr = thresholds[main_model]
two_stage_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob)

module9_summary = pd.DataFrame([
    {
        "model": "cb_heavy_pos_full",
        **evaluate_predictions(oof_df["y_true"].values, cb_dev_prob, threshold=cb_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, cb_test_prob, threshold=cb_thr, prefix="TEST_"),
    },
    {
        "model": "two_stage_residual_full",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob, threshold=two_stage_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob, threshold=two_stage_thr, prefix="TEST_"),
    }
])

module9_summary.to_csv(TABLE_DIR / "module9_full_system_summary.csv", index=False)
display(module9_summary)

# ============================================================
# Module 9G — Optional Blended Two-Stage Variant
# ============================================================

blend_alpha = 0.7

final_dev_prob_blend = oof_df[main_model].values.copy()
final_test_prob_blend = test_df[main_model].values.copy()

final_dev_prob_blend[hard_mask_dev] = (
    blend_alpha * meta_oof_hard +
    (1 - blend_alpha) * oof_df.loc[hard_mask_dev, main_model].values
)

if len(meta_test_hard_pred) > 0:
    final_test_prob_blend[hard_mask_test] = (
        blend_alpha * meta_test_hard_pred +
        (1 - blend_alpha) * test_df.loc[hard_mask_test, main_model].values
    )

blend_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_blend)

blend_summary = pd.DataFrame([
    {
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_blend, threshold=blend_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob_blend, threshold=blend_thr, prefix="TEST_"),
    }
])

blend_summary.to_csv(TABLE_DIR / "module9_blend_summary.csv", index=False)
display(blend_summary)

# ============================================================
# Module 9H — Final Comparison Table
# ============================================================

module9_all = pd.concat([module9_summary, blend_summary], axis=0, ignore_index=True)
module9_all.to_csv(TABLE_DIR / "module9_all_results.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module9_all[keep_cols].sort_values("TEST_PR_AUC", ascending=False))
```

```python
# ============================================================
# Module 9.5A — Error Detector Target
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"
HARD_COL = "hard_uncertainty"

# Defensive checks
assert HARD_COL in oof_df.columns, f"{HARD_COL} missing from oof_df"
assert HARD_COL in test_df.columns, f"{HARD_COL} missing from test_df"

hard_mask_dev = oof_df[HARD_COL].values.astype(bool)
hard_mask_test = test_df[HARD_COL].values.astype(bool)

# CatBoost binary prediction using OOF-selected threshold
cb_pred_dev = (oof_df[main_model].values >= thresholds[main_model]).astype(int)

# Error target: 1 if CatBoost is wrong, 0 if correct
cb_error_target_dev = (cb_pred_dev != oof_df["y_true"].values).astype(int)

# Restrict to hard region
X_err_dev = meta_dev_all.loc[hard_mask_dev].reset_index(drop=True).copy()
y_err_dev = pd.Series(cb_error_target_dev[hard_mask_dev]).reset_index(drop=True).astype(int)

X_err_test = meta_test_all.loc[hard_mask_test].reset_index(drop=True).copy()
y_test_hard = test_df.loc[hard_mask_test, "y_true"].reset_index(drop=True).astype(int).copy()

print("Error detector DEV shape:", X_err_dev.shape, "error rate:", y_err_dev.mean())
print("Error detector TEST shape:", X_err_test.shape)

# ============================================================
# Module 9.5B — OOF Error Detector Training
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold

def build_error_detector_catboost():
    assert HAS_CATBOOST, "CatBoost is required."
    return CatBoostClassifier(
        iterations=250,
        depth=4,
        learning_rate=0.03,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        verbose=False,
        random_seed=CFG["seed"]
    )

err_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=CFG["seed"]
)

err_oof_hard = np.zeros(len(X_err_dev), dtype=float)
err_test_hard_folds = []
err_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(err_cv.split(X_err_dev, y_err_dev)):
    X_tr = X_err_dev.iloc[tr_idx].copy()
    y_tr = y_err_dev.iloc[tr_idx].values
    X_va = X_err_dev.iloc[va_idx].copy()
    y_va = y_err_dev.iloc[va_idx].values

    model = build_error_detector_catboost()
    model.fit(X_tr, y_tr)

    va_prob = model.predict_proba(X_va)[:, 1]
    err_oof_hard[va_idx] = va_prob

    if len(X_err_test) > 0:
        te_prob = model.predict_proba(X_err_test)[:, 1]
        err_test_hard_folds.append(te_prob)

    row = {
        "fold_id": fold_id,
        "ROC_AUC_error_target": roc_auc_score(y_va, va_prob),
        "PR_AUC_error_target": average_precision_score(y_va, va_prob),
        "MeanPred": float(np.mean(va_prob)),
        "TrueRate": float(np.mean(y_va))
    }
    err_rows.append(row)

    print(
        f"Err fold {fold_id:02d} | "
        f"ROC_AUC={row['ROC_AUC_error_target']:.5f} | "
        f"PR_AUC={row['PR_AUC_error_target']:.5f}"
    )

err_cv_df = pd.DataFrame(err_rows)
err_cv_df.to_csv(TABLE_DIR / "module95_error_detector_cv.csv", index=False)
display(err_cv_df.describe())

if len(err_test_hard_folds) > 0:
    err_test_hard = np.mean(np.vstack(err_test_hard_folds), axis=0)
else:
    err_test_hard = np.array([])

# ============================================================
# Module 9.5C — Error Detector Diagnostics
# ============================================================

error_detector_summary = pd.DataFrame([{
    "model": "cb_error_detector_hard",
    "OOF_ROC_AUC_error_target": roc_auc_score(y_err_dev, err_oof_hard),
    "OOF_PR_AUC_error_target": average_precision_score(y_err_dev, err_oof_hard),
    "OOF_error_rate": float(y_err_dev.mean()),
    "OOF_mean_pred": float(err_oof_hard.mean()),
    "N_hard_dev": int(len(y_err_dev))
}])

error_detector_summary.to_csv(TABLE_DIR / "module95_error_detector_summary.csv", index=False)
display(error_detector_summary)
```

```python
# ============================================================
# Module 10A.1 — Adaptive Blending Function
# ============================================================

def adaptive_blend_probs(cb_prob, ftan_prob, err_prob, w_min=0.15, w_max=0.85):
    """
    cb_prob   : CatBoost probability
    ftan_prob : FTAN probability
    err_prob  : probability CatBoost is wrong
    """
    err_prob = np.asarray(err_prob)
    w_ftan = w_min + (w_max - w_min) * err_prob
    w_ftan = np.clip(w_ftan, w_min, w_max)

    final_prob = (1.0 - w_ftan) * np.asarray(cb_prob) + w_ftan * np.asarray(ftan_prob)
    return final_prob, w_ftan

# ============================================================
# Module 10A.2 — Adaptive Two-Stage Predictions
# ============================================================

cb_dev_prob = oof_df[main_model].values.copy()
ftan_dev_prob = oof_df[alt_model].values.copy()

cb_test_prob = test_df[main_model].values.copy()
ftan_test_prob = test_df[alt_model].values.copy()

adaptive_dev_prob = cb_dev_prob.copy()
adaptive_test_prob = cb_test_prob.copy()

adaptive_dev_weight = np.zeros_like(cb_dev_prob, dtype=float)
adaptive_test_weight = np.zeros_like(cb_test_prob, dtype=float)

# Apply adaptive blending only on hard region
adaptive_dev_prob_hard, adaptive_dev_weight_hard = adaptive_blend_probs(
    cb_prob=cb_dev_prob[hard_mask_dev],
    ftan_prob=ftan_dev_prob[hard_mask_dev],
    err_prob=err_oof_hard,
    w_min=0.15,
    w_max=0.85
)

adaptive_dev_prob[hard_mask_dev] = adaptive_dev_prob_hard
adaptive_dev_weight[hard_mask_dev] = adaptive_dev_weight_hard

if len(err_test_hard) > 0:
    adaptive_test_prob_hard, adaptive_test_weight_hard = adaptive_blend_probs(
        cb_prob=cb_test_prob[hard_mask_test],
        ftan_prob=ftan_test_prob[hard_mask_test],
        err_prob=err_test_hard,
        w_min=0.15,
        w_max=0.85
    )
    adaptive_test_prob[hard_mask_test] = adaptive_test_prob_hard
    adaptive_test_weight[hard_mask_test] = adaptive_test_weight_hard

# Save detailed predictions
adaptive_dev_df = pd.DataFrame({
    "y_true": oof_df["y_true"].values,
    "cb_prob": cb_dev_prob,
    "ftan_prob": ftan_dev_prob,
    "adaptive_prob": adaptive_dev_prob,
    "hard_region": hard_mask_dev.astype(int),
    "ftan_weight": adaptive_dev_weight
})

adaptive_test_df = pd.DataFrame({
    "y_true": test_df["y_true"].values,
    "cb_prob": cb_test_prob,
    "ftan_prob": ftan_test_prob,
    "adaptive_prob": adaptive_test_prob,
    "hard_region": hard_mask_test.astype(int),
    "ftan_weight": adaptive_test_weight
})

adaptive_dev_df.to_csv(OOF_DIR / "module10A_adaptive_dev_predictions.csv", index=False)
adaptive_test_df.to_csv(PRED_DIR / "module10A_adaptive_test_predictions.csv", index=False)

display(adaptive_dev_df.head())
display(adaptive_test_df.head())

# ============================================================
# Module 10A.3 — Compare Baseline, Fixed Blend, Adaptive Blend
# ============================================================

# Baseline
cb_thr = thresholds[main_model]

# Adaptive threshold chosen on DEV
adaptive_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, adaptive_dev_prob)

# Fixed-blend threshold if final_dev_prob_blend exists
has_fixed_blend = "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals()

rows = []

rows.append({
    "model": "cb_heavy_pos_full",
    **evaluate_predictions(oof_df["y_true"].values, cb_dev_prob, threshold=cb_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, cb_test_prob, threshold=cb_thr, prefix="TEST_"),
})

if has_fixed_blend:
    fixed_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

rows.append({
    "model": "adaptive_error_guided_blend",
    **evaluate_predictions(oof_df["y_true"].values, adaptive_dev_prob, threshold=adaptive_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, adaptive_test_prob, threshold=adaptive_thr, prefix="TEST_"),
})

module10A_summary = pd.DataFrame(rows)
module10A_summary.to_csv(TABLE_DIR / "module10A_summary.csv", index=False)
display(module10A_summary)

# ============================================================
# Module 10A.4 — Inspect Adaptive Weights
# ============================================================

weight_summary = pd.DataFrame({
    "split": ["dev", "test"],
    "n_hard": [int(hard_mask_dev.sum()), int(hard_mask_test.sum())],
    "mean_ftan_weight_hard": [
        float(adaptive_dev_weight[hard_mask_dev].mean()) if hard_mask_dev.sum() > 0 else np.nan,
        float(adaptive_test_weight[hard_mask_test].mean()) if hard_mask_test.sum() > 0 else np.nan,
    ],
    "min_ftan_weight_hard": [
        float(adaptive_dev_weight[hard_mask_dev].min()) if hard_mask_dev.sum() > 0 else np.nan,
        float(adaptive_test_weight[hard_mask_test].min()) if hard_mask_test.sum() > 0 else np.nan,
    ],
    "max_ftan_weight_hard": [
        float(adaptive_dev_weight[hard_mask_dev].max()) if hard_mask_dev.sum() > 0 else np.nan,
        float(adaptive_test_weight[hard_mask_test].max()) if hard_mask_test.sum() > 0 else np.nan,
    ],
})

weight_summary.to_csv(TABLE_DIR / "module10A_weight_summary.csv", index=False)
display(weight_summary)

plt.figure(figsize=(6,4))
plt.hist(adaptive_dev_weight[hard_mask_dev], bins=40)
plt.title("Adaptive FTAN Weight Distribution (DEV Hard Region)")
plt.xlabel("FTAN weight")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(FIG_DIR / "module10A_dev_ftan_weight_hist.png", dpi=300)
plt.show()

# ============================================================
# Module 10A.5 — Optional Confidence-Adjusted Adaptive Blend
# ============================================================

def adaptive_blend_probs_v2(cb_prob, ftan_prob, err_prob, cb_conf, conf_scale=0.10, w_min=0.15, w_max=0.90):
    """
    More FTAN weight when:
    - CatBoost likely wrong
    - CatBoost is close to its threshold (low confidence)
    """
    cb_conf = np.asarray(cb_conf)
    err_prob = np.asarray(err_prob)

    conf_factor = np.exp(-cb_conf / conf_scale)   # near threshold => closer to 1
    blended_signal = 0.65 * err_prob + 0.35 * conf_factor

    w_ftan = w_min + (w_max - w_min) * blended_signal
    w_ftan = np.clip(w_ftan, w_min, w_max)

    final_prob = (1.0 - w_ftan) * np.asarray(cb_prob) + w_ftan * np.asarray(ftan_prob)
    return final_prob, w_ftan


adaptive_dev_prob_v2 = cb_dev_prob.copy()
adaptive_test_prob_v2 = cb_test_prob.copy()

adaptive_dev_w_v2 = np.zeros_like(cb_dev_prob, dtype=float)
adaptive_test_w_v2 = np.zeros_like(cb_test_prob, dtype=float)

dev_cb_conf = np.abs(cb_dev_prob[hard_mask_dev] - thresholds[main_model])
test_cb_conf = np.abs(cb_test_prob[hard_mask_test] - thresholds[main_model])

dev_prob_v2_hard, dev_w_v2_hard = adaptive_blend_probs_v2(
    cb_prob=cb_dev_prob[hard_mask_dev],
    ftan_prob=ftan_dev_prob[hard_mask_dev],
    err_prob=err_oof_hard,
    cb_conf=dev_cb_conf,
    conf_scale=0.10,
    w_min=0.15,
    w_max=0.90
)
adaptive_dev_prob_v2[hard_mask_dev] = dev_prob_v2_hard
adaptive_dev_w_v2[hard_mask_dev] = dev_w_v2_hard

if len(err_test_hard) > 0:
    test_prob_v2_hard, test_w_v2_hard = adaptive_blend_probs_v2(
        cb_prob=cb_test_prob[hard_mask_test],
        ftan_prob=ftan_test_prob[hard_mask_test],
        err_prob=err_test_hard,
        cb_conf=test_cb_conf,
        conf_scale=0.10,
        w_min=0.15,
        w_max=0.90
    )
    adaptive_test_prob_v2[hard_mask_test] = test_prob_v2_hard
    adaptive_test_w_v2[hard_mask_test] = test_w_v2_hard

adaptive_thr_v2, _ = find_best_f1_threshold(oof_df["y_true"].values, adaptive_dev_prob_v2)

module10A_v2_summary = pd.DataFrame([{
    "model": "adaptive_error_conf_guided_blend",
    **evaluate_predictions(oof_df["y_true"].values, adaptive_dev_prob_v2, threshold=adaptive_thr_v2, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, adaptive_test_prob_v2, threshold=adaptive_thr_v2, prefix="TEST_"),
}])

module10A_v2_summary.to_csv(TABLE_DIR / "module10A_v2_summary.csv", index=False)
display(module10A_v2_summary)
```

```python
# ============================================================
# Module 10C.1 — Build FTAN-Takeover Target
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"
HARD_COL = "hard_uncertainty"

assert HARD_COL in oof_df.columns, f"{HARD_COL} missing from oof_df"
assert HARD_COL in test_df.columns, f"{HARD_COL} missing from test_df"

hard_mask_dev = oof_df[HARD_COL].values.astype(bool)
hard_mask_test = test_df[HARD_COL].values.astype(bool)

# Binary predictions using OOF-derived thresholds
cb_pred_dev = (oof_df[main_model].values >= thresholds[main_model]).astype(int)
ftan_pred_dev = (oof_df[alt_model].values >= thresholds[alt_model]).astype(int)

y_true_dev = oof_df["y_true"].values.astype(int)

# Takeover target:
# 1 if FTAN is correct and CatBoost is wrong
# 0 otherwise
takeover_target_dev = (
    (ftan_pred_dev == y_true_dev) &
    (cb_pred_dev != y_true_dev)
).astype(int)

# Restrict to hard region
X_take_dev = meta_dev_all.loc[hard_mask_dev].reset_index(drop=True).copy()
y_take_dev = pd.Series(takeover_target_dev[hard_mask_dev]).reset_index(drop=True).astype(int)

X_take_test = meta_test_all.loc[hard_mask_test].reset_index(drop=True).copy()
y_test_hard = test_df.loc[hard_mask_test, "y_true"].reset_index(drop=True).astype(int).copy()

print("Takeover DEV shape:", X_take_dev.shape, "positive rate:", y_take_dev.mean())
print("Takeover TEST shape:", X_take_test.shape)

# ============================================================
# Module 10C.2 — OOF Takeover Detector Training
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold

def build_takeover_detector():
    assert HAS_CATBOOST, "CatBoost is required."
    return CatBoostClassifier(
        iterations=250,
        depth=4,
        learning_rate=0.03,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        verbose=False,
        random_seed=CFG["seed"]
    )

take_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=CFG["seed"]
)

take_oof_hard = np.zeros(len(X_take_dev), dtype=float)
take_test_hard_folds = []
take_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(take_cv.split(X_take_dev, y_take_dev)):
    X_tr = X_take_dev.iloc[tr_idx].copy()
    y_tr = y_take_dev.iloc[tr_idx].values
    X_va = X_take_dev.iloc[va_idx].copy()
    y_va = y_take_dev.iloc[va_idx].values

    model = build_takeover_detector()
    model.fit(X_tr, y_tr)

    va_prob = model.predict_proba(X_va)[:, 1]
    take_oof_hard[va_idx] = va_prob

    if len(X_take_test) > 0:
        te_prob = model.predict_proba(X_take_test)[:, 1]
        take_test_hard_folds.append(te_prob)

    row = {
        "fold_id": fold_id,
        "ROC_AUC_takeover_target": roc_auc_score(y_va, va_prob) if len(np.unique(y_va)) > 1 else np.nan,
        "PR_AUC_takeover_target": average_precision_score(y_va, va_prob),
        "MeanPred": float(np.mean(va_prob)),
        "TrueRate": float(np.mean(y_va))
    }
    take_rows.append(row)

    print(
        f"Take fold {fold_id:02d} | "
        f"ROC_AUC={row['ROC_AUC_takeover_target']:.5f} | "
        f"PR_AUC={row['PR_AUC_takeover_target']:.5f}"
    )

take_cv_df = pd.DataFrame(take_rows)
take_cv_df.to_csv(TABLE_DIR / "module10C_takeover_detector_cv.csv", index=False)
display(take_cv_df.describe())

if len(take_test_hard_folds) > 0:
    take_test_hard = np.mean(np.vstack(take_test_hard_folds), axis=0)
else:
    take_test_hard = np.array([])

# ============================================================
# Module 10C.3 — Takeover Detector Summary
# ============================================================

takeover_summary = pd.DataFrame([{
    "model": "ftan_takeover_detector_hard",
    "OOF_ROC_AUC_takeover_target": roc_auc_score(y_take_dev, take_oof_hard) if len(np.unique(y_take_dev)) > 1 else np.nan,
    "OOF_PR_AUC_takeover_target": average_precision_score(y_take_dev, take_oof_hard),
    "OOF_takeover_rate": float(y_take_dev.mean()),
    "OOF_mean_pred": float(take_oof_hard.mean()),
    "N_hard_dev": int(len(y_take_dev))
}])

takeover_summary.to_csv(TABLE_DIR / "module10C_takeover_summary.csv", index=False)
display(takeover_summary)

# ============================================================
# Module 10C.4 — Hard Switch via Takeover Probability
# ============================================================

def evaluate_switch_thresholds(
    y_true_full,
    cb_prob_full,
    ftan_prob_full,
    hard_mask,
    takeover_prob_hard,
    thresholds_to_try=None
):
    if thresholds_to_try is None:
        thresholds_to_try = np.linspace(0.1, 0.9, 17)

    rows = []

    for t in thresholds_to_try:
        pred = cb_prob_full.copy()
        hard_switch = takeover_prob_hard >= t

        pred_hard = pred[hard_mask].copy()
        pred_hard[hard_switch] = ftan_prob_full[hard_mask][hard_switch]
        pred[hard_mask] = pred_hard

        thr_eval, _ = find_best_f1_threshold(y_true_full, pred)
        metrics = evaluate_predictions(y_true_full, pred, threshold=thr_eval)
        metrics["takeover_gate_thr"] = t
        rows.append(metrics)

    return pd.DataFrame(rows)


cb_dev_prob = oof_df[main_model].values.copy()
ftan_dev_prob = oof_df[alt_model].values.copy()

cb_test_prob = test_df[main_model].values.copy()
ftan_test_prob = test_df[alt_model].values.copy()

switch_grid_dev = evaluate_switch_thresholds(
    y_true_full=oof_df["y_true"].values,
    cb_prob_full=cb_dev_prob,
    ftan_prob_full=ftan_dev_prob,
    hard_mask=hard_mask_dev,
    takeover_prob_hard=take_oof_hard,
    thresholds_to_try=np.linspace(0.1, 0.9, 17)
)

switch_grid_dev.to_csv(TABLE_DIR / "module10C_switch_grid_dev.csv", index=False)
display(switch_grid_dev.sort_values(["PR_AUC", "F1", "MCC"], ascending=False).head(10))

best_switch_row = switch_grid_dev.sort_values(["PR_AUC", "F1", "MCC"], ascending=False).iloc[0]
best_switch_gate_thr = float(best_switch_row["takeover_gate_thr"])
print("Best takeover gate threshold:", best_switch_gate_thr)

# ============================================================
# Module 10C.5 — Final Hard-Switch Predictions
# ============================================================

# DEV
switch_dev_prob = cb_dev_prob.copy()
switch_mask_dev_hard = take_oof_hard >= best_switch_gate_thr
tmp_dev = switch_dev_prob[hard_mask_dev].copy()
tmp_dev[switch_mask_dev_hard] = ftan_dev_prob[hard_mask_dev][switch_mask_dev_hard]
switch_dev_prob[hard_mask_dev] = tmp_dev

# TEST
switch_test_prob = cb_test_prob.copy()
if len(take_test_hard) > 0:
    switch_mask_test_hard = take_test_hard >= best_switch_gate_thr
    tmp_test = switch_test_prob[hard_mask_test].copy()
    tmp_test[switch_mask_test_hard] = ftan_test_prob[hard_mask_test][switch_mask_test_hard]
    switch_test_prob[hard_mask_test] = tmp_test
else:
    switch_mask_test_hard = np.array([], dtype=bool)

switch_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, switch_dev_prob)

module10C_switch_summary = pd.DataFrame([{
    "model": "takeover_guided_hard_switch",
    **evaluate_predictions(oof_df["y_true"].values, switch_dev_prob, threshold=switch_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, switch_test_prob, threshold=switch_thr, prefix="TEST_"),
    "gate_threshold": best_switch_gate_thr
}])

module10C_switch_summary.to_csv(TABLE_DIR / "module10C_switch_summary.csv", index=False)
display(module10C_switch_summary)

# ============================================================
# Module 10C.6 — Soft Takeover-Guided Blending
# ============================================================

def takeover_guided_blend(cb_prob, ftan_prob, take_prob, w_min=0.10, w_max=0.90):
    take_prob = np.asarray(take_prob)
    w_ftan = w_min + (w_max - w_min) * take_prob
    w_ftan = np.clip(w_ftan, w_min, w_max)

    final_prob = (1.0 - w_ftan) * np.asarray(cb_prob) + w_ftan * np.asarray(ftan_prob)
    return final_prob, w_ftan


# DEV
take_blend_dev_prob = cb_dev_prob.copy()
take_blend_dev_weight = np.zeros_like(cb_dev_prob, dtype=float)

dev_hard_prob, dev_hard_weight = takeover_guided_blend(
    cb_prob=cb_dev_prob[hard_mask_dev],
    ftan_prob=ftan_dev_prob[hard_mask_dev],
    take_prob=take_oof_hard,
    w_min=0.10,
    w_max=0.90
)
take_blend_dev_prob[hard_mask_dev] = dev_hard_prob
take_blend_dev_weight[hard_mask_dev] = dev_hard_weight

# TEST
take_blend_test_prob = cb_test_prob.copy()
take_blend_test_weight = np.zeros_like(cb_test_prob, dtype=float)

if len(take_test_hard) > 0:
    test_hard_prob, test_hard_weight = takeover_guided_blend(
        cb_prob=cb_test_prob[hard_mask_test],
        ftan_prob=ftan_test_prob[hard_mask_test],
        take_prob=take_test_hard,
        w_min=0.10,
        w_max=0.90
    )
    take_blend_test_prob[hard_mask_test] = test_hard_prob
    take_blend_test_weight[hard_mask_test] = test_hard_weight

take_blend_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, take_blend_dev_prob)

module10C_blend_summary = pd.DataFrame([{
    "model": "takeover_guided_soft_blend",
    **evaluate_predictions(oof_df["y_true"].values, take_blend_dev_prob, threshold=take_blend_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, take_blend_test_prob, threshold=take_blend_thr, prefix="TEST_"),
}])

module10C_blend_summary.to_csv(TABLE_DIR / "module10C_blend_summary.csv", index=False)
display(module10C_blend_summary)

# ============================================================
# Module 10C.7 — Compare Baseline, Fixed Blend, Takeover Models
# ============================================================

rows = []

# Baseline
rows.append({
    "model": "cb_heavy_pos_full",
    **evaluate_predictions(oof_df["y_true"].values, cb_dev_prob, threshold=thresholds[main_model], prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, cb_test_prob, threshold=thresholds[main_model], prefix="TEST_"),
})

# Fixed blend if available
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# Hard switch
rows.append({
    "model": "takeover_guided_hard_switch",
    **evaluate_predictions(oof_df["y_true"].values, switch_dev_prob, threshold=switch_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, switch_test_prob, threshold=switch_thr, prefix="TEST_"),
})

# Soft blend
rows.append({
    "model": "takeover_guided_soft_blend",
    **evaluate_predictions(oof_df["y_true"].values, take_blend_dev_prob, threshold=take_blend_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, take_blend_test_prob, threshold=take_blend_thr, prefix="TEST_"),
})

module10C_compare = pd.DataFrame(rows)
module10C_compare.to_csv(TABLE_DIR / "module10C_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module10C_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 10C.8 — Takeover Diagnostics
# ============================================================

takeover_diag = pd.DataFrame({
    "split": ["dev", "test"],
    "n_hard": [int(hard_mask_dev.sum()), int(hard_mask_test.sum())],
    "takeover_rate_hard_switch": [
        float(np.mean(switch_mask_dev_hard)) if len(switch_mask_dev_hard) > 0 else np.nan,
        float(np.mean(switch_mask_test_hard)) if len(switch_mask_test_hard) > 0 else np.nan,
    ],
    "mean_ftan_weight_soft_blend": [
        float(take_blend_dev_weight[hard_mask_dev].mean()) if hard_mask_dev.sum() > 0 else np.nan,
        float(take_blend_test_weight[hard_mask_test].mean()) if hard_mask_test.sum() > 0 else np.nan,
    ],
    "min_ftan_weight_soft_blend": [
        float(take_blend_dev_weight[hard_mask_dev].min()) if hard_mask_dev.sum() > 0 else np.nan,
        float(take_blend_test_weight[hard_mask_test].min()) if hard_mask_test.sum() > 0 else np.nan,
    ],
    "max_ftan_weight_soft_blend": [
        float(take_blend_dev_weight[hard_mask_dev].max()) if hard_mask_dev.sum() > 0 else np.nan,
        float(take_blend_test_weight[hard_mask_test].max()) if hard_mask_test.sum() > 0 else np.nan,
    ],
})

takeover_diag.to_csv(TABLE_DIR / "module10C_takeover_diag.csv", index=False)
display(takeover_diag)

plt.figure(figsize=(6,4))
plt.hist(take_blend_dev_weight[hard_mask_dev], bins=40)
plt.title("Takeover-Guided FTAN Weight Distribution (DEV Hard Region)")
plt.xlabel("FTAN weight")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(FIG_DIR / "module10C_takeover_weight_hist.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Module 10D.1 — FTAN with Embedding Extraction
# ============================================================

class FTANClassifier(nn.Module):
    """
    Feature-Token Attention Network
    with pooled embedding extraction.
    """
    def __init__(self, num_features, embed_dim=64, num_heads=4, num_blocks=2, dropout=0.1):
        super().__init__()
        self.tokenizer = FeatureTokenizer(num_features=num_features, embed_dim=embed_dim)
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, dropout=dropout)
            for _ in range(num_blocks)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        self.head = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, 1)
        )

    def forward_features(self, x):
        tok = self.tokenizer(x)   # [B, F, D]
        for blk in self.blocks:
            tok = blk(tok)
        tok = self.norm(tok)
        pooled = tok.mean(dim=1)  # [B, D]
        return pooled

    def forward(self, x):
        pooled = self.forward_features(x)
        logits = self.head(pooled).squeeze(1)
        return logits

# ============================================================
# Module 10D.2 — Extract Leak-Safe FTAN Embeddings
# ============================================================

@torch.no_grad()
def extract_ftan_embeddings(model, loader, device):
    model.eval()
    embs = []
    for batch in loader:
        if isinstance(batch, (list, tuple)):
            Xb = batch[0]
        else:
            Xb = batch
        Xb = Xb.to(device)
        feat = model.forward_features(Xb).detach().cpu().numpy()
        embs.append(feat)
    return np.concatenate(embs, axis=0)


def build_dense_arrays_for_fold(X_tr_df, X_va_df, X_te_df):
    preproc = build_dense_preprocessor(
        numeric_cols=[c for c in X_tr_df.columns if pd.api.types.is_numeric_dtype(X_tr_df[c])],
        categorical_cols=[c for c in X_tr_df.columns if not pd.api.types.is_numeric_dtype(X_tr_df[c])],
        scaler="robust"
    )
    X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr_df, X_va_df, X_te_df)
    return X_tr_t, X_va_t, X_te_t


# OOF embeddings and TEST embeddings
ftan_oof_emb = None
ftan_test_emb_folds = []

for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
    print(f"Extracting FTAN embeddings | fold {fold_id}")

    X_tr_df = X_dev_fe.iloc[tr_idx].copy()
    X_va_df = X_dev_fe.iloc[va_idx].copy()
    X_te_df = X_test_fe.copy()

    X_tr_t, X_va_t, X_te_t = build_dense_arrays_for_fold(X_tr_df, X_va_df, X_te_df)
    input_dim = X_tr_t.shape[1]

    model = FTANClassifier(
        num_features=input_dim,
        embed_dim=CFG["attention_embed_dim"],
        num_heads=CFG["attention_heads"],
        num_blocks=CFG["attention_blocks"],
        dropout=CFG["attention_dropout"]
    )
    model_path = MODEL_DIR / f"ftan_ftan_focal_fold{fold_id}.pt"
    model.load_state_dict(torch.load(model_path, map_location=CFG["device"]))
    model.to(CFG["device"])

    va_loader = DataLoader(
        TabularDataset(X_va_t, None),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    te_loader = DataLoader(
        TabularDataset(X_te_t, None),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    va_emb = extract_ftan_embeddings(model, va_loader, CFG["device"])
    te_emb = extract_ftan_embeddings(model, te_loader, CFG["device"])

    if ftan_oof_emb is None:
        ftan_oof_emb = np.zeros((len(X_dev_fe), va_emb.shape[1]), dtype=np.float32)

    ftan_oof_emb[va_idx] = va_emb
    ftan_test_emb_folds.append(te_emb)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

ftan_test_emb = np.mean(np.stack(ftan_test_emb_folds, axis=0), axis=0)

print("OOF embedding shape:", ftan_oof_emb.shape)
print("TEST embedding shape:", ftan_test_emb.shape)

np.save(OOF_DIR / "ftan_oof_embeddings.npy", ftan_oof_emb)
np.save(PRED_DIR / "ftan_test_embeddings.npy", ftan_test_emb)

# ============================================================
# Module 10D.3 — Build Hard-Region Specialist Features
# ============================================================

hard_mask_dev = oof_df["hard_uncertainty"].values.astype(bool)
hard_mask_test = test_df["hard_uncertainty"].values.astype(bool)

# Base meta signals
spec_dev = pd.DataFrame({
    "cb_prob": oof_df["cb_heavy_pos"].values,
    "ftan_prob": oof_df["ftan_focal"].values,
    "prob_diff": oof_df["cb_heavy_pos"].values - oof_df["ftan_focal"].values,
    "prob_mean": (oof_df["cb_heavy_pos"].values + oof_df["ftan_focal"].values) / 2.0,
    "disagreement_std": oof_df["disagreement_std"].values,
    "disagreement_maxdiff": oof_df["disagreement_maxdiff"].values,
    "cb_conf": np.abs(oof_df["cb_heavy_pos"].values - thresholds["cb_heavy_pos"]),
    "ftan_conf": np.abs(oof_df["ftan_focal"].values - thresholds["ftan_focal"]),
})

spec_test = pd.DataFrame({
    "cb_prob": test_df["cb_heavy_pos"].values,
    "ftan_prob": test_df["ftan_focal"].values,
    "prob_diff": test_df["cb_heavy_pos"].values - test_df["ftan_focal"].values,
    "prob_mean": (test_df["cb_heavy_pos"].values + test_df["ftan_focal"].values) / 2.0,
    "disagreement_std": test_df["disagreement_std"].values,
    "disagreement_maxdiff": test_df["disagreement_maxdiff"].values,
    "cb_conf": np.abs(test_df["cb_heavy_pos"].values - thresholds["cb_heavy_pos"]),
    "ftan_conf": np.abs(test_df["ftan_focal"].values - thresholds["ftan_focal"]),
})

# Add FTAN embeddings
ftan_emb_dim = ftan_oof_emb.shape[1]

ftan_oof_emb_df = pd.DataFrame(
    ftan_oof_emb,
    columns=[f"ftan_emb_{i}" for i in range(ftan_emb_dim)]
)
ftan_test_emb_df = pd.DataFrame(
    ftan_test_emb,
    columns=[f"ftan_emb_{i}" for i in range(ftan_emb_dim)]
)

# Optional: add a compact raw-feature subset only
raw_keep = [
    c for c in [
        "Age", "BMI", "GenHlth", "PhysHlth", "MentHlth",
        "HighBP", "HighChol", "Diabetes", "Stroke",
        "DiffWalk", "PhysActivity", "Smoker",
        "RiskFactorCount", "LifestyleScore",
        "Age_x_Diabetes", "GenHlth_x_PhysHlth"
    ] if c in X_dev_fe.columns
]

raw_dev_df = X_dev_fe[raw_keep].reset_index(drop=True).copy()
raw_test_df = X_test_fe[raw_keep].reset_index(drop=True).copy()
raw_dev_df.columns = [f"raw__{c}" for c in raw_dev_df.columns]
raw_test_df.columns = [f"raw__{c}" for c in raw_test_df.columns]

specialist_dev_all = pd.concat(
    [spec_dev.reset_index(drop=True), ftan_oof_emb_df, raw_dev_df],
    axis=1
)
specialist_test_all = pd.concat(
    [spec_test.reset_index(drop=True), ftan_test_emb_df, raw_test_df],
    axis=1
)

X_spec_dev_hard = specialist_dev_all.loc[hard_mask_dev].reset_index(drop=True).copy()
y_spec_dev_hard = oof_df.loc[hard_mask_dev, "y_true"].reset_index(drop=True).astype(int).copy()

X_spec_test_hard = specialist_test_all.loc[hard_mask_test].reset_index(drop=True).copy()
y_spec_test_hard = test_df.loc[hard_mask_test, "y_true"].reset_index(drop=True).astype(int).copy()

print("Specialist DEV hard shape:", X_spec_dev_hard.shape, "pos_rate:", y_spec_dev_hard.mean())
print("Specialist TEST hard shape:", X_spec_test_hard.shape, "pos_rate:", y_spec_test_hard.mean())
display(X_spec_dev_hard.head())

# ============================================================
# Module 10D.4 — Train Hard-Region Specialist on Label
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold

def build_hard_specialist():
    assert HAS_CATBOOST, "CatBoost is required."
    return CatBoostClassifier(
        iterations=350,
        depth=5,
        learning_rate=0.03,
        loss_function="Logloss",
        eval_metric="AUC",
        auto_class_weights="Balanced",
        verbose=False,
        random_seed=CFG["seed"]
    )

spec_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=CFG["seed"]
)

spec_oof_hard = np.zeros(len(X_spec_dev_hard), dtype=float)
spec_test_hard_folds = []
spec_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(spec_cv.split(X_spec_dev_hard, y_spec_dev_hard)):
    X_tr = X_spec_dev_hard.iloc[tr_idx].copy()
    y_tr = y_spec_dev_hard.iloc[tr_idx].values
    X_va = X_spec_dev_hard.iloc[va_idx].copy()
    y_va = y_spec_dev_hard.iloc[va_idx].values

    model = build_hard_specialist()
    model.fit(X_tr, y_tr)

    va_prob = model.predict_proba(X_va)[:, 1]
    spec_oof_hard[va_idx] = va_prob

    if len(X_spec_test_hard) > 0:
        te_prob = model.predict_proba(X_spec_test_hard)[:, 1]
        spec_test_hard_folds.append(te_prob)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "hard_specialist_ftan_embedding"
    row["fold_id"] = fold_id
    spec_rows.append(row)

    print(
        f"Spec fold {fold_id:02d} | "
        f"PR_AUC={row['PR_AUC']:.5f} | "
        f"ROC_AUC={row['ROC_AUC']:.5f} | "
        f"F1={row['F1']:.5f}"
    )

spec_cv_df = pd.DataFrame(spec_rows)
spec_cv_df.to_csv(TABLE_DIR / "module10D_specialist_cv_metrics.csv", index=False)
display(spec_cv_df.describe())

if len(spec_test_hard_folds) > 0:
    spec_test_hard = np.mean(np.vstack(spec_test_hard_folds), axis=0)
else:
    spec_test_hard = np.array([])

# ============================================================
# Module 10D.5 — Hard-Region Comparison
# ============================================================

cb_hard_dev_prob = oof_df.loc[hard_mask_dev, "cb_heavy_pos"].reset_index(drop=True).values
ftan_hard_dev_prob = oof_df.loc[hard_mask_dev, "ftan_focal"].reset_index(drop=True).values

hard_region_compare_10D = pd.DataFrame([
    {"model": "cb_heavy_pos_hard", **evaluate_predictions(y_spec_dev_hard.values, cb_hard_dev_prob, threshold=thresholds["cb_heavy_pos"])},
    {"model": "ftan_focal_hard", **evaluate_predictions(y_spec_dev_hard.values, ftan_hard_dev_prob, threshold=thresholds["ftan_focal"])},
    {"model": "hard_specialist_ftan_embedding_hard", **evaluate_predictions(y_spec_dev_hard.values, spec_oof_hard, threshold=0.5)},
])

hard_region_compare_10D.to_csv(TABLE_DIR / "module10D_hard_region_compare.csv", index=False)
display(hard_region_compare_10D)

# ============================================================
# Module 10D.6 — Full Two-Stage Specialist Replacement
# ============================================================

final_dev_prob_spec = oof_df["cb_heavy_pos"].values.copy()
final_test_prob_spec = test_df["cb_heavy_pos"].values.copy()

final_dev_prob_spec[hard_mask_dev] = spec_oof_hard
if len(spec_test_hard) > 0:
    final_test_prob_spec[hard_mask_test] = spec_test_hard

spec_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_spec)

module10D_replacement_summary = pd.DataFrame([{
    "model": "two_stage_specialist_replacement",
    **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_spec, threshold=spec_thr, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, final_test_prob_spec, threshold=spec_thr, prefix="TEST_"),
}])

module10D_replacement_summary.to_csv(TABLE_DIR / "module10D_replacement_summary.csv", index=False)
display(module10D_replacement_summary)

# ============================================================
# Module 10D.7 — Blended Two-Stage Specialist
# ============================================================

blend_grid = [0.3, 0.5, 0.7, 0.8]

blend_rows = []

for alpha in blend_grid:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        alpha * spec_oof_hard +
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )

    if len(spec_test_hard) > 0:
        test_prob[hard_mask_test] = (
            alpha * spec_test_hard +
            (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
        )

    thr, _ = find_best_f1_threshold(oof_df["y_true"].values, dev_prob)

    row = {
        "model": f"two_stage_specialist_blend_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(oof_df["y_true"].values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, test_prob, threshold=thr, prefix="TEST_"),
    }
    blend_rows.append(row)

module10D_blend_summary = pd.DataFrame(blend_rows)
module10D_blend_summary.to_csv(TABLE_DIR / "module10D_blend_summary.csv", index=False)
display(module10D_blend_summary.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 10D.8 — Final Comparison vs Current Best
# ============================================================

rows = []

# Current best baseline
rows.append({
    "model": "cb_heavy_pos_full",
    **evaluate_predictions(
        oof_df["y_true"].values,
        oof_df["cb_heavy_pos"].values,
        threshold=thresholds["cb_heavy_pos"],
        prefix="OOF_"
    ),
    **evaluate_predictions(
        test_df["y_true"].values,
        test_df["cb_heavy_pos"].values,
        threshold=thresholds["cb_heavy_pos"],
        prefix="TEST_"
    ),
})

# Current best fixed FTAN blend if exists
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# Specialist replacement
rows.append(module10D_replacement_summary.iloc[0].to_dict())

# Best specialist blend
best_spec_blend = module10D_blend_summary.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False
).iloc[0].to_dict()
rows.append(best_spec_blend)

module10D_compare = pd.DataFrame(rows)
module10D_compare.to_csv(TABLE_DIR / "module10D_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module10D_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 10E.1 — FTAN with Feature Output
# ============================================================

class FTANClassifier(nn.Module):
    def __init__(self, num_features, embed_dim=64, num_heads=4, num_blocks=2, dropout=0.1):
        super().__init__()
        self.tokenizer = FeatureTokenizer(num_features=num_features, embed_dim=embed_dim)
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, dropout=dropout)
            for _ in range(num_blocks)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        self.head = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, 1)
        )

    def forward_features(self, x):
        tok = self.tokenizer(x)
        for blk in self.blocks:
            tok = blk(tok)
        tok = self.norm(tok)
        pooled = tok.mean(dim=1)
        return pooled

    def forward(self, x):
        feat = self.forward_features(x)
        logits = self.head(feat).squeeze(1)
        return logits

# ============================================================
# Module 10E.2 — Build Hard-Band Dense Arrays
# ============================================================

main_model = "cb_heavy_pos"
alt_model = "ftan_focal"
HARD_COL = "hard_uncertainty"

hard_mask_dev = oof_df[HARD_COL].values.astype(bool)
hard_mask_test = test_df[HARD_COL].values.astype(bool)

X_dev_full_fe = X_dev_fe.reset_index(drop=True).copy()
X_test_full_fe = X_test_fe.reset_index(drop=True).copy()

y_dev_full_arr = oof_df["y_true"].values.astype(np.float32)
y_test_full_arr = test_df["y_true"].values.astype(np.float32)

print("Hard dev n:", hard_mask_dev.sum())
print("Hard test n:", hard_mask_test.sum())

# ============================================================
# Module 10E.3 — Supervised Contrastive Loss
# ============================================================

class SupConLoss(nn.Module):
    """
    Supervised contrastive loss on normalized embeddings.
    """
    def __init__(self, temperature=0.10):
        super().__init__()
        self.temperature = temperature

    def forward(self, features, labels):
        """
        features: [B, D]
        labels  : [B]
        """
        device = features.device
        labels = labels.view(-1, 1)

        features = F.normalize(features, p=2, dim=1)
        logits = torch.matmul(features, features.T) / self.temperature

        # numerical stability
        logits_max, _ = torch.max(logits, dim=1, keepdim=True)
        logits = logits - logits_max.detach()

        mask = torch.eq(labels, labels.T).float().to(device)
        logits_mask = torch.ones_like(mask) - torch.eye(mask.shape[0], device=device)
        mask = mask * logits_mask

        exp_logits = torch.exp(logits) * logits_mask
        log_prob = logits - torch.log(exp_logits.sum(dim=1, keepdim=True) + 1e-12)

        pos_count = mask.sum(dim=1)
        mean_log_prob_pos = (mask * log_prob).sum(dim=1) / (pos_count + 1e-12)

        loss = -mean_log_prob_pos
        valid = pos_count > 0
        loss = loss[valid].mean() if valid.any() else torch.tensor(0.0, device=device)

        return loss

# ============================================================
# Module 10E.4 — Joint Fine-Tuning Trainer
# ============================================================

class TabularDatasetWithY(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(np.asarray(X), dtype=torch.float32)
        self.y = torch.tensor(np.asarray(y), dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


@torch.no_grad()
def predict_ftan_prob_and_emb(model, loader, device):
    model.eval()
    probs, embs = [], []
    for Xb, _ in loader:
        Xb = Xb.to(device)
        feat = model.forward_features(Xb)
        logits = model.head(feat).squeeze(1)
        prob = torch.sigmoid(logits).detach().cpu().numpy()
        probs.append(prob)
        embs.append(feat.detach().cpu().numpy())
    return np.concatenate(probs), np.concatenate(embs)


def fit_ftan_hardband_supcon(
    model,
    X_train, y_train,
    X_valid, y_valid,
    device,
    batch_size=256,
    max_epochs=15,
    lr=3e-4,
    weight_decay=1e-5,
    patience=4,
    lambda_cls=1.0,
    lambda_supcon=0.2,
    temperature=0.10
):
    train_ds = TabularDatasetWithY(X_train, y_train)
    valid_ds = TabularDatasetWithY(X_valid, y_valid)

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=CFG["num_workers"],
        pin_memory=True,
        drop_last=True   # helps contrastive batches
    )
    valid_loader = DataLoader(
        valid_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
    cls_criterion = nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
    )
    con_criterion = SupConLoss(temperature=temperature)

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        model.train()
        train_loss_sum = 0.0

        for Xb, yb in train_loader:
            Xb = Xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            feat = model.forward_features(Xb)
            logits = model.head(feat).squeeze(1)

            cls_loss = cls_criterion(logits, yb)
            con_loss = con_criterion(feat, yb.long())

            loss = lambda_cls * cls_loss + lambda_supcon * con_loss
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * len(Xb)

        train_loss = train_loss_sum / len(train_loader.dataset)

        # validation
        model.eval()
        val_probs = []
        val_loss_sum = 0.0
        with torch.no_grad():
            for Xb, yb in valid_loader:
                Xb = Xb.to(device)
                yb = yb.to(device)

                feat = model.forward_features(Xb)
                logits = model.head(feat).squeeze(1)

                cls_loss = cls_criterion(logits, yb)
                con_loss = con_criterion(feat, yb.long())
                loss = lambda_cls * cls_loss + lambda_supcon * con_loss

                val_loss_sum += loss.item() * len(Xb)
                val_probs.append(torch.sigmoid(logits).cpu().numpy())

        val_loss = val_loss_sum / len(valid_loader.dataset)
        val_probs = np.concatenate(val_probs)
        val_pr = average_precision_score(y_valid, val_probs)

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "valid_loss": val_loss,
            "valid_pr_auc": val_pr
        })

        print(
            f"Epoch {epoch+1:02d} | "
            f"train_loss={train_loss:.5f} | "
            f"valid_loss={val_loss:.5f} | "
            f"valid_pr_auc={val_pr:.5f}"
        )

        if val_pr > best_pr:
            best_pr = val_pr
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    history_df = pd.DataFrame(history)
    return model, history_df

# ============================================================
# Module 10E.5 — OOF Hard-Band FTAN Fine-Tuning
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold

# Build dense arrays for the full dev/test once using the same fold-safe process later per split
hard_ftan_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=CFG["seed"]
)

# Create full hard-band index map
hard_dev_indices = np.where(hard_mask_dev)[0]
hard_test_indices = np.where(hard_mask_test)[0]

X_hard_dev_df = X_dev_full_fe.iloc[hard_dev_indices].reset_index(drop=True).copy()
y_hard_dev = y_dev_full_arr[hard_dev_indices]

X_hard_test_df = X_test_full_fe.iloc[hard_test_indices].reset_index(drop=True).copy()
y_hard_test = y_test_full_arr[hard_test_indices]

ftan_hard_oof_prob = np.zeros(len(X_hard_dev_df), dtype=float)
ftan_hard_test_folds = []
ftan_hard_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(hard_ftan_cv.split(X_hard_dev_df, y_hard_dev)):
    print(f"\nHard-band FTAN fine-tune fold {fold_id}")

    X_tr_df = X_hard_dev_df.iloc[tr_idx].copy()
    X_va_df = X_hard_dev_df.iloc[va_idx].copy()
    y_tr = y_hard_dev[tr_idx].astype(np.float32)
    y_va = y_hard_dev[va_idx].astype(np.float32)

    # Build fold-safe dense arrays
    preproc = build_dense_preprocessor(
        numeric_cols=[c for c in X_tr_df.columns if pd.api.types.is_numeric_dtype(X_tr_df[c])],
        categorical_cols=[c for c in X_tr_df.columns if not pd.api.types.is_numeric_dtype(X_tr_df[c])],
        scaler="robust"
    )
    X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr_df, X_va_df, X_hard_test_df)
    input_dim = X_tr_t.shape[1]

    # initialize from pretrained global FTAN focal
    model = FTANClassifier(
        num_features=input_dim,
        embed_dim=CFG["attention_embed_dim"],
        num_heads=CFG["attention_heads"],
        num_blocks=CFG["attention_blocks"],
        dropout=CFG["attention_dropout"]
    )

    # Use a matching pretrained fold if available, otherwise cycle
    pretrain_fold_id = fold_id % len(fold_indices)
    pretrained_path = MODEL_DIR / f"ftan_ftan_focal_fold{pretrain_fold_id}.pt"
    model.load_state_dict(torch.load(pretrained_path, map_location=CFG["device"]))

    model, hist_df = fit_ftan_hardband_supcon(
        model=model,
        X_train=X_tr_t, y_train=y_tr,
        X_valid=X_va_t, y_valid=y_va,
        device=CFG["device"],
        batch_size=256,
        max_epochs=15,
        lr=3e-4,
        weight_decay=1e-5,
        patience=4,
        lambda_cls=1.0,
        lambda_supcon=0.2,
        temperature=0.10
    )

    va_loader = DataLoader(
        TabularDatasetWithY(X_va_t, y_va),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )
    te_loader = DataLoader(
        TabularDatasetWithY(X_te_t, np.zeros(len(X_te_t), dtype=np.float32)),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    va_prob, va_emb = predict_ftan_prob_and_emb(model, va_loader, CFG["device"])
    te_prob, te_emb = predict_ftan_prob_and_emb(model, te_loader, CFG["device"])

    ftan_hard_oof_prob[va_idx] = va_prob
    ftan_hard_test_folds.append(te_prob)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "ftan_hard_supcon"
    row["fold_id"] = fold_id
    ftan_hard_rows.append(row)

    hist_df.to_csv(LOG_DIR / f"module10E_ftan_hard_supcon_fold{fold_id}_history.csv", index=False)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

ftan_hard_cv_df = pd.DataFrame(ftan_hard_rows)
ftan_hard_cv_df.to_csv(TABLE_DIR / "module10E_ftan_hard_supcon_cv.csv", index=False)
display(ftan_hard_cv_df.describe())

ftan_hard_test_prob = np.mean(np.vstack(ftan_hard_test_folds), axis=0)

# ============================================================
# Module 10E.6 — Hard-Band Comparison
# ============================================================

cb_hard_dev_prob = oof_df.loc[hard_mask_dev, "cb_heavy_pos"].reset_index(drop=True).values
ftan_orig_hard_dev_prob = oof_df.loc[hard_mask_dev, "ftan_focal"].reset_index(drop=True).values

module10E_hard_compare = pd.DataFrame([
    {"model": "cb_heavy_pos_hard", **evaluate_predictions(y_hard_dev, cb_hard_dev_prob, threshold=thresholds["cb_heavy_pos"])},
    {"model": "ftan_focal_hard", **evaluate_predictions(y_hard_dev, ftan_orig_hard_dev_prob, threshold=thresholds["ftan_focal"])},
    {"model": "ftan_hard_supcon_hard", **evaluate_predictions(y_hard_dev, ftan_hard_oof_prob, threshold=0.5)},
])

module10E_hard_compare.to_csv(TABLE_DIR / "module10E_hard_compare.csv", index=False)
display(module10E_hard_compare)
```

```python
# ============================================================
# Module 10E.7 — Full Two-Stage Replacement
# ============================================================

two_stage_dev_prob_10E = oof_df["cb_heavy_pos"].values.copy()
two_stage_test_prob_10E = test_df["cb_heavy_pos"].values.copy()

two_stage_dev_prob_10E[hard_mask_dev] = ftan_hard_oof_prob
two_stage_test_prob_10E[hard_mask_test] = ftan_hard_test_prob

thr_10E, _ = find_best_f1_threshold(oof_df["y_true"].values, two_stage_dev_prob_10E)

module10E_replacement_summary = pd.DataFrame([{
    "model": "two_stage_ftan_hard_supcon_replacement",
    **evaluate_predictions(oof_df["y_true"].values, two_stage_dev_prob_10E, threshold=thr_10E, prefix="OOF_"),
    **evaluate_predictions(test_df["y_true"].values, two_stage_test_prob_10E, threshold=thr_10E, prefix="TEST_"),
}])

module10E_replacement_summary.to_csv(TABLE_DIR / "module10E_replacement_summary.csv", index=False)
display(module10E_replacement_summary)

# ============================================================
# Module 10E.8 — Full Two-Stage Blending
# ============================================================

alphas_10E = [0.3, 0.5, 0.7, 0.8]
rows_10E = []

for alpha in alphas_10E:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        alpha * ftan_hard_oof_prob +
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )

    test_prob[hard_mask_test] = (
        alpha * ftan_hard_test_prob +
        (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
    )

    thr, _ = find_best_f1_threshold(oof_df["y_true"].values, dev_prob)

    row = {
        "model": f"two_stage_ftan_hard_supcon_blend_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(oof_df["y_true"].values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, test_prob, threshold=thr, prefix="TEST_"),
    }
    rows_10E.append(row)

module10E_blend_summary = pd.DataFrame(rows_10E)
module10E_blend_summary.to_csv(TABLE_DIR / "module10E_blend_summary.csv", index=False)
display(module10E_blend_summary.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 10E.9 — Final Comparison
# ============================================================

rows = []

# Baseline
rows.append({
    "model": "cb_heavy_pos_full",
    **evaluate_predictions(
        oof_df["y_true"].values,
        oof_df["cb_heavy_pos"].values,
        threshold=thresholds["cb_heavy_pos"],
        prefix="OOF_"
    ),
    **evaluate_predictions(
        test_df["y_true"].values,
        test_df["cb_heavy_pos"].values,
        threshold=thresholds["cb_heavy_pos"],
        prefix="TEST_"
    ),
})

# Current best fixed blend
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(oof_df["y_true"].values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(oof_df["y_true"].values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(test_df["y_true"].values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# Hard-band FTAN replacement
rows.append(module10E_replacement_summary.iloc[0].to_dict())

# Best hard-band FTAN blend
best_10E = module10E_blend_summary.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False
).iloc[0].to_dict()
rows.append(best_10E)

module10E_compare = pd.DataFrame(rows)
module10E_compare.to_csv(TABLE_DIR / "module10E_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module10E_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 10F.1 — Prepare Best Model Probabilities
# ============================================================

# Defensive checks
assert "hard_uncertainty" in oof_df.columns, "hard_uncertainty missing from oof_df"
assert "hard_uncertainty" in test_df.columns, "hard_uncertainty missing from test_df"

hard_mask_dev = oof_df["hard_uncertainty"].values.astype(bool)
hard_mask_test = test_df["hard_uncertainty"].values.astype(bool)

# Best current final probabilities: fixed blend alpha=0.7
if "final_dev_prob_blend" not in globals() or "final_test_prob_blend" not in globals():
    # rebuild if missing
    final_dev_prob_blend = oof_df["cb_heavy_pos"].values.copy()
    final_test_prob_blend = test_df["cb_heavy_pos"].values.copy()

    final_dev_prob_blend[hard_mask_dev] = (
        0.7 * oof_df.loc[hard_mask_dev, "ftan_focal"].values +
        0.3 * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )

    final_test_prob_blend[hard_mask_test] = (
        0.7 * test_df.loc[hard_mask_test, "ftan_focal"].values +
        0.3 * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
    )

y_dev_true = oof_df["y_true"].values.astype(int)
y_test_true = test_df["y_true"].values.astype(int)

print("Hard DEV pct:", hard_mask_dev.mean())
print("Hard TEST pct:", hard_mask_test.mean())
print("DEV probs ready:", final_dev_prob_blend.shape)
print("TEST probs ready:", final_test_prob_blend.shape)

# ============================================================
# Module 10F.2 — Region-Aware Thresholding Helper
# ============================================================

def predict_with_region_thresholds(prob, hard_mask, thr_easy, thr_hard):
    prob = np.asarray(prob)
    hard_mask = np.asarray(hard_mask).astype(bool)

    pred = np.zeros(len(prob), dtype=int)
    pred[~hard_mask] = (prob[~hard_mask] >= thr_easy).astype(int)
    pred[hard_mask] = (prob[hard_mask] >= thr_hard).astype(int)
    return pred


def evaluate_region_thresholds(y_true, prob, hard_mask, thr_easy, thr_hard):
    y_true = np.asarray(y_true).astype(int)
    prob = np.asarray(prob).astype(float)
    pred = predict_with_region_thresholds(prob, hard_mask, thr_easy, thr_hard)

    out = {
        "Precision": precision_score(y_true, pred, zero_division=0),
        "Recall": recall_score(y_true, pred, zero_division=0),
        "F1": f1_score(y_true, pred, zero_division=0),
        "Accuracy": accuracy_score(y_true, pred),
        "MCC": matthews_corrcoef(y_true, pred),
        "PosRatePred": pred.mean(),
        "thr_easy": thr_easy,
        "thr_hard": thr_hard
    }
    return out

# ============================================================
# Module 10F.3 — Grid Search for Region-Specific Thresholds
# ============================================================

thr_easy_grid = np.round(np.arange(0.35, 0.66, 0.02), 2)
thr_hard_grid = np.round(np.arange(0.25, 0.66, 0.02), 2)

rows = []

for thr_easy in thr_easy_grid:
    for thr_hard in thr_hard_grid:
        metrics = evaluate_region_thresholds(
            y_true=y_dev_true,
            prob=final_dev_prob_blend,
            hard_mask=hard_mask_dev,
            thr_easy=thr_easy,
            thr_hard=thr_hard
        )
        rows.append(metrics)

threshold_grid_df = pd.DataFrame(rows)

# rank primarily by F1, then MCC, then Precision
threshold_grid_df = threshold_grid_df.sort_values(
    ["F1", "MCC", "Precision", "Recall"],
    ascending=False
).reset_index(drop=True)

threshold_grid_df.to_csv(TABLE_DIR / "module10F_region_threshold_grid.csv", index=False)
display(threshold_grid_df.head(20))

best_thr_easy = float(threshold_grid_df.iloc[0]["thr_easy"])
best_thr_hard = float(threshold_grid_df.iloc[0]["thr_hard"])

print("Best thresholds found on DEV:")
print("thr_easy =", best_thr_easy)
print("thr_hard =", best_thr_hard)

# ============================================================
# Module 10F.4 — Evaluate Best Region-Specific Thresholds
# ============================================================

# Baseline single-threshold model for comparison
baseline_single_thr, _ = find_best_f1_threshold(y_dev_true, final_dev_prob_blend)

baseline_dev = evaluate_predictions(
    y_true=y_dev_true,
    y_prob=final_dev_prob_blend,
    threshold=baseline_single_thr,
    prefix="OOF_"
)
baseline_test = evaluate_predictions(
    y_true=y_test_true,
    y_prob=final_test_prob_blend,
    threshold=baseline_single_thr,
    prefix="TEST_"
)

# Region-specific threshold results
region_dev = evaluate_region_thresholds(
    y_true=y_dev_true,
    prob=final_dev_prob_blend,
    hard_mask=hard_mask_dev,
    thr_easy=best_thr_easy,
    thr_hard=best_thr_hard
)

region_test = evaluate_region_thresholds(
    y_true=y_test_true,
    prob=final_test_prob_blend,
    hard_mask=hard_mask_test,
    thr_easy=best_thr_easy,
    thr_hard=best_thr_hard
)

module10F_summary = pd.DataFrame([
    {
        "model": "fixed_blend_single_threshold",
        **baseline_dev,
        **baseline_test
    },
    {
        "model": "fixed_blend_region_specific_thresholds",
        "OOF_Precision": region_dev["Precision"],
        "OOF_Recall": region_dev["Recall"],
        "OOF_F1": region_dev["F1"],
        "OOF_Accuracy": region_dev["Accuracy"],
        "OOF_MCC": region_dev["MCC"],
        "OOF_PosRatePred": region_dev["PosRatePred"],
        "OOF_thr_easy": region_dev["thr_easy"],
        "OOF_thr_hard": region_dev["thr_hard"],

        "TEST_Precision": region_test["Precision"],
        "TEST_Recall": region_test["Recall"],
        "TEST_F1": region_test["F1"],
        "TEST_Accuracy": region_test["Accuracy"],
        "TEST_MCC": region_test["MCC"],
        "TEST_PosRatePred": region_test["PosRatePred"],
        "TEST_thr_easy": region_test["thr_easy"],
        "TEST_thr_hard": region_test["thr_hard"]
    }
])

module10F_summary.to_csv(TABLE_DIR / "module10F_summary.csv", index=False)
display(module10F_summary)
```

```python
# ============================================================
# Module 10F.5 — Optional Constrained Threshold Selection
# ============================================================

# Example 1: Best MCC
best_mcc_row = threshold_grid_df.sort_values(["MCC", "F1", "Precision"], ascending=False).iloc[0]

# Example 2: Best F1 among rows with Recall >= 0.55
recall_floor = 0.55
eligible = threshold_grid_df[threshold_grid_df["Recall"] >= recall_floor].copy()
if len(eligible) > 0:
    best_recall_constrained = eligible.sort_values(["F1", "MCC", "Precision"], ascending=False).iloc[0]
else:
    best_recall_constrained = None

optional_rows = []

optional_rows.append({
    "strategy": "best_mcc",
    "thr_easy": float(best_mcc_row["thr_easy"]),
    "thr_hard": float(best_mcc_row["thr_hard"]),
    "Precision": float(best_mcc_row["Precision"]),
    "Recall": float(best_mcc_row["Recall"]),
    "F1": float(best_mcc_row["F1"]),
    "MCC": float(best_mcc_row["MCC"]),
})

if best_recall_constrained is not None:
    optional_rows.append({
        "strategy": f"best_f1_recall_ge_{recall_floor}",
        "thr_easy": float(best_recall_constrained["thr_easy"]),
        "thr_hard": float(best_recall_constrained["thr_hard"]),
        "Precision": float(best_recall_constrained["Precision"]),
        "Recall": float(best_recall_constrained["Recall"]),
        "F1": float(best_recall_constrained["F1"]),
        "MCC": float(best_recall_constrained["MCC"]),
    })

optional_thresholds_df = pd.DataFrame(optional_rows)
optional_thresholds_df.to_csv(TABLE_DIR / "module10F_optional_threshold_strategies.csv", index=False)
display(optional_thresholds_df)

# ============================================================
# Module 10F.6 — Per-Region Diagnostics
# ============================================================

def evaluate_subset_threshold(y_true, prob, threshold):
    pred = (np.asarray(prob) >= threshold).astype(int)
    return {
        "n": len(y_true),
        "pos_rate_true": float(np.mean(y_true)),
        "pos_rate_pred": float(np.mean(pred)),
        "precision": precision_score(y_true, pred, zero_division=0),
        "recall": recall_score(y_true, pred, zero_division=0),
        "f1": f1_score(y_true, pred, zero_division=0),
        "mcc": matthews_corrcoef(y_true, pred),
        "accuracy": accuracy_score(y_true, pred),
    }

# DEV diagnostics
easy_dev = ~hard_mask_dev
hard_dev = hard_mask_dev

dev_diag = pd.DataFrame([
    {
        "region": "easy_dev",
        "threshold": best_thr_easy,
        **evaluate_subset_threshold(y_dev_true[easy_dev], final_dev_prob_blend[easy_dev], best_thr_easy)
    },
    {
        "region": "hard_dev",
        "threshold": best_thr_hard,
        **evaluate_subset_threshold(y_dev_true[hard_dev], final_dev_prob_blend[hard_dev], best_thr_hard)
    },
    {
        "region": "easy_test",
        "threshold": best_thr_easy,
        **evaluate_subset_threshold(y_test_true[~hard_mask_test], final_test_prob_blend[~hard_mask_test], best_thr_easy)
    },
    {
        "region": "hard_test",
        "threshold": best_thr_hard,
        **evaluate_subset_threshold(y_test_true[hard_mask_test], final_test_prob_blend[hard_mask_test], best_thr_hard)
    }
])

dev_diag.to_csv(TABLE_DIR / "module10F_region_diagnostics.csv", index=False)
display(dev_diag)

# ============================================================
# Module 10F.7 — Threshold Heatmap
# ============================================================

pivot_f1 = threshold_grid_df.pivot(index="thr_hard", columns="thr_easy", values="F1")
pivot_mcc = threshold_grid_df.pivot(index="thr_hard", columns="thr_easy", values="MCC")

plt.figure(figsize=(8, 6))
import seaborn as sns
sns.heatmap(pivot_f1, cmap="viridis")
plt.title("OOF F1 Heatmap: Region-Specific Thresholds")
plt.xlabel("thr_easy")
plt.ylabel("thr_hard")
plt.tight_layout()
plt.savefig(FIG_DIR / "module10F_f1_heatmap.png", dpi=300)
plt.show()

plt.figure(figsize=(8, 6))
sns.heatmap(pivot_mcc, cmap="magma")
plt.title("OOF MCC Heatmap: Region-Specific Thresholds")
plt.xlabel("thr_easy")
plt.ylabel("thr_hard")
plt.tight_layout()
plt.savefig(FIG_DIR / "module10F_mcc_heatmap.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Module 11.1 — Agreement vs Correctness
# ============================================================

models = ["cb_heavy_pos", "ftan_focal", "mlp_focal"]

hard_df = oof_df.loc[hard_mask_dev].copy().reset_index(drop=True)

# binary predictions using optimal thresholds
for m in models:
    hard_df[f"{m}_pred"] = (hard_df[m].values >= thresholds[m]).astype(int)

# agreement score
hard_df["agreement"] = hard_df[[f"{m}_pred" for m in models]].sum(axis=1)

# correctness
hard_df["correct"] = (hard_df["cb_heavy_pos_pred"] == hard_df["y_true"]).astype(int)

# all models agree (either all 0 or all 1)
hard_df["all_agree"] = (
    (hard_df["agreement"] == 0) |
    (hard_df["agreement"] == len(models))
)

# suspicious = all agree but wrong
hard_df["suspicious"] = ((hard_df["all_agree"] == 1) & (hard_df["correct"] == 0)).astype(int)

summary_11_1 = hard_df[["all_agree", "correct", "suspicious"]].mean()

print("Agreement summary:")
display(summary_11_1)

print("Suspicious samples (potential label noise):", hard_df["suspicious"].sum())

# ============================================================
# Module 11.2 — Confidence vs Correctness
# ============================================================

hard_df["cb_conf"] = np.abs(hard_df["cb_heavy_pos"] - 0.5)
hard_df["ftan_conf"] = np.abs(hard_df["ftan_focal"] - 0.5)

hard_df["mean_conf"] = (hard_df["cb_conf"] + hard_df["ftan_conf"]) / 2

# bin confidence
hard_df["conf_bin"] = pd.qcut(hard_df["mean_conf"], q=10, duplicates="drop")

conf_analysis = hard_df.groupby("conf_bin").agg({
    "correct": "mean",
    "mean_conf": "mean"
}).reset_index()

conf_analysis.rename(columns={"correct": "accuracy"}, inplace=True)

display(conf_analysis)

# Save
conf_analysis.to_csv(TABLE_DIR / "module11_confidence_analysis.csv", index=False)

# ============================================================
# Module 11.3 — Feature Separability
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score

def evaluate_region_separability(X, y):
    model = LogisticRegression(max_iter=1000)
    model.fit(X, y)
    prob = model.predict_proba(X)[:,1]
    return {
        "ROC_AUC": roc_auc_score(y, prob),
        "PR_AUC": average_precision_score(y, prob)
    }

# EASY
easy_mask = ~hard_mask_dev
X_easy = X_dev_fe.loc[easy_mask]
y_easy = y_dev_true[easy_mask]

# HARD
X_hard = X_dev_fe.loc[hard_mask_dev]
y_hard = y_dev_true[hard_mask_dev]

sep_easy = evaluate_region_separability(X_easy, y_easy)
sep_hard = evaluate_region_separability(X_hard, y_hard)

sep_df = pd.DataFrame([
    {"region": "easy", **sep_easy},
    {"region": "hard", **sep_hard}
])

display(sep_df)
sep_df.to_csv(TABLE_DIR / "module11_separability.csv", index=False)

# ============================================================
# Module 11.4 — Hardest Samples
# ============================================================

hard_df["error"] = (hard_df["cb_heavy_pos_pred"] != hard_df["y_true"]).astype(int)

# define hardest:
hard_df["difficulty_score"] = (
    hard_df["disagreement_std"] +
    hard_df["disagreement_maxdiff"]
)

top_hard = hard_df.sort_values("difficulty_score", ascending=False).head(50)

display(top_hard.head())

# Save for manual inspection
top_hard.to_csv(TABLE_DIR / "module11_top_hard_samples.csv", index=False)

# ============================================================
# Module 11.5 — Noise Score
# ============================================================

noise_mask = (
    (hard_df["mean_conf"] > hard_df["mean_conf"].quantile(0.75)) &
    (hard_df["error"] == 1) &
    (hard_df["all_agree"] == 1)
)

noise_ratio = noise_mask.mean()

print("Estimated noise ratio in hard region:", noise_ratio)

noise_df = hard_df[noise_mask].copy()
noise_df.to_csv(TABLE_DIR / "module11_suspected_noise.csv", index=False)
```

```python
# ============================================================
# Module 12A — Build Suspiciousness Scores
# ============================================================

audit_df = oof_df.loc[hard_mask_dev].copy().reset_index(drop=True)

# We assume Module 11 variables already exist; if not, rebuild them
models = ["cb_heavy_pos", "ftan_focal", "mlp_focal"]

for m in models:
    if f"{m}_pred" not in audit_df.columns:
        audit_df[f"{m}_pred"] = (audit_df[m].values >= thresholds[m]).astype(int)

if "agreement" not in audit_df.columns:
    audit_df["agreement"] = audit_df[[f"{m}_pred" for m in models]].sum(axis=1)

if "all_agree" not in audit_df.columns:
    audit_df["all_agree"] = (
        (audit_df["agreement"] == 0) |
        (audit_df["agreement"] == len(models))
    )

if "cb_conf" not in audit_df.columns:
    audit_df["cb_conf"] = np.abs(audit_df["cb_heavy_pos"].values - thresholds["cb_heavy_pos"])

if "ftan_conf" not in audit_df.columns:
    audit_df["ftan_conf"] = np.abs(audit_df["ftan_focal"].values - thresholds["ftan_focal"])

audit_df["mean_conf"] = (audit_df["cb_conf"] + audit_df["ftan_conf"]) / 2.0

# CatBoost correctness
audit_df["cb_error"] = (
    (audit_df["cb_heavy_pos"].values >= thresholds["cb_heavy_pos"]).astype(int) != audit_df["y_true"].values
).astype(int)

# Strong suspiciousness: all agree + cb wrong + high confidence
conf_q75 = audit_df["mean_conf"].quantile(0.75)
audit_df["suspicious_strong"] = (
    (audit_df["all_agree"] == 1) &
    (audit_df["cb_error"] == 1) &
    (audit_df["mean_conf"] >= conf_q75)
).astype(int)

# Medium suspiciousness: all agree + wrong
audit_df["suspicious_medium"] = (
    (audit_df["all_agree"] == 1) &
    (audit_df["cb_error"] == 1)
).astype(int)

# Soft suspiciousness score
audit_df["suspicious_score"] = (
    1.0 * audit_df["all_agree"].astype(float) +
    1.0 * audit_df["cb_error"].astype(float) +
    2.0 * (audit_df["mean_conf"] >= conf_q75).astype(float)
)

print("Hard-region samples:", len(audit_df))
print("Strong suspicious count:", int(audit_df["suspicious_strong"].sum()))
print("Medium suspicious count:", int(audit_df["suspicious_medium"].sum()))

audit_df.to_csv(TABLE_DIR / "module12_audit_df.csv", index=False)
display(audit_df[[
    "y_true", "cb_heavy_pos", "ftan_focal", "all_agree",
    "cb_error", "mean_conf", "suspicious_medium",
    "suspicious_strong", "suspicious_score"
]].head())

# ============================================================
# Module 12B — Map Suspicious Hard Samples to Full DEV Indices
# ============================================================

hard_dev_indices = np.where(hard_mask_dev)[0]

# Full DEV masks
remove_mask_strong_full = np.zeros(len(X_dev_fe), dtype=bool)
remove_mask_medium_full = np.zeros(len(X_dev_fe), dtype=bool)

remove_mask_strong_full[hard_dev_indices] = audit_df["suspicious_strong"].values.astype(bool)
remove_mask_medium_full[hard_dev_indices] = audit_df["suspicious_medium"].values.astype(bool)

# Sample weights for downweighting
sample_weight_soft = np.ones(len(X_dev_fe), dtype=float)
sample_weight_soft[hard_dev_indices] = np.where(
    audit_df["suspicious_medium"].values.astype(bool),
    0.50,   # suspicious samples get half-weight
    1.00
)

sample_weight_strong = np.ones(len(X_dev_fe), dtype=float)
sample_weight_strong[hard_dev_indices] = np.where(
    audit_df["suspicious_strong"].values.astype(bool),
    0.25,   # strong suspicious samples get quarter-weight
    1.00
)

summary_cleaning = pd.DataFrame([
    {
        "strategy": "remove_strong",
        "n_removed": int(remove_mask_strong_full.sum()),
        "pct_removed": float(remove_mask_strong_full.mean())
    },
    {
        "strategy": "remove_medium",
        "n_removed": int(remove_mask_medium_full.sum()),
        "pct_removed": float(remove_mask_medium_full.mean())
    },
    {
        "strategy": "downweight_soft",
        "n_downweighted": int(np.sum(sample_weight_soft < 1.0)),
        "avg_weight": float(sample_weight_soft.mean())
    },
    {
        "strategy": "downweight_strong",
        "n_downweighted": int(np.sum(sample_weight_strong < 1.0)),
        "avg_weight": float(sample_weight_strong.mean())
    }
])

summary_cleaning.to_csv(TABLE_DIR / "module12_cleaning_summary.csv", index=False)
display(summary_cleaning)

# ============================================================
# Module 12C.1 — CatBoost Runner with Filtering / Weights
# ============================================================

def run_cv_catboost_cleaning(
    X_dev_df,
    y_dev_series,
    X_test_df,
    strategy_name,
    remove_mask_full=None,
    sample_weight_full=None
):
    print(f"\n{'='*70}")
    print(f"Running CatBoost cleaning strategy: {strategy_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        y_tr = y_dev_series.iloc[tr_idx].values
        X_va = X_dev_df.iloc[va_idx].copy()
        y_va = y_dev_series.iloc[va_idx].values

        train_mask_local = np.ones(len(tr_idx), dtype=bool)
        weights_local = None

        if remove_mask_full is not None:
            train_mask_local = ~remove_mask_full[tr_idx]

        if sample_weight_full is not None:
            weights_local = sample_weight_full[tr_idx]
            if remove_mask_full is not None:
                weights_local = weights_local[train_mask_local]

        X_tr = X_tr.iloc[train_mask_local].copy()
        y_tr = y_tr[train_mask_local]

        preproc = build_tree_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])]
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)

        model = build_catboost_custom(class_weights=[1.0, 3.0])

        if weights_local is not None:
            model.fit(X_tr_t, y_tr, sample_weight=weights_local)
        else:
            model.fit(X_tr_t, y_tr)

        va_prob = model.predict_proba(X_va_t)[:, 1]
        te_prob = model.predict_proba(X_te_t)[:, 1]

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        row = evaluate_predictions(y_va, va_prob, threshold=0.5)
        row["model"] = "catboost"
        row["strategy"] = strategy_name
        row["fold_id"] = fold_id
        fold_rows.append(row)

        print(
            f"Fold {fold_id:02d} | "
            f"train_n={len(y_tr)} | "
            f"PR_AUC={row['PR_AUC']:.5f} | "
            f"ROC_AUC={row['ROC_AUC']:.5f}"
        )

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    best_thr, _ = find_best_f1_threshold(y_dev_series.values, oof_pred)

    summary_df = pd.DataFrame([{
        "model": "catboost",
        "strategy": strategy_name,
        **evaluate_predictions(y_dev_series.values, oof_pred, threshold=best_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_"),
    }])

    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 12C.2 — FTAN Runner with Filtering
# ============================================================

def run_cv_ftan_cleaning(
    X_dev_df,
    y_dev_series,
    X_test_df,
    strategy_name,
    remove_mask_full=None
):
    print(f"\n{'='*70}")
    print(f"Running FTAN cleaning strategy: {strategy_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        y_tr = y_dev_series.iloc[tr_idx].values.astype(np.float32)

        X_va = X_dev_df.iloc[va_idx].copy()
        y_va = y_dev_series.iloc[va_idx].values.astype(np.float32)

        train_mask_local = np.ones(len(tr_idx), dtype=bool)
        if remove_mask_full is not None:
            train_mask_local = ~remove_mask_full[tr_idx]

        X_tr = X_tr.iloc[train_mask_local].copy()
        y_tr = y_tr[train_mask_local]

        preproc = build_dense_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])],
            scaler="robust"
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)
        input_dim = X_tr_t.shape[1]

        model = FTANClassifier(
            num_features=input_dim,
            embed_dim=CFG["attention_embed_dim"],
            num_heads=CFG["attention_heads"],
            num_blocks=CFG["attention_blocks"],
            dropout=CFG["attention_dropout"]
        )

        model, hist_df = fit_torch_model_v2(
            model=model,
            X_train=X_tr_t,
            y_train=y_tr,
            X_valid=X_va_t,
            y_valid=y_va,
            device=CFG["device"],
            batch_size=CFG["batch_size"],
            max_epochs=CFG["max_epochs"],
            lr=CFG["lr"],
            weight_decay=CFG["weight_decay"],
            patience=CFG["patience"],
            loss_name="focal",
            focal_alpha=0.25,
            focal_gamma=2.0
        )

        valid_loader = DataLoader(
            TabularDataset(X_va_t, y_va),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )
        test_loader = DataLoader(
            TabularDataset(X_te_t, None),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        va_prob = predict_proba_torch(model, valid_loader, CFG["device"])
        te_prob = predict_proba_torch(model, test_loader, CFG["device"])

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        row = evaluate_predictions(y_va, va_prob, threshold=0.5)
        row["model"] = "ftan"
        row["strategy"] = strategy_name
        row["fold_id"] = fold_id
        fold_rows.append(row)

        print(
            f"Fold {fold_id:02d} | "
            f"train_n={len(y_tr)} | "
            f"PR_AUC={row['PR_AUC']:.5f} | "
            f"ROC_AUC={row['ROC_AUC']:.5f}"
        )

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    cv_df = pd.DataFrame(fold_rows)
    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)

    best_thr, _ = find_best_f1_threshold(y_dev_series.values, oof_pred)

    summary_df = pd.DataFrame([{
        "model": "ftan",
        "strategy": strategy_name,
        **evaluate_predictions(y_dev_series.values, oof_pred, threshold=best_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_"),
    }])

    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 12D — Execute Cleaning Experiments
# ============================================================

X_dev_run = X_dev_fe.copy()
X_test_run = X_test_fe.copy()
y_dev_run = y_dev.copy()

strategies = [
    {
        "name": "baseline",
        "remove_mask": None,
        "sample_weight": None
    },
    {
        "name": "remove_strong_suspicious",
        "remove_mask": remove_mask_strong_full,
        "sample_weight": None
    },
    {
        "name": "remove_medium_suspicious",
        "remove_mask": remove_mask_medium_full,
        "sample_weight": None
    },
    {
        "name": "downweight_strong_suspicious",
        "remove_mask": None,
        "sample_weight": sample_weight_strong
    }
]

all_cv_rows_12 = []
all_summary_rows_12 = []

# CatBoost
for s in strategies:
    cv_df, summary_df, oof_pred, test_pred = run_cv_catboost_cleaning(
        X_dev_df=X_dev_run,
        y_dev_series=y_dev_run,
        X_test_df=X_test_run,
        strategy_name=s["name"],
        remove_mask_full=s["remove_mask"],
        sample_weight_full=s["sample_weight"]
    )
    all_cv_rows_12.append(cv_df)
    all_summary_rows_12.append(summary_df)

# FTAN (filtering only for now)
ftan_strategies = [
    {
        "name": "baseline",
        "remove_mask": None
    },
    {
        "name": "remove_strong_suspicious",
        "remove_mask": remove_mask_strong_full
    },
    {
        "name": "remove_medium_suspicious",
        "remove_mask": remove_mask_medium_full
    }
]

for s in ftan_strategies:
    cv_df, summary_df, oof_pred, test_pred = run_cv_ftan_cleaning(
        X_dev_df=X_dev_run,
        y_dev_series=y_dev_run,
        X_test_df=X_test_run,
        strategy_name=s["name"],
        remove_mask_full=s["remove_mask"]
    )
    all_cv_rows_12.append(cv_df)
    all_summary_rows_12.append(summary_df)

module12_cv = pd.concat(all_cv_rows_12, axis=0, ignore_index=True)
module12_summary = pd.concat(all_summary_rows_12, axis=0, ignore_index=True)

module12_cv.to_csv(TABLE_DIR / "module12_cv_results.csv", index=False)
module12_summary.to_csv(TABLE_DIR / "module12_summary_results.csv", index=False)

display(module12_summary.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 12E — Compact Comparison
# ============================================================

keep_cols = [
    "model", "strategy",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

module12_compact = module12_summary[keep_cols].copy()
module12_compact = module12_compact.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"],
    ascending=False
)

module12_compact.to_csv(TABLE_DIR / "module12_compact.csv", index=False)
display(module12_compact)
```

```python
# ============================================================
# Module 13A — Controlled Feature Interaction Builder
# ============================================================

def add_controlled_interactions(df_in, groups=("A", "B", "C", "D")):
    df = df_in.copy()
    cols = set(df.columns)

    def has(*names):
        return all(n in cols for n in names)

    # ----------------------------
    # Group A — metabolic / vascular
    # ----------------------------
    if "A" in groups:
        if has("Age", "BMI"):
            df["Age_x_BMI"] = df["Age"] * df["BMI"]
        if has("Age", "Diabetes"):
            df["Age_x_Diabetes"] = df["Age"] * df["Diabetes"]
        if has("Age", "HighBP"):
            df["Age_x_HighBP"] = df["Age"] * df["HighBP"]
        if has("Age", "HighChol"):
            df["Age_x_HighChol"] = df["Age"] * df["HighChol"]
        if has("BMI", "Diabetes"):
            df["BMI_x_Diabetes"] = df["BMI"] * df["Diabetes"]
        if has("HighBP", "HighChol"):
            df["HighBP_x_HighChol"] = df["HighBP"] * df["HighChol"]

    # ----------------------------
    # Group B — functional health burden
    # ----------------------------
    if "B" in groups:
        if has("GenHlth", "PhysHlth"):
            df["GenHlth_x_PhysHlth"] = df["GenHlth"] * df["PhysHlth"]
        if has("GenHlth", "MentHlth"):
            df["GenHlth_x_MentHlth"] = df["GenHlth"] * df["MentHlth"]
        if has("PhysHlth", "DiffWalk"):
            df["PhysHlth_x_DiffWalk"] = df["PhysHlth"] * df["DiffWalk"]
        if has("Age", "DiffWalk"):
            df["Age_x_DiffWalk"] = df["Age"] * df["DiffWalk"]
        if has("PhysActivity", "BMI"):
            df["PhysActivity_x_BMI"] = df["PhysActivity"] * df["BMI"]

    # ----------------------------
    # Group C — lifestyle risk
    # ----------------------------
    if "C" in groups:
        if has("Age", "Smoker"):
            df["Age_x_Smoker"] = df["Age"] * df["Smoker"]
        if has("Smoker", "HighBP"):
            df["Smoker_x_HighBP"] = df["Smoker"] * df["HighBP"]
        if has("Smoker", "HighChol"):
            df["Smoker_x_HighChol"] = df["Smoker"] * df["HighChol"]
        if has("Smoker", "PhysActivity"):
            df["Smoker_x_PhysActivity"] = df["Smoker"] * df["PhysActivity"]
        if has("SleepTime", "MentHlth"):
            df["SleepTime_x_MentHlth"] = df["SleepTime"] * df["MentHlth"]

    # ----------------------------
    # Group D — composite burden scores
    # ----------------------------
    if "D" in groups:
        metabolic_cols = [c for c in ["HighBP", "HighChol", "Diabetes", "BMI_ge_30"] if c in df.columns]
        if len(metabolic_cols) > 0:
            df["MetabolicScore"] = df[metabolic_cols].sum(axis=1)

        functional_cols = [c for c in ["DiffWalk", "PhysHlth", "MentHlth"] if c in df.columns]
        if len(functional_cols) > 0:
            # scaled count-like burden
            arr = np.zeros(len(df))
            if "DiffWalk" in df.columns:
                arr = arr + df["DiffWalk"].astype(float)
            if "PhysHlth" in df.columns:
                arr = arr + (df["PhysHlth"] > 0).astype(float)
            if "MentHlth" in df.columns:
                arr = arr + (df["MentHlth"] > 0).astype(float)
            df["FunctionalBurdenScore"] = arr

        vascular_cols = [c for c in ["HighBP", "HighChol", "Stroke"] if c in df.columns]
        if len(vascular_cols) > 0:
            df["VascularScore"] = df[vascular_cols].sum(axis=1)

    return df

# ============================================================
# Module 13B — Define Interaction Ablation Sets
# ============================================================

interaction_sets = {
    "baseline": tuple(),
    "group_A": ("A",),
    "group_B": ("B",),
    "group_C": ("C",),
    "group_D": ("D",),
    "all_groups": ("A", "B", "C", "D"),
}

feature_versions = {}

for name, groups in interaction_sets.items():
    if len(groups) == 0:
        X_dev_ver = X_dev.copy()
        X_test_ver = X_test.copy()

        # keep your existing light engineered baseline if desired
        # or use original X_dev/X_test only
        X_dev_ver = add_feature_engineering(X_dev_ver)
        X_test_ver = add_feature_engineering(X_test_ver)
    else:
        X_dev_ver = add_feature_engineering(X_dev.copy())
        X_test_ver = add_feature_engineering(X_test.copy())

        X_dev_ver = add_controlled_interactions(X_dev_ver, groups=groups)
        X_test_ver = add_controlled_interactions(X_test_ver, groups=groups)

    feature_versions[name] = (X_dev_ver, X_test_ver)
    print(name, X_dev_ver.shape, X_test_ver.shape)

# ============================================================
# Module 13C — CatBoost Feature-Ablation Runner
# ============================================================

def run_catboost_ablation(X_dev_df, y_dev_series, X_test_df, version_name):
    print(f"\n{'='*70}")
    print(f"CatBoost ablation: {version_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        y_tr = y_dev_series.iloc[tr_idx].values
        X_va = X_dev_df.iloc[va_idx].copy()
        y_va = y_dev_series.iloc[va_idx].values

        preproc = build_tree_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])]
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)

        model = build_catboost_custom(class_weights=[1.0, 3.0])
        model.fit(X_tr_t, y_tr)

        va_prob = model.predict_proba(X_va_t)[:, 1]
        te_prob = model.predict_proba(X_te_t)[:, 1]

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        row = evaluate_predictions(y_va, va_prob, threshold=0.5)
        row["model"] = "catboost"
        row["feature_version"] = version_name
        row["fold_id"] = fold_id
        fold_rows.append(row)

        print(f"Fold {fold_id:02d} | PR_AUC={row['PR_AUC']:.5f} | ROC_AUC={row['ROC_AUC']:.5f}")

    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)
    best_thr, _ = find_best_f1_threshold(y_dev_series.values, oof_pred)

    summary_df = pd.DataFrame([{
        "model": "catboost",
        "feature_version": version_name,
        **evaluate_predictions(y_dev_series.values, oof_pred, threshold=best_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_"),
    }])

    cv_df = pd.DataFrame(fold_rows)
    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 13D — FTAN Feature-Ablation Runner
# ============================================================

def run_ftan_ablation(X_dev_df, y_dev_series, X_test_df, version_name):
    print(f"\n{'='*70}")
    print(f"FTAN ablation: {version_name}")
    print(f"{'='*70}")

    oof_pred = np.zeros(len(X_dev_df), dtype=float)
    test_pred_folds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
        X_tr = X_dev_df.iloc[tr_idx].copy()
        y_tr = y_dev_series.iloc[tr_idx].values.astype(np.float32)
        X_va = X_dev_df.iloc[va_idx].copy()
        y_va = y_dev_series.iloc[va_idx].values.astype(np.float32)

        preproc = build_dense_preprocessor(
            numeric_cols=[c for c in X_tr.columns if pd.api.types.is_numeric_dtype(X_tr[c])],
            categorical_cols=[c for c in X_tr.columns if not pd.api.types.is_numeric_dtype(X_tr[c])],
            scaler="robust"
        )
        X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr, X_va, X_test_df)
        input_dim = X_tr_t.shape[1]

        model = FTANClassifier(
            num_features=input_dim,
            embed_dim=CFG["attention_embed_dim"],
            num_heads=CFG["attention_heads"],
            num_blocks=CFG["attention_blocks"],
            dropout=CFG["attention_dropout"]
        )

        model, hist_df = fit_torch_model_v2(
            model=model,
            X_train=X_tr_t,
            y_train=y_tr,
            X_valid=X_va_t,
            y_valid=y_va,
            device=CFG["device"],
            batch_size=CFG["batch_size"],
            max_epochs=CFG["max_epochs"],
            lr=CFG["lr"],
            weight_decay=CFG["weight_decay"],
            patience=CFG["patience"],
            loss_name="focal",
            focal_alpha=0.25,
            focal_gamma=2.0
        )

        valid_loader = DataLoader(
            TabularDataset(X_va_t, y_va),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )
        test_loader = DataLoader(
            TabularDataset(X_te_t, None),
            batch_size=CFG["batch_size"],
            shuffle=False,
            num_workers=CFG["num_workers"],
            pin_memory=True
        )

        va_prob = predict_proba_torch(model, valid_loader, CFG["device"])
        te_prob = predict_proba_torch(model, test_loader, CFG["device"])

        oof_pred[va_idx] = va_prob
        test_pred_folds.append(te_prob)

        row = evaluate_predictions(y_va, va_prob, threshold=0.5)
        row["model"] = "ftan"
        row["feature_version"] = version_name
        row["fold_id"] = fold_id
        fold_rows.append(row)

        print(f"Fold {fold_id:02d} | PR_AUC={row['PR_AUC']:.5f} | ROC_AUC={row['ROC_AUC']:.5f}")

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    test_pred = np.mean(np.vstack(test_pred_folds), axis=0)
    best_thr, _ = find_best_f1_threshold(y_dev_series.values, oof_pred)

    summary_df = pd.DataFrame([{
        "model": "ftan",
        "feature_version": version_name,
        **evaluate_predictions(y_dev_series.values, oof_pred, threshold=best_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_pred, threshold=best_thr, prefix="TEST_"),
    }])

    cv_df = pd.DataFrame(fold_rows)
    return cv_df, summary_df, oof_pred, test_pred

# ============================================================
# Module 13E — Execute CatBoost Ablations
# ============================================================

all_cv_13 = []
all_summary_13 = []

for version_name, (X_dev_ver, X_test_ver) in feature_versions.items():
    cv_df, summary_df, oof_pred, test_pred = run_catboost_ablation(
        X_dev_df=X_dev_ver,
        y_dev_series=y_dev,
        X_test_df=X_test_ver,
        version_name=version_name
    )
    all_cv_13.append(cv_df)
    all_summary_13.append(summary_df)

module13_cat_cv = pd.concat(all_cv_13, axis=0, ignore_index=True)
module13_cat_summary = pd.concat(all_summary_13, axis=0, ignore_index=True)

module13_cat_cv.to_csv(TABLE_DIR / "module13_catboost_cv.csv", index=False)
module13_cat_summary.to_csv(TABLE_DIR / "module13_catboost_summary.csv", index=False)

display(module13_cat_summary.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 13F — Select Top Feature Sets for FTAN
# ============================================================

top_versions_for_ftan = (
    module13_cat_summary
    .sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False)
    ["feature_version"]
    .head(3)
    .tolist()
)

print("Top versions for FTAN:", top_versions_for_ftan)

all_cv_13_ftan = []
all_summary_13_ftan = []

for version_name in top_versions_for_ftan:
    X_dev_ver, X_test_ver = feature_versions[version_name]

    cv_df, summary_df, oof_pred, test_pred = run_ftan_ablation(
        X_dev_df=X_dev_ver,
        y_dev_series=y_dev,
        X_test_df=X_test_ver,
        version_name=version_name
    )
    all_cv_13_ftan.append(cv_df)
    all_summary_13_ftan.append(summary_df)

module13_ftan_cv = pd.concat(all_cv_13_ftan, axis=0, ignore_index=True)
module13_ftan_summary = pd.concat(all_summary_13_ftan, axis=0, ignore_index=True)

module13_ftan_cv.to_csv(TABLE_DIR / "module13_ftan_cv.csv", index=False)
module13_ftan_summary.to_csv(TABLE_DIR / "module13_ftan_summary.csv", index=False)

display(module13_ftan_summary.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 13G — Final Feature Interaction Comparison
# ============================================================

module13_compare = pd.concat([module13_cat_summary, module13_ftan_summary], axis=0, ignore_index=True)

keep_cols = [
    "model", "feature_version",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

module13_compare = module13_compare[keep_cols].sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"],
    ascending=False
)

module13_compare.to_csv(TABLE_DIR / "module13_compare.csv", index=False)
display(module13_compare)

# ============================================================
# Module 13H — Optional Follow-up Direction
# ============================================================

best_row_13 = module13_compare.iloc[0]
print("Best interaction setup so far:")
display(best_row_13)
```

```python
# ============================================================
# Module 14A — Dataset with Hard Flags
# ============================================================

class TabularDatasetWithHardFlag(Dataset):
    def __init__(self, X, y, hard_flag):
        self.X = torch.tensor(np.asarray(X), dtype=torch.float32)
        self.y = torch.tensor(np.asarray(y), dtype=torch.float32)
        self.hard_flag = torch.tensor(np.asarray(hard_flag), dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx], self.hard_flag[idx]

# ============================================================
# Module 14B — Hard-Aware Supervised Contrastive Loss
# ============================================================

class HardAwareSupConLoss(nn.Module):
    def __init__(self, temperature=0.10, hard_pair_boost=2.0):
        super().__init__()
        self.temperature = temperature
        self.hard_pair_boost = hard_pair_boost

    def forward(self, features, labels, hard_flags):
        """
        features: [B, D]
        labels:   [B]
        hard_flags: [B]
        """
        device = features.device
        labels = labels.view(-1, 1)
        hard_flags = hard_flags.view(-1, 1)

        features = F.normalize(features, p=2, dim=1)
        logits = torch.matmul(features, features.T) / self.temperature

        logits_max, _ = torch.max(logits, dim=1, keepdim=True)
        logits = logits - logits_max.detach()

        same_class = torch.eq(labels, labels.T).float().to(device)
        not_self = torch.ones_like(same_class) - torch.eye(same_class.shape[0], device=device)

        pos_mask = same_class * not_self

        # hard-hard pair weighting
        hard_pair = torch.matmul(hard_flags, hard_flags.T)
        pair_weight = 1.0 + (self.hard_pair_boost - 1.0) * hard_pair
        pos_weighted_mask = pos_mask * pair_weight

        exp_logits = torch.exp(logits) * not_self
        log_prob = logits - torch.log(exp_logits.sum(dim=1, keepdim=True) + 1e-12)

        pos_count = pos_weighted_mask.sum(dim=1)
        mean_log_prob_pos = (pos_weighted_mask * log_prob).sum(dim=1) / (pos_count + 1e-12)

        loss = -mean_log_prob_pos
        valid = pos_count > 0
        loss = loss[valid].mean() if valid.any() else torch.tensor(0.0, device=device)
        return loss

# ============================================================
# Module 14C — Joint Trainer
# ============================================================

@torch.no_grad()
def predict_ftan_prob(model, loader, device):
    model.eval()
    probs = []
    for batch in loader:
        if len(batch) == 3:
            Xb = batch[0]
        else:
            Xb = batch[0]
        Xb = Xb.to(device)
        logits = model(Xb)
        prob = torch.sigmoid(logits).detach().cpu().numpy()
        probs.append(prob)
    return np.concatenate(probs)


def fit_ftan_global_hardaware(
    model,
    X_train, y_train, hard_train,
    X_valid, y_valid, hard_valid,
    device,
    batch_size=256,
    max_epochs=20,
    lr=3e-4,
    weight_decay=1e-5,
    patience=5,
    lambda_cls=1.0,
    lambda_supcon=0.15,
    temperature=0.10,
    hard_pair_boost=2.0
):
    train_ds = TabularDatasetWithHardFlag(X_train, y_train, hard_train)
    valid_ds = TabularDatasetWithHardFlag(X_valid, y_valid, hard_valid)

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=CFG["num_workers"],
        pin_memory=True,
        drop_last=True
    )
    valid_loader = DataLoader(
        valid_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
    cls_criterion = nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
    )
    con_criterion = HardAwareSupConLoss(
        temperature=temperature,
        hard_pair_boost=hard_pair_boost
    )

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        model.train()
        train_loss_sum = 0.0

        for Xb, yb, hb in train_loader:
            Xb = Xb.to(device)
            yb = yb.to(device)
            hb = hb.to(device)

            optimizer.zero_grad()

            feat = model.forward_features(Xb)
            logits = model.head(feat).squeeze(1)

            cls_loss = cls_criterion(logits, yb)
            con_loss = con_criterion(feat, yb.long(), hb)

            loss = lambda_cls * cls_loss + lambda_supcon * con_loss
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * len(Xb)

        train_loss = train_loss_sum / len(train_loader.dataset)

        valid_probs = predict_ftan_prob(model, valid_loader, device)
        valid_pr = average_precision_score(y_valid, valid_probs)

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "valid_pr_auc": valid_pr
        })

        print(f"Epoch {epoch+1:02d} | train_loss={train_loss:.5f} | valid_pr_auc={valid_pr:.5f}")

        if valid_pr > best_pr:
            best_pr = valid_pr
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, pd.DataFrame(history)

# ============================================================
# Module 14D — OOF Training for Global Hard-Aware FTAN
# ============================================================

hard_flag_dev = oof_df["hard_uncertainty"].values.astype(np.float32)
hard_flag_test = test_df["hard_uncertainty"].values.astype(np.float32)

ftan14_oof = np.zeros(len(X_dev_fe), dtype=float)
ftan14_test_folds = []
ftan14_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
    print(f"\nModule 14 FTAN fold {fold_id}")

    X_tr_df = X_dev_fe.iloc[tr_idx].copy()
    X_va_df = X_dev_fe.iloc[va_idx].copy()

    y_tr = y_dev.iloc[tr_idx].values.astype(np.float32)
    y_va = y_dev.iloc[va_idx].values.astype(np.float32)

    h_tr = hard_flag_dev[tr_idx]
    h_va = hard_flag_dev[va_idx]

    preproc = build_dense_preprocessor(
        numeric_cols=[c for c in X_tr_df.columns if pd.api.types.is_numeric_dtype(X_tr_df[c])],
        categorical_cols=[c for c in X_tr_df.columns if not pd.api.types.is_numeric_dtype(X_tr_df[c])],
        scaler="robust"
    )
    X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr_df, X_va_df, X_test_fe)
    input_dim = X_tr_t.shape[1]

    model = FTANClassifier(
        num_features=input_dim,
        embed_dim=CFG["attention_embed_dim"],
        num_heads=CFG["attention_heads"],
        num_blocks=CFG["attention_blocks"],
        dropout=CFG["attention_dropout"]
    )

    model, hist_df = fit_ftan_global_hardaware(
        model=model,
        X_train=X_tr_t, y_train=y_tr, hard_train=h_tr,
        X_valid=X_va_t, y_valid=y_va, hard_valid=h_va,
        device=CFG["device"],
        batch_size=256,
        max_epochs=20,
        lr=3e-4,
        weight_decay=1e-5,
        patience=5,
        lambda_cls=1.0,
        lambda_supcon=0.15,
        temperature=0.10,
        hard_pair_boost=2.0
    )

    va_loader = DataLoader(
        TabularDatasetWithHardFlag(X_va_t, y_va, h_va),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )
    te_loader = DataLoader(
        TabularDatasetWithHardFlag(X_te_t, np.zeros(len(X_te_t), dtype=np.float32), hard_flag_test),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    va_prob = predict_ftan_prob(model, va_loader, CFG["device"])
    te_prob = predict_ftan_prob(model, te_loader, CFG["device"])

    ftan14_oof[va_idx] = va_prob
    ftan14_test_folds.append(te_prob)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "ftan_global_hardaware"
    row["fold_id"] = fold_id
    ftan14_rows.append(row)

    hist_df.to_csv(LOG_DIR / f"module14_ftan_fold{fold_id}_history.csv", index=False)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

ftan14_cv = pd.DataFrame(ftan14_rows)
ftan14_cv.to_csv(TABLE_DIR / "module14_ftan_cv.csv", index=False)
display(ftan14_cv.describe())

ftan14_test = np.mean(np.vstack(ftan14_test_folds), axis=0)

# ============================================================
# Module 14E — Evaluation
# ============================================================

thr14, _ = find_best_f1_threshold(y_dev.values, ftan14_oof)

module14_summary = pd.DataFrame([{
    "model": "ftan_global_hardaware",
    **evaluate_predictions(y_dev.values, ftan14_oof, threshold=thr14, prefix="OOF_"),
    **evaluate_predictions(y_test.values, ftan14_test, threshold=thr14, prefix="TEST_"),
}])

module14_summary.to_csv(TABLE_DIR / "module14_summary.csv", index=False)
display(module14_summary)

# Hard region check
hard_compare_14 = pd.DataFrame([
    {
        "model": "ftan_focal_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            oof_df.loc[hard_mask_dev, "ftan_focal"].values,
            threshold=thresholds["ftan_focal"]
        )
    },
    {
        "model": "ftan_global_hardaware_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            ftan14_oof[hard_mask_dev],
            threshold=thr14
        )
    }
])

hard_compare_14.to_csv(TABLE_DIR / "module14_hard_compare.csv", index=False)
display(hard_compare_14)

# ============================================================
# Module 14F — Two-Stage Blend with Module 14 FTAN
# ============================================================

blend_rows_14 = []

for alpha in [0.3, 0.5, 0.7]:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        alpha * ftan14_oof[hard_mask_dev] +
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )
    test_prob[hard_mask_test] = (
        alpha * ftan14_test[hard_mask_test] +
        (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
    )

    thr, _ = find_best_f1_threshold(y_dev.values, dev_prob)

    blend_rows_14.append({
        "model": f"two_stage_module14_blend_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(y_dev.values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_prob, threshold=thr, prefix="TEST_"),
    })

module14_blends = pd.DataFrame(blend_rows_14)
module14_blends.to_csv(TABLE_DIR / "module14_blends.csv", index=False)
display(module14_blends.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 14G — Final Comparison
# ============================================================

rows = []

# Current best fixed blend
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(y_dev.values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(y_dev.values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# Module 14 standalone
rows.append(module14_summary.iloc[0].to_dict())

# Best Module 14 blend
best_m14 = module14_blends.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False).iloc[0].to_dict()
rows.append(best_m14)

module14_compare = pd.DataFrame(rows)
module14_compare.to_csv(TABLE_DIR / "module14_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module14_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 15A.1 — FTAN Prototype Model
# ============================================================

class FTANPrototypeClassifier(nn.Module):
    def __init__(self, num_features, embed_dim=64, num_heads=4, num_blocks=2, dropout=0.1):
        super().__init__()
        self.tokenizer = FeatureTokenizer(num_features=num_features, embed_dim=embed_dim)
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, dropout=dropout)
            for _ in range(num_blocks)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        # optional projection head for prototype space
        self.proj = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, embed_dim)
        )

        # learnable class prototypes: [2, D]
        self.prototypes = nn.Parameter(torch.randn(2, embed_dim) * 0.02)

    def forward_features(self, x):
        tok = self.tokenizer(x)
        for blk in self.blocks:
            tok = blk(tok)
        tok = self.norm(tok)
        pooled = tok.mean(dim=1)
        z = self.proj(pooled)
        z = F.normalize(z, p=2, dim=1)
        return z

    def forward(self, x):
        z = self.forward_features(x)  # [B, D]

        proto = F.normalize(self.prototypes, p=2, dim=1)  # [2, D]

        # squared Euclidean distances
        d0 = torch.sum((z - proto[0].unsqueeze(0)) ** 2, dim=1)
        d1 = torch.sum((z - proto[1].unsqueeze(0)) ** 2, dim=1)

        # positive logit: smaller d1 than d0 => more positive
        logits = d0 - d1
        return logits, z, proto

# ============================================================
# Module 15A.2 — Prototype Losses
# ============================================================

def prototype_compactness_loss(z, y, proto):
    """
    Pull embeddings toward their class prototype.
    z: [B, D]
    y: [B]
    proto: [2, D]
    """
    y = y.long()
    assigned_proto = proto[y]  # [B, D]
    loss = torch.mean(torch.sum((z - assigned_proto) ** 2, dim=1))
    return loss


def prototype_separation_loss(proto, margin=1.0):
    """
    Push class prototypes apart.
    """
    d = torch.sum((proto[0] - proto[1]) ** 2)
    loss = F.relu(margin - d)
    return loss

# ============================================================
# Module 15A.3 — Prototype Model Trainer
# ============================================================

class TabularDatasetBasic(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(np.asarray(X), dtype=torch.float32)
        self.y = torch.tensor(np.asarray(y), dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


@torch.no_grad()
def predict_proto_model(model, loader, device):
    model.eval()
    probs, embs = [], []
    for Xb, yb in loader:
        Xb = Xb.to(device)
        logits, z, proto = model(Xb)
        prob = torch.sigmoid(logits).detach().cpu().numpy()
        probs.append(prob)
        embs.append(z.detach().cpu().numpy())
    return np.concatenate(probs), np.concatenate(embs)


def fit_prototype_model(
    model,
    X_train, y_train,
    X_valid, y_valid,
    device,
    batch_size=256,
    max_epochs=20,
    lr=3e-4,
    weight_decay=1e-5,
    patience=5,
    lambda_compact=0.20,
    lambda_sep=0.05,
    proto_margin=1.0
):
    train_ds = TabularDatasetBasic(X_train, y_train)
    valid_ds = TabularDatasetBasic(X_valid, y_valid)

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=CFG["num_workers"],
        pin_memory=True,
        drop_last=False
    )
    valid_loader = DataLoader(
        valid_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
    cls_criterion = nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
    )

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        model.train()
        train_loss_sum = 0.0

        for Xb, yb in train_loader:
            Xb = Xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            logits, z, proto = model(Xb)

            cls_loss = cls_criterion(logits, yb)
            compact_loss = prototype_compactness_loss(z, yb, proto)
            sep_loss = prototype_separation_loss(proto, margin=proto_margin)

            loss = cls_loss + lambda_compact * compact_loss + lambda_sep * sep_loss
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * len(Xb)

        train_loss = train_loss_sum / len(train_loader.dataset)

        val_probs, _ = predict_proto_model(model, valid_loader, device)
        val_pr = average_precision_score(y_valid, val_probs)

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "valid_pr_auc": val_pr
        })

        print(f"Epoch {epoch+1:02d} | train_loss={train_loss:.5f} | valid_pr_auc={val_pr:.5f}")

        if val_pr > best_pr:
            best_pr = val_pr
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, pd.DataFrame(history)

# ============================================================
# Module 15A.4 — OOF Training for Prototype Model
# ============================================================

proto_oof = np.zeros(len(X_dev_fe), dtype=float)
proto_test_folds = []
proto_rows = []
proto_oof_emb = None
proto_test_emb_folds = []

for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
    print(f"\nPrototype model fold {fold_id}")

    X_tr_df = X_dev_fe.iloc[tr_idx].copy()
    X_va_df = X_dev_fe.iloc[va_idx].copy()

    y_tr = y_dev.iloc[tr_idx].values.astype(np.float32)
    y_va = y_dev.iloc[va_idx].values.astype(np.float32)

    preproc = build_dense_preprocessor(
        numeric_cols=[c for c in X_tr_df.columns if pd.api.types.is_numeric_dtype(X_tr_df[c])],
        categorical_cols=[c for c in X_tr_df.columns if not pd.api.types.is_numeric_dtype(X_tr_df[c])],
        scaler="robust"
    )
    X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr_df, X_va_df, X_test_fe)
    input_dim = X_tr_t.shape[1]

    model = FTANPrototypeClassifier(
        num_features=input_dim,
        embed_dim=CFG["attention_embed_dim"],
        num_heads=CFG["attention_heads"],
        num_blocks=CFG["attention_blocks"],
        dropout=CFG["attention_dropout"]
    )

    model, hist_df = fit_prototype_model(
        model=model,
        X_train=X_tr_t, y_train=y_tr,
        X_valid=X_va_t, y_valid=y_va,
        device=CFG["device"],
        batch_size=256,
        max_epochs=20,
        lr=3e-4,
        weight_decay=1e-5,
        patience=5,
        lambda_compact=0.20,
        lambda_sep=0.05,
        proto_margin=1.0
    )

    va_loader = DataLoader(
        TabularDatasetBasic(X_va_t, y_va),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )
    te_loader = DataLoader(
        TabularDatasetBasic(X_te_t, np.zeros(len(X_te_t), dtype=np.float32)),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    va_prob, va_emb = predict_proto_model(model, va_loader, CFG["device"])
    te_prob, te_emb = predict_proto_model(model, te_loader, CFG["device"])

    proto_oof[va_idx] = va_prob
    proto_test_folds.append(te_prob)

    if proto_oof_emb is None:
        proto_oof_emb = np.zeros((len(X_dev_fe), va_emb.shape[1]), dtype=np.float32)
    proto_oof_emb[va_idx] = va_emb
    proto_test_emb_folds.append(te_emb)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "ftan_prototype"
    row["fold_id"] = fold_id
    proto_rows.append(row)

    hist_df.to_csv(LOG_DIR / f"module15A_proto_fold{fold_id}_history.csv", index=False)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

proto_cv = pd.DataFrame(proto_rows)
proto_cv.to_csv(TABLE_DIR / "module15A_proto_cv.csv", index=False)
display(proto_cv.describe())

proto_test = np.mean(np.vstack(proto_test_folds), axis=0)
proto_test_emb = np.mean(np.stack(proto_test_emb_folds, axis=0), axis=0)

# ============================================================
# Module 15A.5 — Global Prototype Evaluation
# ============================================================

proto_thr, _ = find_best_f1_threshold(y_dev.values, proto_oof)

module15A_summary = pd.DataFrame([{
    "model": "ftan_prototype",
    **evaluate_predictions(y_dev.values, proto_oof, threshold=proto_thr, prefix="OOF_"),
    **evaluate_predictions(y_test.values, proto_test, threshold=proto_thr, prefix="TEST_"),
}])

module15A_summary.to_csv(TABLE_DIR / "module15A_summary.csv", index=False)
display(module15A_summary)

# ============================================================
# Module 15A.6 — Hard-Region Prototype Evaluation
# ============================================================

module15A_hard_compare = pd.DataFrame([
    {
        "model": "ftan_focal_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            oof_df.loc[hard_mask_dev, "ftan_focal"].values,
            threshold=thresholds["ftan_focal"]
        )
    },
    {
        "model": "ftan_prototype_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            proto_oof[hard_mask_dev],
            threshold=proto_thr
        )
    }
])

module15A_hard_compare.to_csv(TABLE_DIR / "module15A_hard_compare.csv", index=False)
display(module15A_hard_compare)

# ============================================================
# Module 15A.7 — Two-Stage Blend with Prototype Model
# ============================================================

proto_blend_rows = []

for alpha in [0.3, 0.5, 0.7]:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        alpha * proto_oof[hard_mask_dev] +
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )
    test_prob[hard_mask_test] = (
        alpha * proto_test[hard_mask_test] +
        (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
    )

    thr, _ = find_best_f1_threshold(y_dev.values, dev_prob)

    proto_blend_rows.append({
        "model": f"two_stage_proto_blend_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(y_dev.values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_prob, threshold=thr, prefix="TEST_"),
    })

module15A_blends = pd.DataFrame(proto_blend_rows)
module15A_blends.to_csv(TABLE_DIR / "module15A_blends.csv", index=False)
display(module15A_blends.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 15A.8 — Prototype Geometry Diagnostics
# ============================================================

# Recreate prototype model on one reference fold if needed later, but for now use OOF embeddings only.
# We'll approximate geometry by class-centroid distances in learned embedding space.

proto0 = proto_oof_emb[y_dev.values == 0].mean(axis=0)
proto1 = proto_oof_emb[y_dev.values == 1].mean(axis=0)

def squared_dist(a, b):
    return np.sum((a - b) ** 2, axis=1)

d0 = squared_dist(proto_oof_emb, proto0.reshape(1, -1))
d1 = squared_dist(proto_oof_emb, proto1.reshape(1, -1))
margin = np.abs(d0 - d1)

geom_df = pd.DataFrame({
    "y_true": y_dev.values,
    "hard": hard_mask_dev.astype(int),
    "d0": d0,
    "d1": d1,
    "margin": margin
})

geom_summary = pd.DataFrame([
    {
        "region": "easy",
        "mean_margin": float(geom_df.loc[geom_df["hard"] == 0, "margin"].mean()),
        "median_margin": float(geom_df.loc[geom_df["hard"] == 0, "margin"].median())
    },
    {
        "region": "hard",
        "mean_margin": float(geom_df.loc[geom_df["hard"] == 1, "margin"].mean()),
        "median_margin": float(geom_df.loc[geom_df["hard"] == 1, "margin"].median())
    }
])

geom_summary.to_csv(TABLE_DIR / "module15A_geometry_summary.csv", index=False)
display(geom_summary)

# ============================================================
# Module 15A.9 — Final Comparison
# ============================================================

rows = []

# current best fixed blend
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(y_dev.values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(y_dev.values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# prototype standalone
rows.append(module15A_summary.iloc[0].to_dict())

# best prototype blend
best_proto_blend = module15A_blends.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False
).iloc[0].to_dict()
rows.append(best_proto_blend)

module15A_compare = pd.DataFrame(rows)
module15A_compare.to_csv(TABLE_DIR / "module15A_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module15A_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 15B.1 — FTAN Metric Model
# ============================================================

class FTANMetricClassifier(nn.Module):
    def __init__(self, num_features, embed_dim=64, num_heads=4, num_blocks=2, dropout=0.1):
        super().__init__()
        self.tokenizer = FeatureTokenizer(num_features=num_features, embed_dim=embed_dim)
        self.blocks = nn.ModuleList([
            TransformerBlock(embed_dim, num_heads, dropout=dropout)
            for _ in range(num_blocks)
        ])
        self.norm = nn.LayerNorm(embed_dim)

        self.proj = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, embed_dim)
        )

        self.cls_head = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim, 1)
        )

    def forward_features(self, x):
        tok = self.tokenizer(x)
        for blk in self.blocks:
            tok = blk(tok)
        tok = self.norm(tok)
        pooled = tok.mean(dim=1)
        z = self.proj(pooled)
        z = F.normalize(z, p=2, dim=1)
        return z

    def forward(self, x):
        z = self.forward_features(x)
        logits = self.cls_head(z).squeeze(1)
        return logits, z

# ============================================================
# Module 15B.2 — Batch-Hard Triplet Loss
# ============================================================

class BatchHardTripletLoss(nn.Module):
    def __init__(self, margin=0.2):
        super().__init__()
        self.margin = margin

    def forward(self, embeddings, labels):
        """
        embeddings: [B, D], assumed normalized
        labels: [B]
        """
        device = embeddings.device
        labels = labels.view(-1)

        # pairwise Euclidean distance matrix
        dist = torch.cdist(embeddings, embeddings, p=2)

        loss_vals = []
        for i in range(len(labels)):
            anchor_label = labels[i]

            pos_mask = (labels == anchor_label)
            neg_mask = (labels != anchor_label)

            pos_mask[i] = False  # exclude self

            if pos_mask.sum() == 0 or neg_mask.sum() == 0:
                continue

            hardest_pos = dist[i][pos_mask].max()
            hardest_neg = dist[i][neg_mask].min()

            loss_i = F.relu(hardest_pos - hardest_neg + self.margin)
            loss_vals.append(loss_i)

        if len(loss_vals) == 0:
            return torch.tensor(0.0, device=device)

        return torch.stack(loss_vals).mean()

# ============================================================
# Module 15B.3 — Prediction Helper
# ============================================================

@torch.no_grad()
def predict_metric_model(model, loader, device):
    model.eval()
    probs, embs = [], []
    for Xb, yb in loader:
        Xb = Xb.to(device)
        logits, z = model(Xb)
        prob = torch.sigmoid(logits).detach().cpu().numpy()
        probs.append(prob)
        embs.append(z.detach().cpu().numpy())
    return np.concatenate(probs), np.concatenate(embs)

# ============================================================
# Module 15B.4 — Joint Trainer: BCE + Triplet
# ============================================================

def fit_metric_model(
    model,
    X_train, y_train,
    X_valid, y_valid,
    device,
    batch_size=256,
    max_epochs=20,
    lr=3e-4,
    weight_decay=1e-5,
    patience=5,
    lambda_triplet=0.15,
    triplet_margin=0.2
):
    train_ds = TabularDatasetBasic(X_train, y_train)
    valid_ds = TabularDatasetBasic(X_valid, y_valid)

    train_loader = DataLoader(
        train_ds,
        batch_size=batch_size,
        shuffle=True,
        num_workers=CFG["num_workers"],
        pin_memory=True,
        drop_last=True
    )
    valid_loader = DataLoader(
        valid_ds,
        batch_size=batch_size,
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    pos_weight = (len(y_train) - y_train.sum()) / max(y_train.sum(), 1)
    cls_criterion = nn.BCEWithLogitsLoss(
        pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
    )
    metric_criterion = BatchHardTripletLoss(margin=triplet_margin)

    best_state = None
    best_pr = -np.inf
    patience_counter = 0
    history = []

    model.to(device)

    for epoch in range(max_epochs):
        model.train()
        train_loss_sum = 0.0

        for Xb, yb in train_loader:
            Xb = Xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            logits, z = model(Xb)
            cls_loss = cls_criterion(logits, yb)
            metric_loss = metric_criterion(z, yb.long())

            loss = cls_loss + lambda_triplet * metric_loss
            loss.backward()
            optimizer.step()

            train_loss_sum += loss.item() * len(Xb)

        train_loss = train_loss_sum / len(train_loader.dataset)

        val_probs, _ = predict_metric_model(model, valid_loader, device)
        val_pr = average_precision_score(y_valid, val_probs)

        history.append({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "valid_pr_auc": val_pr
        })

        print(f"Epoch {epoch+1:02d} | train_loss={train_loss:.5f} | valid_pr_auc={val_pr:.5f}")

        if val_pr > best_pr:
            best_pr = val_pr
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("Early stopping triggered.")
                break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, pd.DataFrame(history)

# ============================================================
# Module 15B.5 — OOF Training for Metric Model
# ============================================================

metric_oof = np.zeros(len(X_dev_fe), dtype=float)
metric_test_folds = []
metric_oof_emb = None
metric_test_emb_folds = []
metric_rows = []

for fold_id, (tr_idx, va_idx) in enumerate(fold_indices):
    print(f"\nMetric model fold {fold_id}")

    X_tr_df = X_dev_fe.iloc[tr_idx].copy()
    X_va_df = X_dev_fe.iloc[va_idx].copy()

    y_tr = y_dev.iloc[tr_idx].values.astype(np.float32)
    y_va = y_dev.iloc[va_idx].values.astype(np.float32)

    preproc = build_dense_preprocessor(
        numeric_cols=[c for c in X_tr_df.columns if pd.api.types.is_numeric_dtype(X_tr_df[c])],
        categorical_cols=[c for c in X_tr_df.columns if not pd.api.types.is_numeric_dtype(X_tr_df[c])],
        scaler="robust"
    )
    X_tr_t, X_va_t, X_te_t = fit_transform_fold(preproc, X_tr_df, X_va_df, X_test_fe)
    input_dim = X_tr_t.shape[1]

    model = FTANMetricClassifier(
        num_features=input_dim,
        embed_dim=CFG["attention_embed_dim"],
        num_heads=CFG["attention_heads"],
        num_blocks=CFG["attention_blocks"],
        dropout=CFG["attention_dropout"]
    )

    model, hist_df = fit_metric_model(
        model=model,
        X_train=X_tr_t, y_train=y_tr,
        X_valid=X_va_t, y_valid=y_va,
        device=CFG["device"],
        batch_size=256,
        max_epochs=2,
        lr=3e-4,
        weight_decay=1e-5,
        patience=2,
        lambda_triplet=0.15,
        triplet_margin=0.2
    )

    va_loader = DataLoader(
        TabularDatasetBasic(X_va_t, y_va),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )
    te_loader = DataLoader(
        TabularDatasetBasic(X_te_t, np.zeros(len(X_te_t), dtype=np.float32)),
        batch_size=CFG["batch_size"],
        shuffle=False,
        num_workers=CFG["num_workers"],
        pin_memory=True
    )

    va_prob, va_emb = predict_metric_model(model, va_loader, CFG["device"])
    te_prob, te_emb = predict_metric_model(model, te_loader, CFG["device"])

    metric_oof[va_idx] = va_prob
    metric_test_folds.append(te_prob)

    if metric_oof_emb is None:
        metric_oof_emb = np.zeros((len(X_dev_fe), va_emb.shape[1]), dtype=np.float32)
    metric_oof_emb[va_idx] = va_emb
    metric_test_emb_folds.append(te_emb)

    row = evaluate_predictions(y_va, va_prob, threshold=0.5)
    row["model"] = "ftan_metric"
    row["fold_id"] = fold_id
    metric_rows.append(row)

    hist_df.to_csv(LOG_DIR / f"module15B_metric_fold{fold_id}_history.csv", index=False)

    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

metric_cv = pd.DataFrame(metric_rows)
metric_cv.to_csv(TABLE_DIR / "module15B_metric_cv.csv", index=False)
display(metric_cv.describe())

metric_test = np.mean(np.vstack(metric_test_folds), axis=0)
metric_test_emb = np.mean(np.stack(metric_test_emb_folds, axis=0), axis=0)

# ============================================================
# Module 15B.6 — Metric Model Evaluation
# ============================================================

metric_thr, _ = find_best_f1_threshold(y_dev.values, metric_oof)

module15B_summary = pd.DataFrame([{
    "model": "ftan_metric",
    **evaluate_predictions(y_dev.values, metric_oof, threshold=metric_thr, prefix="OOF_"),
    **evaluate_predictions(y_test.values, metric_test, threshold=metric_thr, prefix="TEST_"),
}])

module15B_summary.to_csv(TABLE_DIR / "module15B_summary.csv", index=False)
display(module15B_summary)

module15B_hard_compare = pd.DataFrame([
    {
        "model": "ftan_focal_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            oof_df.loc[hard_mask_dev, "ftan_focal"].values,
            threshold=thresholds["ftan_focal"]
        )
    },
    {
        "model": "ftan_metric_hard",
        **evaluate_predictions(
            y_dev.values[hard_mask_dev],
            metric_oof[hard_mask_dev],
            threshold=metric_thr
        )
    }
])

module15B_hard_compare.to_csv(TABLE_DIR / "module15B_hard_compare.csv", index=False)
display(module15B_hard_compare)

# ============================================================
# Module 15B.7 — Two-Stage Blend with Metric Model
# ============================================================

metric_blend_rows = []

for alpha in [0.3, 0.5, 0.7]:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        alpha * metric_oof[hard_mask_dev] +
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values
    )
    test_prob[hard_mask_test] = (
        alpha * metric_test[hard_mask_test] +
        (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values
    )

    thr, _ = find_best_f1_threshold(y_dev.values, dev_prob)

    metric_blend_rows.append({
        "model": f"two_stage_metric_blend_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(y_dev.values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_prob, threshold=thr, prefix="TEST_"),
    })

module15B_blends = pd.DataFrame(metric_blend_rows)
module15B_blends.to_csv(TABLE_DIR / "module15B_blends.csv", index=False)
display(module15B_blends.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 15B.8 — Metric Geometry Diagnostics
# ============================================================

c0 = metric_oof_emb[y_dev.values == 0].mean(axis=0)
c1 = metric_oof_emb[y_dev.values == 1].mean(axis=0)

def sqdist(a, b):
    return np.sum((a - b) ** 2, axis=1)

d0 = sqdist(metric_oof_emb, c0.reshape(1, -1))
d1 = sqdist(metric_oof_emb, c1.reshape(1, -1))
margin = np.abs(d0 - d1)

geom15B = pd.DataFrame({
    "y_true": y_dev.values,
    "hard": hard_mask_dev.astype(int),
    "d0": d0,
    "d1": d1,
    "margin": margin
})

geom15B_summary = pd.DataFrame([
    {
        "region": "easy",
        "mean_margin": float(geom15B.loc[geom15B["hard"] == 0, "margin"].mean()),
        "median_margin": float(geom15B.loc[geom15B["hard"] == 0, "margin"].median())
    },
    {
        "region": "hard",
        "mean_margin": float(geom15B.loc[geom15B["hard"] == 1, "margin"].mean()),
        "median_margin": float(geom15B.loc[geom15B["hard"] == 1, "margin"].median())
    }
])

geom15B_summary.to_csv(TABLE_DIR / "module15B_geometry_summary.csv", index=False)
display(geom15B_summary)
```

```python
# ============================================================
# Module 15C.1 — Global Hybrid Prototype + CatBoost Blend
# ============================================================

hybrid_rows = []

for alpha in [0.1, 0.2, 0.3, 0.5]:
    dev_prob = (
        (1 - alpha) * oof_df["cb_heavy_pos"].values +
        alpha * proto_oof
    )
    test_prob = (
        (1 - alpha) * test_df["cb_heavy_pos"].values +
        alpha * proto_test
    )

    thr, _ = find_best_f1_threshold(y_dev.values, dev_prob)

    hybrid_rows.append({
        "model": f"hybrid_proto_global_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(y_dev.values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_prob, threshold=thr, prefix="TEST_"),
    })

module15C_global = pd.DataFrame(hybrid_rows)
module15C_global.to_csv(TABLE_DIR / "module15C_global.csv", index=False)
display(module15C_global.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 15C.2 — Hard-Region Hybrid Prototype + CatBoost Blend
# ============================================================

hybrid_hard_rows = []

for alpha in [0.1, 0.2, 0.3, 0.5]:
    dev_prob = oof_df["cb_heavy_pos"].values.copy()
    test_prob = test_df["cb_heavy_pos"].values.copy()

    dev_prob[hard_mask_dev] = (
        (1 - alpha) * oof_df.loc[hard_mask_dev, "cb_heavy_pos"].values +
        alpha * proto_oof[hard_mask_dev]
    )
    test_prob[hard_mask_test] = (
        (1 - alpha) * test_df.loc[hard_mask_test, "cb_heavy_pos"].values +
        alpha * proto_test[hard_mask_test]
    )

    thr, _ = find_best_f1_threshold(y_dev.values, dev_prob)

    hybrid_hard_rows.append({
        "model": f"hybrid_proto_hard_alpha_{alpha}",
        "alpha": alpha,
        **evaluate_predictions(y_dev.values, dev_prob, threshold=thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, test_prob, threshold=thr, prefix="TEST_"),
    })

module15C_hard = pd.DataFrame(hybrid_hard_rows)
module15C_hard.to_csv(TABLE_DIR / "module15C_hard.csv", index=False)
display(module15C_hard.sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))

# ============================================================
# Module 15C.3 — Final Comparison
# ============================================================

rows = []

# current best blend
if "final_dev_prob_blend" in globals() and "final_test_prob_blend" in globals():
    fixed_thr, _ = find_best_f1_threshold(y_dev.values, final_dev_prob_blend)
    rows.append({
        "model": "two_stage_blend_alpha_0.7",
        **evaluate_predictions(y_dev.values, final_dev_prob_blend, threshold=fixed_thr, prefix="OOF_"),
        **evaluate_predictions(y_test.values, final_test_prob_blend, threshold=fixed_thr, prefix="TEST_"),
    })

# best prototype hybrid global
best_hybrid_global = module15C_global.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False
).iloc[0].to_dict()
rows.append(best_hybrid_global)

# best prototype hybrid hard
best_hybrid_hard = module15C_hard.sort_values(
    ["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False
).iloc[0].to_dict()
rows.append(best_hybrid_hard)

module15C_compare = pd.DataFrame(rows)
module15C_compare.to_csv(TABLE_DIR / "module15C_compare.csv", index=False)

keep_cols = [
    "model",
    "OOF_PR_AUC", "OOF_ROC_AUC", "OOF_Brier", "OOF_NLL", "OOF_ECE",
    "OOF_Precision", "OOF_Recall", "OOF_F1", "OOF_MCC",
    "TEST_PR_AUC", "TEST_ROC_AUC", "TEST_Brier", "TEST_NLL", "TEST_ECE",
    "TEST_Precision", "TEST_Recall", "TEST_F1", "TEST_MCC"
]

display(module15C_compare[keep_cols].sort_values(["TEST_PR_AUC", "TEST_F1", "TEST_MCC"], ascending=False))
```

```python
# ============================================================
# Module 0 — Environment, config, reproducibility, folders
# Fresh notebook starter for final paper-grade pipeline
# ============================================================

from __future__ import annotations

import os
import gc
import json
import time
import math
import random
import platform
import warnings
from copy import deepcopy
from dataclasses import dataclass, asdict, field
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple

import numpy as np
import pandas as pd

from sklearn.model_selection import (
    StratifiedKFold,
    RepeatedStratifiedKFold,
    train_test_split,
)
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    brier_score_loss,
    log_loss,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    matthews_corrcoef,
    confusion_matrix,
)
from sklearn.calibration import calibration_curve

warnings.filterwarnings("ignore")


# -------------------------
# Config
# -------------------------
@dataclass
class ExperimentConfig:
    project_name: str = "trustworthy_cvd_final"

    # Seed / reproducibility
    seed: int = 42

    # Data
    data_path: str = "/kaggle/input/datasets/alexteboul/heart-disease-health-indicators-dataset/heart_disease_health_indicators_BRFSS2015.csv"  # <-- UPDATE THIS
    target_col: str = "HeartDiseaseorAttack"                # <-- UPDATE IF NEEDED
    id_col: Optional[str] = None

    # Split
    holdout_size: float = 0.20

    # Screening CV
    n_splits_screen: int = 3

    # Final CV (used later)
    n_splits_final: int = 5
    n_repeats_final: int = 1
    use_repeated_cv: bool = False

    # Expert selection
    preferred_final_experts: int = 4
    max_final_experts: int = 5
    corr_redundancy_threshold: float = 0.98
    min_keep_per_family: Dict[str, int] = field(default_factory=lambda: {
        "linear": 1,
        "tree": 2,
        "deep": 1,
    })

    # Threshold analysis (used later)
    precision_target: float = 0.30
    recall_target: float = 0.50

    # Output directories
    output_root: str = "/kaggle/working/outputs_final"


CFG = ExperimentConfig()


# -------------------------
# Reproducibility
# -------------------------
def set_global_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

    try:
        import torch
        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    except Exception:
        pass


set_global_seed(CFG.seed)


# -------------------------
# Directories
# -------------------------
OUTPUT_ROOT = Path(CFG.output_root)
TAB_DIR = OUTPUT_ROOT / "tables"
FIG_DIR = OUTPUT_ROOT / "figures"
PRED_DIR = OUTPUT_ROOT / "predictions"
MODEL_DIR = OUTPUT_ROOT / "models"
META_DIR = OUTPUT_ROOT / "meta"

for d in [OUTPUT_ROOT, TAB_DIR, FIG_DIR, PRED_DIR, MODEL_DIR, META_DIR]:
    d.mkdir(parents=True, exist_ok=True)


# -------------------------
# Save helpers
# -------------------------
def save_json(obj: Dict[str, Any], path: Path) -> None:
    with open(path, "w", encoding="utf-8") as f:
        json.dump(obj, f, indent=2, ensure_ascii=False)


def save_text(text: str, path: Path) -> None:
    with open(path, "w", encoding="utf-8") as f:
        f.write(text)


save_json(asdict(CFG), META_DIR / "config_final.json")

save_text(
    "\n".join([
        f"project_name={CFG.project_name}",
        f"platform={platform.platform()}",
        f"python_version={platform.python_version()}",
        f"seed={CFG.seed}",
        f"output_root={str(OUTPUT_ROOT)}",
    ]),
    META_DIR / "environment_manifest.txt"
)

print("Module 0 complete.")
print("Output root:", OUTPUT_ROOT)

# ============================================================
# Module 1 — Data loading and audit
# ============================================================

def load_data(data_path: str) -> pd.DataFrame:
    df = pd.read_csv(data_path)
    return df


def build_feature_list(
    df: pd.DataFrame,
    target_col: str,
    id_col: Optional[str] = None,
) -> List[str]:
    drop_cols = {target_col}
    if id_col is not None and id_col in df.columns:
        drop_cols.add(id_col)
    return [c for c in df.columns if c not in drop_cols]


def audit_dataframe(df: pd.DataFrame, target_col: str) -> Tuple[Dict[str, Any], pd.DataFrame]:
    missing_by_col = df.isna().sum().sort_values(ascending=False)
    dtypes = df.dtypes.astype(str).value_counts().to_dict()

    audit = {
        "n_rows": int(df.shape[0]),
        "n_cols": int(df.shape[1]),
        "n_duplicates": int(df.duplicated().sum()),
        "missing_total": int(df.isna().sum().sum()),
        "target_col": target_col,
        "target_prevalence": float(df[target_col].mean()) if target_col in df.columns else None,
        "dtype_summary": dtypes,
    }

    missing_df = pd.DataFrame({
        "column": missing_by_col.index,
        "missing_n": missing_by_col.values,
        "missing_pct": (missing_by_col.values / len(df)) * 100.0,
    })

    return audit, missing_df


# -------------------------
# Load and inspect
# -------------------------
df = load_data(CFG.data_path)

assert CFG.target_col in df.columns, f"Target column '{CFG.target_col}' not found in dataset."

FEATURES = build_feature_list(df, CFG.target_col, CFG.id_col)
audit, missing_df = audit_dataframe(df, CFG.target_col)

save_json(audit, META_DIR / "data_audit.json")
missing_df.to_csv(TAB_DIR / "missingness_table.csv", index=False)

print("Shape:", df.shape)
print("Target prevalence:", round(df[CFG.target_col].mean(), 6))
print("Number of features:", len(FEATURES))

display(df.head())
display(missing_df.head(20))

print("Module 1 complete.")

# ============================================================
# Module 2 — Fixed dev / holdout split
# ============================================================

def make_dev_holdout_split(
    df: pd.DataFrame,
    target_col: str,
    holdout_size: float,
    seed: int,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    dev_df, holdout_df = train_test_split(
        df,
        test_size=holdout_size,
        stratify=df[target_col],
        random_state=seed,
    )
    return dev_df.reset_index(drop=True), holdout_df.reset_index(drop=True)


def get_cv(config: ExperimentConfig, final_stage: bool = False):
    if final_stage:
        if config.use_repeated_cv and config.n_repeats_final > 1:
            return RepeatedStratifiedKFold(
                n_splits=config.n_splits_final,
                n_repeats=config.n_repeats_final,
                random_state=config.seed,
            )
        return StratifiedKFold(
            n_splits=config.n_splits_final,
            shuffle=True,
            random_state=config.seed,
        )

    return StratifiedKFold(
        n_splits=config.n_splits_screen,
        shuffle=True,
        random_state=config.seed,
    )


dev_df, holdout_df = make_dev_holdout_split(
    df=df,
    target_col=CFG.target_col,
    holdout_size=CFG.holdout_size,
    seed=CFG.seed,
)

X_dev = dev_df[FEATURES].copy()
y_dev = dev_df[CFG.target_col].astype(int).values

X_holdout = holdout_df[FEATURES].copy()
y_holdout = holdout_df[CFG.target_col].astype(int).values

split_summary = {
    "dev_n": int(len(dev_df)),
    "holdout_n": int(len(holdout_df)),
    "dev_prevalence": float(y_dev.mean()),
    "holdout_prevalence": float(y_holdout.mean()),
    "n_features": int(len(FEATURES)),
}

save_json(split_summary, META_DIR / "split_summary.json")

print(split_summary)
print("Module 2 complete.")

# ============================================================
# Module 3 — Metrics and utility helpers
# ============================================================

def clip_probs(p: np.ndarray, eps: float = 1e-6) -> np.ndarray:
    return np.clip(np.asarray(p, dtype=float), eps, 1.0 - eps)


def sigmoid_from_scores(scores: np.ndarray) -> np.ndarray:
    scores = np.asarray(scores, dtype=float)
    return clip_probs(1.0 / (1.0 + np.exp(-scores)))


def expected_calibration_error(
    y_true: np.ndarray,
    y_prob: np.ndarray,
    n_bins: int = 15,
) -> float:
    y_true = np.asarray(y_true).astype(int)
    y_prob = clip_probs(np.asarray(y_prob))

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.digitize(y_prob, bins) - 1

    ece = 0.0
    for b in range(n_bins):
        mask = (bin_ids == b)
        if mask.sum() == 0:
            continue
        acc = y_true[mask].mean()
        conf = y_prob[mask].mean()
        ece += (mask.sum() / len(y_true)) * abs(acc - conf)

    return float(ece)


def evaluate_probs(y_true: np.ndarray, y_prob: np.ndarray, name: str) -> Dict[str, Any]:
    y_true = np.asarray(y_true).astype(int)
    y_prob = clip_probs(np.asarray(y_prob))

    best_thr = 0.50
    best_f1 = -1.0
    best_y_hat = None

    for thr in np.linspace(0.05, 0.95, 181):
        y_hat = (y_prob >= thr).astype(int)
        f1 = f1_score(y_true, y_hat, zero_division=0)
        if f1 > best_f1:
            best_f1 = f1
            best_thr = float(thr)
            best_y_hat = y_hat

    tn, fp, fn, tp = confusion_matrix(y_true, best_y_hat).ravel()

    row = {
        "model": name,
        "PR_AUC": float(average_precision_score(y_true, y_prob)),
        "ROC_AUC": float(roc_auc_score(y_true, y_prob)),
        "Brier": float(brier_score_loss(y_true, y_prob)),
        "NLL": float(log_loss(y_true, y_prob)),
        "ECE": float(expected_calibration_error(y_true, y_prob)),
        "Precision": float(precision_score(y_true, best_y_hat, zero_division=0)),
        "Recall": float(recall_score(y_true, best_y_hat, zero_division=0)),
        "F1": float(f1_score(y_true, best_y_hat, zero_division=0)),
        "Accuracy": float(accuracy_score(y_true, best_y_hat)),
        "MCC": float(matthews_corrcoef(y_true, best_y_hat)),
        "Threshold": float(best_thr),
        "MeanProb": float(np.mean(y_prob)),
        "PosRatePred": float(np.mean(best_y_hat)),
        "N": int(len(y_true)),
        "TN": int(tn),
        "FP": int(fp),
        "FN": int(fn),
        "TP": int(tp),
    }
    return row


def correlation_matrix_from_preds(pred_dict: Dict[str, np.ndarray]) -> pd.DataFrame:
    return pd.DataFrame({k: np.asarray(v) for k, v in pred_dict.items()}).corr(method="pearson")


def summarize_runtime(seconds_value: float) -> str:
    if seconds_value < 60:
        return f"{seconds_value:.2f}s"
    return f"{seconds_value / 60.0:.2f}m"


# quick sanity check
_dummy_y = np.random.binomial(1, 0.1, size=1000)
_dummy_p = clip_probs(np.random.uniform(0.01, 0.99, size=1000))
display(pd.DataFrame([evaluate_probs(_dummy_y, _dummy_p, "sanity_check")]))

print("Module 3 complete.")

# ============================================================
# Module 4 — Preprocessing builders + candidate baseline registry
# We start with sklearn-safe models first.
# Then we optionally extend with LightGBM / XGBoost / CatBoost.
# ============================================================

from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
)

# optional imports
HAS_LIGHTGBM = False
HAS_XGBOOST = False
HAS_CATBOOST = False

try:
    from lightgbm import LGBMClassifier
    HAS_LIGHTGBM = True
except Exception:
    pass

try:
    from xgboost import XGBClassifier
    HAS_XGBOOST = True
except Exception:
    pass

try:
    from catboost import CatBoostClassifier
    HAS_CATBOOST = True
except Exception:
    pass


def build_linear_preprocessor(numeric_cols: List[str]) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", RobustScaler()),
    ])
    return ColumnTransformer([
        ("num", num_pipe, numeric_cols),
    ], remainder="drop")


def build_tree_preprocessor(numeric_cols: List[str]) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
    ])
    return ColumnTransformer([
        ("num", num_pipe, numeric_cols),
    ], remainder="drop")


def make_candidate_model_registry(seed: int) -> Dict[str, Dict[str, Any]]:
    registry: Dict[str, Dict[str, Any]] = {}

    # -------------------------
    # Linear
    # -------------------------
    registry["logreg"] = {
        "family": "linear",
        "preproc": "linear",
        "needs_calibration": False,
        "estimator": LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=seed,
            solver="saga",
            penalty="elasticnet",
            l1_ratio=0.5,
            n_jobs=-1,
        ),
    }

    registry["ridge"] = {
        "family": "linear",
        "preproc": "linear",
        "needs_calibration": True,
        "estimator": RidgeClassifier(
            class_weight="balanced",
            random_state=seed,
        ),
    }

    # -------------------------
    # Bagging / tree
    # -------------------------
    registry["rf"] = {
        "family": "tree",
        "preproc": "tree",
        "needs_calibration": False,
        "estimator": RandomForestClassifier(
            n_estimators=300,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced_subsample",
            random_state=seed,
            n_jobs=-1,
        ),
    }

    registry["et"] = {
        "family": "tree",
        "preproc": "tree",
        "needs_calibration": False,
        "estimator": ExtraTreesClassifier(
            n_estimators=400,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=seed,
            n_jobs=-1,
        ),
    }

    registry["hgb"] = {
        "family": "tree",
        "preproc": "tree",
        "needs_calibration": False,
        "estimator": HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_depth=8,
            max_iter=300,
            random_state=seed,
        ),
    }

    # -------------------------
    # Boosting (optional)
    # -------------------------
    if HAS_LIGHTGBM:
        registry["lgbm"] = {
            "family": "tree",
            "preproc": "tree",
            "needs_calibration": False,
            "estimator": LGBMClassifier(
                n_estimators=500,
                learning_rate=0.03,
                num_leaves=63,
                subsample=0.85,
                colsample_bytree=0.85,
                random_state=seed,
                n_jobs=-1,
                objective="binary",
            ),
        }

    if HAS_XGBOOST:
        registry["xgb"] = {
            "family": "tree",
            "preproc": "tree",
            "needs_calibration": False,
            "estimator": XGBClassifier(
                n_estimators=500,
                learning_rate=0.03,
                max_depth=6,
                subsample=0.85,
                colsample_bytree=0.85,
                eval_metric="logloss",
                random_state=seed,
                n_jobs=-1,
                tree_method="hist",
            ),
        }

    if HAS_CATBOOST:
        registry["catboost"] = {
            "family": "tree",
            "preproc": "tree",
            "needs_calibration": False,
            "estimator": CatBoostClassifier(
                iterations=50,
                learning_rate=0.5,
                depth=5,
                loss_function="Logloss",
                eval_metric="AUC",
                verbose=False,
                random_seed=seed,
            ),
        }

    return registry


CANDIDATE_MODEL_REGISTRY = make_candidate_model_registry(CFG.seed)

registry_view = pd.DataFrame([
    {
        "model_name": k,
        "family": v["family"],
        "preproc": v["preproc"],
        "needs_calibration": v["needs_calibration"],
        "estimator": str(v["estimator"])[:150],
    }
    for k, v in CANDIDATE_MODEL_REGISTRY.items()
])

registry_view.to_csv(TAB_DIR / "candidate_model_registry.csv", index=False)

display(registry_view)
print("Number of registered candidate models:", len(CANDIDATE_MODEL_REGISTRY))
print("Module 4 complete.")

# ============================================================
# Module 5 — Candidate baseline screening engine
# Run baselines first, evaluate them, then choose experts later
# ============================================================

def build_preprocessor_by_tag(tag: str, numeric_cols: List[str]) -> ColumnTransformer:
    if tag == "linear":
        return build_linear_preprocessor(numeric_cols)
    return build_tree_preprocessor(numeric_cols)


def get_model_probabilities(fitted_pipeline, X_val: pd.DataFrame) -> np.ndarray:
    model = fitted_pipeline

    if hasattr(model, "predict_proba"):
        p = model.predict_proba(X_val)[:, 1]
        return clip_probs(p)

    if hasattr(model, "decision_function"):
        scores = model.decision_function(X_val)
        return sigmoid_from_scores(scores)

    preds = model.predict(X_val)
    return clip_probs(np.asarray(preds, dtype=float))


def run_baseline_screening(
    X_df: pd.DataFrame,
    y: np.ndarray,
    registry: Dict[str, Dict[str, Any]],
    config: ExperimentConfig,
) -> Tuple[pd.DataFrame, Dict[str, np.ndarray], pd.DataFrame]:
    cv = get_cv(config, final_stage=False)
    numeric_cols = list(X_df.columns)

    screening_rows: List[Dict[str, Any]] = []
    oof_pred_store: Dict[str, np.ndarray] = {}
    runtime_rows: List[Dict[str, Any]] = []

    for model_name, meta in registry.items():
        print("=" * 80)
        print(f"Screening model: {model_name}")
        print("=" * 80)

        start_t = time.time()
        oof = np.zeros(len(y), dtype=float)
        fold_rows: List[Dict[str, Any]] = []

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_df, y), start=1):
            X_tr_df = X_df.iloc[tr_idx].copy()
            X_va_df = X_df.iloc[va_idx].copy()
            y_tr = y[tr_idx]
            y_va = y[va_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)
            estimator = deepcopy(meta["estimator"])

            pipe = Pipeline([
                ("pre", preproc),
                ("model", estimator),
            ])

            pipe.fit(X_tr_df, y_tr)
            p_va = get_model_probabilities(pipe, X_va_df)

            oof[va_idx] = p_va

            fold_row = evaluate_probs(y_va, p_va, name=f"{model_name}_fold{fold_id}")
            fold_row["model_name"] = model_name
            fold_row["family"] = meta["family"]
            fold_row["fold"] = fold_id
            fold_rows.append(fold_row)

            print(
                f"Fold {fold_id} | "
                f"PR_AUC={fold_row['PR_AUC']:.6f} | "
                f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
                f"Brier={fold_row['Brier']:.6f} | "
                f"ECE={fold_row['ECE']:.6f}"
            )

            del pipe, estimator, preproc, X_tr_df, X_va_df, y_tr, y_va
            gc.collect()

        runtime_seconds = time.time() - start_t

        summary = evaluate_probs(y, oof, name=model_name)
        summary["model_name"] = model_name
        summary["family"] = meta["family"]
        summary["runtime_seconds"] = float(runtime_seconds)
        summary["runtime_pretty"] = summarize_runtime(runtime_seconds)
        summary["needs_calibration"] = bool(meta.get("needs_calibration", False))

        screening_rows.append(summary)
        oof_pred_store[model_name] = oof

        runtime_rows.append({
            "model_name": model_name,
            "family": meta["family"],
            "runtime_seconds": float(runtime_seconds),
            "runtime_pretty": summarize_runtime(runtime_seconds),
        })

        fold_df = pd.DataFrame(fold_rows)
        fold_df.to_csv(TAB_DIR / f"screen_{model_name}_folds.csv", index=False)

        pd.DataFrame({
            "y_true": y.astype(int),
            "y_prob": oof,
        }).to_csv(PRED_DIR / f"screen_{model_name}_oof.csv", index=False)

        print("-" * 80)
        print(
            f"OOF summary for {model_name} | "
            f"PR_AUC={summary['PR_AUC']:.6f} | "
            f"ROC_AUC={summary['ROC_AUC']:.6f} | "
            f"Brier={summary['Brier']:.6f} | "
            f"ECE={summary['ECE']:.6f} | "
            f"Runtime={summary['runtime_pretty']}"
        )
        print()

    screening_df = pd.DataFrame(screening_rows).sort_values(
        ["PR_AUC", "Brier", "ROC_AUC"],
        ascending=[False, True, False]
    ).reset_index(drop=True)

    runtime_df = pd.DataFrame(runtime_rows).sort_values("runtime_seconds").reset_index(drop=True)
    corr_df = correlation_matrix_from_preds(oof_pred_store)

    screening_df.to_csv(TAB_DIR / "screening_results.csv", index=False)
    runtime_df.to_csv(TAB_DIR / "screening_runtime.csv", index=False)
    corr_df.to_csv(TAB_DIR / "screening_oof_correlation.csv")

    return screening_df, oof_pred_store, corr_df


# -------------------------
# Run baseline screening
# -------------------------
screening_df, screening_oof_preds, screening_corr_df = run_baseline_screening(
    X_df=X_dev,
    y=y_dev,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

display(screening_df[[
    "model_name", "family", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"
]])

display(screening_corr_df.round(4))

print("Module 5 complete.")
print("Next: choose experts based on these actual screening results.")
```

```python
# ============================================================
# Module 4 — Extended preprocessing + candidate baseline registry
# Includes ML + DL candidate models for screening
# ============================================================

from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier,
)

# Optional classical boosters
HAS_LIGHTGBM = False
HAS_XGBOOST = False
HAS_CATBOOST = False

try:
    from lightgbm import LGBMClassifier
    HAS_LIGHTGBM = True
except Exception:
    pass

try:
    from xgboost import XGBClassifier
    HAS_XGBOOST = True
except Exception:
    pass

try:
    from catboost import CatBoostClassifier
    HAS_CATBOOST = True
except Exception:
    pass

# PyTorch
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Torch device:", DEVICE)

# Optional TabNet
HAS_TABNET = False
try:
    from pytorch_tabnet.tab_model import TabNetClassifier
    HAS_TABNET = True
except Exception:
    pass


# -------------------------
# Preprocessing builders
# -------------------------
def build_linear_preprocessor(numeric_cols: List[str]) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", RobustScaler()),
    ])
    return ColumnTransformer([
        ("num", num_pipe, numeric_cols),
    ], remainder="drop")


def build_tree_preprocessor(numeric_cols: List[str]) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
    ])
    return ColumnTransformer([
        ("num", num_pipe, numeric_cols),
    ], remainder="drop")


def build_dl_preprocessor(numeric_cols: List[str]) -> ColumnTransformer:
    num_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", RobustScaler()),
    ])
    return ColumnTransformer([
        ("num", num_pipe, numeric_cols),
    ], remainder="drop")


# -------------------------
# PyTorch dataset
# -------------------------
class TabularTorchDataset(Dataset):
    def __init__(self, X: np.ndarray, y: Optional[np.ndarray] = None):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = None if y is None else torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx: int):
        if self.y is None:
            return self.X[idx]
        return self.X[idx], self.y[idx]


# -------------------------
# Simple deep models for screening
# -------------------------
class MLPBinaryClassifier(nn.Module):
    def __init__(self, in_dim: int, hidden_dims: List[int] = [256, 128], dropout: float = 0.20):
        super().__init__()

        layers: List[nn.Module] = []
        prev = in_dim
        for h in hidden_dims:
            layers.extend([
                nn.Linear(prev, h),
                nn.BatchNorm1d(h),
                nn.ReLU(),
                nn.Dropout(dropout),
            ])
            prev = h

        layers.append(nn.Linear(prev, 1))
        self.net = nn.Sequential(*layers)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x).squeeze(1)


class TabResNetBinaryClassifier(nn.Module):
    def __init__(self, in_dim: int, hidden_dim: int = 256, n_blocks: int = 3, dropout: float = 0.15):
        super().__init__()
        self.input_proj = nn.Linear(in_dim, hidden_dim)

        self.blocks = nn.ModuleList()
        for _ in range(n_blocks):
            self.blocks.append(nn.Sequential(
                nn.BatchNorm1d(hidden_dim),
                nn.ReLU(),
                nn.Linear(hidden_dim, hidden_dim),
                nn.BatchNorm1d(hidden_dim),
                nn.ReLU(),
                nn.Dropout(dropout),
                nn.Linear(hidden_dim, hidden_dim),
            ))

        self.head = nn.Linear(hidden_dim, 1)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        h = self.input_proj(x)
        for block in self.blocks:
            h = h + block(h)
        out = self.head(h).squeeze(1)
        return out


# -------------------------
# Torch training helpers
# -------------------------
def make_torch_loaders(
    X_tr: np.ndarray,
    y_tr: np.ndarray,
    X_va: np.ndarray,
    y_va: np.ndarray,
    batch_size: int = 1024,
) -> Tuple[DataLoader, DataLoader]:
    ds_tr = TabularTorchDataset(X_tr, y_tr)
    ds_va = TabularTorchDataset(X_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va


def train_torch_binary_model(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 1e-3,
    weight_decay: float = 1e-5,
    max_epochs: int = 20,
    patience: int = 4,
    pos_weight: Optional[float] = None,
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        train_losses = []

        for xb, yb in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            optimizer.step()
            train_losses.append(loss.item())

        model.eval()
        valid_losses = []

        with torch.no_grad():
            for xb, yb in valid_loader:
                xb = xb.to(device)
                yb = yb.to(device)
                logits = model(xb)
                loss = criterion(logits, yb)
                valid_losses.append(loss.item())

        mean_train = float(np.mean(train_losses))
        mean_valid = float(np.mean(valid_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_train:.5f} | valid={mean_valid:.5f} | best={best_val:.5f}")

        if mean_valid < best_val - 1e-6:
            best_val = mean_valid
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    model = model.to(device)
    return model


@torch.no_grad()
def predict_torch_binary_model(
    model: nn.Module,
    X: np.ndarray,
    batch_size: int = 2048,
    device: str = DEVICE,
) -> np.ndarray:
    model.eval()
    ds = TabularTorchDataset(X, None)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs = []
    for xb in dl:
        xb = xb.to(device)
        logits = model(xb)
        probs = torch.sigmoid(logits).detach().cpu().numpy()
        all_probs.append(probs)

    probs = np.concatenate(all_probs)
    return clip_probs(probs)


# -------------------------
# Candidate registry
# -------------------------
def make_candidate_model_registry(seed: int) -> Dict[str, Dict[str, Any]]:
    registry: Dict[str, Dict[str, Any]] = {}

    # Linear
    registry["logreg"] = {
        "family": "linear",
        "preproc": "linear",
        "kind": "sklearn",
        "needs_calibration": False,
        "estimator": LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=seed,
            solver="saga",
            penalty="elasticnet",
            l1_ratio=0.5,
            n_jobs=-1,
        ),
    }

    registry["ridge"] = {
        "family": "linear",
        "preproc": "linear",
        "kind": "sklearn",
        "needs_calibration": True,
        "estimator": RidgeClassifier(
            class_weight="balanced",
            random_state=seed,
        ),
    }

    # Tree / bagging
    registry["rf"] = {
        "family": "tree",
        "preproc": "tree",
        "kind": "sklearn",
        "needs_calibration": False,
        "estimator": RandomForestClassifier(
            n_estimators=300,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced_subsample",
            random_state=seed,
            n_jobs=-1,
        ),
    }

    registry["et"] = {
        "family": "tree",
        "preproc": "tree",
        "kind": "sklearn",
        "needs_calibration": False,
        "estimator": ExtraTreesClassifier(
            n_estimators=400,
            max_depth=None,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=seed,
            n_jobs=-1,
        ),
    }

    registry["hgb"] = {
        "family": "tree",
        "preproc": "tree",
        "kind": "sklearn",
        "needs_calibration": False,
        "estimator": HistGradientBoostingClassifier(
            learning_rate=0.05,
            max_depth=8,
            max_iter=300,
            random_state=seed,
        ),
    }

    if HAS_LIGHTGBM:
        registry["lgbm"] = {
            "family": "tree",
            "preproc": "tree",
            "kind": "sklearn",
            "needs_calibration": False,
            "estimator": LGBMClassifier(
                n_estimators=500,
                learning_rate=0.03,
                num_leaves=63,
                subsample=0.85,
                colsample_bytree=0.85,
                random_state=seed,
                n_jobs=-1,
                objective="binary",
            ),
        }

    if HAS_XGBOOST:
        registry["xgb"] = {
            "family": "tree",
            "preproc": "tree",
            "kind": "sklearn",
            "needs_calibration": False,
            "estimator": XGBClassifier(
                n_estimators=500,
                learning_rate=0.03,
                max_depth=6,
                subsample=0.85,
                colsample_bytree=0.85,
                eval_metric="logloss",
                random_state=seed,
                n_jobs=-1,
                tree_method="hist",
            ),
        }

    if HAS_CATBOOST:
        registry["catboost"] = {
            "family": "tree",
            "preproc": "tree",
            "kind": "sklearn",
            "needs_calibration": False,
            "estimator": CatBoostClassifier(
                iterations=500,
                learning_rate=0.03,
                depth=6,
                loss_function="Logloss",
                eval_metric="AUC",
                verbose=False,
                random_seed=seed,
            ),
        }

    # Deep tabular candidates for screening
    registry["mlp"] = {
        "family": "deep",
        "preproc": "dl",
        "kind": "torch",
        "needs_calibration": False,
        "builder": lambda in_dim: MLPBinaryClassifier(
            in_dim=in_dim,
            hidden_dims=[256, 128],
            dropout=0.20,
        ),
        "train_params": {
            "lr": 1e-3,
            "weight_decay": 1e-5,
            "max_epochs": 8,
            "patience": 3,
            "batch_size": 1024,
        },
    }

    registry["tabresnet"] = {
        "family": "deep",
        "preproc": "dl",
        "kind": "torch",
        "needs_calibration": False,
        "builder": lambda in_dim: TabResNetBinaryClassifier(
            in_dim=in_dim,
            hidden_dim=256,
            n_blocks=3,
            dropout=0.15,
        ),
        "train_params": {
            "lr": 1e-3,
            "weight_decay": 1e-5,
            "max_epochs": 8,
            "patience": 4,
            "batch_size": 1024,
        },
    }

    if HAS_TABNET:
        registry["tabnet"] = {
            "family": "deep",
            "preproc": "dl",
            "kind": "tabnet",
            "needs_calibration": False,
            "builder": None,
            "train_params": {
                "max_epochs": 8,
                "patience": 3,
                "batch_size": 1024,
                "virtual_batch_size": 128,
            },
        }

    # Placeholders for later expansion
    # DeepFM, FT-Transformer, AutoInt, DCN-v2
    # Add them after the first DL screening pipeline is stable

    return registry


CANDIDATE_MODEL_REGISTRY = make_candidate_model_registry(CFG.seed)

registry_view = pd.DataFrame([
    {
        "model_name": k,
        "family": v["family"],
        "kind": v["kind"],
        "preproc": v["preproc"],
        "needs_calibration": v["needs_calibration"],
    }
    for k, v in CANDIDATE_MODEL_REGISTRY.items()
])

registry_view.to_csv(TAB_DIR / "candidate_model_registry_extended.csv", index=False)
display(registry_view)

print("Number of registered candidates:", len(CANDIDATE_MODEL_REGISTRY))
```

```python
# ============================================================
# Module 5 — Unified baseline screening engine
# Supports sklearn + torch + optional TabNet
# ============================================================

def build_preprocessor_by_tag(tag: str, numeric_cols: List[str]) -> ColumnTransformer:
    if tag == "linear":
        return build_linear_preprocessor(numeric_cols)
    if tag == "tree":
        return build_tree_preprocessor(numeric_cols)
    if tag == "dl":
        return build_dl_preprocessor(numeric_cols)
    raise ValueError(f"Unknown preproc tag: {tag}")


def get_sklearn_probabilities(fitted_pipeline, X_val: pd.DataFrame) -> np.ndarray:
    if hasattr(fitted_pipeline, "predict_proba"):
        p = fitted_pipeline.predict_proba(X_val)[:, 1]
        return clip_probs(p)

    if hasattr(fitted_pipeline, "decision_function"):
        scores = fitted_pipeline.decision_function(X_val)
        return sigmoid_from_scores(scores)

    preds = fitted_pipeline.predict(X_val)
    return clip_probs(np.asarray(preds, dtype=float))


def run_one_sklearn_fold(
    estimator,
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
) -> np.ndarray:
    pipe = Pipeline([
        ("pre", preproc),
        ("model", deepcopy(estimator)),
    ])
    pipe.fit(X_tr_df, y_tr)
    p_va = get_sklearn_probabilities(pipe, X_va_df)
    return p_va


def run_one_torch_fold(
    model_builder,
    train_params: Dict[str, Any],
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
    y_va: np.ndarray,
) -> np.ndarray:
    X_tr = preproc.fit_transform(X_tr_df)
    X_va = preproc.transform(X_va_df)

    if hasattr(X_tr, "toarray"):
        X_tr = X_tr.toarray()
    if hasattr(X_va, "toarray"):
        X_va = X_va.toarray()

    X_tr = np.asarray(X_tr, dtype=np.float32)
    X_va = np.asarray(X_va, dtype=np.float32)

    pos = float((y_tr == 1).sum())
    neg = float((y_tr == 0).sum())
    pos_weight = neg / max(pos, 1.0)

    dl_tr, dl_va = make_torch_loaders(
        X_tr=X_tr,
        y_tr=y_tr.astype(np.float32),
        X_va=X_va,
        y_va=y_va.astype(np.float32),
        batch_size=int(train_params.get("batch_size", 1024)),
    )

    model = model_builder(X_tr.shape[1])
    model = train_torch_binary_model(
        model=model,
        train_loader=dl_tr,
        valid_loader=dl_va,
        device=DEVICE,
        lr=float(train_params.get("lr", 1e-3)),
        weight_decay=float(train_params.get("weight_decay", 1e-5)),
        max_epochs=int(train_params.get("max_epochs", 20)),
        patience=int(train_params.get("patience", 4)),
        pos_weight=pos_weight,
    )

    p_va = predict_torch_binary_model(
        model=model,
        X=X_va,
        batch_size=int(train_params.get("batch_size", 2048)),
        device=DEVICE,
    )

    del model, dl_tr, dl_va, X_tr, X_va
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return p_va


def run_one_tabnet_fold(
    train_params: Dict[str, Any],
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
    y_va: np.ndarray,
) -> np.ndarray:
    X_tr = preproc.fit_transform(X_tr_df)
    X_va = preproc.transform(X_va_df)

    if hasattr(X_tr, "toarray"):
        X_tr = X_tr.toarray()
    if hasattr(X_va, "toarray"):
        X_va = X_va.toarray()

    X_tr = np.asarray(X_tr, dtype=np.float32)
    X_va = np.asarray(X_va, dtype=np.float32)

    clf = TabNetClassifier(
        n_d=16,
        n_a=16,
        n_steps=4,
        gamma=1.5,
        lambda_sparse=1e-4,
        optimizer_fn=torch.optim.Adam,
        optimizer_params=dict(lr=2e-2),
        mask_type="entmax",
        seed=CFG.seed,
        verbose=0,
    )

    clf.fit(
        X_train=X_tr,
        y_train=y_tr.astype(int),
        eval_set=[(X_va, y_va.astype(int))],
        eval_name=["valid"],
        eval_metric=["auc"],
        max_epochs=int(train_params.get("max_epochs", 30)),
        patience=int(train_params.get("patience", 5)),
        batch_size=int(train_params.get("batch_size", 1024)),
        virtual_batch_size=int(train_params.get("virtual_batch_size", 128)),
        num_workers=0,
        drop_last=False,
    )

    p_va = clf.predict_proba(X_va)[:, 1]
    p_va = clip_probs(p_va)

    del clf, X_tr, X_va
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return p_va


def run_baseline_screening(
    X_df: pd.DataFrame,
    y: np.ndarray,
    registry: Dict[str, Dict[str, Any]],
    config: ExperimentConfig,
) -> Tuple[pd.DataFrame, Dict[str, np.ndarray], pd.DataFrame]:
    cv = get_cv(config, final_stage=False)
    numeric_cols = list(X_df.columns)

    screening_rows: List[Dict[str, Any]] = []
    oof_pred_store: Dict[str, np.ndarray] = {}
    runtime_rows: List[Dict[str, Any]] = []

    for model_name, meta in registry.items():
        print("=" * 90)
        print(f"Screening model: {model_name} | family={meta['family']} | kind={meta['kind']}")
        print("=" * 90)

        start_t = time.time()
        oof = np.zeros(len(y), dtype=float)
        fold_rows: List[Dict[str, Any]] = []

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_df, y), start=1):
            print(f"[{model_name}] Fold {fold_id}")

            X_tr_df = X_df.iloc[tr_idx].copy()
            X_va_df = X_df.iloc[va_idx].copy()
            y_tr = y[tr_idx]
            y_va = y[va_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            if meta["kind"] == "sklearn":
                p_va = run_one_sklearn_fold(
                    estimator=meta["estimator"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                )

            elif meta["kind"] == "torch":
                p_va = run_one_torch_fold(
                    model_builder=meta["builder"],
                    train_params=meta["train_params"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                    y_va=y_va,
                )

            elif meta["kind"] == "tabnet":
                p_va = run_one_tabnet_fold(
                    train_params=meta["train_params"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                    y_va=y_va,
                )

            else:
                raise ValueError(f"Unsupported model kind: {meta['kind']}")

            oof[va_idx] = p_va

            fold_row = evaluate_probs(y_va, p_va, name=f"{model_name}_fold{fold_id}")
            fold_row["model_name"] = model_name
            fold_row["family"] = meta["family"]
            fold_row["kind"] = meta["kind"]
            fold_row["fold"] = fold_id
            fold_rows.append(fold_row)

            print(
                f"Fold {fold_id} | "
                f"PR_AUC={fold_row['PR_AUC']:.6f} | "
                f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
                f"Brier={fold_row['Brier']:.6f} | "
                f"ECE={fold_row['ECE']:.6f}"
            )

            del X_tr_df, X_va_df, y_tr, y_va, preproc
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        runtime_seconds = time.time() - start_t

        summary = evaluate_probs(y, oof, name=model_name)
        summary["model_name"] = model_name
        summary["family"] = meta["family"]
        summary["kind"] = meta["kind"]
        summary["runtime_seconds"] = float(runtime_seconds)
        summary["runtime_pretty"] = summarize_runtime(runtime_seconds)
        summary["needs_calibration"] = bool(meta.get("needs_calibration", False))

        screening_rows.append(summary)
        oof_pred_store[model_name] = oof

        runtime_rows.append({
            "model_name": model_name,
            "family": meta["family"],
            "kind": meta["kind"],
            "runtime_seconds": float(runtime_seconds),
            "runtime_pretty": summarize_runtime(runtime_seconds),
        })

        pd.DataFrame(fold_rows).to_csv(TAB_DIR / f"screen_{model_name}_folds.csv", index=False)

        pd.DataFrame({
            "y_true": y.astype(int),
            "y_prob": oof,
        }).to_csv(PRED_DIR / f"screen_{model_name}_oof.csv", index=False)

        print("-" * 90)
        print(
            f"OOF summary for {model_name} | "
            f"PR_AUC={summary['PR_AUC']:.6f} | "
            f"ROC_AUC={summary['ROC_AUC']:.6f} | "
            f"Brier={summary['Brier']:.6f} | "
            f"ECE={summary['ECE']:.6f} | "
            f"Runtime={summary['runtime_pretty']}"
        )
        print()

    screening_df = pd.DataFrame(screening_rows).sort_values(
        ["PR_AUC", "Brier", "ROC_AUC"],
        ascending=[False, True, False]
    ).reset_index(drop=True)

    runtime_df = pd.DataFrame(runtime_rows).sort_values("runtime_seconds").reset_index(drop=True)
    corr_df = correlation_matrix_from_preds(oof_pred_store)

    screening_df.to_csv(TAB_DIR / "screening_results.csv", index=False)
    runtime_df.to_csv(TAB_DIR / "screening_runtime.csv", index=False)
    corr_df.to_csv(TAB_DIR / "screening_oof_correlation.csv")

    return screening_df, oof_pred_store, corr_df


# -------------------------
# Run screening
# -------------------------
screening_df, screening_oof_preds, screening_corr_df = run_baseline_screening(
    X_df=X_dev,
    y=y_dev,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

display(
    screening_df[
        ["model_name", "family", "kind", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"]
    ]
)

display(screening_corr_df.round(4))

print("Extended Module 5 complete.")
print("Now we will choose experts only after reviewing these baseline results.")
```

```python
# ============================================================
# Module 4B — Add DeepFM + FT-Transformer screening models
# ============================================================

class DeepFMBinaryClassifier(nn.Module):
    """
    Simplified dense-input DeepFM for continuous / already-processed tabular features.
    Uses:
      - linear term
      - FM second-order interaction term
      - deep MLP branch
    """
    def __init__(
        self,
        in_dim: int,
        embed_dim: int = 16,
        deep_hidden_dims: List[int] = [256, 128],
        dropout: float = 0.20,
    ):
        super().__init__()
        self.in_dim = in_dim
        self.embed_dim = embed_dim

        # linear term
        self.linear = nn.Linear(in_dim, 1)

        # FM embeddings: one embedding vector per feature
        self.feature_embed = nn.Parameter(torch.randn(in_dim, embed_dim) * 0.01)

        # deep branch
        deep_layers: List[nn.Module] = []
        prev_dim = in_dim
        for h in deep_hidden_dims:
            deep_layers.extend([
                nn.Linear(prev_dim, h),
                nn.BatchNorm1d(h),
                nn.ReLU(),
                nn.Dropout(dropout),
            ])
            prev_dim = h
        self.deep = nn.Sequential(*deep_layers)
        self.deep_out = nn.Linear(prev_dim, 1)

    def fm_interaction(self, x: torch.Tensor) -> torch.Tensor:
        # x: [B, D]
        # embeddings: [D, K]
        xv = x.unsqueeze(-1) * self.feature_embed.unsqueeze(0)   # [B, D, K]
        sum_square = torch.sum(xv, dim=1) ** 2                   # [B, K]
        square_sum = torch.sum(xv ** 2, dim=1)                   # [B, K]
        interactions = 0.5 * torch.sum(sum_square - square_sum, dim=1, keepdim=True)  # [B, 1]
        return interactions

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        linear_term = self.linear(x)                  # [B, 1]
        fm_term = self.fm_interaction(x)              # [B, 1]
        deep_term = self.deep_out(self.deep(x))       # [B, 1]
        logits = linear_term + fm_term + deep_term
        return logits.squeeze(1)

# ============================================================
# FT-Transformer building blocks
# ============================================================

class TransformerBlock(nn.Module):
    def __init__(
        self,
        dim: int,
        n_heads: int = 8,
        attn_dropout: float = 0.10,
        ff_dropout: float = 0.10,
        ff_mult: int = 4,
    ):
        super().__init__()
        self.norm1 = nn.LayerNorm(dim)
        self.attn = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=n_heads,
            dropout=attn_dropout,
            batch_first=True,
        )
        self.norm2 = nn.LayerNorm(dim)

        self.ff = nn.Sequential(
            nn.Linear(dim, dim * ff_mult),
            nn.GELU(),
            nn.Dropout(ff_dropout),
            nn.Linear(dim * ff_mult, dim),
            nn.Dropout(ff_dropout),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        h = self.norm1(x)
        attn_out, _ = self.attn(h, h, h, need_weights=False)
        x = x + attn_out
        x = x + self.ff(self.norm2(x))
        return x


class FTTransformerBinaryClassifier(nn.Module):
    """
    Simplified FT-Transformer for dense numeric features.
    Each numeric feature gets a token via learned scale projection.
    """
    def __init__(
        self,
        in_dim: int,
        token_dim: int = 32,
        n_blocks: int = 3,
        n_heads: int = 8,
        dropout: float = 0.10,
    ):
        super().__init__()
        self.in_dim = in_dim
        self.token_dim = token_dim

        # CLS token
        self.cls_token = nn.Parameter(torch.zeros(1, 1, token_dim))

        # One learnable vector per feature to convert scalar -> token
        self.feature_token_weight = nn.Parameter(torch.randn(in_dim, token_dim) * 0.02)
        self.feature_token_bias = nn.Parameter(torch.zeros(in_dim, token_dim))

        self.blocks = nn.ModuleList([
            TransformerBlock(
                dim=token_dim,
                n_heads=n_heads,
                attn_dropout=dropout,
                ff_dropout=dropout,
                ff_mult=4,
            )
            for _ in range(n_blocks)
        ])

        self.head = nn.Sequential(
            nn.LayerNorm(token_dim),
            nn.Linear(token_dim, token_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(token_dim, 1),
        )

    def numeric_to_tokens(self, x: torch.Tensor) -> torch.Tensor:
        # x: [B, D]
        # token_weight: [D, T]
        tokens = x.unsqueeze(-1) * self.feature_token_weight.unsqueeze(0) + self.feature_token_bias.unsqueeze(0)
        # output: [B, D, T]
        return tokens

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        bsz = x.size(0)
        feat_tokens = self.numeric_to_tokens(x)                     # [B, D, T]
        cls = self.cls_token.expand(bsz, -1, -1)                   # [B, 1, T]
        h = torch.cat([cls, feat_tokens], dim=1)                   # [B, 1+D, T]

        for block in self.blocks:
            h = block(h)

        cls_out = h[:, 0, :]                                       # [B, T]
        logits = self.head(cls_out).squeeze(1)
        return logits

# ============================================================
# Register DeepFM + FT-Transformer
# ============================================================

CANDIDATE_MODEL_REGISTRY["deepfm"] = {
    "family": "deep",
    "preproc": "dl",
    "kind": "torch",
    "needs_calibration": False,
    "builder": lambda in_dim: DeepFMBinaryClassifier(
        in_dim=in_dim,
        embed_dim=16,
        deep_hidden_dims=[256, 128],
        dropout=0.20,
    ),
    "train_params": {
        "lr": 8e-4,
        "weight_decay": 1e-5,
        "max_epochs": 18,
        "patience": 4,
        "batch_size": 1024,
    },
}

CANDIDATE_MODEL_REGISTRY["fttransformer"] = {
    "family": "deep",
    "preproc": "dl",
    "kind": "torch",
    "needs_calibration": False,
    "builder": lambda in_dim: FTTransformerBinaryClassifier(
        in_dim=in_dim,
        token_dim=32,
        n_blocks=3,
        n_heads=8,
        dropout=0.10,
    ),
    "train_params": {
        "lr": 7e-4,
        "weight_decay": 1e-5,
        "max_epochs": 16,
        "patience": 4,
        "batch_size": 768,
    },
}

registry_view = pd.DataFrame([
    {
        "model_name": k,
        "family": v["family"],
        "kind": v["kind"],
        "preproc": v["preproc"],
        "needs_calibration": v["needs_calibration"],
    }
    for k, v in CANDIDATE_MODEL_REGISTRY.items()
]).sort_values(["family", "model_name"]).reset_index(drop=True)

registry_view.to_csv(TAB_DIR / "candidate_model_registry_extended_v2.csv", index=False)
display(registry_view)

print("DeepFM and FT-Transformer added.")

# ============================================================
# Module 5B — Run screening with expanded candidate pool
# Includes ML + DL + DeepFM + FT-Transformer
# ============================================================

screening_df, screening_oof_preds, screening_corr_df = run_baseline_screening(
    X_df=X_dev,
    y=y_dev,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

screening_df = screening_df.sort_values(
    ["PR_AUC", "Brier", "ROC_AUC"],
    ascending=[False, True, False]
).reset_index(drop=True)

screening_df.to_csv(TAB_DIR / "screening_results_full_pool.csv", index=False)
screening_corr_df.to_csv(TAB_DIR / "screening_oof_correlation_full_pool.csv")

display(
    screening_df[
        ["model_name", "family", "kind", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"]
    ]
)

display(screening_corr_df.round(4))

print("Full screening complete.")
print("Next step: expert selection based on actual ML + DL screening results.")
```

```python
# ============================================================
# Module 6 — Expert selection from screened ML + DL baselines
# ============================================================

def select_final_experts(
    screening_df: pd.DataFrame,
    pred_store: Dict[str, np.ndarray],
    corr_df: pd.DataFrame,
    registry: Dict[str, Dict[str, Any]],
    config: ExperimentConfig,
) -> Tuple[List[str], pd.DataFrame]:
    selected: List[str] = []
    selection_rows: List[Dict[str, Any]] = []

    screening_df = screening_df.copy().sort_values(
        ["PR_AUC", "Brier", "ROC_AUC"],
        ascending=[False, True, False]
    ).reset_index(drop=True)

    def is_redundant(candidate: str, selected_models: List[str]) -> Tuple[bool, Optional[str], Optional[float]]:
        for s in selected_models:
            corr_val = float(corr_df.loc[candidate, s])
            if corr_val >= config.corr_redundancy_threshold:
                return True, s, corr_val
        return False, None, None

    # Step 1: enforce family coverage
    for family, min_keep in config.min_keep_per_family.items():
        family_df = screening_df[screening_df["family"] == family].copy()

        kept_in_family = 0
        for _, row in family_df.iterrows():
            model_name = row["model_name"]
            redundant, against, corr_val = is_redundant(model_name, selected)

            if redundant:
                selection_rows.append({
                    "model_name": model_name,
                    "family": family,
                    "decision": "rejected_redundant",
                    "reason": f"corr_with_{against}={corr_val:.4f}",
                })
                continue

            selected.append(model_name)
            kept_in_family += 1
            selection_rows.append({
                "model_name": model_name,
                "family": family,
                "decision": "selected_family_requirement",
                "reason": f"top_available_{family}",
            })

            if kept_in_family >= min_keep:
                break

    # Step 2: fill remaining preferred slots by strength + non-redundancy
    for _, row in screening_df.iterrows():
        model_name = row["model_name"]
        family = row["family"]

        if model_name in selected:
            continue
        if len(selected) >= config.preferred_final_experts:
            break

        redundant, against, corr_val = is_redundant(model_name, selected)
        if redundant:
            selection_rows.append({
                "model_name": model_name,
                "family": family,
                "decision": "rejected_redundant",
                "reason": f"corr_with_{against}={corr_val:.4f}",
            })
            continue

        selected.append(model_name)
        selection_rows.append({
            "model_name": model_name,
            "family": family,
            "decision": "selected_fill",
            "reason": "strong_non_redundant_candidate",
        })

    # Step 3: optional expansion up to max_final_experts
    for _, row in screening_df.iterrows():
        model_name = row["model_name"]
        family = row["family"]

        if model_name in selected:
            continue
        if len(selected) >= config.max_final_experts:
            break

        redundant, against, corr_val = is_redundant(model_name, selected)
        if redundant:
            continue

        selected.append(model_name)
        selection_rows.append({
            "model_name": model_name,
            "family": family,
            "decision": "selected_optional",
            "reason": "non_redundant_optional_expansion",
        })

    selection_df = pd.DataFrame(selection_rows)

    final_expert_df = screening_df[screening_df["model_name"].isin(selected)].copy()
    final_expert_df["selected_order"] = final_expert_df["model_name"].apply(lambda x: selected.index(x) + 1)
    final_expert_df = final_expert_df.sort_values("selected_order").reset_index(drop=True)

    final_expert_df.to_csv(TAB_DIR / "final_expert_registry.csv", index=False)
    selection_df.to_csv(TAB_DIR / "expert_selection_log.csv", index=False)

    return selected, final_expert_df


selected_experts, final_expert_df = select_final_experts(
    screening_df=screening_df,
    pred_store=screening_oof_preds,
    corr_df=screening_corr_df,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

print("Selected experts:", selected_experts)
display(final_expert_df[[
    "selected_order", "model_name", "family", "kind", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"
]])
```

```python
# ============================================================
# Module 6B — Expert Selection Evidence Module
# ============================================================

import seaborn as sns
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Family-wise winners (best per family)
# ------------------------------------------------------------
def get_family_winners(screening_df: pd.DataFrame) -> pd.DataFrame:
    winners = (
        screening_df.sort_values(
            ["family", "PR_AUC", "Brier"],
            ascending=[True, False, True]
        )
        .groupby("family")
        .head(1)
        .reset_index(drop=True)
    )
    winners.to_csv(TAB_DIR / "family_winners.csv", index=False)
    return winners


family_winners_df = get_family_winners(screening_df)

print("Family-wise best models:")
display(family_winners_df[
    ["model_name", "family", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
])


# ------------------------------------------------------------
# 2. Correlation Heatmap (OOF predictions)
# ------------------------------------------------------------
def plot_correlation_heatmap(corr_df: pd.DataFrame, save_path=None):
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        corr_df,
        annot=False,
        cmap="coolwarm",
        center=0.9,   # important for highlighting high correlations
        square=True
    )
    plt.title("OOF Prediction Correlation Matrix")
    plt.tight_layout()

    if save_path is not None:
        plt.savefig(save_path, dpi=300)

    plt.show()


plot_correlation_heatmap(
    screening_corr_df,
    save_path=FIG_DIR / "oof_correlation_heatmap.png"
)


# ------------------------------------------------------------
# 3. Selected vs Excluded Models Table
# ------------------------------------------------------------
def build_selection_rationale(
    screening_df: pd.DataFrame,
    selected_models: List[str],
    corr_df: pd.DataFrame,
) -> pd.DataFrame:
    rows = []

    for _, row in screening_df.iterrows():
        model = row["model_name"]
        family = row["family"]

        if model in selected_models:
            decision = "selected"
            reason = "Top-performing + contributes diversity"
        else:
            # find most correlated selected model
            if len(selected_models) > 0:
                corr_vals = [(m, corr_df.loc[model, m]) for m in selected_models]
                most_corr_model, corr_val = sorted(corr_vals, key=lambda x: -x[1])[0]

                if corr_val > 0.98:
                    reason = f"Highly redundant with {most_corr_model} (corr={corr_val:.3f})"
                else:
                    reason = "Lower performance than selected alternatives"
            else:
                reason = "Not selected"

            decision = "excluded"

        rows.append({
            "model_name": model,
            "family": family,
            "PR_AUC": row["PR_AUC"],
            "ROC_AUC": row["ROC_AUC"],
            "Brier": row["Brier"],
            "ECE": row["ECE"],
            "decision": decision,
            "reason": reason,
        })

    df = pd.DataFrame(rows).sort_values(
        ["decision", "PR_AUC"],
        ascending=[True, False]
    )

    df.to_csv(TAB_DIR / "expert_selection_rationale.csv", index=False)
    return df


selection_rationale_df = build_selection_rationale(
    screening_df,
    selected_models=selected_experts,
    corr_df=screening_corr_df,
)

print("Expert selection rationale:")
display(selection_rationale_df)


# ------------------------------------------------------------
# 4. Compact Screening Table (for paper)
# ------------------------------------------------------------
def build_paper_screening_table(screening_df: pd.DataFrame) -> pd.DataFrame:
    cols = [
        "model_name",
        "family",
        "PR_AUC",
        "ROC_AUC",
        "Brier",
        "ECE",
        "runtime_pretty",
    ]
    df = screening_df[cols].copy()
    df = df.sort_values("PR_AUC", ascending=False)

    df.to_csv(TAB_DIR / "paper_screening_table.csv", index=False)
    return df


paper_screening_df = build_paper_screening_table(screening_df)

print("Paper-ready screening table:")
display(paper_screening_df.head(10))


# ------------------------------------------------------------
# 5. Expert summary table (final shortlist)
# ------------------------------------------------------------
def build_expert_summary(final_expert_df: pd.DataFrame) -> pd.DataFrame:
    cols = [
        "selected_order",
        "model_name",
        "family",
        "PR_AUC",
        "ROC_AUC",
        "Brier",
        "ECE",
        "runtime_pretty",
    ]
    df = final_expert_df[cols].copy()

    df.to_csv(TAB_DIR / "final_expert_summary.csv", index=False)
    return df


expert_summary_df = build_expert_summary(final_expert_df)

print("Final expert summary:")
display(expert_summary_df)


# ------------------------------------------------------------
# 6. (Optional) Diversity score per model
# ------------------------------------------------------------
def compute_diversity_scores(corr_df: pd.DataFrame) -> pd.DataFrame:
    scores = []

    for model in corr_df.columns:
        mean_corr = corr_df[model].drop(model).mean()
        scores.append({
            "model_name": model,
            "mean_correlation": float(mean_corr),
            "diversity_score": float(1 - mean_corr),
        })

    df = pd.DataFrame(scores).sort_values("diversity_score", ascending=False)
    df.to_csv(TAB_DIR / "model_diversity_scores.csv", index=False)
    return df


diversity_df = compute_diversity_scores(screening_corr_df)

print("Model diversity scores (higher is better):")
display(diversity_df.head(10))


print("Module 6B complete — Expert selection is now fully justified with evidence.")
```

```python
# ============================================================
# Module 6C — Fixed Expert Selection Logic
# ============================================================

def select_final_experts_fixed(
    screening_df: pd.DataFrame,
    corr_df: pd.DataFrame,
    registry: Dict[str, Dict[str, Any]],
    config: ExperimentConfig,
    perf_drop_tol: float = 0.01,          # allow up to 0.01 PR-AUC drop from best model
    within_family_corr_thr: float = 0.985 # redundancy filter only within family
) -> Tuple[List[str], pd.DataFrame, pd.DataFrame]:
    """
    Expert selection policy:
      1. Always keep family winners (linear, tree, deep) if available.
      2. Apply redundancy filtering only WITHIN the same family.
      3. Add one strong complementary model if:
         - not redundant within family
         - performance is within tolerance of global best
      4. Never add weak models just for diversity.
    """

    df = screening_df.copy().sort_values(
        ["PR_AUC", "Brier", "ROC_AUC"],
        ascending=[False, True, False]
    ).reset_index(drop=True)

    global_best_pr = float(df["PR_AUC"].max())
    selected: List[str] = []
    log_rows: List[Dict[str, Any]] = []

    # --------------------------------------------------------
    # Step 1 — Keep family winners
    # --------------------------------------------------------
    family_winners = (
        df.sort_values(["family", "PR_AUC", "Brier"], ascending=[True, False, True])
          .groupby("family")
          .head(1)
          .reset_index(drop=True)
    )

    for _, row in family_winners.iterrows():
        model = row["model_name"]
        family = row["family"]
        selected.append(model)
        log_rows.append({
            "model_name": model,
            "family": family,
            "decision": "selected_family_winner",
            "reason": "Best model within family",
            "PR_AUC": row["PR_AUC"],
            "ROC_AUC": row["ROC_AUC"],
            "Brier": row["Brier"],
            "ECE": row["ECE"],
        })

    # --------------------------------------------------------
    # Step 2 — Add strong complementary models
    # Only if not redundant within family and not too weak
    # --------------------------------------------------------
    def is_within_family_redundant(candidate: str, family: str, already_selected: List[str]) -> Tuple[bool, Optional[str], Optional[float]]:
        for s in already_selected:
            if registry[s]["family"] != family:
                continue
            corr_val = float(corr_df.loc[candidate, s])
            if corr_val >= within_family_corr_thr:
                return True, s, corr_val
        return False, None, None

    for _, row in df.iterrows():
        model = row["model_name"]
        family = row["family"]
        pr_auc = float(row["PR_AUC"])

        if model in selected:
            continue

        # performance floor
        if pr_auc < (global_best_pr - perf_drop_tol):
            log_rows.append({
                "model_name": model,
                "family": family,
                "decision": "excluded_low_performance",
                "reason": f"PR_AUC below best-{perf_drop_tol:.3f}",
                "PR_AUC": row["PR_AUC"],
                "ROC_AUC": row["ROC_AUC"],
                "Brier": row["Brier"],
                "ECE": row["ECE"],
            })
            continue

        # within-family redundancy only
        redundant, against, corr_val = is_within_family_redundant(model, family, selected)
        if redundant:
            log_rows.append({
                "model_name": model,
                "family": family,
                "decision": "excluded_within_family_redundant",
                "reason": f"Highly redundant with {against} (corr={corr_val:.3f})",
                "PR_AUC": row["PR_AUC"],
                "ROC_AUC": row["ROC_AUC"],
                "Brier": row["Brier"],
                "ECE": row["ECE"],
            })
            continue

        if len(selected) < config.preferred_final_experts:
            selected.append(model)
            log_rows.append({
                "model_name": model,
                "family": family,
                "decision": "selected_complementary",
                "reason": "Strong non-redundant complementary model",
                "PR_AUC": row["PR_AUC"],
                "ROC_AUC": row["ROC_AUC"],
                "Brier": row["Brier"],
                "ECE": row["ECE"],
            })

    # --------------------------------------------------------
    # Final outputs
    # --------------------------------------------------------
    selected = selected[:config.max_final_experts]

    selected_df = df[df["model_name"].isin(selected)].copy()
    selected_df["selected_order"] = selected_df["model_name"].apply(lambda x: selected.index(x) + 1)
    selected_df = selected_df.sort_values("selected_order").reset_index(drop=True)

    decision_df = pd.DataFrame(log_rows).sort_values(
        ["decision", "PR_AUC"], ascending=[True, False]
    ).reset_index(drop=True)

    selected_df.to_csv(TAB_DIR / "final_expert_registry_fixed.csv", index=False)
    decision_df.to_csv(TAB_DIR / "expert_selection_log_fixed.csv", index=False)

    return selected, selected_df, decision_df


# -------------------------
# Run fixed selection
# -------------------------
selected_experts, final_expert_df, selection_log_df = select_final_experts_fixed(
    screening_df=screening_df,
    corr_df=screening_corr_df,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
    perf_drop_tol=0.01,
    within_family_corr_thr=0.985,
)

print("Selected experts:", selected_experts)
display(final_expert_df[[
    "selected_order", "model_name", "family", "kind", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"
]])
display(selection_log_df.head(20))
```

```python
selected_experts = ["logreg", "catboost", "hgb", "deepfm"]
```

```python
# ============================================================
# Module 7 — Final shortlisted expert training (OOF + HOLDOUT)
# ============================================================

def run_one_sklearn_fold_final(
    estimator,
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
    X_holdout_df: pd.DataFrame,
) -> Tuple[np.ndarray, np.ndarray]:
    pipe = Pipeline([
        ("pre", preproc),
        ("model", deepcopy(estimator)),
    ])
    pipe.fit(X_tr_df, y_tr)

    p_va = get_sklearn_probabilities(pipe, X_va_df)
    p_holdout = get_sklearn_probabilities(pipe, X_holdout_df)
    return p_va, p_holdout


def run_one_torch_fold_final(
    model_builder,
    train_params: Dict[str, Any],
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
    y_va: np.ndarray,
    X_holdout_df: pd.DataFrame,
) -> Tuple[np.ndarray, np.ndarray]:
    X_tr = preproc.fit_transform(X_tr_df)
    X_va = preproc.transform(X_va_df)
    X_holdout = preproc.transform(X_holdout_df)

    if hasattr(X_tr, "toarray"):
        X_tr = X_tr.toarray()
    if hasattr(X_va, "toarray"):
        X_va = X_va.toarray()
    if hasattr(X_holdout, "toarray"):
        X_holdout = X_holdout.toarray()

    X_tr = np.asarray(X_tr, dtype=np.float32)
    X_va = np.asarray(X_va, dtype=np.float32)
    X_holdout = np.asarray(X_holdout, dtype=np.float32)

    pos = float((y_tr == 1).sum())
    neg = float((y_tr == 0).sum())
    pos_weight = neg / max(pos, 1.0)

    dl_tr, dl_va = make_torch_loaders(
        X_tr=X_tr,
        y_tr=y_tr.astype(np.float32),
        X_va=X_va,
        y_va=y_va.astype(np.float32),
        batch_size=int(train_params.get("batch_size", 1024)),
    )

    model = model_builder(X_tr.shape[1])
    model = train_torch_binary_model(
        model=model,
        train_loader=dl_tr,
        valid_loader=dl_va,
        device=DEVICE,
        lr=float(train_params.get("lr", 1e-3)),
        weight_decay=float(train_params.get("weight_decay", 1e-5)),
        max_epochs=int(train_params.get("max_epochs", 20)),
        patience=int(train_params.get("patience", 4)),
        pos_weight=pos_weight,
    )

    p_va = predict_torch_binary_model(
        model=model,
        X=X_va,
        batch_size=int(train_params.get("batch_size", 2048)),
        device=DEVICE,
    )
    p_holdout = predict_torch_binary_model(
        model=model,
        X=X_holdout,
        batch_size=int(train_params.get("batch_size", 2048)),
        device=DEVICE,
    )

    del model, dl_tr, dl_va, X_tr, X_va, X_holdout
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return p_va, p_holdout


def run_one_tabnet_fold_final(
    train_params: Dict[str, Any],
    preproc,
    X_tr_df: pd.DataFrame,
    y_tr: np.ndarray,
    X_va_df: pd.DataFrame,
    y_va: np.ndarray,
    X_holdout_df: pd.DataFrame,
) -> Tuple[np.ndarray, np.ndarray]:
    X_tr = preproc.fit_transform(X_tr_df)
    X_va = preproc.transform(X_va_df)
    X_holdout = preproc.transform(X_holdout_df)

    if hasattr(X_tr, "toarray"):
        X_tr = X_tr.toarray()
    if hasattr(X_va, "toarray"):
        X_va = X_va.toarray()
    if hasattr(X_holdout, "toarray"):
        X_holdout = X_holdout.toarray()

    X_tr = np.asarray(X_tr, dtype=np.float32)
    X_va = np.asarray(X_va, dtype=np.float32)
    X_holdout = np.asarray(X_holdout, dtype=np.float32)

    clf = TabNetClassifier(
        n_d=16,
        n_a=16,
        n_steps=4,
        gamma=1.5,
        lambda_sparse=1e-4,
        optimizer_fn=torch.optim.Adam,
        optimizer_params=dict(lr=2e-2),
        mask_type="entmax",
        seed=CFG.seed,
        verbose=0,
    )

    clf.fit(
        X_train=X_tr,
        y_train=y_tr.astype(int),
        eval_set=[(X_va, y_va.astype(int))],
        eval_name=["valid"],
        eval_metric=["auc"],
        max_epochs=int(train_params.get("max_epochs", 30)),
        patience=int(train_params.get("patience", 5)),
        batch_size=int(train_params.get("batch_size", 1024)),
        virtual_batch_size=int(train_params.get("virtual_batch_size", 128)),
        num_workers=0,
        drop_last=False,
    )

    p_va = clip_probs(clf.predict_proba(X_va)[:, 1])
    p_holdout = clip_probs(clf.predict_proba(X_holdout)[:, 1])

    del clf, X_tr, X_va, X_holdout
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return p_va, p_holdout


def run_final_expert_models(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    X_holdout: pd.DataFrame,
    y_holdout: np.ndarray,
    selected_models: List[str],
    registry: Dict[str, Dict[str, Any]],
    config: ExperimentConfig,
) -> Tuple[pd.DataFrame, Dict[str, np.ndarray], Dict[str, np.ndarray], pd.DataFrame]:
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    result_rows: List[Dict[str, Any]] = []
    fold_rows_all: List[Dict[str, Any]] = []
    dev_pred_store: Dict[str, np.ndarray] = {}
    holdout_pred_store: Dict[str, np.ndarray] = {}

    for model_name in selected_models:
        meta = registry[model_name]
        print("=" * 100)
        print(f"Final expert run: {model_name} | family={meta['family']} | kind={meta['kind']}")
        print("=" * 100)

        start_t = time.time()
        oof = np.zeros(len(y_dev), dtype=float)
        holdout_fold_preds: List[np.ndarray] = []

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
            print(f"[{model_name}] Final fold {fold_id}")

            X_tr_df = X_dev.iloc[tr_idx].copy()
            X_va_df = X_dev.iloc[va_idx].copy()
            y_tr = y_dev[tr_idx]
            y_va = y_dev[va_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            if meta["kind"] == "sklearn":
                p_va, p_holdout = run_one_sklearn_fold_final(
                    estimator=meta["estimator"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                    X_holdout_df=X_holdout,
                )

            elif meta["kind"] == "torch":
                p_va, p_holdout = run_one_torch_fold_final(
                    model_builder=meta["builder"],
                    train_params=meta["train_params"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                    y_va=y_va,
                    X_holdout_df=X_holdout,
                )

            elif meta["kind"] == "tabnet":
                p_va, p_holdout = run_one_tabnet_fold_final(
                    train_params=meta["train_params"],
                    preproc=preproc,
                    X_tr_df=X_tr_df,
                    y_tr=y_tr,
                    X_va_df=X_va_df,
                    y_va=y_va,
                    X_holdout_df=X_holdout,
                )

            else:
                raise ValueError(f"Unsupported kind: {meta['kind']}")

            oof[va_idx] = p_va
            holdout_fold_preds.append(p_holdout)

            fold_row = evaluate_probs(y_va, p_va, name=f"{model_name}_final_fold{fold_id}")
            fold_row["model_name"] = model_name
            fold_row["family"] = meta["family"]
            fold_row["kind"] = meta["kind"]
            fold_row["fold"] = fold_id
            fold_rows_all.append(fold_row)

            print(
                f"Fold {fold_id} | "
                f"PR_AUC={fold_row['PR_AUC']:.6f} | "
                f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
                f"Brier={fold_row['Brier']:.6f} | "
                f"ECE={fold_row['ECE']:.6f}"
            )

            del X_tr_df, X_va_df, y_tr, y_va, preproc
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        runtime_seconds = time.time() - start_t

        holdout_prob = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))

        dev_row = evaluate_probs(y_dev, oof, name=f"{model_name}_DEV_OOF")
        dev_row["split"] = "DEV_OOF"
        dev_row["model_name"] = model_name
        dev_row["family"] = meta["family"]
        dev_row["kind"] = meta["kind"]
        dev_row["runtime_seconds"] = float(runtime_seconds)
        dev_row["runtime_pretty"] = summarize_runtime(runtime_seconds)

        holdout_row = evaluate_probs(y_holdout, holdout_prob, name=f"{model_name}_HOLDOUT")
        holdout_row["split"] = "HOLDOUT"
        holdout_row["model_name"] = model_name
        holdout_row["family"] = meta["family"]
        holdout_row["kind"] = meta["kind"]
        holdout_row["runtime_seconds"] = float(runtime_seconds)
        holdout_row["runtime_pretty"] = summarize_runtime(runtime_seconds)

        result_rows.extend([dev_row, holdout_row])
        dev_pred_store[model_name] = oof
        holdout_pred_store[model_name] = holdout_prob

        pd.DataFrame({
            "y_true": y_dev.astype(int),
            "y_prob": oof,
        }).to_csv(PRED_DIR / f"final_{model_name}_dev_oof.csv", index=False)

        pd.DataFrame({
            "y_true": y_holdout.astype(int),
            "y_prob": holdout_prob,
        }).to_csv(PRED_DIR / f"final_{model_name}_holdout.csv", index=False)

        print("-" * 100)
        print(
            f"{model_name} summary | "
            f"DEV PR_AUC={dev_row['PR_AUC']:.6f} | "
            f"HOLDOUT PR_AUC={holdout_row['PR_AUC']:.6f} | "
            f"Runtime={dev_row['runtime_pretty']}"
        )
        print()

    results_df = pd.DataFrame(result_rows)
    fold_df = pd.DataFrame(fold_rows_all)

    results_df.to_csv(TAB_DIR / "final_expert_results.csv", index=False)
    fold_df.to_csv(TAB_DIR / "final_expert_fold_results.csv", index=False)

    return results_df, dev_pred_store, holdout_pred_store, fold_df


# -------------------------
# Run Module 7
# -------------------------
final_expert_results_df, final_dev_preds, final_holdout_preds, final_fold_results_df = run_final_expert_models(
    X_dev=X_dev,
    y_dev=y_dev,
    X_holdout=X_holdout,
    y_holdout=y_holdout,
    selected_models=selected_experts,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

print("Final expert results:")
display(final_expert_results_df[[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty"
]].sort_values(["split", "PR_AUC"], ascending=[True, False]))
```

```python
# ============================================================
# Module 8 — Ensemble Baselines
# ============================================================

from sklearn.linear_model import LogisticRegression

# ------------------------------------------------------------
# Helper: stack predictions into matrix
# ------------------------------------------------------------
def stack_predictions(pred_dict: Dict[str, np.ndarray], model_list: List[str]) -> np.ndarray:
    return np.column_stack([pred_dict[m] for m in model_list])


# ------------------------------------------------------------
# 1. Simple Average Ensemble
# ------------------------------------------------------------
def simple_average_ensemble(pred_dict, model_list):
    preds = stack_predictions(pred_dict, model_list)
    return clip_probs(np.mean(preds, axis=1))


# ------------------------------------------------------------
# 2. Performance-Weighted Ensemble
# ------------------------------------------------------------
def performance_weighted_ensemble(pred_dict, model_list, results_df):
    weights = []
    for m in model_list:
        pr_auc = results_df[
            (results_df["model_name"] == m) &
            (results_df["split"] == "DEV_OOF")
        ]["PR_AUC"].values[0]
        weights.append(pr_auc)

    weights = np.array(weights)
    weights = weights / weights.sum()

    preds = stack_predictions(pred_dict, model_list)
    weighted_preds = np.dot(preds, weights)

    return clip_probs(weighted_preds), weights


# ------------------------------------------------------------
# 3. Stacking Ensemble (Logistic Regression)
# ------------------------------------------------------------
def stacking_ensemble(
    dev_preds,
    holdout_preds,
    y_dev,
    y_holdout,
    model_list
):
    X_dev_stack = stack_predictions(dev_preds, model_list)
    X_holdout_stack = stack_predictions(holdout_preds, model_list)

    meta_model = LogisticRegression(
        max_iter=2000,
        class_weight="balanced"
    )

    meta_model.fit(X_dev_stack, y_dev)

    dev_pred = clip_probs(meta_model.predict_proba(X_dev_stack)[:, 1])
    holdout_pred = clip_probs(meta_model.predict_proba(X_holdout_stack)[:, 1])

    return dev_pred, holdout_pred, meta_model


# ------------------------------------------------------------
# Run Ensembles
# ------------------------------------------------------------

# Simple average
avg_dev = simple_average_ensemble(final_dev_preds, selected_experts)
avg_hold = simple_average_ensemble(final_holdout_preds, selected_experts)

# Weighted
w_dev, weights = performance_weighted_ensemble(
    final_dev_preds,
    selected_experts,
    final_expert_results_df
)
w_hold = simple_average_ensemble(final_holdout_preds, selected_experts)

# Stacking
stack_dev, stack_hold, stack_model = stacking_ensemble(
    final_dev_preds,
    final_holdout_preds,
    y_dev,
    y_holdout,
    selected_experts
)

print("Weights used for weighted ensemble:")
for m, w in zip(selected_experts, weights):
    print(f"{m}: {w:.4f}")

# ============================================================
# Evaluate Ensemble Models
# ============================================================

ensemble_results = []

def eval_and_store(name, dev_pred, hold_pred):
    dev_row = evaluate_probs(y_dev, dev_pred, name=name + "_DEV")
    dev_row["model_name"] = name
    dev_row["split"] = "DEV_OOF"

    hold_row = evaluate_probs(y_holdout, hold_pred, name=name + "_HOLDOUT")
    hold_row["model_name"] = name
    hold_row["split"] = "HOLDOUT"

    ensemble_results.extend([dev_row, hold_row])


eval_and_store("SimpleAverage", avg_dev, avg_hold)
eval_and_store("WeightedEnsemble", w_dev, w_hold)
eval_and_store("Stacking", stack_dev, stack_hold)

ensemble_df = pd.DataFrame(ensemble_results)

display(
    ensemble_df[
        ["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["split", "PR_AUC"], ascending=[True, False])
)

ensemble_df.to_csv(TAB_DIR / "ensemble_results.csv", index=False)
```

```python
# ============================================================
# Module 9 — Trust-Aware MoE
# Uses selected expert predictions + trust features + optional context
# ============================================================

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader


# ------------------------------------------------------------
# 1. Trust-feature construction
# ------------------------------------------------------------
def binary_entropy_np(p: np.ndarray) -> np.ndarray:
    p = clip_probs(p)
    return -(p * np.log(p) + (1.0 - p) * np.log(1.0 - p))


def build_trust_features_from_pred_dict(
    X_df: pd.DataFrame,
    pred_dict: Dict[str, np.ndarray],
    expert_list: List[str],
    context_cols: Optional[List[str]] = None,
) -> pd.DataFrame:
    """
    Builds trust/meta features from expert predictions.
    """
    pred_mat = np.column_stack([clip_probs(pred_dict[m]) for m in expert_list])
    pred_df = pd.DataFrame(pred_mat, columns=[f"p_{m}" for m in expert_list])

    meta = pd.DataFrame(index=np.arange(len(pred_df)))
    meta["mean_pred"] = pred_mat.mean(axis=1)
    meta["std_pred"] = pred_mat.std(axis=1)
    meta["min_pred"] = pred_mat.min(axis=1)
    meta["max_pred"] = pred_mat.max(axis=1)
    meta["range_pred"] = meta["max_pred"] - meta["min_pred"]

    # expert entropies
    for i, m in enumerate(expert_list):
        p = pred_mat[:, i]
        meta[f"ent_{m}"] = binary_entropy_np(p)
        meta[f"margin_{m}"] = np.abs(p - 0.5)

    # pairwise disagreements
    for i in range(len(expert_list)):
        for j in range(i + 1, len(expert_list)):
            mi = expert_list[i]
            mj = expert_list[j]
            meta[f"dis_{mi}_{mj}"] = np.abs(pred_mat[:, i] - pred_mat[:, j])

    # sorted margins
    sorted_preds = np.sort(pred_mat, axis=1)
    meta["top1_minus_top2"] = sorted_preds[:, -1] - sorted_preds[:, -2]
    meta["top2_minus_top3"] = sorted_preds[:, -2] - sorted_preds[:, -3]

    # raw expert predictions
    meta = pd.concat([meta, pred_df], axis=1)

    # optional context features
    if context_cols is not None:
        for c in context_cols:
            if c in X_df.columns:
                meta[f"ctx_{c}"] = pd.to_numeric(X_df[c], errors="coerce").fillna(0.0).astype(np.float32)

    return meta.astype(np.float32)


# ------------------------------------------------------------
# 2. Dataset
# ------------------------------------------------------------
class TrustMoEDataset(Dataset):
    def __init__(self, X_meta: np.ndarray, expert_probs: np.ndarray, y: np.ndarray):
        self.X_meta = torch.tensor(X_meta, dtype=torch.float32)
        self.expert_probs = torch.tensor(expert_probs, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx: int):
        return self.X_meta[idx], self.expert_probs[idx], self.y[idx]


# ------------------------------------------------------------
# 3. Gating model
# ------------------------------------------------------------
class TrustAwareMoE(nn.Module):
    def __init__(
        self,
        in_dim: int,
        n_experts: int,
        hidden_dim: int = 64,
        dropout: float = 0.10,
    ):
        super().__init__()
        self.gate = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, n_experts),
        )

    def forward(self, x_meta: torch.Tensor, expert_probs: torch.Tensor):
        gate_logits = self.gate(x_meta)                    # [B, E]
        weights = torch.softmax(gate_logits, dim=1)       # [B, E]
        fused_prob = torch.sum(weights * expert_probs, dim=1)
        fused_prob = torch.clamp(fused_prob, 1e-6, 1 - 1e-6)
        return fused_prob, weights, gate_logits


# ------------------------------------------------------------
# 4. Training helpers
# ------------------------------------------------------------
def make_moe_loaders(
    X_meta_tr: np.ndarray,
    E_tr: np.ndarray,
    y_tr: np.ndarray,
    X_meta_va: np.ndarray,
    E_va: np.ndarray,
    y_va: np.ndarray,
    batch_size: int = 2048,
) -> Tuple[DataLoader, DataLoader]:
    ds_tr = TrustMoEDataset(X_meta_tr, E_tr, y_tr)
    ds_va = TrustMoEDataset(X_meta_va, E_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va


def train_trust_moe(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 1e-3,
    weight_decay: float = 1e-4,
    max_epochs: int = 24,
    patience: int = 5,
    pos_weight: Optional[float] = None,
    entropy_reg: float = 0.01,
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)
            p_hat, weights, _ = model(xb_meta, xb_exp)

            logits = torch.logit(torch.clamp(p_hat, 1e-6, 1 - 1e-6))
            loss = criterion(logits, yb)

            # encourage non-collapsed gate during screening/final baseline
            gate_entropy = -(weights * torch.log(torch.clamp(weights, 1e-8, 1.0))).sum(dim=1).mean()
            loss = loss - entropy_reg * gate_entropy

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()
            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                yb = yb.to(device)

                p_hat, weights, _ = model(xb_meta, xb_exp)
                logits = torch.logit(torch.clamp(p_hat, 1e-6, 1 - 1e-6))
                loss = criterion(logits, yb)
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model


@torch.no_grad()
def predict_trust_moe(
    model: nn.Module,
    X_meta: np.ndarray,
    expert_probs: np.ndarray,
    batch_size: int = 4096,
    device: str = DEVICE,
) -> Tuple[np.ndarray, np.ndarray]:
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    ds = torch.utils.data.TensorDataset(X_meta_t, E_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs = []
    all_weights = []

    for xb_meta, xb_exp in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        p_hat, weights, _ = model(xb_meta, xb_exp)

        all_probs.append(p_hat.detach().cpu().numpy())
        all_weights.append(weights.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    weights = np.concatenate(all_weights, axis=0)
    return probs, weights


# ------------------------------------------------------------
# 5. Prepare trust-feature tables
# ------------------------------------------------------------
selected_experts = ["logreg", "catboost", "hgb", "deepfm"]

# lightweight context columns: only keep if present
candidate_context_cols = [
    "Age", "Sex", "BMI", "GenHlth", "HighBP", "HighChol", "Diabetes", "DiffWalk"
]
context_cols = [c for c in candidate_context_cols if c in X_dev.columns]

dev_meta_df = build_trust_features_from_pred_dict(
    X_df=X_dev,
    pred_dict=final_dev_preds,
    expert_list=selected_experts,
    context_cols=context_cols,
)

holdout_meta_df = build_trust_features_from_pred_dict(
    X_df=X_holdout,
    pred_dict=final_holdout_preds,
    expert_list=selected_experts,
    context_cols=context_cols,
)

E_dev = np.column_stack([clip_probs(final_dev_preds[m]) for m in selected_experts]).astype(np.float32)
E_holdout = np.column_stack([clip_probs(final_holdout_preds[m]) for m in selected_experts]).astype(np.float32)

print("Trust meta feature shape (DEV):", dev_meta_df.shape)
print("Expert prediction matrix shape (DEV):", E_dev.shape)

dev_meta_df.to_csv(PRED_DIR / "trust_moe_dev_meta_features.csv", index=False)
holdout_meta_df.to_csv(PRED_DIR / "trust_moe_holdout_meta_features.csv", index=False)
# ============================================================
# Module 9B — Final OOF + HOLDOUT run for Trust-Aware MoE
# ============================================================

def run_trust_aware_moe_final(
    X_meta_dev_df: pd.DataFrame,
    E_dev: np.ndarray,
    y_dev: np.ndarray,
    X_meta_holdout_df: pd.DataFrame,
    E_holdout: np.ndarray,
    y_holdout: np.ndarray,
    expert_list: List[str],
    config: ExperimentConfig,
) -> Tuple[pd.DataFrame, pd.DataFrame, np.ndarray, np.ndarray, np.ndarray]:
    cv = get_cv(config, final_stage=True)

    X_meta_dev = X_meta_dev_df.values.astype(np.float32)
    X_meta_holdout = X_meta_holdout_df.values.astype(np.float32)

    oof_pred = np.zeros(len(y_dev), dtype=np.float32)
    oof_weights = np.zeros((len(y_dev), len(expert_list)), dtype=np.float32)
    holdout_fold_preds = []
    holdout_fold_weights = []

    fold_rows = []
    start_t = time.time()

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
        print("=" * 90)
        print(f"[TrustAwareMoE] Final fold {fold_id}")
        print("=" * 90)

        X_tr = X_meta_dev[tr_idx]
        X_va = X_meta_dev[va_idx]
        E_tr = E_dev[tr_idx]
        E_va = E_dev[va_idx]
        y_tr = y_dev[tr_idx].astype(np.float32)
        y_va = y_dev[va_idx].astype(np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_moe_loaders(
            X_meta_tr=X_tr,
            E_tr=E_tr,
            y_tr=y_tr,
            X_meta_va=X_va,
            E_va=E_va,
            y_va=y_va,
            batch_size=2048,
        )

        model = TrustAwareMoE(
            in_dim=X_meta_dev.shape[1],
            n_experts=len(expert_list),
            hidden_dim=64,
            dropout=0.10,
        )

        model = train_trust_moe(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=1e-3,
            weight_decay=1e-4,
            max_epochs=24,
            patience=5,
            pos_weight=pos_weight,
            entropy_reg=0.01,
        )

        p_va, w_va = predict_trust_moe(
            model=model,
            X_meta=X_va,
            expert_probs=E_va,
            batch_size=4096,
            device=DEVICE,
        )

        p_hold, w_hold = predict_trust_moe(
            model=model,
            X_meta=X_meta_holdout,
            expert_probs=E_holdout,
            batch_size=4096,
            device=DEVICE,
        )

        oof_pred[va_idx] = p_va
        oof_weights[va_idx] = w_va
        holdout_fold_preds.append(p_hold)
        holdout_fold_weights.append(w_hold)

        fold_row = evaluate_probs(y_va, p_va, name=f"TrustAwareMoE_fold{fold_id}")
        fold_row["model_name"] = "TrustAwareMoE"
        fold_row["split"] = f"fold_{fold_id}"
        fold_row["mean_w_" + expert_list[0]] = float(w_va[:, 0].mean())
        fold_row["mean_w_" + expert_list[1]] = float(w_va[:, 1].mean())
        fold_row["mean_w_" + expert_list[2]] = float(w_va[:, 2].mean())
        fold_row["mean_w_" + expert_list[3]] = float(w_va[:, 3].mean())
        fold_rows.append(fold_row)

        print(
            f"Fold {fold_id} | "
            f"PR_AUC={fold_row['PR_AUC']:.6f} | "
            f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
            f"Brier={fold_row['Brier']:.6f} | "
            f"ECE={fold_row['ECE']:.6f}"
        )

        del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, y_tr, y_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    runtime_seconds = time.time() - start_t

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
    holdout_weights = np.mean(np.stack(holdout_fold_weights, axis=0), axis=0)

    dev_row = evaluate_probs(y_dev, oof_pred, name="TrustAwareMoE_DEV_OOF")
    dev_row["model_name"] = "TrustAwareMoE"
    dev_row["split"] = "DEV_OOF"
    dev_row["runtime_seconds"] = float(runtime_seconds)
    dev_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    for i, m in enumerate(expert_list):
        dev_row[f"mean_w_{m}"] = float(oof_weights[:, i].mean())

    holdout_row = evaluate_probs(y_holdout, holdout_pred, name="TrustAwareMoE_HOLDOUT")
    holdout_row["model_name"] = "TrustAwareMoE"
    holdout_row["split"] = "HOLDOUT"
    holdout_row["runtime_seconds"] = float(runtime_seconds)
    holdout_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    for i, m in enumerate(expert_list):
        holdout_row[f"mean_w_{m}"] = float(holdout_weights[:, i].mean())

    results_df = pd.DataFrame([dev_row, holdout_row])
    folds_df = pd.DataFrame(fold_rows)

    results_df.to_csv(TAB_DIR / "trust_aware_moe_results.csv", index=False)
    folds_df.to_csv(TAB_DIR / "trust_aware_moe_fold_results.csv", index=False)

    pd.DataFrame({
        "y_true": y_dev.astype(int),
        "p_moe": oof_pred,
        **{f"w_{m}": oof_weights[:, i] for i, m in enumerate(expert_list)}
    }).to_csv(PRED_DIR / "trust_aware_moe_dev_oof.csv", index=False)

    pd.DataFrame({
        "y_true": y_holdout.astype(int),
        "p_moe": holdout_pred,
        **{f"w_{m}": holdout_weights[:, i] for i, m in enumerate(expert_list)}
    }).to_csv(PRED_DIR / "trust_aware_moe_holdout.csv", index=False)

    return results_df, folds_df, oof_pred, holdout_pred, holdout_weights


trust_moe_results_df, trust_moe_folds_df, trust_moe_dev_pred, trust_moe_holdout_pred, trust_moe_holdout_weights = run_trust_aware_moe_final(
    X_meta_dev_df=dev_meta_df,
    E_dev=E_dev,
    y_dev=y_dev,
    X_meta_holdout_df=holdout_meta_df,
    E_holdout=E_holdout,
    y_holdout=y_holdout,
    expert_list=selected_experts,
    config=CFG,
)

print("Trust-Aware MoE results:")
display(trust_moe_results_df[[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE", "runtime_pretty",
    "mean_w_logreg", "mean_w_catboost", "mean_w_hgb", "mean_w_deepfm"
]])
# ============================================================
# Module 9C — Quick comparison with experts + ensembles
# ============================================================

comparison_rows = []

# best single experts
comparison_rows.append(
    final_expert_results_df[final_expert_results_df["model_name"] == "catboost"].copy()
)
comparison_rows.append(
    final_expert_results_df[final_expert_results_df["model_name"] == "hgb"].copy()
)
comparison_rows.append(
    final_expert_results_df[final_expert_results_df["model_name"] == "deepfm"].copy()
)
comparison_rows.append(
    final_expert_results_df[final_expert_results_df["model_name"] == "logreg"].copy()
)

# ensemble baselines
comparison_rows.append(ensemble_df.copy())

# moe
comparison_rows.append(trust_moe_results_df.copy())

comparison_df = pd.concat(comparison_rows, axis=0, ignore_index=True)

display(
    comparison_df[
        ["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["split", "PR_AUC"], ascending=[True, False])
)

comparison_df.to_csv(TAB_DIR / "comparison_up_to_moe.csv", index=False)
```

```python
# ============================================================
# Module 10 — TRF-Net
# Trust-Aware Residual Fusion Network
# Backbone + bounded residual correction in logit space
# ============================================================

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader


# ------------------------------------------------------------
# 1. TRF dataset
# ------------------------------------------------------------
class TRFDataset(Dataset):
    def __init__(
        self,
        X_meta: np.ndarray,
        expert_probs: np.ndarray,
        backbone_prob: np.ndarray,
        y: np.ndarray,
    ):
        self.X_meta = torch.tensor(X_meta, dtype=torch.float32)
        self.expert_probs = torch.tensor(expert_probs, dtype=torch.float32)
        self.backbone_prob = torch.tensor(backbone_prob, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx: int):
        return (
            self.X_meta[idx],
            self.expert_probs[idx],
            self.backbone_prob[idx],
            self.y[idx],
        )

# ============================================================
# 2. TRF model
# ============================================================

class TRFNet(nn.Module):
    """
    Trust-Aware Residual Fusion Network

    Final logit:
        logit(p_final) = logit(p_backbone) + delta

    where delta is bounded via tanh to avoid uncontrolled correction.
    """
    def __init__(
        self,
        meta_dim: int,
        n_experts: int,
        hidden_dim: int = 64,
        dropout: float = 0.10,
        max_delta: float = 1.25,
    ):
        super().__init__()
        self.max_delta = max_delta

        # input includes:
        #   meta trust features
        #   raw expert probabilities
        #   backbone probability
        in_dim = meta_dim + n_experts + 1

        self.residual_net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 1),
        )

    def forward(
        self,
        x_meta: torch.Tensor,
        expert_probs: torch.Tensor,
        backbone_prob: torch.Tensor,
    ):
        x = torch.cat(
            [
                x_meta,
                expert_probs,
                backbone_prob.unsqueeze(1),
            ],
            dim=1,
        )

        raw_delta = self.residual_net(x).squeeze(1)
        delta = self.max_delta * torch.tanh(raw_delta)

        base_logit = torch.logit(torch.clamp(backbone_prob, 1e-6, 1 - 1e-6))
        final_logit = base_logit + delta
        final_prob = torch.sigmoid(final_logit)
        final_prob = torch.clamp(final_prob, 1e-6, 1 - 1e-6)

        return final_prob, delta, final_logit

# ============================================================
# 3. Loader helpers
# ============================================================

def make_trf_loaders(
    X_meta_tr: np.ndarray,
    E_tr: np.ndarray,
    B_tr: np.ndarray,
    y_tr: np.ndarray,
    X_meta_va: np.ndarray,
    E_va: np.ndarray,
    B_va: np.ndarray,
    y_va: np.ndarray,
    batch_size: int = 2048,
) -> Tuple[DataLoader, DataLoader]:
    ds_tr = TRFDataset(X_meta_tr, E_tr, B_tr, y_tr)
    ds_va = TRFDataset(X_meta_va, E_va, B_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va

# ============================================================
# 4. Training helpers
# ============================================================

def train_trf_model(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 1e-3,
    weight_decay: float = 1e-4,
    max_epochs: int = 24,
    patience: int = 5,
    pos_weight: Optional[float] = None,
    delta_l2: float = 0.02,   # keeps correction controlled
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, xb_back, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            xb_back = xb_back.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)

            p_hat, delta, final_logit = model(xb_meta, xb_exp, xb_back)
            loss = criterion(final_logit, yb)

            # penalize overly large residuals
            loss = loss + delta_l2 * torch.mean(delta ** 2)

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()
            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, xb_back, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                xb_back = xb_back.to(device)
                yb = yb.to(device)

                p_hat, delta, final_logit = model(xb_meta, xb_exp, xb_back)
                loss = criterion(final_logit, yb)
                loss = loss + delta_l2 * torch.mean(delta ** 2)
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model

# ============================================================
# 5. Prediction helper
# ============================================================

@torch.no_grad()
def predict_trf_model(
    model: nn.Module,
    X_meta: np.ndarray,
    expert_probs: np.ndarray,
    backbone_prob: np.ndarray,
    batch_size: int = 4096,
    device: str = DEVICE,
) -> Tuple[np.ndarray, np.ndarray]:
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    B_t = torch.tensor(backbone_prob, dtype=torch.float32)

    ds = torch.utils.data.TensorDataset(X_meta_t, E_t, B_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs = []
    all_delta = []

    for xb_meta, xb_exp, xb_back in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        xb_back = xb_back.to(device)

        p_hat, delta, _ = model(xb_meta, xb_exp, xb_back)
        all_probs.append(p_hat.detach().cpu().numpy())
        all_delta.append(delta.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    deltas = np.concatenate(all_delta)
    return probs, deltas

# ============================================================
# Module 10B — Prepare TRF inputs
# ============================================================

selected_experts = ["logreg", "catboost", "hgb", "deepfm"]

X_meta_dev_trf = dev_meta_df.values.astype(np.float32)
X_meta_holdout_trf = holdout_meta_df.values.astype(np.float32)

E_dev_trf = np.column_stack([clip_probs(final_dev_preds[m]) for m in selected_experts]).astype(np.float32)
E_holdout_trf = np.column_stack([clip_probs(final_holdout_preds[m]) for m in selected_experts]).astype(np.float32)

# Reliable backbone
backbone_name = "catboost"
B_dev_trf = clip_probs(final_dev_preds[backbone_name]).astype(np.float32)
B_holdout_trf = clip_probs(final_holdout_preds[backbone_name]).astype(np.float32)

print("TRF meta shape:", X_meta_dev_trf.shape)
print("TRF expert matrix shape:", E_dev_trf.shape)
print("TRF backbone shape:", B_dev_trf.shape)

# ============================================================
# Module 10C — Final OOF + HOLDOUT run for TRF-Net
# ============================================================

def run_trf_final(
    X_meta_dev: np.ndarray,
    E_dev: np.ndarray,
    B_dev: np.ndarray,
    y_dev: np.ndarray,
    X_meta_holdout: np.ndarray,
    E_holdout: np.ndarray,
    B_holdout: np.ndarray,
    y_holdout: np.ndarray,
    config: ExperimentConfig,
    backbone_name: str = "catboost",
) -> Tuple[pd.DataFrame, pd.DataFrame, np.ndarray, np.ndarray, np.ndarray]:
    cv = get_cv(config, final_stage=True)

    oof_pred = np.zeros(len(y_dev), dtype=np.float32)
    oof_delta = np.zeros(len(y_dev), dtype=np.float32)

    holdout_fold_preds = []
    holdout_fold_deltas = []

    fold_rows = []
    start_t = time.time()

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
        print("=" * 90)
        print(f"[TRF-Net] Final fold {fold_id}")
        print("=" * 90)

        X_tr = X_meta_dev[tr_idx]
        X_va = X_meta_dev[va_idx]

        E_tr = E_dev[tr_idx]
        E_va = E_dev[va_idx]

        B_tr = B_dev[tr_idx]
        B_va = B_dev[va_idx]

        y_tr = y_dev[tr_idx].astype(np.float32)
        y_va = y_dev[va_idx].astype(np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_trf_loaders(
            X_meta_tr=X_tr,
            E_tr=E_tr,
            B_tr=B_tr,
            y_tr=y_tr,
            X_meta_va=X_va,
            E_va=E_va,
            B_va=B_va,
            y_va=y_va,
            batch_size=2048,
        )

        model = TRFNet(
            meta_dim=X_meta_dev.shape[1],
            n_experts=E_dev.shape[1],
            hidden_dim=64,
            dropout=0.10,
            max_delta=1.25,
        )

        model = train_trf_model(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=1e-3,
            weight_decay=1e-4,
            max_epochs=24,
            patience=5,
            pos_weight=pos_weight,
            delta_l2=0.02,
        )

        p_va, d_va = predict_trf_model(
            model=model,
            X_meta=X_va,
            expert_probs=E_va,
            backbone_prob=B_va,
            batch_size=4096,
            device=DEVICE,
        )

        p_hold, d_hold = predict_trf_model(
            model=model,
            X_meta=X_meta_holdout,
            expert_probs=E_holdout,
            backbone_prob=B_holdout,
            batch_size=4096,
            device=DEVICE,
        )

        oof_pred[va_idx] = p_va
        oof_delta[va_idx] = d_va

        holdout_fold_preds.append(p_hold)
        holdout_fold_deltas.append(d_hold)

        fold_row = evaluate_probs(y_va, p_va, name=f"TRFNet_fold{fold_id}")
        fold_row["model_name"] = "TRFNet"
        fold_row["split"] = f"fold_{fold_id}"
        fold_row["mean_abs_delta"] = float(np.mean(np.abs(d_va)))
        fold_row["mean_delta"] = float(np.mean(d_va))
        fold_rows.append(fold_row)

        print(
            f"Fold {fold_id} | "
            f"PR_AUC={fold_row['PR_AUC']:.6f} | "
            f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
            f"Brier={fold_row['Brier']:.6f} | "
            f"ECE={fold_row['ECE']:.6f} | "
            f"|delta|={fold_row['mean_abs_delta']:.6f}"
        )

        del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, B_tr, B_va, y_tr, y_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    runtime_seconds = time.time() - start_t

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
    holdout_delta = np.mean(np.vstack(holdout_fold_deltas), axis=0)

    dev_row = evaluate_probs(y_dev, oof_pred, name="TRFNet_DEV_OOF")
    dev_row["model_name"] = "TRFNet"
    dev_row["split"] = "DEV_OOF"
    dev_row["runtime_seconds"] = float(runtime_seconds)
    dev_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    dev_row["backbone"] = backbone_name
    dev_row["mean_abs_delta"] = float(np.mean(np.abs(oof_delta)))
    dev_row["mean_delta"] = float(np.mean(oof_delta))

    holdout_row = evaluate_probs(y_holdout, holdout_pred, name="TRFNet_HOLDOUT")
    holdout_row["model_name"] = "TRFNet"
    holdout_row["split"] = "HOLDOUT"
    holdout_row["runtime_seconds"] = float(runtime_seconds)
    holdout_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    holdout_row["backbone"] = backbone_name
    holdout_row["mean_abs_delta"] = float(np.mean(np.abs(holdout_delta)))
    holdout_row["mean_delta"] = float(np.mean(holdout_delta))

    results_df = pd.DataFrame([dev_row, holdout_row])
    folds_df = pd.DataFrame(fold_rows)

    results_df.to_csv(TAB_DIR / "trfnet_results.csv", index=False)
    folds_df.to_csv(TAB_DIR / "trfnet_fold_results.csv", index=False)

    pd.DataFrame({
        "y_true": y_dev.astype(int),
        "p_backbone": B_dev,
        "p_trf": oof_pred,
        "delta_trf": oof_delta,
    }).to_csv(PRED_DIR / "trfnet_dev_oof.csv", index=False)

    pd.DataFrame({
        "y_true": y_holdout.astype(int),
        "p_backbone": B_holdout,
        "p_trf": holdout_pred,
        "delta_trf": holdout_delta,
    }).to_csv(PRED_DIR / "trfnet_holdout.csv", index=False)

    return results_df, folds_df, oof_pred, holdout_pred, holdout_delta


trf_results_df, trf_folds_df, trf_dev_pred, trf_holdout_pred, trf_holdout_delta = run_trf_final(
    X_meta_dev=X_meta_dev_trf,
    E_dev=E_dev_trf,
    B_dev=B_dev_trf,
    y_dev=y_dev,
    X_meta_holdout=X_meta_holdout_trf,
    E_holdout=E_holdout_trf,
    B_holdout=B_holdout_trf,
    y_holdout=y_holdout,
    config=CFG,
    backbone_name=backbone_name,
)

print("TRF-Net results:")
display(trf_results_df[[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE",
    "runtime_pretty", "backbone", "mean_abs_delta", "mean_delta"
]])

# ============================================================
# Module 10D — Quick comparison up to TRF
# ============================================================

comparison_rows = []

# single experts
comparison_rows.append(final_expert_results_df.copy())

# ensembles
comparison_rows.append(ensemble_df.copy())

# moe
comparison_rows.append(trust_moe_results_df.copy())

# trf
comparison_rows.append(trf_results_df.copy())

comparison_trf_df = pd.concat(comparison_rows, axis=0, ignore_index=True)

display(
    comparison_trf_df[
        ["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["split", "PR_AUC"], ascending=[True, False])
)

comparison_trf_df.to_csv(TAB_DIR / "comparison_up_to_trf.csv", index=False)
```

```python
# ============================================================
# Module 10 v2 — TRF-Net (fixed)
# Key fixes:
#   1) smaller max_delta
#   2) zero-centered delta penalty
#   3) stronger delta regularization
#   4) trust-gated residual shrinkage
# ============================================================

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader


class TRFDataset(Dataset):
    def __init__(
        self,
        X_meta: np.ndarray,
        expert_probs: np.ndarray,
        backbone_prob: np.ndarray,
        y: np.ndarray,
    ):
        self.X_meta = torch.tensor(X_meta, dtype=torch.float32)
        self.expert_probs = torch.tensor(expert_probs, dtype=torch.float32)
        self.backbone_prob = torch.tensor(backbone_prob, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx: int):
        return (
            self.X_meta[idx],
            self.expert_probs[idx],
            self.backbone_prob[idx],
            self.y[idx],
        )


class TRFNetV2(nn.Module):
    """
    Final logit:
        logit(p_final) = logit(p_backbone) + gated_delta

    gated_delta is:
        max_delta * tanh(raw_delta) * trust_scale

    trust_scale shrinks correction when disagreement is high.
    """
    def __init__(
        self,
        meta_dim: int,
        n_experts: int,
        hidden_dim: int = 64,
        dropout: float = 0.10,
        max_delta: float = 0.50,
    ):
        super().__init__()
        self.max_delta = max_delta

        in_dim = meta_dim + n_experts + 1

        self.residual_net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 1),
        )

        # small trust gate from meta features only
        self.trust_gate = nn.Sequential(
            nn.Linear(meta_dim, 32),
            nn.GELU(),
            nn.Linear(32, 1),
            nn.Sigmoid(),
        )

    def forward(
        self,
        x_meta: torch.Tensor,
        expert_probs: torch.Tensor,
        backbone_prob: torch.Tensor,
    ):
        x = torch.cat(
            [x_meta, expert_probs, backbone_prob.unsqueeze(1)],
            dim=1,
        )

        raw_delta = self.residual_net(x).squeeze(1)
        base_delta = self.max_delta * torch.tanh(raw_delta)

        # trust_scale in [0,1]
        trust_scale = self.trust_gate(x_meta).squeeze(1)

        # final bounded correction
        delta = base_delta * trust_scale

        base_logit = torch.logit(torch.clamp(backbone_prob, 1e-6, 1 - 1e-6))
        final_logit = base_logit + delta
        final_prob = torch.sigmoid(final_logit)
        final_prob = torch.clamp(final_prob, 1e-6, 1 - 1e-6)

        return final_prob, delta, final_logit, trust_scale


def make_trf_loaders(
    X_meta_tr: np.ndarray,
    E_tr: np.ndarray,
    B_tr: np.ndarray,
    y_tr: np.ndarray,
    X_meta_va: np.ndarray,
    E_va: np.ndarray,
    B_va: np.ndarray,
    y_va: np.ndarray,
    batch_size: int = 2048,
) -> Tuple[DataLoader, DataLoader]:
    ds_tr = TRFDataset(X_meta_tr, E_tr, B_tr, y_tr)
    ds_va = TRFDataset(X_meta_va, E_va, B_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va


def train_trf_model_v2(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 8e-4,
    weight_decay: float = 1e-4,
    max_epochs: int = 24,
    patience: int = 5,
    pos_weight: Optional[float] = None,
    delta_l2: float = 0.10,
    mean_delta_penalty: float = 0.05,
    trust_reg: float = 0.01,
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, xb_back, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            xb_back = xb_back.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)

            p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)
            loss_main = criterion(final_logit, yb)

            # strong control of residual magnitude
            loss_delta = delta_l2 * torch.mean(delta ** 2)

            # enforce zero-centered corrections
            loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)

            # prevent trust gate from saturating too hard
            loss_trust = trust_reg * (torch.mean((trust_scale - 0.5) ** 2))

            loss = loss_main + loss_delta + loss_mean_delta + loss_trust
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, xb_back, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                xb_back = xb_back.to(device)
                yb = yb.to(device)

                p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)
                loss_main = criterion(final_logit, yb)
                loss_delta = delta_l2 * torch.mean(delta ** 2)
                loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
                loss_trust = trust_reg * (torch.mean((trust_scale - 0.5) ** 2))
                loss = loss_main + loss_delta + loss_mean_delta + loss_trust
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model


@torch.no_grad()
def predict_trf_model_v2(
    model: nn.Module,
    X_meta: np.ndarray,
    expert_probs: np.ndarray,
    backbone_prob: np.ndarray,
    batch_size: int = 4096,
    device: str = DEVICE,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    B_t = torch.tensor(backbone_prob, dtype=torch.float32)

    ds = torch.utils.data.TensorDataset(X_meta_t, E_t, B_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs = []
    all_delta = []
    all_trust = []

    for xb_meta, xb_exp, xb_back in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        xb_back = xb_back.to(device)

        p_hat, delta, _, trust_scale = model(xb_meta, xb_exp, xb_back)
        all_probs.append(p_hat.detach().cpu().numpy())
        all_delta.append(delta.detach().cpu().numpy())
        all_trust.append(trust_scale.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    deltas = np.concatenate(all_delta)
    trust = np.concatenate(all_trust)
    return probs, deltas, trust

# ============================================================
# Module 10 v2 — Final OOF + HOLDOUT run for TRF-Net
# ============================================================

def run_trf_final_v2(
    X_meta_dev: np.ndarray,
    E_dev: np.ndarray,
    B_dev: np.ndarray,
    y_dev: np.ndarray,
    X_meta_holdout: np.ndarray,
    E_holdout: np.ndarray,
    B_holdout: np.ndarray,
    y_holdout: np.ndarray,
    config: ExperimentConfig,
    backbone_name: str = "catboost",
) -> Tuple[pd.DataFrame, pd.DataFrame, np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
    cv = get_cv(config, final_stage=True)

    oof_pred = np.zeros(len(y_dev), dtype=np.float32)
    oof_delta = np.zeros(len(y_dev), dtype=np.float32)
    oof_trust = np.zeros(len(y_dev), dtype=np.float32)

    holdout_fold_preds = []
    holdout_fold_deltas = []
    holdout_fold_trust = []

    fold_rows = []
    start_t = time.time()

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
        print("=" * 90)
        print(f"[TRF-Net v2] Final fold {fold_id}")
        print("=" * 90)

        X_tr = X_meta_dev[tr_idx]
        X_va = X_meta_dev[va_idx]

        E_tr = E_dev[tr_idx]
        E_va = E_dev[va_idx]

        B_tr = B_dev[tr_idx]
        B_va = B_dev[va_idx]

        y_tr = y_dev[tr_idx].astype(np.float32)
        y_va = y_dev[va_idx].astype(np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_trf_loaders(
            X_meta_tr=X_tr,
            E_tr=E_tr,
            B_tr=B_tr,
            y_tr=y_tr,
            X_meta_va=X_va,
            E_va=E_va,
            B_va=B_va,
            y_va=y_va,
            batch_size=2048,
        )

        model = TRFNetV2(
            meta_dim=X_meta_dev.shape[1],
            n_experts=E_dev.shape[1],
            hidden_dim=64,
            dropout=0.10,
            max_delta=0.50,
        )

        model = train_trf_model_v2(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=8e-4,
            weight_decay=1e-4,
            max_epochs=24,
            patience=5,
            pos_weight=pos_weight,
            delta_l2=0.10,
            mean_delta_penalty=0.05,
            trust_reg=0.01,
        )

        p_va, d_va, t_va = predict_trf_model_v2(
            model=model,
            X_meta=X_va,
            expert_probs=E_va,
            backbone_prob=B_va,
            batch_size=4096,
            device=DEVICE,
        )

        p_hold, d_hold, t_hold = predict_trf_model_v2(
            model=model,
            X_meta=X_meta_holdout,
            expert_probs=E_holdout,
            backbone_prob=B_holdout,
            batch_size=4096,
            device=DEVICE,
        )

        oof_pred[va_idx] = p_va
        oof_delta[va_idx] = d_va
        oof_trust[va_idx] = t_va

        holdout_fold_preds.append(p_hold)
        holdout_fold_deltas.append(d_hold)
        holdout_fold_trust.append(t_hold)

        fold_row = evaluate_probs(y_va, p_va, name=f"TRFNetV2_fold{fold_id}")
        fold_row["model_name"] = "TRFNetV2"
        fold_row["split"] = f"fold_{fold_id}"
        fold_row["mean_abs_delta"] = float(np.mean(np.abs(d_va)))
        fold_row["mean_delta"] = float(np.mean(d_va))
        fold_row["mean_trust"] = float(np.mean(t_va))
        fold_rows.append(fold_row)

        print(
            f"Fold {fold_id} | "
            f"PR_AUC={fold_row['PR_AUC']:.6f} | "
            f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
            f"Brier={fold_row['Brier']:.6f} | "
            f"ECE={fold_row['ECE']:.6f} | "
            f"|delta|={fold_row['mean_abs_delta']:.6f} | "
            f"mean_delta={fold_row['mean_delta']:.6f}"
        )

        del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, B_tr, B_va, y_tr, y_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    runtime_seconds = time.time() - start_t

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
    holdout_delta = np.mean(np.vstack(holdout_fold_deltas), axis=0)
    holdout_trust = np.mean(np.vstack(holdout_fold_trust), axis=0)

    dev_row = evaluate_probs(y_dev, oof_pred, name="TRFNetV2_DEV_OOF")
    dev_row["model_name"] = "TRFNetV2"
    dev_row["split"] = "DEV_OOF"
    dev_row["runtime_seconds"] = float(runtime_seconds)
    dev_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    dev_row["backbone"] = backbone_name
    dev_row["mean_abs_delta"] = float(np.mean(np.abs(oof_delta)))
    dev_row["mean_delta"] = float(np.mean(oof_delta))
    dev_row["mean_trust"] = float(np.mean(oof_trust))

    holdout_row = evaluate_probs(y_holdout, holdout_pred, name="TRFNetV2_HOLDOUT")
    holdout_row["model_name"] = "TRFNetV2"
    holdout_row["split"] = "HOLDOUT"
    holdout_row["runtime_seconds"] = float(runtime_seconds)
    holdout_row["runtime_pretty"] = summarize_runtime(runtime_seconds)
    holdout_row["backbone"] = backbone_name
    holdout_row["mean_abs_delta"] = float(np.mean(np.abs(holdout_delta)))
    holdout_row["mean_delta"] = float(np.mean(holdout_delta))
    holdout_row["mean_trust"] = float(np.mean(holdout_trust))

    results_df = pd.DataFrame([dev_row, holdout_row])
    folds_df = pd.DataFrame(fold_rows)

    results_df.to_csv(TAB_DIR / "trfnet_v2_results.csv", index=False)
    folds_df.to_csv(TAB_DIR / "trfnet_v2_fold_results.csv", index=False)

    pd.DataFrame({
        "y_true": y_dev.astype(int),
        "p_backbone": B_dev,
        "p_trf": oof_pred,
        "delta_trf": oof_delta,
        "trust_trf": oof_trust,
    }).to_csv(PRED_DIR / "trfnet_v2_dev_oof.csv", index=False)

    pd.DataFrame({
        "y_true": y_holdout.astype(int),
        "p_backbone": B_holdout,
        "p_trf": holdout_pred,
        "delta_trf": holdout_delta,
        "trust_trf": holdout_trust,
    }).to_csv(PRED_DIR / "trfnet_v2_holdout.csv", index=False)

    return results_df, folds_df, oof_pred, holdout_pred, holdout_delta, holdout_trust


trf_v2_results_df, trf_v2_folds_df, trf_v2_dev_pred, trf_v2_holdout_pred, trf_v2_holdout_delta, trf_v2_holdout_trust = run_trf_final_v2(
    X_meta_dev=X_meta_dev_trf,
    E_dev=E_dev_trf,
    B_dev=B_dev_trf,
    y_dev=y_dev,
    X_meta_holdout=X_meta_holdout_trf,
    E_holdout=E_holdout_trf,
    B_holdout=B_holdout_trf,
    y_holdout=y_holdout,
    config=CFG,
    backbone_name=backbone_name,
)

display(trf_v2_results_df[[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE",
    "runtime_pretty", "backbone", "mean_abs_delta", "mean_delta", "mean_trust"
]])
```

```python
# ============================================================
# Module 10E — TRF v3 controlled experiment block
# Small, safe search over TRF settings
# ============================================================

import itertools
import gc
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader


# ------------------------------------------------------------
# 1. Backbone builder
# ------------------------------------------------------------
def build_backbone_predictions(
    final_dev_preds: Dict[str, np.ndarray],
    final_holdout_preds: Dict[str, np.ndarray],
    backbone_name: str,
) -> Tuple[np.ndarray, np.ndarray]:
    if backbone_name == "catboost":
        b_dev = clip_probs(final_dev_preds["catboost"]).astype(np.float32)
        b_hold = clip_probs(final_holdout_preds["catboost"]).astype(np.float32)
    elif backbone_name == "hgb":
        b_dev = clip_probs(final_dev_preds["hgb"]).astype(np.float32)
        b_hold = clip_probs(final_holdout_preds["hgb"]).astype(np.float32)
    elif backbone_name == "cat_hgb_avg":
        b_dev = clip_probs(
            0.5 * clip_probs(final_dev_preds["catboost"]) +
            0.5 * clip_probs(final_dev_preds["hgb"])
        ).astype(np.float32)
        b_hold = clip_probs(
            0.5 * clip_probs(final_holdout_preds["catboost"]) +
            0.5 * clip_probs(final_holdout_preds["hgb"])
        ).astype(np.float32)
    else:
        raise ValueError(f"Unknown backbone_name: {backbone_name}")
    return b_dev, b_hold


# ------------------------------------------------------------
# 2. TRF v3 model
# ------------------------------------------------------------
class TRFNetV3(nn.Module):
    """
    TRF with:
      - smaller bounded residual
      - trust gate
      - optional disagreement-aware scaling
    """
    def __init__(
        self,
        meta_dim: int,
        n_experts: int,
        hidden_dim: int = 64,
        dropout: float = 0.10,
        max_delta: float = 0.40,
        use_disagreement_scaling: bool = True,
        std_feature_index: Optional[int] = None,
    ):
        super().__init__()
        self.max_delta = max_delta
        self.use_disagreement_scaling = use_disagreement_scaling
        self.std_feature_index = std_feature_index

        in_dim = meta_dim + n_experts + 1

        self.residual_net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 1),
        )

        self.trust_gate = nn.Sequential(
            nn.Linear(meta_dim, 32),
            nn.GELU(),
            nn.Linear(32, 1),
            nn.Sigmoid(),
        )

    def forward(
        self,
        x_meta: torch.Tensor,
        expert_probs: torch.Tensor,
        backbone_prob: torch.Tensor,
    ):
        x = torch.cat(
            [x_meta, expert_probs, backbone_prob.unsqueeze(1)],
            dim=1,
        )

        raw_delta = self.residual_net(x).squeeze(1)
        base_delta = self.max_delta * torch.tanh(raw_delta)

        trust_scale = self.trust_gate(x_meta).squeeze(1)

        if self.use_disagreement_scaling and self.std_feature_index is not None:
            # std_pred is already in meta features; clamp to a safe range
            std_pred = torch.clamp(x_meta[:, self.std_feature_index], 0.0, 1.0)
            disagreement_scale = 0.50 + std_pred
        else:
            disagreement_scale = 1.0

        delta = base_delta * trust_scale * disagreement_scale

        base_logit = torch.logit(torch.clamp(backbone_prob, 1e-6, 1 - 1e-6))
        final_logit = base_logit + delta
        final_prob = torch.sigmoid(final_logit)
        final_prob = torch.clamp(final_prob, 1e-6, 1 - 1e-6)

        return final_prob, delta, final_logit, trust_scale


# ------------------------------------------------------------
# 3. Training / prediction
# ------------------------------------------------------------
def train_trf_model_v3(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 8e-4,
    weight_decay: float = 1e-4,
    max_epochs: int = 8,
    patience: int = 3,
    pos_weight: Optional[float] = None,
    delta_l2: float = 0.12,
    mean_delta_penalty: float = 0.10,
    trust_reg: float = 0.01,
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, xb_back, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            xb_back = xb_back.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)

            p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)
            loss_main = criterion(final_logit, yb)
            loss_delta = delta_l2 * torch.mean(delta ** 2)
            loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
            loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)

            loss = loss_main + loss_delta + loss_mean_delta + loss_trust
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, xb_back, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                xb_back = xb_back.to(device)
                yb = yb.to(device)

                p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)
                loss_main = criterion(final_logit, yb)
                loss_delta = delta_l2 * torch.mean(delta ** 2)
                loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
                loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)
                loss = loss_main + loss_delta + loss_mean_delta + loss_trust
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model


@torch.no_grad()
def predict_trf_model_v3(
    model: nn.Module,
    X_meta: np.ndarray,
    expert_probs: np.ndarray,
    backbone_prob: np.ndarray,
    batch_size: int = 4096,
    device: str = DEVICE,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    B_t = torch.tensor(backbone_prob, dtype=torch.float32)

    ds = torch.utils.data.TensorDataset(X_meta_t, E_t, B_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs, all_delta, all_trust = [], [], []

    for xb_meta, xb_exp, xb_back in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        xb_back = xb_back.to(device)

        p_hat, delta, _, trust_scale = model(xb_meta, xb_exp, xb_back)
        all_probs.append(p_hat.detach().cpu().numpy())
        all_delta.append(delta.detach().cpu().numpy())
        all_trust.append(trust_scale.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    deltas = np.concatenate(all_delta)
    trust = np.concatenate(all_trust)
    return probs, deltas, trust


# ------------------------------------------------------------
# 4. Single experiment runner
# ------------------------------------------------------------
def run_trf_v3_single_experiment(
    X_meta_dev: np.ndarray,
    E_dev: np.ndarray,
    y_dev: np.ndarray,
    X_meta_holdout: np.ndarray,
    E_holdout: np.ndarray,
    y_holdout: np.ndarray,
    B_dev: np.ndarray,
    B_holdout: np.ndarray,
    config: ExperimentConfig,
    backbone_name: str,
    max_delta: float,
    mean_delta_penalty: float,
    delta_l2: float,
    use_disagreement_scaling: bool,
    hidden_dim: int = 64,
    dropout: float = 0.10,
) -> Dict[str, Any]:
    cv = get_cv(config, final_stage=True)

    oof_pred = np.zeros(len(y_dev), dtype=np.float32)
    oof_delta = np.zeros(len(y_dev), dtype=np.float32)
    oof_trust = np.zeros(len(y_dev), dtype=np.float32)

    holdout_fold_preds = []
    holdout_fold_deltas = []
    holdout_fold_trust = []

    # locate std_pred feature in meta table
    std_idx = None
    if "std_pred" in dev_meta_df.columns:
        std_idx = dev_meta_df.columns.get_loc("std_pred")

    start_t = time.time()

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
        print("=" * 90)
        print(
            f"[TRF v3] fold={fold_id} | backbone={backbone_name} | "
            f"max_delta={max_delta} | mean_pen={mean_delta_penalty} | "
            f"delta_l2={delta_l2} | disagree_scale={use_disagreement_scaling}"
        )
        print("=" * 90)

        X_tr = X_meta_dev[tr_idx]
        X_va = X_meta_dev[va_idx]
        E_tr = E_dev[tr_idx]
        E_va = E_dev[va_idx]
        Bb_tr = B_dev[tr_idx]
        Bb_va = B_dev[va_idx]
        y_tr = y_dev[tr_idx].astype(np.float32)
        y_va = y_dev[va_idx].astype(np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_trf_loaders(
            X_meta_tr=X_tr,
            E_tr=E_tr,
            B_tr=Bb_tr,
            y_tr=y_tr,
            X_meta_va=X_va,
            E_va=E_va,
            B_va=Bb_va,
            y_va=y_va,
            batch_size=2048,
        )

        model = TRFNetV3(
            meta_dim=X_meta_dev.shape[1],
            n_experts=E_dev.shape[1],
            hidden_dim=hidden_dim,
            dropout=dropout,
            max_delta=max_delta,
            use_disagreement_scaling=use_disagreement_scaling,
            std_feature_index=std_idx,
        )

        model = train_trf_model_v3(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=8e-4,
            weight_decay=1e-4,
            max_epochs=8,
            patience=3,
            pos_weight=pos_weight,
            delta_l2=delta_l2,
            mean_delta_penalty=mean_delta_penalty,
            trust_reg=0.01,
        )

        p_va, d_va, t_va = predict_trf_model_v3(
            model=model,
            X_meta=X_va,
            expert_probs=E_va,
            backbone_prob=Bb_va,
            batch_size=4096,
            device=DEVICE,
        )
        p_hold, d_hold, t_hold = predict_trf_model_v3(
            model=model,
            X_meta=X_meta_holdout,
            expert_probs=E_holdout,
            backbone_prob=B_holdout,
            batch_size=4096,
            device=DEVICE,
        )

        oof_pred[va_idx] = p_va
        oof_delta[va_idx] = d_va
        oof_trust[va_idx] = t_va

        holdout_fold_preds.append(p_hold)
        holdout_fold_deltas.append(d_hold)
        holdout_fold_trust.append(t_hold)

        del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, Bb_tr, Bb_va, y_tr, y_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    runtime_seconds = time.time() - start_t

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
    holdout_delta = np.mean(np.vstack(holdout_fold_deltas), axis=0)
    holdout_trust = np.mean(np.vstack(holdout_fold_trust), axis=0)

    dev_row = evaluate_probs(y_dev, oof_pred, name="TRFv3_DEV")
    hold_row = evaluate_probs(y_holdout, holdout_pred, name="TRFv3_HOLDOUT")

    row = {
        "backbone": backbone_name,
        "max_delta": max_delta,
        "mean_delta_penalty": mean_delta_penalty,
        "delta_l2": delta_l2,
        "use_disagreement_scaling": use_disagreement_scaling,
        "runtime_seconds": float(runtime_seconds),
        "runtime_pretty": summarize_runtime(runtime_seconds),

        "DEV_PR_AUC": dev_row["PR_AUC"],
        "DEV_ROC_AUC": dev_row["ROC_AUC"],
        "DEV_Brier": dev_row["Brier"],
        "DEV_ECE": dev_row["ECE"],
        "DEV_mean_abs_delta": float(np.mean(np.abs(oof_delta))),
        "DEV_mean_delta": float(np.mean(oof_delta)),
        "DEV_mean_trust": float(np.mean(oof_trust)),

        "HOLDOUT_PR_AUC": hold_row["PR_AUC"],
        "HOLDOUT_ROC_AUC": hold_row["ROC_AUC"],
        "HOLDOUT_Brier": hold_row["Brier"],
        "HOLDOUT_ECE": hold_row["ECE"],
        "HOLDOUT_mean_abs_delta": float(np.mean(np.abs(holdout_delta))),
        "HOLDOUT_mean_delta": float(np.mean(holdout_delta)),
        "HOLDOUT_mean_trust": float(np.mean(holdout_trust)),
    }

    return row


# ------------------------------------------------------------
# 5. Small search space
# ------------------------------------------------------------
trf_v3_grid = list(itertools.product(
    ["catboost", "hgb", "cat_hgb_avg"],   # backbones
    [0.35, 0.40],                         # max_delta
    [0.05, 0.10],                         # mean_delta_penalty
    [0.10, 0.12],                         # delta_l2
    [True, False],                        # disagreement scaling
))

print(f"Total TRF v3 experiments: {len(trf_v3_grid)}")

# ============================================================
# Module 10F — Run TRF v3 search
# ============================================================

trf_v3_rows = []

X_meta_dev_v3 = X_meta_dev_trf
X_meta_holdout_v3 = X_meta_holdout_trf
E_dev_v3 = E_dev_trf
E_holdout_v3 = E_holdout_trf

for backbone_name, max_delta, mean_delta_penalty, delta_l2, use_disagreement_scaling in trf_v3_grid:
    B_dev_v3, B_holdout_v3 = build_backbone_predictions(
        final_dev_preds=final_dev_preds,
        final_holdout_preds=final_holdout_preds,
        backbone_name=backbone_name,
    )

    row = run_trf_v3_single_experiment(
        X_meta_dev=X_meta_dev_v3,
        E_dev=E_dev_v3,
        y_dev=y_dev,
        X_meta_holdout=X_meta_holdout_v3,
        E_holdout=E_holdout_v3,
        y_holdout=y_holdout,
        B_dev=B_dev_v3,
        B_holdout=B_holdout_v3,
        config=CFG,
        backbone_name=backbone_name,
        max_delta=max_delta,
        mean_delta_penalty=mean_delta_penalty,
        delta_l2=delta_l2,
        use_disagreement_scaling=use_disagreement_scaling,
        hidden_dim=64,
        dropout=0.10,
    )
    trf_v3_rows.append(row)

    tmp_df = pd.DataFrame(trf_v3_rows)
    tmp_df.to_csv(TAB_DIR / "trf_v3_search_results_partial.csv", index=False)

trf_v3_df = pd.DataFrame(trf_v3_rows)
trf_v3_df.to_csv(TAB_DIR / "trf_v3_search_results.csv", index=False)

print("TRF v3 search complete.")
display(trf_v3_df.head())

# ============================================================
# Module 10G — Rank TRF v3 candidates
# Balanced ranking:
#   higher PR-AUC is good
#   lower Brier/ECE is good
#   lower abs(mean_delta) is good
# ============================================================

rank_df = trf_v3_df.copy()

# normalize components
rank_df["score_pr"] = rank_df["HOLDOUT_PR_AUC"].rank(method="average", ascending=False)
rank_df["score_brier"] = rank_df["HOLDOUT_Brier"].rank(method="average", ascending=True)
rank_df["score_ece"] = rank_df["HOLDOUT_ECE"].rank(method="average", ascending=True)
rank_df["score_delta"] = rank_df["HOLDOUT_mean_abs_delta"].rank(method="average", ascending=True)
rank_df["score_mean_delta"] = rank_df["HOLDOUT_mean_delta"].abs().rank(method="average", ascending=True)

# composite rank: lower is better
rank_df["composite_rank"] = (
    0.35 * rank_df["score_pr"] +
    0.25 * rank_df["score_brier"] +
    0.20 * rank_df["score_ece"] +
    0.10 * rank_df["score_delta"] +
    0.10 * rank_df["score_mean_delta"]
)

rank_df = rank_df.sort_values(
    ["composite_rank", "HOLDOUT_PR_AUC", "HOLDOUT_Brier"],
    ascending=[True, False, True]
).reset_index(drop=True)

rank_df.to_csv(TAB_DIR / "trf_v3_search_ranked.csv", index=False)

display(rank_df[[
    "backbone", "max_delta", "mean_delta_penalty", "delta_l2", "use_disagreement_scaling",
    "HOLDOUT_PR_AUC", "HOLDOUT_ROC_AUC", "HOLDOUT_Brier", "HOLDOUT_ECE",
    "HOLDOUT_mean_abs_delta", "HOLDOUT_mean_delta", "HOLDOUT_mean_trust",
    "runtime_pretty", "composite_rank"
]].head(10))

# ============================================================
# Module 10H — Recommended winner
# ============================================================

best_trf_v3 = rank_df.iloc[0].to_dict()
print("Best TRF v3 configuration:")
for k, v in best_trf_v3.items():
    print(f"{k}: {v}")
```

```python
# ============================================================
# Audit Module — Recheck CatBoost Brier + ECE carefully
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import brier_score_loss, log_loss
from sklearn.calibration import calibration_curve


# ------------------------------------------------------------
# 1. Load saved CatBoost predictions
# Adjust paths if needed
# ------------------------------------------------------------
cb_dev_path = PRED_DIR / "final_catboost_dev_oof.csv"
cb_hold_path = PRED_DIR / "final_catboost_holdout.csv"

cb_dev_df = pd.read_csv(cb_dev_path)
cb_hold_df = pd.read_csv(cb_hold_path)

print("DEV file shape:", cb_dev_df.shape)
print("HOLDOUT file shape:", cb_hold_df.shape)
display(cb_dev_df.head())
display(cb_hold_df.head())

# ============================================================
# 2. Robust helpers
# ============================================================

def clip_probs(p, eps=1e-12):
    p = np.asarray(p, dtype=float)
    return np.clip(p, eps, 1.0 - eps)


def ece_fixed_bins(y_true, y_prob, n_bins=10):
    """
    Standard weighted ECE with equal-width bins.
    """
    y_true = np.asarray(y_true).astype(int)
    y_prob = clip_probs(y_prob)

    bin_edges = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.digitize(y_prob, bin_edges, right=True) - 1
    bin_ids = np.clip(bin_ids, 0, n_bins - 1)

    ece = 0.0
    rows = []

    for b in range(n_bins):
        mask = (bin_ids == b)
        n_b = int(mask.sum())
        if n_b == 0:
            rows.append({
                "bin": b,
                "n": 0,
                "conf_mean": np.nan,
                "acc_mean": np.nan,
                "gap": np.nan,
            })
            continue

        conf_mean = float(y_prob[mask].mean())
        acc_mean = float(y_true[mask].mean())
        gap = abs(acc_mean - conf_mean)

        ece += (n_b / len(y_true)) * gap

        rows.append({
            "bin": b,
            "n": n_b,
            "conf_mean": conf_mean,
            "acc_mean": acc_mean,
            "gap": gap,
        })

    return float(ece), pd.DataFrame(rows)


def ece_quantile_bins(y_true, y_prob, n_bins=10):
    """
    Weighted ECE with equal-frequency bins.
    Often more stable for imbalanced data.
    """
    y_true = np.asarray(y_true).astype(int)
    y_prob = clip_probs(y_prob)

    # unique quantile edges
    bin_edges = np.quantile(y_prob, np.linspace(0, 1, n_bins + 1))
    bin_edges[0] = 0.0
    bin_edges[-1] = 1.0
    bin_edges = np.unique(bin_edges)

    if len(bin_edges) < 3:
        return np.nan, pd.DataFrame()

    bin_ids = np.digitize(y_prob, bin_edges, right=True) - 1
    bin_ids = np.clip(bin_ids, 0, len(bin_edges) - 2)

    ece = 0.0
    rows = []

    for b in range(len(bin_edges) - 1):
        mask = (bin_ids == b)
        n_b = int(mask.sum())
        if n_b == 0:
            continue

        conf_mean = float(y_prob[mask].mean())
        acc_mean = float(y_true[mask].mean())
        gap = abs(acc_mean - conf_mean)

        ece += (n_b / len(y_true)) * gap

        rows.append({
            "bin": b,
            "left": bin_edges[b],
            "right": bin_edges[b + 1],
            "n": n_b,
            "conf_mean": conf_mean,
            "acc_mean": acc_mean,
            "gap": gap,
        })

    return float(ece), pd.DataFrame(rows)


def audit_probabilities(name, y_true, y_prob):
    y_true = np.asarray(y_true).astype(int)
    y_prob = clip_probs(y_prob)

    out = {
        "name": name,
        "n": int(len(y_true)),
        "prevalence": float(y_true.mean()),
        "prob_min": float(y_prob.min()),
        "prob_max": float(y_prob.max()),
        "prob_mean": float(y_prob.mean()),
        "brier": float(brier_score_loss(y_true, y_prob)),
        "nll": float(log_loss(y_true, y_prob)),
    }

    for bins in [10, 15, 20, 30, 50]:
        ece_fw, _ = ece_fixed_bins(y_true, y_prob, n_bins=bins)
        ece_q, _ = ece_quantile_bins(y_true, y_prob, n_bins=bins)
        out[f"ece_fixed_{bins}"] = ece_fw
        out[f"ece_quantile_{bins}"] = ece_q

    return out

# ============================================================
# 3. Recompute CatBoost audit stats
# ============================================================

cb_dev_y = cb_dev_df["y_true"].values
cb_dev_p = cb_dev_df["y_prob"].values

cb_hold_y = cb_hold_df["y_true"].values
cb_hold_p = cb_hold_df["y_prob"].values

audit_rows = [
    audit_probabilities("CatBoost_DEV_OOF", cb_dev_y, cb_dev_p),
    audit_probabilities("CatBoost_HOLDOUT", cb_hold_y, cb_hold_p),
]

audit_df = pd.DataFrame(audit_rows)
display(audit_df.T)

# ============================================================
# 4. Bin-level reliability tables
# ============================================================

ece10_dev, ece10_dev_bins = ece_fixed_bins(cb_dev_y, cb_dev_p, n_bins=10)
ece20_dev, ece20_dev_bins = ece_fixed_bins(cb_dev_y, cb_dev_p, n_bins=20)

ece10_hold, ece10_hold_bins = ece_fixed_bins(cb_hold_y, cb_hold_p, n_bins=10)
ece20_hold, ece20_hold_bins = ece_fixed_bins(cb_hold_y, cb_hold_p, n_bins=20)

print("DEV ECE (10 bins):", ece10_dev)
print("DEV ECE (20 bins):", ece20_dev)
print("HOLDOUT ECE (10 bins):", ece10_hold)
print("HOLDOUT ECE (20 bins):", ece20_hold)

print("\nDEV 10-bin reliability table:")
display(ece10_dev_bins)

print("\nHOLDOUT 10-bin reliability table:")
display(ece10_hold_bins)

# ============================================================
# 5. Reliability plots
# ============================================================

def plot_reliability(y_true, y_prob, title, n_bins=10):
    frac_pos, mean_pred = calibration_curve(
        y_true,
        y_prob,
        n_bins=n_bins,
        strategy="uniform"
    )

    plt.figure(figsize=(5.5, 5.5))
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.plot(mean_pred, frac_pos, marker="o")
    plt.xlabel("Mean predicted probability")
    plt.ylabel("Observed positive fraction")
    plt.title(title)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()


plot_reliability(cb_dev_y, cb_dev_p, "CatBoost DEV OOF Reliability (10 bins)", n_bins=10)
plot_reliability(cb_hold_y, cb_hold_p, "CatBoost HOLDOUT Reliability (10 bins)", n_bins=10)

plot_reliability(cb_dev_y, cb_dev_p, "CatBoost DEV OOF Reliability (20 bins)", n_bins=20)
plot_reliability(cb_hold_y, cb_hold_p, "CatBoost HOLDOUT Reliability (20 bins)", n_bins=20)

# ============================================================
# 6. Compare your original ECE function vs robust versions
# ============================================================

# Replace this with your earlier function if needed
def old_expected_calibration_error(y_true, y_prob, n_bins=15):
    y_true = np.asarray(y_true).astype(int)
    y_prob = np.clip(np.asarray(y_prob), 1e-6, 1.0 - 1e-6)

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_ids = np.digitize(y_prob, bins) - 1

    ece = 0.0
    for b in range(n_bins):
        mask = (bin_ids == b)
        if mask.sum() == 0:
            continue
        acc = y_true[mask].mean()
        conf = y_prob[mask].mean()
        ece += (mask.sum() / len(y_true)) * abs(acc - conf)
    return float(ece)


compare_df = pd.DataFrame([
    {
        "split": "DEV_OOF",
        "old_ece_15": old_expected_calibration_error(cb_dev_y, cb_dev_p, 15),
        "ece_fixed_10": ece_fixed_bins(cb_dev_y, cb_dev_p, 10)[0],
        "ece_fixed_15": ece_fixed_bins(cb_dev_y, cb_dev_p, 15)[0],
        "ece_fixed_20": ece_fixed_bins(cb_dev_y, cb_dev_p, 20)[0],
        "ece_quantile_10": ece_quantile_bins(cb_dev_y, cb_dev_p, 10)[0],
        "ece_quantile_15": ece_quantile_bins(cb_dev_y, cb_dev_p, 15)[0],
        "ece_quantile_20": ece_quantile_bins(cb_dev_y, cb_dev_p, 20)[0],
        "brier": brier_score_loss(cb_dev_y, cb_dev_p),
    },
    {
        "split": "HOLDOUT",
        "old_ece_15": old_expected_calibration_error(cb_hold_y, cb_hold_p, 15),
        "ece_fixed_10": ece_fixed_bins(cb_hold_y, cb_hold_p, 10)[0],
        "ece_fixed_15": ece_fixed_bins(cb_hold_y, cb_hold_p, 15)[0],
        "ece_fixed_20": ece_fixed_bins(cb_hold_y, cb_hold_p, 20)[0],
        "ece_quantile_10": ece_quantile_bins(cb_hold_y, cb_hold_p, 10)[0],
        "ece_quantile_15": ece_quantile_bins(cb_hold_y, cb_hold_p, 15)[0],
        "ece_quantile_20": ece_quantile_bins(cb_hold_y, cb_hold_p, 20)[0],
        "brier": brier_score_loss(cb_hold_y, cb_hold_p),
    }
])

display(compare_df)
```

```python
# ============================================================
# Targeted TRF Improvement Block — Step 1
# Enhanced meta features for TRF
# ============================================================

def build_trf_enhanced_meta_features(
    base_meta_df: pd.DataFrame,
    final_dev_preds: Dict[str, np.ndarray],
    final_holdout_preds: Dict[str, np.ndarray],
    expert_list: List[str],
    backbone_name: str = "cat_hgb_avg",
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Adds backbone-relative features that help TRF learn when the backbone is challenged.
    """

    dev_df = base_meta_df.copy()
    hold_df = holdout_meta_df.copy()

    def get_backbone(pred_dict: Dict[str, np.ndarray], name: str) -> np.ndarray:
        if name == "catboost":
            return clip_probs(pred_dict["catboost"])
        elif name == "hgb":
            return clip_probs(pred_dict["hgb"])
        elif name == "cat_hgb_avg":
            return clip_probs(0.5 * clip_probs(pred_dict["catboost"]) + 0.5 * clip_probs(pred_dict["hgb"]))
        else:
            raise ValueError(f"Unknown backbone: {name}")

    b_dev = get_backbone(final_dev_preds, backbone_name)
    b_hold = get_backbone(final_holdout_preds, backbone_name)

    dev_df["backbone_prob"] = b_dev
    hold_df["backbone_prob"] = b_hold

    dev_df["backbone_margin"] = np.abs(b_dev - 0.5)
    hold_df["backbone_margin"] = np.abs(b_hold - 0.5)

    dev_df["backbone_entropy"] = binary_entropy_np(b_dev)
    hold_df["backbone_entropy"] = binary_entropy_np(b_hold)

    dev_df["mean_minus_backbone"] = dev_df["mean_pred"] - b_dev
    hold_df["mean_minus_backbone"] = hold_df["mean_pred"] - b_hold

    dev_df["max_minus_backbone"] = dev_df["max_pred"] - b_dev
    hold_df["max_minus_backbone"] = hold_df["max_pred"] - b_hold

    dev_df["backbone_minus_min"] = b_dev - dev_df["min_pred"]
    hold_df["backbone_minus_min"] = b_hold - hold_df["min_pred"]

    for m in expert_list:
        pcol = f"p_{m}"
        if pcol in dev_df.columns:
            dev_df[f"{pcol}_minus_backbone"] = dev_df[pcol] - b_dev
            hold_df[f"{pcol}_minus_backbone"] = hold_df[pcol] - b_hold

    return dev_df.astype(np.float32), hold_df.astype(np.float32)

# ============================================================
# Targeted TRF Improvement Block — Step 2
# Improved TRF model
# ============================================================

class TRFNetImproved(nn.Module):
    """
    Improvements:
      - smaller or configurable hidden layer
      - stronger disagreement-aware scaling
      - bounded residual
      - trust gate
    """
    def __init__(
        self,
        meta_dim: int,
        n_experts: int,
        hidden_dim: int = 32,
        dropout: float = 0.10,
        max_delta: float = 0.30,
        use_disagreement_scaling: bool = True,
        std_feature_index: Optional[int] = None,
        disagreement_strength: float = 2.5,
        disagreement_bias: float = 0.10,
    ):
        super().__init__()
        self.max_delta = max_delta
        self.use_disagreement_scaling = use_disagreement_scaling
        self.std_feature_index = std_feature_index
        self.disagreement_strength = disagreement_strength
        self.disagreement_bias = disagreement_bias

        in_dim = meta_dim + n_experts + 1

        self.residual_net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 1),
        )

        self.trust_gate = nn.Sequential(
            nn.Linear(meta_dim, 32),
            nn.GELU(),
            nn.Linear(32, 1),
            nn.Sigmoid(),
        )

    def forward(
        self,
        x_meta: torch.Tensor,
        expert_probs: torch.Tensor,
        backbone_prob: torch.Tensor,
    ):
        x = torch.cat([x_meta, expert_probs, backbone_prob.unsqueeze(1)], dim=1)

        raw_delta = self.residual_net(x).squeeze(1)
        base_delta = self.max_delta * torch.tanh(raw_delta)

        trust_scale = self.trust_gate(x_meta).squeeze(1)

        if self.use_disagreement_scaling and self.std_feature_index is not None:
            std_pred = torch.clamp(x_meta[:, self.std_feature_index], 0.0, 1.0)
            disagreement_scale = self.disagreement_bias + self.disagreement_strength * std_pred
        else:
            disagreement_scale = 1.0

        delta = base_delta * trust_scale * disagreement_scale

        base_logit = torch.logit(torch.clamp(backbone_prob, 1e-6, 1 - 1e-6))
        final_logit = base_logit + delta
        final_prob = torch.sigmoid(final_logit)
        final_prob = torch.clamp(final_prob, 1e-6, 1 - 1e-6)

        return final_prob, delta, final_logit, trust_scale

# ============================================================
# Targeted TRF Improvement Block — Step 3
# Improved training
# ============================================================

def train_trf_improved(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 8e-4,
    weight_decay: float = 1e-4,
    max_epochs: int = 24,
    patience: int = 5,
    pos_weight: Optional[float] = None,
    delta_l2: float = 0.10,
    mean_delta_penalty: float = 0.15,
    trust_reg: float = 0.02,
) -> nn.Module:
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, xb_back, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            xb_back = xb_back.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)

            p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)

            loss_main = criterion(final_logit, yb)
            loss_delta = delta_l2 * torch.mean(delta ** 2)
            loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
            loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)

            loss = loss_main + loss_delta + loss_mean_delta + loss_trust
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, xb_back, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                xb_back = xb_back.to(device)
                yb = yb.to(device)

                p_hat, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)
                loss_main = criterion(final_logit, yb)
                loss_delta = delta_l2 * torch.mean(delta ** 2)
                loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
                loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)

                loss = loss_main + loss_delta + loss_mean_delta + loss_trust
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model

# ============================================================
# Targeted TRF Improvement Block — Step 4
# Prediction helper
# ============================================================

@torch.no_grad()
def predict_trf_improved(
    model: nn.Module,
    X_meta: np.ndarray,
    expert_probs: np.ndarray,
    backbone_prob: np.ndarray,
    batch_size: int = 4096,
    device: str = DEVICE,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    B_t = torch.tensor(backbone_prob, dtype=torch.float32)

    ds = torch.utils.data.TensorDataset(X_meta_t, E_t, B_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs, all_delta, all_trust = [], [], []

    for xb_meta, xb_exp, xb_back in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        xb_back = xb_back.to(device)

        p_hat, delta, _, trust_scale = model(xb_meta, xb_exp, xb_back)
        all_probs.append(p_hat.detach().cpu().numpy())
        all_delta.append(delta.detach().cpu().numpy())
        all_trust.append(trust_scale.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    deltas = np.concatenate(all_delta)
    trust = np.concatenate(all_trust)
    return probs, deltas, trust

# ============================================================
# Targeted TRF Improvement Block — Step 5
# Single experiment runner
# ============================================================

def run_trf_improved_single_experiment(
    X_meta_dev_df: pd.DataFrame,
    X_meta_holdout_df: pd.DataFrame,
    E_dev: np.ndarray,
    E_holdout: np.ndarray,
    y_dev: np.ndarray,
    y_holdout: np.ndarray,
    B_dev: np.ndarray,
    B_holdout: np.ndarray,
    config: ExperimentConfig,
    backbone_name: str,
    hidden_dim: int,
    mean_delta_penalty: float,
    use_enhanced_meta: bool,
    disagreement_strength: float,
    disagreement_bias: float,
    seed_offset: int = 0,
) -> Dict[str, Any]:
    cv = get_cv(config, final_stage=True)

    X_meta_dev = X_meta_dev_df.values.astype(np.float32)
    X_meta_holdout = X_meta_holdout_df.values.astype(np.float32)

    oof_pred = np.zeros(len(y_dev), dtype=np.float32)
    oof_delta = np.zeros(len(y_dev), dtype=np.float32)
    oof_trust = np.zeros(len(y_dev), dtype=np.float32)

    holdout_fold_preds = []
    holdout_fold_deltas = []
    holdout_fold_trust = []

    std_idx = None
    if "std_pred" in X_meta_dev_df.columns:
        std_idx = X_meta_dev_df.columns.get_loc("std_pred")

    start_t = time.time()

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
        print("=" * 90)
        print(
            f"[TRF improved] fold={fold_id} | backbone={backbone_name} | "
            f"hidden_dim={hidden_dim} | mean_pen={mean_delta_penalty} | "
            f"enhanced_meta={use_enhanced_meta} | ds={disagreement_strength}"
        )
        print("=" * 90)

        X_tr = X_meta_dev[tr_idx]
        X_va = X_meta_dev[va_idx]
        E_tr = E_dev[tr_idx]
        E_va = E_dev[va_idx]
        Bb_tr = B_dev[tr_idx]
        Bb_va = B_dev[va_idx]
        y_tr = y_dev[tr_idx].astype(np.float32)
        y_va = y_dev[va_idx].astype(np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_trf_loaders(
            X_meta_tr=X_tr,
            E_tr=E_tr,
            B_tr=Bb_tr,
            y_tr=y_tr,
            X_meta_va=X_va,
            E_va=E_va,
            B_va=Bb_va,
            y_va=y_va,
            batch_size=2048,
        )

        set_global_seed(CFG.seed + seed_offset + fold_id)

        model = TRFNetImproved(
            meta_dim=X_meta_dev.shape[1],
            n_experts=E_dev.shape[1],
            hidden_dim=hidden_dim,
            dropout=0.10,
            max_delta=0.30,
            use_disagreement_scaling=True,
            std_feature_index=std_idx,
            disagreement_strength=disagreement_strength,
            disagreement_bias=disagreement_bias,
        )

        model = train_trf_improved(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=8e-4,
            weight_decay=1e-4,
            max_epochs=10,
            patience=5,
            pos_weight=pos_weight,
            delta_l2=0.10,
            mean_delta_penalty=mean_delta_penalty,
            trust_reg=0.02,
        )

        p_va, d_va, t_va = predict_trf_improved(
            model=model,
            X_meta=X_va,
            expert_probs=E_va,
            backbone_prob=Bb_va,
            batch_size=4096,
            device=DEVICE,
        )
        p_hold, d_hold, t_hold = predict_trf_improved(
            model=model,
            X_meta=X_meta_holdout,
            expert_probs=E_holdout,
            backbone_prob=B_holdout,
            batch_size=4096,
            device=DEVICE,
        )

        oof_pred[va_idx] = p_va
        oof_delta[va_idx] = d_va
        oof_trust[va_idx] = t_va

        holdout_fold_preds.append(p_hold)
        holdout_fold_deltas.append(d_hold)
        holdout_fold_trust.append(t_hold)

        del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, Bb_tr, Bb_va, y_tr, y_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    runtime_seconds = time.time() - start_t

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
    holdout_delta = np.mean(np.vstack(holdout_fold_deltas), axis=0)
    holdout_trust = np.mean(np.vstack(holdout_fold_trust), axis=0)

    dev_row = evaluate_probs(y_dev, oof_pred, name="TRFImproved_DEV")
    hold_row = evaluate_probs(y_holdout, holdout_pred, name="TRFImproved_HOLDOUT")

    row = {
        "backbone": backbone_name,
        "hidden_dim": hidden_dim,
        "mean_delta_penalty": mean_delta_penalty,
        "use_enhanced_meta": use_enhanced_meta,
        "disagreement_strength": disagreement_strength,
        "disagreement_bias": disagreement_bias,
        "seed_offset": seed_offset,
        "runtime_seconds": float(runtime_seconds),
        "runtime_pretty": summarize_runtime(runtime_seconds),

        "DEV_PR_AUC": dev_row["PR_AUC"],
        "DEV_ROC_AUC": dev_row["ROC_AUC"],
        "DEV_Brier": dev_row["Brier"],
        "DEV_ECE": dev_row["ECE"],
        "DEV_mean_abs_delta": float(np.mean(np.abs(oof_delta))),
        "DEV_mean_delta": float(np.mean(oof_delta)),
        "DEV_mean_trust": float(np.mean(oof_trust)),

        "HOLDOUT_PR_AUC": hold_row["PR_AUC"],
        "HOLDOUT_ROC_AUC": hold_row["ROC_AUC"],
        "HOLDOUT_Brier": hold_row["Brier"],
        "HOLDOUT_ECE": hold_row["ECE"],
        "HOLDOUT_mean_abs_delta": float(np.mean(np.abs(holdout_delta))),
        "HOLDOUT_mean_delta": float(np.mean(holdout_delta)),
        "HOLDOUT_mean_trust": float(np.mean(holdout_trust)),
    }

    return row, holdout_pred

# ============================================================
# Targeted TRF Improvement Block — Step 6
# Small promising search
# ============================================================

best_backbone = "cat_hgb_avg"
B_dev_best, B_hold_best = build_backbone_predictions(
    final_dev_preds=final_dev_preds,
    final_holdout_preds=final_holdout_preds,
    backbone_name=best_backbone,
)

base_meta_dev = dev_meta_df.copy()
base_meta_hold = holdout_meta_df.copy()

enh_meta_dev, enh_meta_hold = build_trf_enhanced_meta_features(
    base_meta_df=base_meta_dev,
    final_dev_preds=final_dev_preds,
    final_holdout_preds=final_holdout_preds,
    expert_list=selected_experts,
    backbone_name=best_backbone,
)

targeted_configs = [
    {"hidden_dim": 64, "mean_delta_penalty": 0.15, "use_enhanced_meta": False, "disagreement_strength": 2.5, "disagreement_bias": 0.10},
    {"hidden_dim": 64, "mean_delta_penalty": 0.20, "use_enhanced_meta": False, "disagreement_strength": 2.5, "disagreement_bias": 0.10},
    {"hidden_dim": 32, "mean_delta_penalty": 0.15, "use_enhanced_meta": False, "disagreement_strength": 2.5, "disagreement_bias": 0.10},
    {"hidden_dim": 32, "mean_delta_penalty": 0.20, "use_enhanced_meta": False, "disagreement_strength": 2.5, "disagreement_bias": 0.10},
    {"hidden_dim": 64, "mean_delta_penalty": 0.15, "use_enhanced_meta": True,  "disagreement_strength": 2.5, "disagreement_bias": 0.10},
    {"hidden_dim": 32, "mean_delta_penalty": 0.15, "use_enhanced_meta": True,  "disagreement_strength": 2.5, "disagreement_bias": 0.10},
]

targeted_rows = []
targeted_holdout_preds = {}

for i, cfg_i in enumerate(targeted_configs, start=1):
    print("\n" + "#" * 100)
    print(f"Running targeted TRF config {i}/{len(targeted_configs)}")
    print(cfg_i)
    print("#" * 100)

    X_dev_use = enh_meta_dev if cfg_i["use_enhanced_meta"] else base_meta_dev
    X_hold_use = enh_meta_hold if cfg_i["use_enhanced_meta"] else base_meta_hold

    row, hold_pred = run_trf_improved_single_experiment(
        X_meta_dev_df=X_dev_use,
        X_meta_holdout_df=X_hold_use,
        E_dev=E_dev_trf,
        E_holdout=E_holdout_trf,
        y_dev=y_dev,
        y_holdout=y_holdout,
        B_dev=B_dev_best,
        B_holdout=B_hold_best,
        config=CFG_TRF_SEARCH if "CFG_TRF_SEARCH" in globals() else CFG,
        backbone_name=best_backbone,
        hidden_dim=cfg_i["hidden_dim"],
        mean_delta_penalty=cfg_i["mean_delta_penalty"],
        use_enhanced_meta=cfg_i["use_enhanced_meta"],
        disagreement_strength=cfg_i["disagreement_strength"],
        disagreement_bias=cfg_i["disagreement_bias"],
        seed_offset=0,
    )

    key = f"cfg_{i}"
    targeted_rows.append(row)
    targeted_holdout_preds[key] = hold_pred

    pd.DataFrame(targeted_rows).to_csv(TAB_DIR / "trf_targeted_improvement_results_partial.csv", index=False)

targeted_df = pd.DataFrame(targeted_rows)
targeted_df.to_csv(TAB_DIR / "trf_targeted_improvement_results.csv", index=False)

display(targeted_df.sort_values(
    ["HOLDOUT_PR_AUC", "HOLDOUT_Brier", "HOLDOUT_ECE"],
    ascending=[False, True, True]
))
```

```python
# ============================================================
# FINAL LOCKED MODEL — TRF Final
# Uses fixed best configuration, no tuning/search
# ============================================================

import gc
import time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader


# ------------------------------------------------------------
# 0. Locked final config
# ------------------------------------------------------------
FINAL_TRF_CONFIG = {
    "backbone_name": "cat_hgb_avg",
    "hidden_dim": 32,
    "max_delta": 0.30,
    "mean_delta_penalty": 0.15,
    "delta_l2": 0.10,
    "trust_reg": 0.02,
    "use_enhanced_meta": True,
    "use_disagreement_scaling": True,
    "disagreement_strength": 2.5,
    "disagreement_bias": 0.10,
    "dropout": 0.10,
    "lr": 8e-4,
    "weight_decay": 1e-4,
    "max_epochs": 24,
    "patience": 5,
    "batch_size": 2048,
    "pred_batch_size": 4096,
    "seed_offsets": [0, 10, 20],  # 3-seed ensemble
}


# ------------------------------------------------------------
# 1. Backbone builder
# ------------------------------------------------------------
def build_backbone_predictions_locked(
    final_dev_preds,
    final_holdout_preds,
    backbone_name: str,
):
    if backbone_name == "catboost":
        b_dev = clip_probs(final_dev_preds["catboost"]).astype(np.float32)
        b_hold = clip_probs(final_holdout_preds["catboost"]).astype(np.float32)
    elif backbone_name == "hgb":
        b_dev = clip_probs(final_dev_preds["hgb"]).astype(np.float32)
        b_hold = clip_probs(final_holdout_preds["hgb"]).astype(np.float32)
    elif backbone_name == "cat_hgb_avg":
        b_dev = clip_probs(
            0.5 * clip_probs(final_dev_preds["catboost"]) +
            0.5 * clip_probs(final_dev_preds["hgb"])
        ).astype(np.float32)
        b_hold = clip_probs(
            0.5 * clip_probs(final_holdout_preds["catboost"]) +
            0.5 * clip_probs(final_holdout_preds["hgb"])
        ).astype(np.float32)
    else:
        raise ValueError(f"Unknown backbone_name: {backbone_name}")
    return b_dev, b_hold


# ------------------------------------------------------------
# 2. Enhanced meta feature builder
# ------------------------------------------------------------
def build_trf_final_meta_features(
    base_dev_meta_df: pd.DataFrame,
    base_holdout_meta_df: pd.DataFrame,
    final_dev_preds,
    final_holdout_preds,
    expert_list,
    backbone_name: str,
):
    dev_df = base_dev_meta_df.copy()
    hold_df = base_holdout_meta_df.copy()

    b_dev, b_hold = build_backbone_predictions_locked(
        final_dev_preds=final_dev_preds,
        final_holdout_preds=final_holdout_preds,
        backbone_name=backbone_name,
    )

    dev_df["backbone_prob"] = b_dev
    hold_df["backbone_prob"] = b_hold

    dev_df["backbone_margin"] = np.abs(b_dev - 0.5)
    hold_df["backbone_margin"] = np.abs(b_hold - 0.5)

    dev_df["backbone_entropy"] = binary_entropy_np(b_dev)
    hold_df["backbone_entropy"] = binary_entropy_np(b_hold)

    dev_df["mean_minus_backbone"] = dev_df["mean_pred"] - b_dev
    hold_df["mean_minus_backbone"] = hold_df["mean_pred"] - b_hold

    dev_df["max_minus_backbone"] = dev_df["max_pred"] - b_dev
    hold_df["max_minus_backbone"] = hold_df["max_pred"] - b_hold

    dev_df["backbone_minus_min"] = b_dev - dev_df["min_pred"]
    hold_df["backbone_minus_min"] = b_hold - hold_df["min_pred"]

    for m in expert_list:
        pcol = f"p_{m}"
        if pcol in dev_df.columns:
            dev_df[f"{pcol}_minus_backbone"] = dev_df[pcol] - b_dev
            hold_df[f"{pcol}_minus_backbone"] = hold_df[pcol] - b_hold

    return dev_df.astype(np.float32), hold_df.astype(np.float32)


# ------------------------------------------------------------
# 3. Dataset
# ------------------------------------------------------------
class TRFFinalDataset(Dataset):
    def __init__(self, X_meta, expert_probs, backbone_prob, y):
        self.X_meta = torch.tensor(X_meta, dtype=torch.float32)
        self.expert_probs = torch.tensor(expert_probs, dtype=torch.float32)
        self.backbone_prob = torch.tensor(backbone_prob, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return (
            self.X_meta[idx],
            self.expert_probs[idx],
            self.backbone_prob[idx],
            self.y[idx],
        )


def make_trf_final_loaders(
    X_meta_tr,
    E_tr,
    B_tr,
    y_tr,
    X_meta_va,
    E_va,
    B_va,
    y_va,
    batch_size=2048,
):
    ds_tr = TRFFinalDataset(X_meta_tr, E_tr, B_tr, y_tr)
    ds_va = TRFFinalDataset(X_meta_va, E_va, B_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va


# ------------------------------------------------------------
# 4. Final locked model
# ------------------------------------------------------------
class TRFFinalModel(nn.Module):
    def __init__(
        self,
        meta_dim: int,
        n_experts: int,
        hidden_dim: int = 32,
        dropout: float = 0.10,
        max_delta: float = 0.30,
        use_disagreement_scaling: bool = True,
        std_feature_index: int = None,
        disagreement_strength: float = 2.5,
        disagreement_bias: float = 0.10,
    ):
        super().__init__()
        self.max_delta = max_delta
        self.use_disagreement_scaling = use_disagreement_scaling
        self.std_feature_index = std_feature_index
        self.disagreement_strength = disagreement_strength
        self.disagreement_bias = disagreement_bias

        in_dim = meta_dim + n_experts + 1

        self.residual_net = nn.Sequential(
            nn.Linear(in_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, hidden_dim),
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(hidden_dim, 1),
        )

        self.trust_gate = nn.Sequential(
            nn.Linear(meta_dim, 32),
            nn.GELU(),
            nn.Linear(32, 1),
            nn.Sigmoid(),
        )

    def forward(self, x_meta, expert_probs, backbone_prob):
        x = torch.cat([x_meta, expert_probs, backbone_prob.unsqueeze(1)], dim=1)

        raw_delta = self.residual_net(x).squeeze(1)
        base_delta = self.max_delta * torch.tanh(raw_delta)

        trust_scale = self.trust_gate(x_meta).squeeze(1)

        if self.use_disagreement_scaling and self.std_feature_index is not None:
            std_pred = torch.clamp(x_meta[:, self.std_feature_index], 0.0, 1.0)
            disagreement_scale = self.disagreement_bias + self.disagreement_strength * std_pred
        else:
            disagreement_scale = 1.0

        delta = base_delta * trust_scale * disagreement_scale

        base_logit = torch.logit(torch.clamp(backbone_prob, 1e-6, 1 - 1e-6))
        final_logit = base_logit + delta
        final_prob = torch.sigmoid(final_logit)
        final_prob = torch.clamp(final_prob, 1e-6, 1 - 1e-6)

        return final_prob, delta, final_logit, trust_scale


# ------------------------------------------------------------
# 5. Train / predict helpers
# ------------------------------------------------------------
def train_trf_final_model(
    model,
    train_loader,
    valid_loader,
    device=DEVICE,
    lr=8e-4,
    weight_decay=1e-4,
    max_epochs=10,
    patience=5,
    pos_weight=None,
    delta_l2=0.10,
    mean_delta_penalty=0.15,
    trust_reg=0.02,
):
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        tr_losses = []

        for xb_meta, xb_exp, xb_back, yb in train_loader:
            xb_meta = xb_meta.to(device)
            xb_exp = xb_exp.to(device)
            xb_back = xb_back.to(device)
            yb = yb.to(device)

            optimizer.zero_grad(set_to_none=True)

            _, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)

            loss_main = criterion(final_logit, yb)
            loss_delta = delta_l2 * torch.mean(delta ** 2)
            loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
            loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)

            loss = loss_main + loss_delta + loss_mean_delta + loss_trust
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            tr_losses.append(loss.item())

        model.eval()
        va_losses = []

        with torch.no_grad():
            for xb_meta, xb_exp, xb_back, yb in valid_loader:
                xb_meta = xb_meta.to(device)
                xb_exp = xb_exp.to(device)
                xb_back = xb_back.to(device)
                yb = yb.to(device)

                _, delta, final_logit, trust_scale = model(xb_meta, xb_exp, xb_back)

                loss_main = criterion(final_logit, yb)
                loss_delta = delta_l2 * torch.mean(delta ** 2)
                loss_mean_delta = mean_delta_penalty * (torch.mean(delta) ** 2)
                loss_trust = trust_reg * torch.mean((trust_scale - 0.5) ** 2)

                loss = loss_main + loss_delta + loss_mean_delta + loss_trust
                va_losses.append(loss.item())

        mean_tr = float(np.mean(tr_losses))
        mean_va = float(np.mean(va_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model


@torch.no_grad()
def predict_trf_final_model(
    model,
    X_meta,
    expert_probs,
    backbone_prob,
    batch_size=4096,
    device=DEVICE,
):
    model.eval()

    X_meta_t = torch.tensor(X_meta, dtype=torch.float32)
    E_t = torch.tensor(expert_probs, dtype=torch.float32)
    B_t = torch.tensor(backbone_prob, dtype=torch.float32)

    ds = torch.utils.data.TensorDataset(X_meta_t, E_t, B_t)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs, all_delta, all_trust = [], [], []

    for xb_meta, xb_exp, xb_back in dl:
        xb_meta = xb_meta.to(device)
        xb_exp = xb_exp.to(device)
        xb_back = xb_back.to(device)

        p_hat, delta, _, trust_scale = model(xb_meta, xb_exp, xb_back)
        all_probs.append(p_hat.detach().cpu().numpy())
        all_delta.append(delta.detach().cpu().numpy())
        all_trust.append(trust_scale.detach().cpu().numpy())

    probs = clip_probs(np.concatenate(all_probs))
    deltas = np.concatenate(all_delta)
    trust = np.concatenate(all_trust)
    return probs, deltas, trust


# ------------------------------------------------------------
# 6. Final runner (3-seed ensemble)
# ------------------------------------------------------------
def run_final_trf_locked(
    X_dev,
    X_holdout,
    y_dev,
    y_holdout,
    dev_meta_df,
    holdout_meta_df,
    final_dev_preds,
    final_holdout_preds,
    selected_experts,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
):
    backbone_name = final_cfg["backbone_name"]

    # backbone
    B_dev, B_hold = build_backbone_predictions_locked(
        final_dev_preds=final_dev_preds,
        final_holdout_preds=final_holdout_preds,
        backbone_name=backbone_name,
    )

    # expert matrices
    E_dev = np.column_stack([clip_probs(final_dev_preds[m]) for m in selected_experts]).astype(np.float32)
    E_hold = np.column_stack([clip_probs(final_holdout_preds[m]) for m in selected_experts]).astype(np.float32)

    # meta features
    if final_cfg["use_enhanced_meta"]:
        meta_dev_df, meta_hold_df = build_trf_final_meta_features(
            base_dev_meta_df=dev_meta_df,
            base_holdout_meta_df=holdout_meta_df,
            final_dev_preds=final_dev_preds,
            final_holdout_preds=final_holdout_preds,
            expert_list=selected_experts,
            backbone_name=backbone_name,
        )
    else:
        meta_dev_df, meta_hold_df = dev_meta_df.copy(), holdout_meta_df.copy()

    X_meta_dev = meta_dev_df.values.astype(np.float32)
    X_meta_hold = meta_hold_df.values.astype(np.float32)

    std_idx = meta_dev_df.columns.get_loc("std_pred") if "std_pred" in meta_dev_df.columns else None
    cv = get_cv(config, final_stage=True)

    seed_offsets = final_cfg["seed_offsets"]

    all_seed_dev_preds = []
    all_seed_hold_preds = []
    all_seed_dev_deltas = []
    all_seed_hold_deltas = []
    all_seed_dev_trust = []
    all_seed_hold_trust = []

    seed_rows = []
    fold_rows = []

    total_start = time.time()

    for seed_offset in seed_offsets:
        print("=" * 100)
        print(f"Running final TRF seed ensemble member with seed_offset={seed_offset}")
        print("=" * 100)

        oof_pred = np.zeros(len(y_dev), dtype=np.float32)
        oof_delta = np.zeros(len(y_dev), dtype=np.float32)
        oof_trust = np.zeros(len(y_dev), dtype=np.float32)

        holdout_fold_preds = []
        holdout_fold_deltas = []
        holdout_fold_trust = []

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
            print("-" * 90)
            print(f"[Final TRF] seed_offset={seed_offset} | fold={fold_id}")
            print("-" * 90)

            X_tr = X_meta_dev[tr_idx]
            X_va = X_meta_dev[va_idx]
            E_tr = E_dev[tr_idx]
            E_va = E_dev[va_idx]
            Bb_tr = B_dev[tr_idx]
            Bb_va = B_dev[va_idx]
            y_tr = y_dev[tr_idx].astype(np.float32)
            y_va = y_dev[va_idx].astype(np.float32)

            pos = float((y_tr == 1).sum())
            neg = float((y_tr == 0).sum())
            pos_weight = neg / max(pos, 1.0)

            dl_tr, dl_va = make_trf_final_loaders(
                X_meta_tr=X_tr,
                E_tr=E_tr,
                B_tr=Bb_tr,
                y_tr=y_tr,
                X_meta_va=X_va,
                E_va=E_va,
                B_va=Bb_va,
                y_va=y_va,
                batch_size=final_cfg["batch_size"],
            )

            set_global_seed(config.seed + seed_offset + fold_id)

            model = TRFFinalModel(
                meta_dim=X_meta_dev.shape[1],
                n_experts=E_dev.shape[1],
                hidden_dim=final_cfg["hidden_dim"],
                dropout=final_cfg["dropout"],
                max_delta=final_cfg["max_delta"],
                use_disagreement_scaling=final_cfg["use_disagreement_scaling"],
                std_feature_index=std_idx,
                disagreement_strength=final_cfg["disagreement_strength"],
                disagreement_bias=final_cfg["disagreement_bias"],
            )

            model = train_trf_final_model(
                model=model,
                train_loader=dl_tr,
                valid_loader=dl_va,
                device=DEVICE,
                lr=final_cfg["lr"],
                weight_decay=final_cfg["weight_decay"],
                max_epochs=final_cfg["max_epochs"],
                patience=final_cfg["patience"],
                pos_weight=pos_weight,
                delta_l2=final_cfg["delta_l2"],
                mean_delta_penalty=final_cfg["mean_delta_penalty"],
                trust_reg=final_cfg["trust_reg"],
            )

            p_va, d_va, t_va = predict_trf_final_model(
                model=model,
                X_meta=X_va,
                expert_probs=E_va,
                backbone_prob=Bb_va,
                batch_size=final_cfg["pred_batch_size"],
                device=DEVICE,
            )
            p_hold, d_hold, t_hold = predict_trf_final_model(
                model=model,
                X_meta=X_meta_hold,
                expert_probs=E_hold,
                backbone_prob=B_hold,
                batch_size=final_cfg["pred_batch_size"],
                device=DEVICE,
            )

            oof_pred[va_idx] = p_va
            oof_delta[va_idx] = d_va
            oof_trust[va_idx] = t_va

            holdout_fold_preds.append(p_hold)
            holdout_fold_deltas.append(d_hold)
            holdout_fold_trust.append(t_hold)

            fold_row = evaluate_probs(y_dev[va_idx], p_va, name=f"TRF_final_seed{seed_offset}_fold{fold_id}")
            fold_row["seed_offset"] = seed_offset
            fold_row["fold_id"] = fold_id
            fold_row["mean_abs_delta"] = float(np.mean(np.abs(d_va)))
            fold_row["mean_delta"] = float(np.mean(d_va))
            fold_row["mean_trust"] = float(np.mean(t_va))
            fold_rows.append(fold_row)

            del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, Bb_tr, Bb_va, y_tr, y_va
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))
        holdout_delta = np.mean(np.vstack(holdout_fold_deltas), axis=0)
        holdout_trust = np.mean(np.vstack(holdout_fold_trust), axis=0)

        dev_row = evaluate_probs(y_dev, oof_pred, name=f"TRF_final_seed{seed_offset}_DEV")
        hold_row = evaluate_probs(y_holdout, holdout_pred, name=f"TRF_final_seed{seed_offset}_HOLDOUT")

        seed_rows.append({
            "seed_offset": seed_offset,
            "DEV_PR_AUC": dev_row["PR_AUC"],
            "DEV_ROC_AUC": dev_row["ROC_AUC"],
            "DEV_Brier": dev_row["Brier"],
            "DEV_ECE": dev_row["ECE"],
            "DEV_mean_abs_delta": float(np.mean(np.abs(oof_delta))),
            "DEV_mean_delta": float(np.mean(oof_delta)),
            "DEV_mean_trust": float(np.mean(oof_trust)),
            "HOLDOUT_PR_AUC": hold_row["PR_AUC"],
            "HOLDOUT_ROC_AUC": hold_row["ROC_AUC"],
            "HOLDOUT_Brier": hold_row["Brier"],
            "HOLDOUT_ECE": hold_row["ECE"],
            "HOLDOUT_mean_abs_delta": float(np.mean(np.abs(holdout_delta))),
            "HOLDOUT_mean_delta": float(np.mean(holdout_delta)),
            "HOLDOUT_mean_trust": float(np.mean(holdout_trust)),
        })

        all_seed_dev_preds.append(oof_pred)
        all_seed_hold_preds.append(holdout_pred)
        all_seed_dev_deltas.append(oof_delta)
        all_seed_hold_deltas.append(holdout_delta)
        all_seed_dev_trust.append(oof_trust)
        all_seed_hold_trust.append(holdout_trust)

    # ensemble over seeds
    final_dev_pred = clip_probs(np.mean(np.vstack(all_seed_dev_preds), axis=0))
    final_hold_pred = clip_probs(np.mean(np.vstack(all_seed_hold_preds), axis=0))
    final_dev_delta = np.mean(np.vstack(all_seed_dev_deltas), axis=0)
    final_hold_delta = np.mean(np.vstack(all_seed_hold_deltas), axis=0)
    final_dev_trust = np.mean(np.vstack(all_seed_dev_trust), axis=0)
    final_hold_trust = np.mean(np.vstack(all_seed_hold_trust), axis=0)

    runtime_seconds = time.time() - total_start

    dev_final_row = evaluate_probs(y_dev, final_dev_pred, name="TRF_FINAL_DEV_OOF")
    dev_final_row.update({
        "model_name": "TRF_FINAL",
        "split": "DEV_OOF",
        "runtime_seconds": float(runtime_seconds),
        "runtime_pretty": summarize_runtime(runtime_seconds),
        "backbone": backbone_name,
        "mean_abs_delta": float(np.mean(np.abs(final_dev_delta))),
        "mean_delta": float(np.mean(final_dev_delta)),
        "mean_trust": float(np.mean(final_dev_trust)),
    })

    hold_final_row = evaluate_probs(y_holdout, final_hold_pred, name="TRF_FINAL_HOLDOUT")
    hold_final_row.update({
        "model_name": "TRF_FINAL",
        "split": "HOLDOUT",
        "runtime_seconds": float(runtime_seconds),
        "runtime_pretty": summarize_runtime(runtime_seconds),
        "backbone": backbone_name,
        "mean_abs_delta": float(np.mean(np.abs(final_hold_delta))),
        "mean_delta": float(np.mean(final_hold_delta)),
        "mean_trust": float(np.mean(final_hold_trust)),
    })

    results_df = pd.DataFrame([dev_final_row, hold_final_row])
    seed_df = pd.DataFrame(seed_rows)
    fold_df = pd.DataFrame(fold_rows)

    # save
    results_df.to_csv(TAB_DIR / "trf_final_results.csv", index=False)
    seed_df.to_csv(TAB_DIR / "trf_final_seed_results.csv", index=False)
    fold_df.to_csv(TAB_DIR / "trf_final_fold_results.csv", index=False)

    pd.DataFrame({
        "y_true": y_dev.astype(int),
        "p_backbone": B_dev,
        "p_trf_final": final_dev_pred,
        "delta_trf_final": final_dev_delta,
        "trust_trf_final": final_dev_trust,
    }).to_csv(PRED_DIR / "trf_final_dev_oof.csv", index=False)

    pd.DataFrame({
        "y_true": y_holdout.astype(int),
        "p_backbone": B_hold,
        "p_trf_final": final_hold_pred,
        "delta_trf_final": final_hold_delta,
        "trust_trf_final": final_hold_trust,
    }).to_csv(PRED_DIR / "trf_final_holdout.csv", index=False)

    return {
        "config": final_cfg,
        "results_df": results_df,
        "seed_df": seed_df,
        "fold_df": fold_df,
        "final_dev_pred": final_dev_pred,
        "final_hold_pred": final_hold_pred,
        "final_dev_delta": final_dev_delta,
        "final_hold_delta": final_hold_delta,
        "final_dev_trust": final_dev_trust,
        "final_hold_trust": final_hold_trust,
        "meta_dev_df": meta_dev_df,
        "meta_hold_df": meta_hold_df,
    }


# ------------------------------------------------------------
# 7. Run final model
# ------------------------------------------------------------
FINAL_TRF_OUTPUT = run_final_trf_locked(
    X_dev=X_dev,
    X_holdout=X_holdout,
    y_dev=y_dev,
    y_holdout=y_holdout,
    dev_meta_df=dev_meta_df,
    holdout_meta_df=holdout_meta_df,
    final_dev_preds=final_dev_preds,
    final_holdout_preds=final_holdout_preds,
    selected_experts=selected_experts,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

display(FINAL_TRF_OUTPUT["results_df"][[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE",
    "runtime_pretty", "backbone", "mean_abs_delta", "mean_delta", "mean_trust"
]])

display(FINAL_TRF_OUTPUT["seed_df"])
```

```python
# ============================================================
# Fix Module 8 outputs so Module 11 can use them
# ============================================================

from sklearn.linear_model import LogisticRegression

def stack_predictions(pred_dict: Dict[str, np.ndarray], model_list: List[str]) -> np.ndarray:
    return np.column_stack([pred_dict[m] for m in model_list])

def simple_average_ensemble(pred_dict, model_list):
    preds = stack_predictions(pred_dict, model_list)
    return clip_probs(np.mean(preds, axis=1))

def performance_weighted_ensemble(dev_pred_dict, hold_pred_dict, model_list, results_df):
    weights = []
    for m in model_list:
        pr_auc = results_df[
            (results_df["model_name"] == m) &
            (results_df["split"] == "DEV_OOF")
        ]["PR_AUC"].values[0]
        weights.append(pr_auc)

    weights = np.array(weights, dtype=float)
    weights = weights / weights.sum()

    dev_preds = stack_predictions(dev_pred_dict, model_list)
    hold_preds = stack_predictions(hold_pred_dict, model_list)

    dev_weighted = clip_probs(np.dot(dev_preds, weights))
    hold_weighted = clip_probs(np.dot(hold_preds, weights))

    return dev_weighted, hold_weighted, weights

def stacking_ensemble(dev_preds, holdout_preds, y_dev, model_list):
    X_dev_stack = stack_predictions(dev_preds, model_list)
    X_holdout_stack = stack_predictions(holdout_preds, model_list)

    meta_model = LogisticRegression(
        max_iter=2000,
        class_weight="balanced"
    )
    meta_model.fit(X_dev_stack, y_dev)

    dev_pred = clip_probs(meta_model.predict_proba(X_dev_stack)[:, 1])
    holdout_pred = clip_probs(meta_model.predict_proba(X_holdout_stack)[:, 1])

    return dev_pred, holdout_pred, meta_model


# Rebuild ensemble predictions correctly
avg_dev = simple_average_ensemble(final_dev_preds, selected_experts)
avg_hold = simple_average_ensemble(final_holdout_preds, selected_experts)

w_dev, w_hold, weights = performance_weighted_ensemble(
    final_dev_preds,
    final_holdout_preds,
    selected_experts,
    final_expert_results_df
)

stack_dev, stack_hold, stack_model = stacking_ensemble(
    final_dev_preds,
    final_holdout_preds,
    y_dev,
    selected_experts
)

print("Weights used for weighted ensemble:")
for m, w in zip(selected_experts, weights):
    print(f"{m}: {w:.4f}")


# Save ensemble predictions into dictionaries for later modules
final_dev_preds["simple_average"] = avg_dev
final_holdout_preds["simple_average"] = avg_hold

final_dev_preds["weighted_ensemble"] = w_dev
final_holdout_preds["weighted_ensemble"] = w_hold

final_dev_preds["stacking"] = stack_dev
final_holdout_preds["stacking"] = stack_hold


# Rebuild ensemble summary table
ensemble_results = []

def eval_and_store(name, dev_pred, hold_pred):
    dev_row = evaluate_probs(y_dev, dev_pred, name=name + "_DEV")
    dev_row["model_name"] = name
    dev_row["split"] = "DEV_OOF"

    hold_row = evaluate_probs(y_holdout, hold_pred, name=name + "_HOLDOUT")
    hold_row["model_name"] = name
    hold_row["split"] = "HOLDOUT"

    ensemble_results.extend([dev_row, hold_row])

eval_and_store("SimpleAverage", avg_dev, avg_hold)
eval_and_store("WeightedEnsemble", w_dev, w_hold)
eval_and_store("Stacking", stack_dev, stack_hold)

ensemble_df = pd.DataFrame(ensemble_results)
ensemble_df.to_csv(TAB_DIR / "ensemble_results.csv", index=False)

display(
    ensemble_df[
        ["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["split", "PR_AUC"], ascending=[True, False])
)

models = {
    "CatBoost": final_dev_preds["catboost"],
    "WeightedEnsemble": final_dev_preds["weighted_ensemble"],
    "TrustAwareMoE": trust_moe_dev_pred,   # or FINAL_MOE_OUTPUT["final_dev_pred"] if you stored it that way
    "TRF_Final": FINAL_TRF_OUTPUT["final_dev_pred"],
}
```

```python
# pick the TRF dev prediction variable that actually exists
if "FINAL_TRF_OUTPUT" in globals():
    trf_dev_for_analysis = FINAL_TRF_OUTPUT["final_dev_pred"]
elif "trf_v2_dev_pred" in globals():
    trf_dev_for_analysis = trf_v2_dev_pred
elif "trf_dev_pred" in globals():
    trf_dev_for_analysis = trf_dev_pred
else:
    raise NameError("No TRF dev prediction variable found.")

# pick the MoE dev prediction variable that actually exists
if "FINAL_MOE_OUTPUT" in globals():
    moe_dev_for_analysis = FINAL_MOE_OUTPUT["final_dev_pred"]
elif "trust_moe_dev_pred" in globals():
    moe_dev_for_analysis = trust_moe_dev_pred
else:
    raise NameError("No MoE dev prediction variable found.")

models = {
    "CatBoost": final_dev_preds["catboost"],
    "WeightedEnsemble": final_dev_preds["weighted_ensemble"],
    "TrustAwareMoE": moe_dev_for_analysis,
    "TRF_Final": trf_dev_for_analysis,
}
```

```python
# ============================================================
# Module 11 — Decision / Threshold Analysis
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve


# ------------------------------------------------------------
# 1. Helper functions
# ------------------------------------------------------------
def compute_metrics_at_threshold(y_true, y_prob, thr):
    y_pred = (y_prob >= thr).astype(int)

    tp = ((y_pred == 1) & (y_true == 1)).sum()
    fp = ((y_pred == 1) & (y_true == 0)).sum()
    fn = ((y_pred == 0) & (y_true == 1)).sum()
    tn = ((y_pred == 0) & (y_true == 0)).sum()

    precision = tp / (tp + fp + 1e-12)
    recall = tp / (tp + fn + 1e-12)

    f1 = 2 * precision * recall / (precision + recall + 1e-12)
    accuracy = (tp + tn) / (tp + tn + fp + fn)

    return {
        "threshold": thr,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "accuracy": accuracy,
        "tp": tp,
        "fp": fp,
        "fn": fn,
        "tn": tn,
    }


# ------------------------------------------------------------
# 2. Threshold sweep
# ------------------------------------------------------------
def threshold_sweep(y_true, y_prob, name):
    thresholds = np.linspace(0.01, 0.99, 50)

    rows = []
    for t in thresholds:
        r = compute_metrics_at_threshold(y_true, y_prob, t)
        r["model"] = name
        rows.append(r)

    df = pd.DataFrame(rows)
    return df


# ------------------------------------------------------------
# 3. Recall @ Precision
# ------------------------------------------------------------
def recall_at_precision(y_true, y_prob, target_precision):
    precision, recall, thresholds = precision_recall_curve(y_true, y_prob)

    mask = precision >= target_precision
    if mask.sum() == 0:
        return np.nan

    return recall[mask].max()


# ------------------------------------------------------------
# 4. Precision @ Recall
# ------------------------------------------------------------
def precision_at_recall(y_true, y_prob, target_recall):
    precision, recall, thresholds = precision_recall_curve(y_true, y_prob)

    mask = recall >= target_recall
    if mask.sum() == 0:
        return np.nan

    return precision[mask].max()


# ------------------------------------------------------------
# 5. Top-K analysis
# ------------------------------------------------------------
def topk_analysis(y_true, y_prob, k_list=[0.01, 0.02, 0.05, 0.10]):
    N = len(y_true)
    order = np.argsort(-y_prob)

    rows = []

    for k in k_list:
        top_n = int(N * k)
        idx = order[:top_n]

        tp = y_true[idx].sum()
        precision = tp / top_n
        recall = tp / y_true.sum()

        rows.append({
            "k_frac": k,
            "top_n": top_n,
            "precision": precision,
            "recall": recall,
            "tp": tp,
        })

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# 6. Run analysis
# ------------------------------------------------------------
# pick the TRF dev prediction variable that actually exists
if "FINAL_TRF_OUTPUT" in globals():
    trf_dev_for_analysis = FINAL_TRF_OUTPUT["final_dev_pred"]
elif "trf_v2_dev_pred" in globals():
    trf_dev_for_analysis = trf_v2_dev_pred
elif "trf_dev_pred" in globals():
    trf_dev_for_analysis = trf_dev_pred
else:
    raise NameError("No TRF dev prediction variable found.")

models = {
    "CatBoost": final_dev_preds["catboost"],
    "HGB": final_dev_preds["hgb"],
    "DeepFM": final_dev_preds["deepfm"],
    "WeightedEnsemble": final_dev_preds["weighted_ensemble"],
    "SimpleAverage": final_dev_preds["simple_average"],
    "Stacking": final_dev_preds["stacking"],
    "TRF_Final": trf_dev_for_analysis,
}

threshold_tables = []
recall_precision_rows = []
topk_tables = []

for name, prob in models.items():
    prob = clip_probs(prob)

    # Threshold sweep
    df_thr = threshold_sweep(y_dev, prob, name)
    threshold_tables.append(df_thr)

    # Recall @ Precision
    for p in [0.7, 0.8, 0.9]:
        r = recall_at_precision(y_dev, prob, p)
        recall_precision_rows.append({
            "model": name,
            "target_precision": p,
            "recall": r,
        })

    # Precision @ Recall
    for r_target in [0.6, 0.7, 0.8]:
        p_val = precision_at_recall(y_dev, prob, r_target)
        recall_precision_rows.append({
            "model": name,
            "target_recall": r_target,
            "precision": p_val,
        })

    # Top-k
    df_topk = topk_analysis(y_dev, prob)
    df_topk["model"] = name
    topk_tables.append(df_topk)


threshold_df = pd.concat(threshold_tables, ignore_index=True)
recall_precision_df = pd.DataFrame(recall_precision_rows)
topk_df = pd.concat(topk_tables, ignore_index=True)

# Save
threshold_df.to_csv(TAB_DIR / "threshold_sweep.csv", index=False)
recall_precision_df.to_csv(TAB_DIR / "precision_recall_targets.csv", index=False)
topk_df.to_csv(TAB_DIR / "topk_analysis.csv", index=False)

display(threshold_df.head())
display(recall_precision_df)
display(topk_df.head())

# PR curve
plt.figure(figsize=(6,5))
for name, prob in models.items():
    precision, recall, _ = precision_recall_curve(y_dev, prob)
    plt.plot(recall, precision, label=name)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid(True)
plt.savefig(FIG_DIR / "pr_curve.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Module 11 — Decision / Threshold Analysis
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve


# ------------------------------------------------------------
# 1. Helper functions
# ------------------------------------------------------------
def compute_metrics_at_threshold(y_true, y_prob, thr):
    y_pred = (y_prob >= thr).astype(int)

    tp = ((y_pred == 1) & (y_true == 1)).sum()
    fp = ((y_pred == 1) & (y_true == 0)).sum()
    fn = ((y_pred == 0) & (y_true == 1)).sum()
    tn = ((y_pred == 0) & (y_true == 0)).sum()

    precision = tp / (tp + fp + 1e-12)
    recall = tp / (tp + fn + 1e-12)

    f1 = 2 * precision * recall / (precision + recall + 1e-12)
    accuracy = (tp + tn) / (tp + tn + fp + fn)

    return {
        "threshold": thr,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "accuracy": accuracy,
        "tp": tp,
        "fp": fp,
        "fn": fn,
        "tn": tn,
    }


# ------------------------------------------------------------
# 2. Threshold sweep
# ------------------------------------------------------------
def threshold_sweep(y_true, y_prob, name):
    thresholds = np.linspace(0.01, 0.99, 50)

    rows = []
    for t in thresholds:
        r = compute_metrics_at_threshold(y_true, y_prob, t)
        r["model"] = name
        rows.append(r)

    df = pd.DataFrame(rows)
    return df


# ------------------------------------------------------------
# 3. Recall @ Precision
# ------------------------------------------------------------
def recall_at_precision(y_true, y_prob, target_precision):
    precision, recall, thresholds = precision_recall_curve(y_true, y_prob)

    mask = precision >= target_precision
    if mask.sum() == 0:
        return np.nan

    return recall[mask].max()


# ------------------------------------------------------------
# 4. Precision @ Recall
# ------------------------------------------------------------
def precision_at_recall(y_true, y_prob, target_recall):
    precision, recall, thresholds = precision_recall_curve(y_true, y_prob)

    mask = recall >= target_recall
    if mask.sum() == 0:
        return np.nan

    return precision[mask].max()


# ------------------------------------------------------------
# 5. Top-K analysis
# ------------------------------------------------------------
def topk_analysis(y_true, y_prob, k_list=[0.01, 0.02, 0.05, 0.10]):
    N = len(y_true)
    order = np.argsort(-y_prob)

    rows = []

    for k in k_list:
        top_n = int(N * k)
        idx = order[:top_n]

        tp = y_true[idx].sum()
        precision = tp / top_n
        recall = tp / y_true.sum()

        rows.append({
            "k_frac": k,
            "top_n": top_n,
            "precision": precision,
            "recall": recall,
            "tp": tp,
        })

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# 6. Run analysis
# ------------------------------------------------------------
# pick the TRF holdout prediction variable that actually exists
if "FINAL_TRF_OUTPUT" in globals():
    trf_hold_for_analysis = FINAL_TRF_OUTPUT["final_hold_pred"]
elif "trf_v2_holdout_pred" in globals():
    trf_hold_for_analysis = trf_v2_holdout_pred
elif "trf_holdout_pred" in globals():
    trf_hold_for_analysis = trf_holdout_pred
else:
    raise NameError("No TRF holdout prediction variable found.")

holdout_models = {
    "CatBoost": final_holdout_preds["catboost"],
    "HGB": final_holdout_preds["hgb"],
    "DeepFM": final_holdout_preds["deepfm"],
    "WeightedEnsemble": final_holdout_preds["weighted_ensemble"],
    "SimpleAverage": final_holdout_preds["simple_average"],
    "Stacking": final_holdout_preds["stacking"],
    "TRF_Final": trf_hold_for_analysis,
}

threshold_tables = []
recall_precision_rows = []
topk_tables = []

for name, prob in models.items():
    prob = clip_probs(prob)

    # Threshold sweep
    df_thr = threshold_sweep(y_dev, prob, name)
    threshold_tables.append(df_thr)

    # Recall @ Precision
    for p in [0.7, 0.8, 0.9]:
        r = recall_at_precision(y_dev, prob, p)
        recall_precision_rows.append({
            "model": name,
            "target_precision": p,
            "recall": r,
        })

    # Precision @ Recall
    for r_target in [0.6, 0.7, 0.8]:
        p_val = precision_at_recall(y_dev, prob, r_target)
        recall_precision_rows.append({
            "model": name,
            "target_recall": r_target,
            "precision": p_val,
        })

    # Top-k
    df_topk = topk_analysis(y_dev, prob)
    df_topk["model"] = name
    topk_tables.append(df_topk)


threshold_df = pd.concat(threshold_tables, ignore_index=True)
recall_precision_df = pd.DataFrame(recall_precision_rows)
topk_df = pd.concat(topk_tables, ignore_index=True)

# Save
threshold_df.to_csv(TAB_DIR / "threshold_sweep.csv", index=False)
recall_precision_df.to_csv(TAB_DIR / "precision_recall_targets.csv", index=False)
topk_df.to_csv(TAB_DIR / "topk_analysis.csv", index=False)

display(threshold_df.head())
display(recall_precision_df)
display(topk_df.head())

# PR curve
plt.figure(figsize=(6,5))
for name, prob in models.items():
    precision, recall, _ = precision_recall_curve(y_dev, prob)
    plt.plot(recall, precision, label=name)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid(True)
plt.savefig(FIG_DIR / "pr_curve.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Module 12 — Selective Prediction / Uncertainty
# ============================================================

import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Uncertainty function
# ------------------------------------------------------------
def compute_uncertainty(p):
    return p * (1 - p)


# ------------------------------------------------------------
# 2. Selective evaluation
# ------------------------------------------------------------
def selective_metrics(y_true, y_prob, coverage_list=np.linspace(0.1, 1.0, 10)):
    uncertainty = compute_uncertainty(y_prob)

    order = np.argsort(uncertainty)  # lowest uncertainty first

    rows = []
    N = len(y_true)

    for c in coverage_list:
        k = int(N * c)
        idx = order[:k]

        y_sub = y_true[idx]
        p_sub = y_prob[idx]

        metrics = evaluate_probs(y_sub, p_sub, name="selective")

        rows.append({
            "coverage": c,
            "n_samples": k,
            "PR_AUC": metrics["PR_AUC"],
            "ROC_AUC": metrics["ROC_AUC"],
            "Brier": metrics["Brier"],
            "ECE": metrics["ECE"],
        })

    return pd.DataFrame(rows)

# ------------------------------------------------------------
# 3. Run for all models (HOLDOUT)
# ------------------------------------------------------------
if "FINAL_TRF_OUTPUT" in globals():
    trf_hold = FINAL_TRF_OUTPUT["final_hold_pred"]
elif "trf_v2_holdout_pred" in globals():
    trf_hold = trf_v2_holdout_pred
elif "trf_holdout_pred" in globals():
    trf_hold = trf_holdout_pred
else:
    raise NameError("No TRF holdout prediction found.")

models = {
    "CatBoost": final_holdout_preds["catboost"],
    "HGB": final_holdout_preds["hgb"],
    "DeepFM": final_holdout_preds["deepfm"],
    "WeightedEnsemble": final_holdout_preds["weighted_ensemble"],
    "Stacking": final_holdout_preds["stacking"],
    "Proposed Model": trf_hold,
}

selective_results = []

for name, prob in models.items():
    prob = clip_probs(prob)

    df_sel = selective_metrics(y_holdout, prob)
    df_sel["model"] = name

    selective_results.append(df_sel)

selective_df = pd.concat(selective_results, ignore_index=True)

# Save
selective_df.to_csv(TAB_DIR / "selective_prediction.csv", index=False)

display(selective_df.head())

import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

for name in selective_df["model"].unique():
    df = selective_df[selective_df["model"] == name]
    plt.plot(df["coverage"], df["Brier"], label=name)

plt.xlabel("Coverage")
plt.ylabel("Brier Score (↓ better)")
plt.title("Risk–Coverage Curve")
plt.legend()
plt.grid(True)

plt.savefig(FIG_DIR / "risk_coverage.png", dpi=300)
plt.show()
```

```python
def high_confidence_error(y_true, y_prob, quantile=0.2):
    unc = y_prob * (1 - y_prob)
    thr = np.quantile(unc, quantile)
    idx = unc <= thr

    y_pred = (y_prob[idx] >= 0.5).astype(int)
    error = (y_pred != y_true[idx]).mean()

    return error

for name, prob in models.items():
    err = high_confidence_error(y_holdout, prob, 0.2)
    print(name, err)
```

```python
# ============================================================
# Module 13 — Robustness / Stress Testing
# Holdout-only perturbation study
# ============================================================

import gc
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from copy import deepcopy
from sklearn.pipeline import Pipeline

# ------------------------------------------------------------
# 0. Config
# ------------------------------------------------------------
ROBUSTNESS_CONFIG = {
    "scenarios": [
        {"name": "clean",          "kind": "clean",         "strength": 0.00},
        {"name": "noise_0.25",     "kind": "gaussian",      "strength": 0.25},
        {"name": "noise_0.50",     "kind": "gaussian",      "strength": 0.50},
        {"name": "mask_0.10",      "kind": "feature_mask",  "strength": 0.10},
        {"name": "mask_0.20",      "kind": "feature_mask",  "strength": 0.20},
        {"name": "corrupt_top3",   "kind": "corrupt_cols",  "strength": 1.00},
    ],
    # adjust these to match your dataset if needed
    "top_corrupt_cols_candidates": [
        "BMI", "GenHlth", "Age", "HighBP", "HighChol", "PhysHlth", "MentHlth"
    ],
    "use_cfg_final_cv": True,
}

print("Robustness scenarios:")
for s in ROBUSTNESS_CONFIG["scenarios"]:
    print(s)

# ============================================================
# 1. Scenario builders
# ============================================================

def get_numeric_cols(df: pd.DataFrame):
    return [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]

def get_existing_corrupt_cols(df: pd.DataFrame, candidates):
    return [c for c in candidates if c in df.columns][:3]

def make_holdout_scenarios(X_dev: pd.DataFrame, X_holdout: pd.DataFrame, cfg: dict):
    rng = np.random.default_rng(CFG.seed)
    scenarios = {}

    numeric_cols = get_numeric_cols(X_holdout)
    dev_std = X_dev[numeric_cols].std().replace(0, 1.0)
    dev_median = X_dev[numeric_cols].median()

    corrupt_cols = get_existing_corrupt_cols(X_holdout, cfg["top_corrupt_cols_candidates"])

    for sc in cfg["scenarios"]:
        name = sc["name"]
        kind = sc["kind"]
        strength = sc["strength"]

        Xp = X_holdout.copy()

        if kind == "clean":
            pass

        elif kind == "gaussian":
            noise = rng.normal(
                loc=0.0,
                scale=(dev_std.values * strength),
                size=(len(Xp), len(numeric_cols))
            )
            Xp.loc[:, numeric_cols] = Xp[numeric_cols].values + noise

        elif kind == "feature_mask":
            mask = rng.random((len(Xp), len(numeric_cols))) < strength
            arr = Xp[numeric_cols].values.copy()
            med = dev_median.values.reshape(1, -1)
            arr[mask] = np.broadcast_to(med, arr.shape)[mask]
            Xp.loc[:, numeric_cols] = arr

        elif kind == "corrupt_cols":
            for c in corrupt_cols:
                Xp[c] = rng.permutation(Xp[c].values)

        else:
            raise ValueError(f"Unknown scenario kind: {kind}")

        scenarios[name] = Xp

    return scenarios


X_HOLDOUT_SCENARIOS = make_holdout_scenarios(X_dev, X_holdout, ROBUSTNESS_CONFIG)
print("Scenario keys:", list(X_HOLDOUT_SCENARIOS.keys()))

# ============================================================
# 2. Expert holdout predictions for all scenarios
# Train clean on each dev fold, infer on every perturbed holdout scenario
# ============================================================

def get_expert_holdout_preds_all_scenarios(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    holdout_scenarios: dict,
    selected_models: list,
    registry: dict,
    config: ExperimentConfig,
):
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    scenario_model_fold_preds = {
        sc_name: {m: [] for m in selected_models}
        for sc_name in holdout_scenarios.keys()
    }

    for model_name in selected_models:
        meta = registry[model_name]
        print("=" * 100)
        print(f"[Robustness] Expert training for model={model_name}")
        print("=" * 100)

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
            print(f"fold {fold_id}")

            X_tr_df = X_dev.iloc[tr_idx].copy()
            y_tr = y_dev[tr_idx]
            y_va = y_dev[va_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            if meta["kind"] == "sklearn":
                pipe = Pipeline([
                    ("pre", preproc),
                    ("model", deepcopy(meta["estimator"])),
                ])
                pipe.fit(X_tr_df, y_tr)

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    p = get_sklearn_probabilities(pipe, X_hold_sc)
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del pipe

            elif meta["kind"] == "torch":
                # fit preprocessor once on this training fold
                X_tr = preproc.fit_transform(X_tr_df)
                if hasattr(X_tr, "toarray"):
                    X_tr = X_tr.toarray()
                X_tr = np.asarray(X_tr, dtype=np.float32)

                # small inner split from training fold for early stopping
                inner_rng = np.random.default_rng(CFG.seed + fold_id)
                idx = np.arange(len(X_tr))
                inner_rng.shuffle(idx)
                cut = int(0.9 * len(idx))
                tr2, va2 = idx[:cut], idx[cut:]

                X_fit = X_tr[tr2]
                y_fit = y_tr[tr2].astype(np.float32)
                X_es = X_tr[va2]
                y_es = y_tr[va2].astype(np.float32)

                pos = float((y_fit == 1).sum())
                neg = float((y_fit == 0).sum())
                pos_weight = neg / max(pos, 1.0)

                dl_tr, dl_va = make_torch_loaders(
                    X_tr=X_fit,
                    y_tr=y_fit,
                    X_va=X_es,
                    y_va=y_es,
                    batch_size=int(meta["train_params"].get("batch_size", 1024)),
                )

                model = meta["builder"](X_fit.shape[1])
                model = train_torch_binary_model(
                    model=model,
                    train_loader=dl_tr,
                    valid_loader=dl_va,
                    device=DEVICE,
                    lr=float(meta["train_params"].get("lr", 1e-3)),
                    weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
                    max_epochs=int(meta["train_params"].get("max_epochs", 10)),
                    patience=int(meta["train_params"].get("patience", 4)),
                    pos_weight=pos_weight,
                )

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    Xh = preproc.transform(X_hold_sc)
                    if hasattr(Xh, "toarray"):
                        Xh = Xh.toarray()
                    Xh = np.asarray(Xh, dtype=np.float32)

                    p = predict_torch_binary_model(
                        model=model,
                        X=Xh,
                        batch_size=int(meta["train_params"].get("batch_size", 2048)),
                        device=DEVICE,
                    )
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del model, dl_tr, dl_va, X_tr, X_fit, X_es

            else:
                raise ValueError(f"Unsupported expert kind for robustness: {meta['kind']}")

            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    # average across folds
    scenario_model_preds = {}
    for sc_name in holdout_scenarios.keys():
        scenario_model_preds[sc_name] = {}
        for model_name in selected_models:
            scenario_model_preds[sc_name][model_name] = clip_probs(
                np.mean(np.vstack(scenario_model_fold_preds[sc_name][model_name]), axis=0)
            )

    return scenario_model_preds


ROBUST_EXPERT_HOLDOUT_PREDS = get_expert_holdout_preds_all_scenarios(
    X_dev=X_dev,
    y_dev=y_dev,
    holdout_scenarios=X_HOLDOUT_SCENARIOS,
    selected_models=selected_experts,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

# ============================================================
# 3. Ensemble predictions for all scenarios
# ============================================================

def compute_weighted_holdout_from_scenario_preds(scenario_pred_dict, selected_experts, final_expert_results_df):
    w = []
    for m in selected_experts:
        pr_auc = final_expert_results_df[
            (final_expert_results_df["model_name"] == m) &
            (final_expert_results_df["split"] == "DEV_OOF")
        ]["PR_AUC"].values[0]
        w.append(pr_auc)
    w = np.array(w, dtype=float)
    w = w / w.sum()

    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts])
    return clip_probs(np.dot(mat, w))

def compute_simpleavg_holdout_from_scenario_preds(scenario_pred_dict, selected_experts):
    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts])
    return clip_probs(np.mean(mat, axis=1))

def compute_stacking_holdout_from_scenario_preds(scenario_pred_dict, selected_experts, stack_model):
    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts])
    return clip_probs(stack_model.predict_proba(mat)[:, 1])


ROBUST_HOLDOUT_MODELS = {}

for sc_name, pred_dict in ROBUST_EXPERT_HOLDOUT_PREDS.items():
    out = {}

    # expert baselines
    out["CatBoost"] = pred_dict["catboost"]
    out["HGB"] = pred_dict["hgb"]
    out["DeepFM"] = pred_dict["deepfm"]

    # ensemble baselines
    out["WeightedEnsemble"] = compute_weighted_holdout_from_scenario_preds(
        pred_dict, selected_experts, final_expert_results_df
    )
    out["SimpleAverage"] = compute_simpleavg_holdout_from_scenario_preds(
        pred_dict, selected_experts
    )
    out["Stacking"] = compute_stacking_holdout_from_scenario_preds(
        pred_dict, selected_experts, stack_model
    )

    ROBUST_HOLDOUT_MODELS[sc_name] = out

print("Built scenario predictions for baselines.")
```

```python
# ============================================================
# 4. Proposed Model under perturbation
# Train TRF on clean dev once per seed/fold, infer on every perturbed scenario
# ============================================================

def build_backbone_from_pred_dict(pred_dict, backbone_name="cat_hgb_avg"):
    if backbone_name == "catboost":
        return clip_probs(pred_dict["catboost"])
    elif backbone_name == "hgb":
        return clip_probs(pred_dict["hgb"])
    elif backbone_name == "cat_hgb_avg":
        return clip_probs(0.5 * clip_probs(pred_dict["catboost"]) + 0.5 * clip_probs(pred_dict["hgb"]))
    else:
        raise ValueError(backbone_name)

def build_scenario_meta_df(X_hold_sc, scenario_pred_dict, selected_experts, backbone_name):
    # base trust/meta features
    meta_df = build_trust_features_from_pred_dict(
        X_df=X_hold_sc,
        pred_dict=scenario_pred_dict,
        expert_list=selected_experts,
        context_cols=[c for c in ["Age", "Sex", "BMI", "GenHlth", "HighBP", "HighChol", "Diabetes", "DiffWalk"] if c in X_hold_sc.columns]
    )

    # enhanced backbone-relative features
    b = build_backbone_from_pred_dict(scenario_pred_dict, backbone_name=backbone_name)
    meta_df["backbone_prob"] = b
    meta_df["backbone_margin"] = np.abs(b - 0.5)
    meta_df["backbone_entropy"] = binary_entropy_np(b)
    meta_df["mean_minus_backbone"] = meta_df["mean_pred"] - b
    meta_df["max_minus_backbone"] = meta_df["max_pred"] - b
    meta_df["backbone_minus_min"] = b - meta_df["min_pred"]

    for m in selected_experts:
        pcol = f"p_{m}"
        if pcol in meta_df.columns:
            meta_df[f"{pcol}_minus_backbone"] = meta_df[pcol] - b

    return meta_df.astype(np.float32)

def get_proposed_model_preds_all_scenarios(
    y_dev,
    dev_meta_df,
    E_dev,
    B_dev,
    holdout_scenarios,
    robust_expert_holdout_preds,
    selected_experts,
    config,
    final_cfg,
):
    cv = get_cv(config, final_stage=True)
    seed_offsets = final_cfg["seed_offsets"]
    backbone_name = final_cfg["backbone_name"]

    # clean dev enhanced meta for training
    meta_dev_df, _dummy_hold = build_trf_final_meta_features(
        base_dev_meta_df=dev_meta_df,
        base_holdout_meta_df=holdout_meta_df,
        final_dev_preds=final_dev_preds,
        final_holdout_preds=final_holdout_preds,
        expert_list=selected_experts,
        backbone_name=backbone_name,
    )
    X_meta_dev = meta_dev_df.values.astype(np.float32)
    std_idx = meta_dev_df.columns.get_loc("std_pred") if "std_pred" in meta_dev_df.columns else None

    # scenario meta matrices
    scenario_meta = {}
    scenario_backbone = {}
    scenario_E = {}
    for sc_name, X_hold_sc in holdout_scenarios.items():
        pred_dict = robust_expert_holdout_preds[sc_name]
        meta_hold_sc = build_scenario_meta_df(
            X_hold_sc=X_hold_sc,
            scenario_pred_dict=pred_dict,
            selected_experts=selected_experts,
            backbone_name=backbone_name,
        )
        scenario_meta[sc_name] = meta_hold_sc.values.astype(np.float32)
        scenario_backbone[sc_name] = build_backbone_from_pred_dict(pred_dict, backbone_name=backbone_name).astype(np.float32)
        scenario_E[sc_name] = np.column_stack([clip_probs(pred_dict[m]) for m in selected_experts]).astype(np.float32)

    scenario_seed_preds = {sc_name: [] for sc_name in holdout_scenarios.keys()}

    for seed_offset in seed_offsets:
        print("=" * 100)
        print(f"[Robustness] Proposed Model seed_offset={seed_offset}")
        print("=" * 100)

        fold_preds_per_scenario = {sc_name: [] for sc_name in holdout_scenarios.keys()}

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
            print(f"seed_offset={seed_offset} | fold={fold_id}")

            X_tr = X_meta_dev[tr_idx]
            X_va = X_meta_dev[va_idx]
            E_tr = E_dev[tr_idx]
            E_va = E_dev[va_idx]
            Bb_tr = B_dev[tr_idx]
            Bb_va = B_dev[va_idx]
            y_tr = y_dev[tr_idx].astype(np.float32)
            y_va = y_dev[va_idx].astype(np.float32)

            pos = float((y_tr == 1).sum())
            neg = float((y_tr == 0).sum())
            pos_weight = neg / max(pos, 1.0)

            dl_tr, dl_va = make_trf_final_loaders(
                X_meta_tr=X_tr,
                E_tr=E_tr,
                B_tr=Bb_tr,
                y_tr=y_tr,
                X_meta_va=X_va,
                E_va=E_va,
                B_va=Bb_va,
                y_va=y_va,
                batch_size=final_cfg["batch_size"],
            )

            set_global_seed(config.seed + seed_offset + fold_id)

            model = TRFFinalModel(
                meta_dim=X_meta_dev.shape[1],
                n_experts=E_dev.shape[1],
                hidden_dim=final_cfg["hidden_dim"],
                dropout=final_cfg["dropout"],
                max_delta=final_cfg["max_delta"],
                use_disagreement_scaling=final_cfg["use_disagreement_scaling"],
                std_feature_index=std_idx,
                disagreement_strength=final_cfg["disagreement_strength"],
                disagreement_bias=final_cfg["disagreement_bias"],
            )

            model = train_trf_final_model(
                model=model,
                train_loader=dl_tr,
                valid_loader=dl_va,
                device=DEVICE,
                lr=final_cfg["lr"],
                weight_decay=final_cfg["weight_decay"],
                max_epochs=final_cfg["max_epochs"],
                patience=final_cfg["patience"],
                pos_weight=pos_weight,
                delta_l2=final_cfg["delta_l2"],
                mean_delta_penalty=final_cfg["mean_delta_penalty"],
                trust_reg=final_cfg["trust_reg"],
            )

            for sc_name in holdout_scenarios.keys():
                p_sc, _, _ = predict_trf_final_model(
                    model=model,
                    X_meta=scenario_meta[sc_name],
                    expert_probs=scenario_E[sc_name],
                    backbone_prob=scenario_backbone[sc_name],
                    batch_size=final_cfg["pred_batch_size"],
                    device=DEVICE,
                )
                fold_preds_per_scenario[sc_name].append(p_sc)

            del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, Bb_tr, Bb_va, y_tr, y_va
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        for sc_name in holdout_scenarios.keys():
            seed_pred = clip_probs(np.mean(np.vstack(fold_preds_per_scenario[sc_name]), axis=0))
            scenario_seed_preds[sc_name].append(seed_pred)

    scenario_final_preds = {}
    for sc_name in holdout_scenarios.keys():
        scenario_final_preds[sc_name] = clip_probs(
            np.mean(np.vstack(scenario_seed_preds[sc_name]), axis=0)
        )

    return scenario_final_preds


# clean dev inputs for TRF training
B_dev_clean, _ = build_backbone_predictions_locked(
    final_dev_preds=final_dev_preds,
    final_holdout_preds=final_holdout_preds,
    backbone_name=FINAL_TRF_CONFIG["backbone_name"],
)
E_dev_clean = np.column_stack([clip_probs(final_dev_preds[m]) for m in selected_experts]).astype(np.float32)

ROBUST_PROPOSED_PREDS = get_proposed_model_preds_all_scenarios(
    y_dev=y_dev,
    dev_meta_df=dev_meta_df,
    E_dev=E_dev_clean,
    B_dev=B_dev_clean,
    holdout_scenarios=X_HOLDOUT_SCENARIOS,
    robust_expert_holdout_preds=ROBUST_EXPERT_HOLDOUT_PREDS,
    selected_experts=selected_experts,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

for sc_name in ROBUST_HOLDOUT_MODELS.keys():
    ROBUST_HOLDOUT_MODELS[sc_name]["Proposed Model"] = ROBUST_PROPOSED_PREDS[sc_name]

print("Built Proposed Model predictions for all robustness scenarios.")

# ============================================================
# 5. Evaluate all scenarios
# ============================================================

robustness_rows = []

for sc_name, model_dict in ROBUST_HOLDOUT_MODELS.items():
    for model_name, p in model_dict.items():
        row = evaluate_probs(y_holdout, clip_probs(p), name=f"{model_name}_{sc_name}")
        row["scenario"] = sc_name
        row["model_name"] = model_name
        robustness_rows.append(row)

robustness_df = pd.DataFrame(robustness_rows)
robustness_df.to_csv(TAB_DIR / "robustness_results.csv", index=False)

display(
    robustness_df[
        ["scenario", "model_name", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["scenario", "PR_AUC"], ascending=[True, False])
)

# ============================================================
# 6. Degradation table vs clean
# ============================================================

clean_df = robustness_df[robustness_df["scenario"] == "clean"].copy()
clean_df = clean_df.set_index("model_name")

degradation_rows = []

for _, row in robustness_df.iterrows():
    model_name = row["model_name"]
    sc_name = row["scenario"]

    base = clean_df.loc[model_name]

    degradation_rows.append({
        "scenario": sc_name,
        "model_name": model_name,
        "PR_AUC": row["PR_AUC"],
        "Brier": row["Brier"],
        "ECE": row["ECE"],
        "delta_PR_AUC": row["PR_AUC"] - base["PR_AUC"],
        "delta_Brier": row["Brier"] - base["Brier"],
        "delta_ECE": row["ECE"] - base["ECE"],
    })

degradation_df = pd.DataFrame(degradation_rows)
degradation_df.to_csv(TAB_DIR / "robustness_degradation.csv", index=False)

display(
    degradation_df.sort_values(["scenario", "delta_PR_AUC"], ascending=[True, False])
)

# ============================================================
# 7. Plot degradation curves
# ============================================================

plot_order = ["clean", "noise_0.25", "noise_0.50", "mask_0.10", "mask_0.20", "corrupt_top3"]
plot_models = ["CatBoost", "HGB", "WeightedEnsemble", "Stacking", "Proposed Model"]

plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_df[degradation_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["PR_AUC"], marker="o", label=model_name)

plt.ylabel("PR-AUC")
plt.xlabel("Scenario")
plt.title("Robustness: PR-AUC under Perturbations")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_pr_auc.png", dpi=300)
plt.show()


plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_df[degradation_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["Brier"], marker="o", label=model_name)

plt.ylabel("Brier Score")
plt.xlabel("Scenario")
plt.title("Robustness: Brier under Perturbations")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_brier.png", dpi=300)
plt.show()
```

```python
# ============================================================
# Adversarial DeepFM Module
# Robust DeepFM for tabular screening / expert use
# ============================================================

import gc
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader


# ------------------------------------------------------------
# 1. Dataset
# ------------------------------------------------------------
class TabularTorchDataset(Dataset):
    def __init__(self, X: np.ndarray, y: np.ndarray = None):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = None if y is None else torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        if self.y is None:
            return self.X[idx]
        return self.X[idx], self.y[idx]


# ------------------------------------------------------------
# 2. DeepFM backbone
# Dense-input simplified DeepFM for already-preprocessed numeric features
# ------------------------------------------------------------
class DeepFMBinaryClassifier(nn.Module):
    def __init__(
        self,
        in_dim: int,
        embed_dim: int = 16,
        deep_hidden_dims = [256, 128],
        dropout: float = 0.20,
    ):
        super().__init__()
        self.in_dim = in_dim
        self.embed_dim = embed_dim

        # linear term
        self.linear = nn.Linear(in_dim, 1)

        # FM feature embeddings
        self.feature_embed = nn.Parameter(torch.randn(in_dim, embed_dim) * 0.01)

        # deep branch
        layers = []
        prev_dim = in_dim
        for h in deep_hidden_dims:
            layers.extend([
                nn.Linear(prev_dim, h),
                nn.BatchNorm1d(h),
                nn.ReLU(),
                nn.Dropout(dropout),
            ])
            prev_dim = h

        self.deep = nn.Sequential(*layers)
        self.deep_out = nn.Linear(prev_dim, 1)

    def fm_interaction(self, x: torch.Tensor) -> torch.Tensor:
        xv = x.unsqueeze(-1) * self.feature_embed.unsqueeze(0)   # [B, D, K]
        sum_square = torch.sum(xv, dim=1) ** 2
        square_sum = torch.sum(xv ** 2, dim=1)
        interactions = 0.5 * torch.sum(sum_square - square_sum, dim=1, keepdim=True)
        return interactions

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        linear_term = self.linear(x)
        fm_term = self.fm_interaction(x)
        deep_term = self.deep_out(self.deep(x))
        logits = linear_term + fm_term + deep_term
        return logits.squeeze(1)

# ============================================================
# 3. Adversarial / robust perturbation helpers
# ============================================================

def make_noisy_view(
    x: torch.Tensor,
    noise_std: float = 0.02,
) -> torch.Tensor:
    if noise_std <= 0:
        return x
    noise = torch.randn_like(x) * noise_std
    return x + noise


def make_masked_view(
    x: torch.Tensor,
    mask_prob: float = 0.05,
) -> torch.Tensor:
    if mask_prob <= 0:
        return x
    mask = (torch.rand_like(x) < mask_prob).float()
    return x * (1.0 - mask)


def make_fgsm_view(
    model: nn.Module,
    x: torch.Tensor,
    y: torch.Tensor,
    eps: float = 0.01,
    pos_weight: float = None,
) -> torch.Tensor:
    """
    Input-space FGSM perturbation for tabular features.
    """
    if eps <= 0:
        return x

    x_adv = x.detach().clone().requires_grad_(True)
    model.eval()

    logits = model(x_adv)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=x.device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    loss = criterion(logits, y)
    grad = torch.autograd.grad(loss, x_adv, retain_graph=False, create_graph=False)[0]

    x_adv = x_adv + eps * grad.sign()
    return x_adv.detach()

# ============================================================
# 4. Adversarial DeepFM training
# ============================================================

def make_torch_loaders(
    X_tr: np.ndarray,
    y_tr: np.ndarray,
    X_va: np.ndarray,
    y_va: np.ndarray,
    batch_size: int = 1024,
):
    ds_tr = TabularTorchDataset(X_tr, y_tr)
    ds_va = TabularTorchDataset(X_va, y_va)

    dl_tr = DataLoader(ds_tr, batch_size=batch_size, shuffle=True, drop_last=False)
    dl_va = DataLoader(ds_va, batch_size=batch_size, shuffle=False, drop_last=False)
    return dl_tr, dl_va


def train_adversarial_deepfm(
    model: nn.Module,
    train_loader: DataLoader,
    valid_loader: DataLoader,
    device: str = DEVICE,
    lr: float = 8e-4,
    weight_decay: float = 1e-5,
    max_epochs: int = 8,
    patience: int = 4,
    pos_weight: float = None,
    noise_std: float = 0.02,
    mask_prob: float = 0.05,
    fgsm_eps: float = 0.01,
    adv_bce_weight: float = 0.50,
    consistency_weight: float = 0.25,
):
    model = model.to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    if pos_weight is not None:
        criterion = nn.BCEWithLogitsLoss(
            pos_weight=torch.tensor([pos_weight], dtype=torch.float32, device=device)
        )
    else:
        criterion = nn.BCEWithLogitsLoss()

    best_val = float("inf")
    best_state = None
    bad_epochs = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        train_losses = []

        for xb, yb in train_loader:
            xb = xb.to(device)
            yb = yb.to(device)

            # clean forward
            logits_clean = model(xb)
            loss_clean = criterion(logits_clean, yb)

            # robust/noisy view
            xb_pert = xb.detach().clone()
            xb_pert = make_noisy_view(xb_pert, noise_std=noise_std)
            xb_pert = make_masked_view(xb_pert, mask_prob=mask_prob)

            # FGSM view (optional)
            if fgsm_eps > 0:
                xb_fgsm = make_fgsm_view(
                    model=model,
                    x=xb_pert,
                    y=yb,
                    eps=fgsm_eps,
                    pos_weight=pos_weight,
                )
                xb_pert = xb_fgsm

            logits_adv = model(xb_pert)
            loss_adv = criterion(logits_adv, yb)

            p_clean = torch.sigmoid(logits_clean)
            p_adv = torch.sigmoid(logits_adv)

            # symmetric consistency
            loss_consistency = F.mse_loss(p_clean, p_adv)

            loss = (
                loss_clean
                + adv_bce_weight * loss_adv
                + consistency_weight * loss_consistency
            )

            optimizer.zero_grad(set_to_none=True)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
            optimizer.step()

            train_losses.append(loss.item())

        # validation on clean data
        model.eval()
        valid_losses = []

        with torch.no_grad():
            for xb, yb in valid_loader:
                xb = xb.to(device)
                yb = yb.to(device)
                logits = model(xb)
                loss = criterion(logits, yb)
                valid_losses.append(loss.item())

        mean_tr = float(np.mean(train_losses))
        mean_va = float(np.mean(valid_losses))

        if epoch == 1 or epoch % 2 == 0:
            print(f"ep {epoch:02d} | train={mean_tr:.5f} | valid={mean_va:.5f} | best={best_val:.5f}")

        if mean_va < best_val - 1e-6:
            best_val = mean_va
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            bad_epochs = 0
        else:
            bad_epochs += 1

        if bad_epochs >= patience:
            break

    if best_state is not None:
        model.load_state_dict(best_state)

    return model


@torch.no_grad()
def predict_torch_binary_model(
    model: nn.Module,
    X: np.ndarray,
    batch_size: int = 2048,
    device: str = DEVICE,
):
    model.eval()
    ds = TabularTorchDataset(X, None)
    dl = DataLoader(ds, batch_size=batch_size, shuffle=False, drop_last=False)

    all_probs = []
    for xb in dl:
        xb = xb.to(device)
        logits = model(xb)
        probs = torch.sigmoid(logits).detach().cpu().numpy()
        all_probs.append(probs)

    return clip_probs(np.concatenate(all_probs))

# ============================================================
# 5. CV runner for adversarial DeepFM
# ============================================================

def run_adversarial_deepfm_cv(
    X_df: pd.DataFrame,
    y: np.ndarray,
    config: ExperimentConfig,
    batch_size: int = 1024,
    embed_dim: int = 16,
    hidden_dims = [256, 128],
    dropout: float = 0.20,
    lr: float = 8e-4,
    weight_decay: float = 1e-5,
    max_epochs: int = 8,
    patience: int = 4,
    noise_std: float = 0.02,
    mask_prob: float = 0.05,
    fgsm_eps: float = 0.01,
    adv_bce_weight: float = 0.50,
    consistency_weight: float = 0.25,
):
    cv = get_cv(config, final_stage=False)
    numeric_cols = list(X_df.columns)

    oof = np.zeros(len(y), dtype=float)
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_df, y), start=1):
        print("=" * 80)
        print(f"[AdvDeepFM] fold {fold_id}")
        print("=" * 80)

        X_tr_df = X_df.iloc[tr_idx].copy()
        X_va_df = X_df.iloc[va_idx].copy()
        y_tr = y[tr_idx]
        y_va = y[va_idx]

        preproc = build_dl_preprocessor(numeric_cols)

        X_tr = preproc.fit_transform(X_tr_df)
        X_va = preproc.transform(X_va_df)

        if hasattr(X_tr, "toarray"):
            X_tr = X_tr.toarray()
        if hasattr(X_va, "toarray"):
            X_va = X_va.toarray()

        X_tr = np.asarray(X_tr, dtype=np.float32)
        X_va = np.asarray(X_va, dtype=np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_torch_loaders(
            X_tr=X_tr,
            y_tr=y_tr.astype(np.float32),
            X_va=X_va,
            y_va=y_va.astype(np.float32),
            batch_size=batch_size,
        )

        model = DeepFMBinaryClassifier(
            in_dim=X_tr.shape[1],
            embed_dim=embed_dim,
            deep_hidden_dims=hidden_dims,
            dropout=dropout,
        )

        model = train_adversarial_deepfm(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=lr,
            weight_decay=weight_decay,
            max_epochs=max_epochs,
            patience=patience,
            pos_weight=pos_weight,
            noise_std=noise_std,
            mask_prob=mask_prob,
            fgsm_eps=fgsm_eps,
            adv_bce_weight=adv_bce_weight,
            consistency_weight=consistency_weight,
        )

        p_va = predict_torch_binary_model(
            model=model,
            X=X_va,
            batch_size=batch_size,
            device=DEVICE,
        )

        oof[va_idx] = p_va

        fold_row = evaluate_probs(y_va, p_va, name=f"AdvDeepFM_fold{fold_id}")
        fold_row["fold"] = fold_id
        fold_rows.append(fold_row)

        print(
            f"fold={fold_id} | "
            f"PR_AUC={fold_row['PR_AUC']:.6f} | "
            f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
            f"Brier={fold_row['Brier']:.6f} | "
            f"ECE={fold_row['ECE']:.6f}"
        )

        del model, dl_tr, dl_va, X_tr, X_va
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    summary = evaluate_probs(y, oof, name="AdvDeepFM_OOF")
    fold_df = pd.DataFrame(fold_rows)

    return summary, oof, fold_df

# ============================================================
# 6. Register adversarial DeepFM
# ============================================================

CANDIDATE_MODEL_REGISTRY["deepfm_adv"] = {
    "family": "deep",
    "preproc": "dl",
    "kind": "torch_adv_deepfm",
    "needs_calibration": False,
    "builder": lambda in_dim: DeepFMBinaryClassifier(
        in_dim=in_dim,
        embed_dim=16,
        deep_hidden_dims=[256, 128],
        dropout=0.20,
    ),
    "train_params": {
        "lr": 8e-4,
        "weight_decay": 1e-5,
        "max_epochs": 8,
        "patience": 4,
        "batch_size": 1024,
        "noise_std": 0.02,
        "mask_prob": 0.05,
        "fgsm_eps": 0.01,
        "adv_bce_weight": 0.50,
        "consistency_weight": 0.25,
    },
}
```

```python
# ============================================================
# Minimal Integration — Step 1
# Final rerun for AdvDeepFM only
# ============================================================

def run_final_adv_deepfm_only(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    X_holdout: pd.DataFrame,
    y_holdout: np.ndarray,
    config: ExperimentConfig,
    batch_size: int = 1024,
    embed_dim: int = 16,
    hidden_dims = [256, 128],
    dropout: float = 0.20,
    lr: float = 8e-4,
    weight_decay: float = 1e-5,
    max_epochs: int = 8,
    patience: int = 4,
    noise_std: float = 0.02,
    mask_prob: float = 0.05,
    fgsm_eps: float = 0.01,
    adv_bce_weight: float = 0.50,
    consistency_weight: float = 0.25,
):
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    oof = np.zeros(len(y_dev), dtype=float)
    holdout_fold_preds = []
    fold_rows = []

    for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
        print("=" * 90)
        print(f"[AdvDeepFM final] fold {fold_id}")
        print("=" * 90)

        X_tr_df = X_dev.iloc[tr_idx].copy()
        X_va_df = X_dev.iloc[va_idx].copy()
        y_tr = y_dev[tr_idx]
        y_va = y_dev[va_idx]

        preproc = build_dl_preprocessor(numeric_cols)

        X_tr = preproc.fit_transform(X_tr_df)
        X_va = preproc.transform(X_va_df)
        X_hold = preproc.transform(X_holdout)

        if hasattr(X_tr, "toarray"):
            X_tr = X_tr.toarray()
        if hasattr(X_va, "toarray"):
            X_va = X_va.toarray()
        if hasattr(X_hold, "toarray"):
            X_hold = X_hold.toarray()

        X_tr = np.asarray(X_tr, dtype=np.float32)
        X_va = np.asarray(X_va, dtype=np.float32)
        X_hold = np.asarray(X_hold, dtype=np.float32)

        pos = float((y_tr == 1).sum())
        neg = float((y_tr == 0).sum())
        pos_weight = neg / max(pos, 1.0)

        dl_tr, dl_va = make_torch_loaders(
            X_tr=X_tr,
            y_tr=y_tr.astype(np.float32),
            X_va=X_va,
            y_va=y_va.astype(np.float32),
            batch_size=batch_size,
        )

        model = DeepFMBinaryClassifier(
            in_dim=X_tr.shape[1],
            embed_dim=embed_dim,
            deep_hidden_dims=hidden_dims,
            dropout=dropout,
        )

        model = train_adversarial_deepfm(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=lr,
            weight_decay=weight_decay,
            max_epochs=max_epochs,
            patience=patience,
            pos_weight=pos_weight,
            noise_std=noise_std,
            mask_prob=mask_prob,
            fgsm_eps=fgsm_eps,
            adv_bce_weight=adv_bce_weight,
            consistency_weight=consistency_weight,
        )

        p_va = predict_torch_binary_model(
            model=model,
            X=X_va,
            batch_size=batch_size,
            device=DEVICE,
        )
        p_hold = predict_torch_binary_model(
            model=model,
            X=X_hold,
            batch_size=batch_size,
            device=DEVICE,
        )

        oof[va_idx] = p_va
        holdout_fold_preds.append(p_hold)

        fold_row = evaluate_probs(y_va, p_va, name=f"AdvDeepFM_final_fold{fold_id}")
        fold_row["model_name"] = "deepfm_adv"
        fold_row["fold"] = fold_id
        fold_rows.append(fold_row)

        print(
            f"fold={fold_id} | "
            f"PR_AUC={fold_row['PR_AUC']:.6f} | "
            f"ROC_AUC={fold_row['ROC_AUC']:.6f} | "
            f"Brier={fold_row['Brier']:.6f} | "
            f"ECE={fold_row['ECE']:.6f}"
        )

        del model, dl_tr, dl_va, X_tr, X_va, X_hold
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    holdout_pred = clip_probs(np.mean(np.vstack(holdout_fold_preds), axis=0))

    dev_row = evaluate_probs(y_dev, oof, name="AdvDeepFM_DEV_OOF")
    dev_row["model_name"] = "deepfm_adv"
    dev_row["split"] = "DEV_OOF"

    hold_row = evaluate_probs(y_holdout, holdout_pred, name="AdvDeepFM_HOLDOUT")
    hold_row["model_name"] = "deepfm_adv"
    hold_row["split"] = "HOLDOUT"

    results_df = pd.DataFrame([dev_row, hold_row])
    fold_df = pd.DataFrame(fold_rows)

    return results_df, oof, holdout_pred, fold_df


adv_deepfm_results_df, adv_deepfm_dev_pred, adv_deepfm_hold_pred, adv_deepfm_fold_df = run_final_adv_deepfm_only(
    X_dev=X_dev,
    y_dev=y_dev,
    X_holdout=X_holdout,
    y_holdout=y_holdout,
    config=CFG,
)

display(adv_deepfm_results_df[["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]])

# ============================================================
# Minimal Integration — Step 2
# Replace DeepFM with AdvDeepFM
# ============================================================

final_dev_preds_adv = dict(final_dev_preds)
final_holdout_preds_adv = dict(final_holdout_preds)

final_dev_preds_adv["deepfm_adv"] = adv_deepfm_dev_pred
final_holdout_preds_adv["deepfm_adv"] = adv_deepfm_hold_pred

selected_experts_adv = ["logreg", "catboost", "hgb", "deepfm_adv"]

print("New expert set:", selected_experts_adv)

# ============================================================
# Minimal Integration — Step 3
# Rebuild ensemble baselines using AdvDeepFM
# ============================================================

# update expert results table by swapping deepfm -> deepfm_adv
final_expert_results_adv_df = final_expert_results_df.copy()
final_expert_results_adv_df = final_expert_results_adv_df[final_expert_results_adv_df["model_name"] != "deepfm"].copy()
final_expert_results_adv_df = pd.concat([final_expert_results_adv_df, adv_deepfm_results_df], ignore_index=True)

# Simple average
avg_dev_adv = simple_average_ensemble(final_dev_preds_adv, selected_experts_adv)
avg_hold_adv = simple_average_ensemble(final_holdout_preds_adv, selected_experts_adv)

# Weighted
w_dev_adv, w_hold_adv, weights_adv = performance_weighted_ensemble(
    final_dev_preds_adv,
    final_holdout_preds_adv,
    selected_experts_adv,
    final_expert_results_adv_df
)

# Stacking
stack_dev_adv, stack_hold_adv, stack_model_adv = stacking_ensemble(
    final_dev_preds_adv,
    final_holdout_preds_adv,
    y_dev,
    selected_experts_adv
)

print("New weights with AdvDeepFM:")
for m, w in zip(selected_experts_adv, weights_adv):
    print(f"{m}: {w:.4f}")

ensemble_adv_rows = []

def eval_and_store_adv(name, dev_pred, hold_pred):
    dev_row = evaluate_probs(y_dev, dev_pred, name=name + "_DEV")
    dev_row["model_name"] = name
    dev_row["split"] = "DEV_OOF"

    hold_row = evaluate_probs(y_holdout, hold_pred, name=name + "_HOLDOUT")
    hold_row["model_name"] = name
    hold_row["split"] = "HOLDOUT"

    ensemble_adv_rows.extend([dev_row, hold_row])

eval_and_store_adv("SimpleAverage_adv", avg_dev_adv, avg_hold_adv)
eval_and_store_adv("WeightedEnsemble_adv", w_dev_adv, w_hold_adv)
eval_and_store_adv("Stacking_adv", stack_dev_adv, stack_hold_adv)

ensemble_adv_df = pd.DataFrame(ensemble_adv_rows)

display(ensemble_adv_df[["model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE"]])

# ============================================================
# Minimal Integration — Step 4
# Rerun Proposed Model with AdvDeepFM
# ============================================================

FINAL_TRF_OUTPUT_ADV = run_final_trf_locked(
    X_dev=X_dev,
    X_holdout=X_holdout,
    y_dev=y_dev,
    y_holdout=y_holdout,
    dev_meta_df=dev_meta_df,
    holdout_meta_df=holdout_meta_df,
    final_dev_preds=final_dev_preds_adv,
    final_holdout_preds=final_holdout_preds_adv,
    selected_experts=selected_experts_adv,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

display(FINAL_TRF_OUTPUT_ADV["results_df"][[
    "model_name", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE",
    "mean_abs_delta", "mean_delta", "mean_trust"
]])

# ============================================================
# Minimal Integration — Step 5
# Compare old vs new Proposed Model
# ============================================================

base_trf_results = FINAL_TRF_OUTPUT["results_df"].copy()
base_trf_results["version"] = "Current_Proposed_Model"

adv_trf_results = FINAL_TRF_OUTPUT_ADV["results_df"].copy()
adv_trf_results["version"] = "Proposed_Model_with_AdvDeepFM"

compare_trf_df = pd.concat([base_trf_results, adv_trf_results], ignore_index=True)

display(compare_trf_df[[
    "version", "split", "PR_AUC", "ROC_AUC", "Brier", "ECE",
    "mean_abs_delta", "mean_delta", "mean_trust"
]])
```

```python
# ============================================================
# Module 13 (Focused)
# Robustness comparison:
#   1) Current_Proposed_Model
#   2) Proposed_Model_with_AdvDeepFM
# ============================================================

import gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 0. Config
# ------------------------------------------------------------
ROBUSTNESS_COMPARE_CONFIG = {
    "scenarios": [
        {"name": "clean",        "kind": "clean",        "strength": 0.00},
        {"name": "noise_0.25",   "kind": "gaussian",     "strength": 0.25},
        {"name": "noise_0.50",   "kind": "gaussian",     "strength": 0.50},
        {"name": "mask_0.10",    "kind": "feature_mask", "strength": 0.10},
        {"name": "mask_0.20",    "kind": "feature_mask", "strength": 0.20},
        {"name": "corrupt_top3", "kind": "corrupt_cols", "strength": 1.00},
    ],
    "top_corrupt_cols_candidates": [
        "BMI", "GenHlth", "Age", "HighBP", "HighChol", "PhysHlth", "MentHlth"
    ],
}

print("Scenarios:")
for s in ROBUSTNESS_COMPARE_CONFIG["scenarios"]:
    print(s)

# ============================================================
# 1. Scenario builder
# ============================================================

def get_numeric_cols(df: pd.DataFrame):
    return [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]

def get_existing_corrupt_cols(df: pd.DataFrame, candidates):
    return [c for c in candidates if c in df.columns][:3]

def make_holdout_scenarios_for_compare(X_dev: pd.DataFrame, X_holdout: pd.DataFrame, cfg: dict):
    rng = np.random.default_rng(CFG.seed)
    scenarios = {}

    numeric_cols = get_numeric_cols(X_holdout)
    dev_std = X_dev[numeric_cols].std().replace(0, 1.0)
    dev_median = X_dev[numeric_cols].median()
    corrupt_cols = get_existing_corrupt_cols(X_holdout, cfg["top_corrupt_cols_candidates"])

    for sc in cfg["scenarios"]:
        name = sc["name"]
        kind = sc["kind"]
        strength = sc["strength"]

        Xp = X_holdout.copy()

        if kind == "clean":
            pass

        elif kind == "gaussian":
            noise = rng.normal(
                loc=0.0,
                scale=(dev_std.values * strength),
                size=(len(Xp), len(numeric_cols))
            )
            Xp.loc[:, numeric_cols] = Xp[numeric_cols].values + noise

        elif kind == "feature_mask":
            mask = rng.random((len(Xp), len(numeric_cols))) < strength
            arr = Xp[numeric_cols].values.copy()
            med = dev_median.values.reshape(1, -1)
            arr[mask] = np.broadcast_to(med, arr.shape)[mask]
            Xp.loc[:, numeric_cols] = arr

        elif kind == "corrupt_cols":
            for c in corrupt_cols:
                Xp[c] = rng.permutation(Xp[c].values)

        else:
            raise ValueError(f"Unknown scenario kind: {kind}")

        scenarios[name] = Xp

    return scenarios


X_HOLDOUT_SCENARIOS_COMPARE = make_holdout_scenarios_for_compare(
    X_dev=X_dev,
    X_holdout=X_holdout,
    cfg=ROBUSTNESS_COMPARE_CONFIG
)

print("Built scenarios:", list(X_HOLDOUT_SCENARIOS_COMPARE.keys()))

# ============================================================
# 2. Helper: backbone from prediction dict
# ============================================================

def build_backbone_from_pred_dict_for_compare(pred_dict, backbone_name="cat_hgb_avg"):
    if backbone_name == "catboost":
        return clip_probs(pred_dict["catboost"])
    elif backbone_name == "hgb":
        return clip_probs(pred_dict["hgb"])
    elif backbone_name == "cat_hgb_avg":
        return clip_probs(
            0.5 * clip_probs(pred_dict["catboost"]) +
            0.5 * clip_probs(pred_dict["hgb"])
        )
    else:
        raise ValueError(backbone_name)

# ============================================================
# 3. Helper: build scenario meta features
# ============================================================

def build_scenario_meta_df_for_compare(
    X_hold_sc: pd.DataFrame,
    scenario_pred_dict: dict,
    selected_experts_local: list,
    backbone_name: str,
):
    meta_df = build_trust_features_from_pred_dict(
        X_df=X_hold_sc,
        pred_dict=scenario_pred_dict,
        expert_list=selected_experts_local,
        context_cols=[c for c in ["Age", "Sex", "BMI", "GenHlth", "HighBP", "HighChol", "Diabetes", "DiffWalk"] if c in X_hold_sc.columns]
    )

    b = build_backbone_from_pred_dict_for_compare(
        scenario_pred_dict,
        backbone_name=backbone_name
    )

    meta_df["backbone_prob"] = b
    meta_df["backbone_margin"] = np.abs(b - 0.5)
    meta_df["backbone_entropy"] = binary_entropy_np(b)
    meta_df["mean_minus_backbone"] = meta_df["mean_pred"] - b
    meta_df["max_minus_backbone"] = meta_df["max_pred"] - b
    meta_df["backbone_minus_min"] = b - meta_df["min_pred"]

    for m in selected_experts_local:
        pcol = f"p_{m}"
        if pcol in meta_df.columns:
            meta_df[f"{pcol}_minus_backbone"] = meta_df[pcol] - b

    return meta_df.astype(np.float32)

# ============================================================
# 4. Build expert predictions for all scenarios
#    for a given expert set + prediction dicts
#    No retraining needed here because we reuse clean holdout
#    per-expert scenario predictions from existing pred dicts only
#    if you want strict perturbation-aware reruns for experts too,
#    that requires retraining per fold. Here we compare proposed
#    model variants fairly using their own scenario expert inputs.
# ============================================================

def build_scenario_pred_dicts_from_base(
    base_holdout_preds: dict,
    scenario_names: list,
):
    """
    Placeholder container builder.
    We fill this later after computing scenario expert predictions.
    """
    return {sc: {} for sc in scenario_names}

# ============================================================
# 5. Recompute expert predictions on perturbed holdout scenarios
#    for a specific expert set
# ============================================================

def get_expert_holdout_preds_all_scenarios_for_expert_set(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    holdout_scenarios: dict,
    selected_models: list,
    registry: dict,
    config: ExperimentConfig,
):
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    scenario_model_fold_preds = {
        sc_name: {m: [] for m in selected_models}
        for sc_name in holdout_scenarios.keys()
    }

    for model_name in selected_models:
        meta = registry[model_name]
        print("=" * 100)
        print(f"[Robustness compare] expert={model_name}")
        print("=" * 100)

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
            print(f"fold {fold_id}")

            X_tr_df = X_dev.iloc[tr_idx].copy()
            y_tr = y_dev[tr_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            if meta["kind"] == "sklearn":
                pipe = Pipeline([
                    ("pre", preproc),
                    ("model", deepcopy(meta["estimator"])),
                ])
                pipe.fit(X_tr_df, y_tr)

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    p = get_sklearn_probabilities(pipe, X_hold_sc)
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del pipe

            elif meta["kind"] in ["torch", "torch_adv_deepfm"]:
    X_tr = preproc.fit_transform(X_tr_df)
    if hasattr(X_tr, "toarray"):
        X_tr = X_tr.toarray()
    X_tr = np.asarray(X_tr, dtype=np.float32)

    inner_rng = np.random.default_rng(CFG.seed + fold_id)
    idx = np.arange(len(X_tr))
    inner_rng.shuffle(idx)
    cut = int(0.9 * len(idx))
    tr2, va2 = idx[:cut], idx[cut:]

    X_fit = X_tr[tr2]
    y_fit = y_tr[tr2].astype(np.float32)
    X_es = X_tr[va2]
    y_es = y_tr[va2].astype(np.float32)

    pos = float((y_fit == 1).sum())
    neg = float((y_fit == 0).sum())
    pos_weight = neg / max(pos, 1.0)

    dl_tr, dl_va = make_torch_loaders(
        X_tr=X_fit,
        y_tr=y_fit,
        X_va=X_es,
        y_va=y_es,
        batch_size=int(meta["train_params"].get("batch_size", 1024)),
    )

    model = meta["builder"](X_fit.shape[1])

    if meta["kind"] == "torch_adv_deepfm":
        model = train_adversarial_deepfm(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=float(meta["train_params"].get("lr", 8e-4)),
            weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
            max_epochs=int(meta["train_params"].get("max_epochs", 18)),
            patience=int(meta["train_params"].get("patience", 4)),
            pos_weight=pos_weight,
            noise_std=float(meta["train_params"].get("noise_std", 0.02)),
            mask_prob=float(meta["train_params"].get("mask_prob", 0.05)),
            fgsm_eps=float(meta["train_params"].get("fgsm_eps", 0.01)),
            adv_bce_weight=float(meta["train_params"].get("adv_bce_weight", 0.50)),
            consistency_weight=float(meta["train_params"].get("consistency_weight", 0.25)),
        )
    else:
        model = train_torch_binary_model(
            model=model,
            train_loader=dl_tr,
            valid_loader=dl_va,
            device=DEVICE,
            lr=float(meta["train_params"].get("lr", 1e-3)),
            weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
            max_epochs=int(meta["train_params"].get("max_epochs", 20)),
            patience=int(meta["train_params"].get("patience", 4)),
            pos_weight=pos_weight,
        )

    for sc_name, X_hold_sc in holdout_scenarios.items():
        Xh = preproc.transform(X_hold_sc)
        if hasattr(Xh, "toarray"):
            Xh = Xh.toarray()
        Xh = np.asarray(Xh, dtype=np.float32)

        p = predict_torch_binary_model(
            model=model,
            X=Xh,
            batch_size=int(meta["train_params"].get("batch_size", 2048)),
            device=DEVICE,
        )
        scenario_model_fold_preds[sc_name][model_name].append(p)

    del model, dl_tr, dl_va, X_tr, X_fit, X_es

            else:
                raise ValueError(f"Unsupported kind for robustness compare: {meta['kind']}")

            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    scenario_model_preds = {}
    for sc_name in holdout_scenarios.keys():
        scenario_model_preds[sc_name] = {}
        for model_name in selected_models:
            scenario_model_preds[sc_name][model_name] = clip_probs(
                np.mean(np.vstack(scenario_model_fold_preds[sc_name][model_name]), axis=0)
            )

    return scenario_model_preds

# ============================================================
# 6. Ensure deepfm_adv is registered
# ============================================================

if "deepfm_adv" not in CANDIDATE_MODEL_REGISTRY:
    CANDIDATE_MODEL_REGISTRY["deepfm_adv"] = {
        "family": "deep",
        "preproc": "dl",
        "kind": "torch",
        "needs_calibration": False,
        "builder": lambda in_dim: DeepFMBinaryClassifier(
            in_dim=in_dim,
            embed_dim=16,
            deep_hidden_dims=[256, 128],
            dropout=0.20,
        ),
        "train_params": {
            "lr": 8e-4,
            "weight_decay": 1e-5,
            "max_epochs": 18,
            "patience": 4,
            "batch_size": 1024,
            "noise_std": 0.02,
            "mask_prob": 0.05,
            "fgsm_eps": 0.01,
            "adv_bce_weight": 0.50,
            "consistency_weight": 0.25,
        },
    }

print("deepfm_adv ready in registry.")

# ============================================================
# 7. Build scenario expert predictions
#    A) current expert set
#    B) adv expert set
# ============================================================

# A) current expert set
ROBUST_EXPERT_PREDS_CURRENT = get_expert_holdout_preds_all_scenarios_for_expert_set(
    X_dev=X_dev,
    y_dev=y_dev,
    holdout_scenarios=X_HOLDOUT_SCENARIOS_COMPARE,
    selected_models=selected_experts,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

# B) adv expert set
ROBUST_EXPERT_PREDS_ADV = get_expert_holdout_preds_all_scenarios_for_expert_set(
    X_dev=X_dev,
    y_dev=y_dev,
    holdout_scenarios=X_HOLDOUT_SCENARIOS_COMPARE,
    selected_models=selected_experts_adv,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

# ============================================================
# 8. Proposed model inference across scenarios
# ============================================================

def get_proposed_model_preds_all_scenarios_compare(
    y_dev,
    base_dev_meta_df,
    final_dev_preds_local,
    robust_expert_holdout_preds_local,
    selected_experts_local,
    config,
    final_cfg,
):
    cv = get_cv(config, final_stage=True)
    backbone_name = final_cfg["backbone_name"]
    seed_offsets = final_cfg["seed_offsets"]

    # clean dev backbone and clean dev expert matrices
    B_dev_clean, _ = build_backbone_predictions_locked(
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds_local if final_dev_preds_local is final_dev_preds else final_holdout_preds_adv,
        backbone_name=backbone_name,
    )
    E_dev_clean = np.column_stack([clip_probs(final_dev_preds_local[m]) for m in selected_experts_local]).astype(np.float32)

    # clean dev enhanced meta
    meta_dev_df, _ = build_trf_final_meta_features(
        base_dev_meta_df=base_dev_meta_df,
        base_holdout_meta_df=holdout_meta_df,
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds if final_dev_preds_local is final_dev_preds else final_holdout_preds_adv,
        expert_list=selected_experts_local,
        backbone_name=backbone_name,
    )
    X_meta_dev = meta_dev_df.values.astype(np.float32)
    std_idx = meta_dev_df.columns.get_loc("std_pred") if "std_pred" in meta_dev_df.columns else None

    # scenario-side inputs
    scenario_meta = {}
    scenario_backbone = {}
    scenario_E = {}

    for sc_name, X_hold_sc in X_HOLDOUT_SCENARIOS_COMPARE.items():
        pred_dict = robust_expert_holdout_preds_local[sc_name]

        meta_hold_sc = build_scenario_meta_df_for_compare(
            X_hold_sc=X_hold_sc,
            scenario_pred_dict=pred_dict,
            selected_experts_local=selected_experts_local,
            backbone_name=backbone_name,
        )

        scenario_meta[sc_name] = meta_hold_sc.values.astype(np.float32)
        scenario_backbone[sc_name] = build_backbone_from_pred_dict_for_compare(
            pred_dict,
            backbone_name=backbone_name
        ).astype(np.float32)
        scenario_E[sc_name] = np.column_stack(
            [clip_probs(pred_dict[m]) for m in selected_experts_local]
        ).astype(np.float32)

    scenario_seed_preds = {sc_name: [] for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys()}

    for seed_offset in seed_offsets:
        print("=" * 100)
        print(f"[Scenario compare] seed_offset={seed_offset}")
        print("=" * 100)

        fold_preds_per_scenario = {sc_name: [] for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys()}

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
            print(f"seed_offset={seed_offset} | fold={fold_id}")

            X_tr = X_meta_dev[tr_idx]
            X_va = X_meta_dev[va_idx]
            E_tr = E_dev_clean[tr_idx]
            E_va = E_dev_clean[va_idx]
            Bb_tr = B_dev_clean[tr_idx]
            Bb_va = B_dev_clean[va_idx]
            y_tr = y_dev[tr_idx].astype(np.float32)
            y_va = y_dev[va_idx].astype(np.float32)

            pos = float((y_tr == 1).sum())
            neg = float((y_tr == 0).sum())
            pos_weight = neg / max(pos, 1.0)

            dl_tr, dl_va = make_trf_final_loaders(
                X_meta_tr=X_tr,
                E_tr=E_tr,
                B_tr=Bb_tr,
                y_tr=y_tr,
                X_meta_va=X_va,
                E_va=E_va,
                B_va=Bb_va,
                y_va=y_va,
                batch_size=final_cfg["batch_size"],
            )

            set_global_seed(config.seed + seed_offset + fold_id)

            model = TRFFinalModel(
                meta_dim=X_meta_dev.shape[1],
                n_experts=E_dev_clean.shape[1],
                hidden_dim=final_cfg["hidden_dim"],
                dropout=final_cfg["dropout"],
                max_delta=final_cfg["max_delta"],
                use_disagreement_scaling=final_cfg["use_disagreement_scaling"],
                std_feature_index=std_idx,
                disagreement_strength=final_cfg["disagreement_strength"],
                disagreement_bias=final_cfg["disagreement_bias"],
            )

            model = train_trf_final_model(
                model=model,
                train_loader=dl_tr,
                valid_loader=dl_va,
                device=DEVICE,
                lr=final_cfg["lr"],
                weight_decay=final_cfg["weight_decay"],
                max_epochs=final_cfg["max_epochs"],
                patience=final_cfg["patience"],
                pos_weight=pos_weight,
                delta_l2=final_cfg["delta_l2"],
                mean_delta_penalty=final_cfg["mean_delta_penalty"],
                trust_reg=final_cfg["trust_reg"],
            )

            for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys():
                p_sc, _, _ = predict_trf_final_model(
                    model=model,
                    X_meta=scenario_meta[sc_name],
                    expert_probs=scenario_E[sc_name],
                    backbone_prob=scenario_backbone[sc_name],
                    batch_size=final_cfg["pred_batch_size"],
                    device=DEVICE,
                )
                fold_preds_per_scenario[sc_name].append(p_sc)

            del model, dl_tr, dl_va, X_tr, X_va, E_tr, E_va, Bb_tr, Bb_va, y_tr, y_va
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys():
            seed_pred = clip_probs(np.mean(np.vstack(fold_preds_per_scenario[sc_name]), axis=0))
            scenario_seed_preds[sc_name].append(seed_pred)

    scenario_final_preds = {}
    for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys():
        scenario_final_preds[sc_name] = clip_probs(
            np.mean(np.vstack(scenario_seed_preds[sc_name]), axis=0)
        )

    return scenario_final_preds

# ============================================================
# 9. Run proposed model scenario predictions
# ============================================================

# Current Proposed Model
ROBUST_PROPOSED_CURRENT = get_proposed_model_preds_all_scenarios_compare(
    y_dev=y_dev,
    base_dev_meta_df=dev_meta_df,
    final_dev_preds_local=final_dev_preds,
    robust_expert_holdout_preds_local=ROBUST_EXPERT_PREDS_CURRENT,
    selected_experts_local=selected_experts,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

# Proposed Model with AdvDeepFM
ROBUST_PROPOSED_ADV = get_proposed_model_preds_all_scenarios_compare(
    y_dev=y_dev,
    base_dev_meta_df=dev_meta_df,
    final_dev_preds_local=final_dev_preds_adv,
    robust_expert_holdout_preds_local=ROBUST_EXPERT_PREDS_ADV,
    selected_experts_local=selected_experts_adv,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

# ============================================================
# 10. Evaluate only the two proposed models
# ============================================================

compare_rows = []

for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys():
    p_curr = ROBUST_PROPOSED_CURRENT[sc_name]
    row_curr = evaluate_probs(y_holdout, p_curr, name=f"Current_Proposed_Model_{sc_name}")
    row_curr["scenario"] = sc_name
    row_curr["model_name"] = "Current_Proposed_Model"
    compare_rows.append(row_curr)

    p_adv = ROBUST_PROPOSED_ADV[sc_name]
    row_adv = evaluate_probs(y_holdout, p_adv, name=f"Proposed_Model_with_AdvDeepFM_{sc_name}")
    row_adv["scenario"] = sc_name
    row_adv["model_name"] = "Proposed_Model_with_AdvDeepFM"
    compare_rows.append(row_adv)

robust_compare_df = pd.DataFrame(compare_rows)
robust_compare_df.to_csv(TAB_DIR / "robustness_compare_current_vs_adv.csv", index=False)

display(
    robust_compare_df[
        ["scenario", "model_name", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["scenario", "model_name"])
)

# ============================================================
# 11. Compute degradation relative to each model's own clean baseline
# ============================================================

clean_compare_df = robust_compare_df[robust_compare_df["scenario"] == "clean"].copy()
clean_compare_df = clean_compare_df.set_index("model_name")

deg_rows = []

for _, row in robust_compare_df.iterrows():
    model_name = row["model_name"]
    sc_name = row["scenario"]

    base = clean_compare_df.loc[model_name]

    deg_rows.append({
        "scenario": sc_name,
        "model_name": model_name,
        "PR_AUC": row["PR_AUC"],
        "Brier": row["Brier"],
        "ECE": row["ECE"],
        "delta_PR_AUC": row["PR_AUC"] - base["PR_AUC"],
        "delta_Brier": row["Brier"] - base["Brier"],
        "delta_ECE": row["ECE"] - base["ECE"],
    })

degradation_compare_df = pd.DataFrame(deg_rows)
degradation_compare_df.to_csv(TAB_DIR / "robustness_degradation_current_vs_adv.csv", index=False)

display(
    degradation_compare_df.sort_values(["scenario", "model_name"])
)

# ============================================================
# 12. Plot direct comparison
# ============================================================

plot_order = ["clean", "noise_0.25", "noise_0.50", "mask_0.10", "mask_0.20", "corrupt_top3"]
plot_models = ["Current_Proposed_Model", "Proposed_Model_with_AdvDeepFM"]

plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_compare_df[degradation_compare_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["PR_AUC"], marker="o", label=model_name)

plt.ylabel("PR-AUC")
plt.xlabel("Scenario")
plt.title("Robustness Comparison: Current vs AdvDeepFM Proposed Model")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_compare_pr_auc_current_vs_adv.png", dpi=300)
plt.show()


plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_compare_df[degradation_compare_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["ECE"], marker="o", label=model_name)

plt.ylabel("ECE")
plt.xlabel("Scenario")
plt.title("Robustness Comparison: ECE under Perturbations")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_compare_ece_current_vs_adv.png", dpi=300)
plt.show()
```

```python
def get_expert_holdout_preds_all_scenarios_for_expert_set(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    holdout_scenarios: dict,
    selected_models: list,
    registry: dict,
    config: ExperimentConfig,
):
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    scenario_model_fold_preds = {
        sc_name: {m: [] for m in selected_models}
        for sc_name in holdout_scenarios.keys()
    }

    for model_name in selected_models:
        meta = registry[model_name]
        print("=" * 100)
        print(f"[Robustness compare] expert={model_name}")
        print("=" * 100)

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
            print(f"fold {fold_id}")

            X_tr_df = X_dev.iloc[tr_idx].copy()
            y_tr = y_dev[tr_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            # --------------------------------------------------
            # sklearn models
            # --------------------------------------------------
            if meta["kind"] == "sklearn":
                pipe = Pipeline([
                    ("pre", preproc),
                    ("model", deepcopy(meta["estimator"])),
                ])
                pipe.fit(X_tr_df, y_tr)

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    p = get_sklearn_probabilities(pipe, X_hold_sc)
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del pipe

            # --------------------------------------------------
            # torch models + adversarial torch models
            # --------------------------------------------------
            elif meta["kind"] in ["torch", "torch_adv_deepfm"]:
                X_tr = preproc.fit_transform(X_tr_df)
                if hasattr(X_tr, "toarray"):
                    X_tr = X_tr.toarray()
                X_tr = np.asarray(X_tr, dtype=np.float32)

                # simple inner split for early stopping
                inner_rng = np.random.default_rng(CFG.seed + fold_id)
                idx = np.arange(len(X_tr))
                inner_rng.shuffle(idx)
                cut = int(0.9 * len(idx))
                tr2, va2 = idx[:cut], idx[cut:]

                X_fit = X_tr[tr2]
                y_fit = y_tr[tr2].astype(np.float32)
                X_es = X_tr[va2]
                y_es = y_tr[va2].astype(np.float32)

                pos = float((y_fit == 1).sum())
                neg = float((y_fit == 0).sum())
                pos_weight = neg / max(pos, 1.0)

                dl_tr, dl_va = make_torch_loaders(
                    X_tr=X_fit,
                    y_tr=y_fit,
                    X_va=X_es,
                    y_va=y_es,
                    batch_size=int(meta["train_params"].get("batch_size", 1024)),
                )

                model = meta["builder"](X_fit.shape[1])

                # normal torch model
                if meta["kind"] == "torch":
                    model = train_torch_binary_model(
                        model=model,
                        train_loader=dl_tr,
                        valid_loader=dl_va,
                        device=DEVICE,
                        lr=float(meta["train_params"].get("lr", 1e-3)),
                        weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
                        max_epochs=int(meta["train_params"].get("max_epochs", 20)),
                        patience=int(meta["train_params"].get("patience", 4)),
                        pos_weight=pos_weight,
                    )

                # adversarial DeepFM
                else:
                    model = train_adversarial_deepfm(
                        model=model,
                        train_loader=dl_tr,
                        valid_loader=dl_va,
                        device=DEVICE,
                        lr=float(meta["train_params"].get("lr", 8e-4)),
                        weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
                        max_epochs=int(meta["train_params"].get("max_epochs", 8)),
                        patience=int(meta["train_params"].get("patience", 4)),
                        pos_weight=pos_weight,
                        noise_std=float(meta["train_params"].get("noise_std", 0.02)),
                        mask_prob=float(meta["train_params"].get("mask_prob", 0.05)),
                        fgsm_eps=float(meta["train_params"].get("fgsm_eps", 0.01)),
                        adv_bce_weight=float(meta["train_params"].get("adv_bce_weight", 0.50)),
                        consistency_weight=float(meta["train_params"].get("consistency_weight", 0.25)),
                    )

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    Xh = preproc.transform(X_hold_sc)
                    if hasattr(Xh, "toarray"):
                        Xh = Xh.toarray()
                    Xh = np.asarray(Xh, dtype=np.float32)

                    p = predict_torch_binary_model(
                        model=model,
                        X=Xh,
                        batch_size=int(meta["train_params"].get("batch_size", 2048)),
                        device=DEVICE,
                    )
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del model, dl_tr, dl_va, X_tr, X_fit, X_es

            # --------------------------------------------------
            # unsupported
            # --------------------------------------------------
            else:
                raise ValueError(f"Unsupported kind for robustness compare: {meta['kind']}")

            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    # average across folds
    scenario_model_preds = {}
    for sc_name in holdout_scenarios.keys():
        scenario_model_preds[sc_name] = {}
        for model_name in selected_models:
            scenario_model_preds[sc_name][model_name] = clip_probs(
                np.mean(np.vstack(scenario_model_fold_preds[sc_name][model_name]), axis=0)
            )

    return scenario_model_preds

ROBUST_EXPERT_PREDS_ADV = get_expert_holdout_preds_all_scenarios_for_expert_set(
    X_dev=X_dev,
    y_dev=y_dev,
    holdout_scenarios=X_HOLDOUT_SCENARIOS_COMPARE,
    selected_models=selected_experts_adv,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

ROBUST_PROPOSED_ADV = get_proposed_model_preds_all_scenarios_compare(
    y_dev=y_dev,
    base_dev_meta_df=dev_meta_df,
    final_dev_preds_local=final_dev_preds_adv,
    robust_expert_holdout_preds_local=ROBUST_EXPERT_PREDS_ADV,
    selected_experts_local=selected_experts_adv,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

compare_rows = []

for sc_name in X_HOLDOUT_SCENARIOS_COMPARE.keys():
    p_curr = ROBUST_PROPOSED_CURRENT[sc_name]
    row_curr = evaluate_probs(y_holdout, p_curr, name=f"Current_Proposed_Model_{sc_name}")
    row_curr["scenario"] = sc_name
    row_curr["model_name"] = "Current_Proposed_Model"
    compare_rows.append(row_curr)

    p_adv = ROBUST_PROPOSED_ADV[sc_name]
    row_adv = evaluate_probs(y_holdout, p_adv, name=f"Proposed_Model_with_AdvDeepFM_{sc_name}")
    row_adv["scenario"] = sc_name
    row_adv["model_name"] = "Proposed_Model_with_AdvDeepFM"
    compare_rows.append(row_adv)

robust_compare_df = pd.DataFrame(compare_rows)
display(robust_compare_df[["scenario", "model_name", "PR_AUC", "ROC_AUC", "Brier", "ECE"]])
```

```python
# ============================================================
# Simple Proposed Model scenario inference (for comparison)
# ============================================================

def get_proposed_model_preds_all_scenarios_compare(
    y_dev,
    base_dev_meta_df,
    final_dev_preds_local,
    robust_expert_holdout_preds_local,
    selected_experts_local,
    config,
    final_cfg,
):
    cv = get_cv(config, final_stage=True)

    backbone_name = final_cfg["backbone_name"]
    seed_offsets = final_cfg["seed_offsets"]

    # -------------------------
    # Clean DEV inputs
    # -------------------------
    B_dev, _ = build_backbone_predictions_locked(
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds,
        backbone_name=backbone_name,
    )

    E_dev = np.column_stack(
        [clip_probs(final_dev_preds_local[m]) for m in selected_experts_local]
    ).astype(np.float32)

    meta_dev_df, _ = build_trf_final_meta_features(
        base_dev_meta_df=base_dev_meta_df,
        base_holdout_meta_df=holdout_meta_df,
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds,
        expert_list=selected_experts_local,
        backbone_name=backbone_name,
    )

    X_meta_dev = meta_dev_df.values.astype(np.float32)

    std_idx = meta_dev_df.columns.get_loc("std_pred") if "std_pred" in meta_dev_df.columns else None

    # -------------------------
    # Scenario inputs
    # -------------------------
    scenario_meta = {}
    scenario_backbone = {}
    scenario_E = {}

    for sc_name, pred_dict in robust_expert_holdout_preds_local.items():

        meta_df = build_scenario_meta_df_for_compare(
            X_hold_sc=X_HOLDOUT_SCENARIOS_COMPARE[sc_name],
            scenario_pred_dict=pred_dict,
            selected_experts_local=selected_experts_local,
            backbone_name=backbone_name,
        )

        scenario_meta[sc_name] = meta_df.values.astype(np.float32)

        b = build_backbone_from_pred_dict_for_compare(
            pred_dict,
            backbone_name=backbone_name
        )
        scenario_backbone[sc_name] = b.astype(np.float32)

        scenario_E[sc_name] = np.column_stack(
            [clip_probs(pred_dict[m]) for m in selected_experts_local]
        ).astype(np.float32)

    # -------------------------
    # Run TRF
    # -------------------------
    scenario_preds = {sc: [] for sc in robust_expert_holdout_preds_local.keys()}

    for seed_offset in seed_offsets:
        print(f"seed_offset={seed_offset}")

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
            print(f"  fold={fold_id}")

            X_tr = X_meta_dev[tr_idx]
            X_va = X_meta_dev[va_idx]

            E_tr = E_dev[tr_idx]
            E_va = E_dev[va_idx]

            B_tr = B_dev[tr_idx]
            B_va = B_dev[va_idx]

            y_tr = y_dev[tr_idx].astype(np.float32)
            y_va = y_dev[va_idx].astype(np.float32)

            pos = float((y_tr == 1).sum())
            neg = float((y_tr == 0).sum())
            pos_weight = neg / max(pos, 1.0)

            dl_tr, dl_va = make_trf_final_loaders(
                X_meta_tr=X_tr,
                E_tr=E_tr,
                B_tr=B_tr,
                y_tr=y_tr,
                X_meta_va=X_va,
                E_va=E_va,
                B_va=B_va,
                y_va=y_va,
                batch_size=final_cfg["batch_size"],
            )

            set_global_seed(config.seed + seed_offset + fold_id)

            model = TRFFinalModel(
                meta_dim=X_meta_dev.shape[1],
                n_experts=E_dev.shape[1],
                hidden_dim=final_cfg["hidden_dim"],
                dropout=final_cfg["dropout"],
                max_delta=final_cfg["max_delta"],
                use_disagreement_scaling=final_cfg["use_disagreement_scaling"],
                std_feature_index=std_idx,
                disagreement_strength=final_cfg["disagreement_strength"],
                disagreement_bias=final_cfg["disagreement_bias"],
            )

            model = train_trf_final_model(
                model=model,
                train_loader=dl_tr,
                valid_loader=dl_va,
                device=DEVICE,
                lr=final_cfg["lr"],
                weight_decay=final_cfg["weight_decay"],
                max_epochs=final_cfg["max_epochs"],
                patience=final_cfg["patience"],
                pos_weight=pos_weight,
                delta_l2=final_cfg["delta_l2"],
                mean_delta_penalty=final_cfg["mean_delta_penalty"],
                trust_reg=final_cfg["trust_reg"],
            )

            for sc_name in scenario_preds.keys():
                p, _, _ = predict_trf_final_model(
                    model=model,
                    X_meta=scenario_meta[sc_name],
                    expert_probs=scenario_E[sc_name],
                    backbone_prob=scenario_backbone[sc_name],
                    batch_size=final_cfg["pred_batch_size"],
                    device=DEVICE,
                )
                scenario_preds[sc_name].append(p)

            del model, dl_tr, dl_va
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    # average across folds + seeds
    final_preds = {}
    for sc_name in scenario_preds.keys():
        final_preds[sc_name] = clip_probs(
            np.mean(np.vstack(scenario_preds[sc_name]), axis=0)
        )

    return final_preds
```

```python
ROBUST_PROPOSED_ADV = get_proposed_model_preds_all_scenarios_compare(
    y_dev=y_dev,
    base_dev_meta_df=dev_meta_df,
    final_dev_preds_local=final_dev_preds_adv,
    robust_expert_holdout_preds_local=ROBUST_EXPERT_PREDS_ADV,
    selected_experts_local=selected_experts_adv,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)
```

```python
# ============================================================
# Module 13 — Robustness / Stress Testing
# Proposed Model = adversarially trained version
# ============================================================

import gc
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from copy import deepcopy
from sklearn.pipeline import Pipeline

# ============================================================
# 0. Config
# ============================================================

ROBUSTNESS_CONFIG = {
    "scenarios": [
        {"name": "clean",          "kind": "clean",         "strength": 0.00},
        {"name": "noise_0.25",     "kind": "gaussian",      "strength": 0.25},
        {"name": "noise_0.50",     "kind": "gaussian",      "strength": 0.50},
        {"name": "mask_0.10",      "kind": "feature_mask",  "strength": 0.10},
        {"name": "mask_0.20",      "kind": "feature_mask",  "strength": 0.20},
        {"name": "corrupt_top3",   "kind": "corrupt_cols",  "strength": 1.00},
    ],
    "top_corrupt_cols_candidates": [
        "BMI", "GenHlth", "Age", "HighBP", "HighChol", "PhysHlth", "MentHlth"
    ],
}

# ============================================================
# 1. Scenario builders
# ============================================================

def get_numeric_cols(df: pd.DataFrame):
    return [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]

def get_existing_corrupt_cols(df: pd.DataFrame, candidates):
    return [c for c in candidates if c in df.columns][:3]

def make_holdout_scenarios(X_dev: pd.DataFrame, X_holdout: pd.DataFrame, cfg: dict):
    rng = np.random.default_rng(CFG.seed)
    scenarios = {}

    numeric_cols = get_numeric_cols(X_holdout)
    dev_std = X_dev[numeric_cols].std().replace(0, 1.0)
    dev_median = X_dev[numeric_cols].median()

    corrupt_cols = get_existing_corrupt_cols(X_holdout, cfg["top_corrupt_cols_candidates"])

    for sc in cfg["scenarios"]:
        name = sc["name"]
        kind = sc["kind"]
        strength = sc["strength"]

        Xp = X_holdout.copy()

        if kind == "clean":
            pass

        elif kind == "gaussian":
            noise = rng.normal(
                loc=0.0,
                scale=(dev_std.values * strength),
                size=(len(Xp), len(numeric_cols))
            )
            Xp.loc[:, numeric_cols] = Xp[numeric_cols].values + noise

        elif kind == "feature_mask":
            mask = rng.random((len(Xp), len(numeric_cols))) < strength
            arr = Xp[numeric_cols].values.copy()
            med = dev_median.values.reshape(1, -1)
            arr[mask] = np.broadcast_to(med, arr.shape)[mask]
            Xp.loc[:, numeric_cols] = arr

        elif kind == "corrupt_cols":
            for c in corrupt_cols:
                Xp[c] = rng.permutation(Xp[c].values)

        else:
            raise ValueError(f"Unknown scenario kind: {kind}")

        scenarios[name] = Xp

    return scenarios


X_HOLDOUT_SCENARIOS = make_holdout_scenarios(X_dev, X_holdout, ROBUSTNESS_CONFIG)
print("Scenario keys:", list(X_HOLDOUT_SCENARIOS.keys()))

# ============================================================
# 2. Expert holdout predictions for all scenarios
# ============================================================

def get_expert_holdout_preds_all_scenarios(
    X_dev: pd.DataFrame,
    y_dev: np.ndarray,
    holdout_scenarios: dict,
    selected_models: list,
    registry: dict,
    config: ExperimentConfig,
):
    cv = get_cv(config, final_stage=True)
    numeric_cols = list(X_dev.columns)

    scenario_model_fold_preds = {
        sc_name: {m: [] for m in selected_models}
        for sc_name in holdout_scenarios.keys()
    }

    for model_name in selected_models:
        meta = registry[model_name]
        print("=" * 100)
        print(f"[Robustness] Expert training for model={model_name}")
        print("=" * 100)

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_dev, y_dev), start=1):
            print(f"fold {fold_id}")

            X_tr_df = X_dev.iloc[tr_idx].copy()
            y_tr = y_dev[tr_idx]

            preproc = build_preprocessor_by_tag(meta["preproc"], numeric_cols)

            if meta["kind"] == "sklearn":
                pipe = Pipeline([
                    ("pre", preproc),
                    ("model", deepcopy(meta["estimator"])),
                ])
                pipe.fit(X_tr_df, y_tr)

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    p = get_sklearn_probabilities(pipe, X_hold_sc)
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del pipe

            elif meta["kind"] in ["torch", "torch_adv_deepfm"]:
                X_tr = preproc.fit_transform(X_tr_df)
                if hasattr(X_tr, "toarray"):
                    X_tr = X_tr.toarray()
                X_tr = np.asarray(X_tr, dtype=np.float32)

                inner_rng = np.random.default_rng(CFG.seed + fold_id)
                idx = np.arange(len(X_tr))
                inner_rng.shuffle(idx)
                cut = int(0.9 * len(idx))
                tr2, va2 = idx[:cut], idx[cut:]

                X_fit = X_tr[tr2]
                y_fit = y_tr[tr2].astype(np.float32)
                X_es = X_tr[va2]
                y_es = y_tr[va2].astype(np.float32)

                pos = float((y_fit == 1).sum())
                neg = float((y_fit == 0).sum())
                pos_weight = neg / max(pos, 1.0)

                dl_tr, dl_va = make_torch_loaders(
                    X_tr=X_fit,
                    y_tr=y_fit,
                    X_va=X_es,
                    y_va=y_es,
                    batch_size=int(meta["train_params"].get("batch_size", 1024)),
                )

                model = meta["builder"](X_fit.shape[1])

                if meta["kind"] == "torch_adv_deepfm":
                    model = train_adversarial_deepfm(
                        model=model,
                        train_loader=dl_tr,
                        valid_loader=dl_va,
                        device=DEVICE,
                        lr=float(meta["train_params"].get("lr", 8e-4)),
                        weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
                        max_epochs=int(meta["train_params"].get("max_epochs", 8)),
                        patience=int(meta["train_params"].get("patience", 4)),
                        pos_weight=pos_weight,
                        noise_std=float(meta["train_params"].get("noise_std", 0.02)),
                        mask_prob=float(meta["train_params"].get("mask_prob", 0.05)),
                        fgsm_eps=float(meta["train_params"].get("fgsm_eps", 0.01)),
                        adv_bce_weight=float(meta["train_params"].get("adv_bce_weight", 0.50)),
                        consistency_weight=float(meta["train_params"].get("consistency_weight", 0.25)),
                    )
                else:
                    model = train_torch_binary_model(
                        model=model,
                        train_loader=dl_tr,
                        valid_loader=dl_va,
                        device=DEVICE,
                        lr=float(meta["train_params"].get("lr", 1e-3)),
                        weight_decay=float(meta["train_params"].get("weight_decay", 1e-5)),
                        max_epochs=int(meta["train_params"].get("max_epochs", 8)),
                        patience=int(meta["train_params"].get("patience", 3)),
                        pos_weight=pos_weight,
                    )

                for sc_name, X_hold_sc in holdout_scenarios.items():
                    Xh = preproc.transform(X_hold_sc)
                    if hasattr(Xh, "toarray"):
                        Xh = Xh.toarray()
                    Xh = np.asarray(Xh, dtype=np.float32)

                    p = predict_torch_binary_model(
                        model=model,
                        X=Xh,
                        batch_size=int(meta["train_params"].get("batch_size", 2048)),
                        device=DEVICE,
                    )
                    scenario_model_fold_preds[sc_name][model_name].append(p)

                del model, dl_tr, dl_va, X_tr, X_fit, X_es

            else:
                raise ValueError(f"Unsupported kind for robustness: {meta['kind']}")

            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    scenario_model_preds = {}
    for sc_name in holdout_scenarios.keys():
        scenario_model_preds[sc_name] = {}
        for model_name in selected_models:
            scenario_model_preds[sc_name][model_name] = clip_probs(
                np.mean(np.vstack(scenario_model_fold_preds[sc_name][model_name]), axis=0)
            )

    return scenario_model_preds

# ============================================================
# 3. Baseline scenario predictions
# ============================================================

# make sure deepfm_adv exists in registry if selected_experts_adv uses it
if "deepfm_adv" not in CANDIDATE_MODEL_REGISTRY:
    CANDIDATE_MODEL_REGISTRY["deepfm_adv"] = {
        "family": "deep",
        "preproc": "dl",
        "kind": "torch_adv_deepfm",
        "builder": lambda in_dim: DeepFMBinaryClassifier(
            in_dim=in_dim,
            embed_dim=16,
            deep_hidden_dims=[256, 128],
            dropout=0.20,
        ),
        "train_params": {
            "lr": 8e-4,
            "weight_decay": 1e-5,
            "max_epochs": 8,
            "patience": 3,
            "batch_size": 1024,
            "noise_std": 0.02,
            "mask_prob": 0.05,
            "fgsm_eps": 0.01,
            "adv_bce_weight": 0.50,
            "consistency_weight": 0.25,
        },
    }

ROBUST_EXPERT_HOLDOUT_PREDS = get_expert_holdout_preds_all_scenarios(
    X_dev=X_dev,
    y_dev=y_dev,
    holdout_scenarios=X_HOLDOUT_SCENARIOS,
    selected_models=selected_experts_adv,
    registry=CANDIDATE_MODEL_REGISTRY,
    config=CFG,
)

# ============================================================
# 4. Ensemble predictions for all scenarios
# ============================================================

def compute_weighted_holdout_from_scenario_preds(scenario_pred_dict, selected_experts_local, final_expert_results_local_df):
    w = []
    for m in selected_experts_local:
        pr_auc = final_expert_results_local_df[
            (final_expert_results_local_df["model_name"] == m) &
            (final_expert_results_local_df["split"] == "DEV_OOF")
        ]["PR_AUC"].values[0]
        w.append(pr_auc)

    w = np.array(w, dtype=float)
    w = w / w.sum()

    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts_local])
    return clip_probs(np.dot(mat, w))

def compute_simpleavg_holdout_from_scenario_preds(scenario_pred_dict, selected_experts_local):
    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts_local])
    return clip_probs(np.mean(mat, axis=1))

def compute_stacking_holdout_from_scenario_preds(scenario_pred_dict, selected_experts_local, stack_model_local):
    mat = np.column_stack([scenario_pred_dict[m] for m in selected_experts_local])
    return clip_probs(stack_model_local.predict_proba(mat)[:, 1])


ROBUST_HOLDOUT_MODELS = {}

for sc_name, pred_dict in ROBUST_EXPERT_HOLDOUT_PREDS.items():
    out = {}

    # top baselines
    out["CatBoost"] = pred_dict["catboost"]
    out["HGB"] = pred_dict["hgb"]

    # use adversarial DeepFM here
    out["DeepFM_adv"] = pred_dict["deepfm_adv"]

    # ensemble baselines rebuilt with adv expert set
    out["WeightedEnsemble"] = compute_weighted_holdout_from_scenario_preds(
        pred_dict, selected_experts_adv, final_expert_results_adv_df
    )
    out["SimpleAverage"] = compute_simpleavg_holdout_from_scenario_preds(
        pred_dict, selected_experts_adv
    )
    out["Stacking"] = compute_stacking_holdout_from_scenario_preds(
        pred_dict, selected_experts_adv, stack_model_adv
    )

    ROBUST_HOLDOUT_MODELS[sc_name] = out

print("Built scenario predictions for baselines.")

# ============================================================
# 5. Proposed Model (adversarial version) under perturbation
# ============================================================

def build_backbone_from_pred_dict(pred_dict, backbone_name="cat_hgb_avg"):
    if backbone_name == "catboost":
        return clip_probs(pred_dict["catboost"])
    elif backbone_name == "hgb":
        return clip_probs(pred_dict["hgb"])
    elif backbone_name == "cat_hgb_avg":
        return clip_probs(0.5 * clip_probs(pred_dict["catboost"]) + 0.5 * clip_probs(pred_dict["hgb"]))
    else:
        raise ValueError(backbone_name)

def build_scenario_meta_df_aligned(
    X_hold_sc: pd.DataFrame,
    scenario_pred_dict: dict,
    selected_experts_local: list,
    backbone_name: str,
    reference_columns: list,
):
    meta_df = build_trust_features_from_pred_dict(
        X_df=X_hold_sc,
        pred_dict=scenario_pred_dict,
        expert_list=selected_experts_local,
        context_cols=[c for c in ["Age", "Sex", "BMI", "GenHlth", "HighBP", "HighChol", "Diabetes", "DiffWalk"] if c in X_hold_sc.columns]
    )

    b = build_backbone_from_pred_dict(
        scenario_pred_dict,
        backbone_name=backbone_name
    )

    meta_df["backbone_prob"] = b
    meta_df["backbone_margin"] = np.abs(b - 0.5)
    meta_df["backbone_entropy"] = binary_entropy_np(b)
    meta_df["mean_minus_backbone"] = meta_df["mean_pred"] - b
    meta_df["max_minus_backbone"] = meta_df["max_pred"] - b
    meta_df["backbone_minus_min"] = b - meta_df["min_pred"]

    for m in selected_experts_local:
        pcol = f"p_{m}"
        if pcol in meta_df.columns:
            meta_df[f"{pcol}_minus_backbone"] = meta_df[pcol] - b

    for col in reference_columns:
        if col not in meta_df.columns:
            meta_df[col] = 0.0

    meta_df = meta_df[reference_columns].copy()
    return meta_df.astype(np.float32)

def get_proposed_model_preds_all_scenarios_adv(
    y_dev,
    base_dev_meta_df,
    final_dev_preds_local,
    final_holdout_preds_local,
    robust_expert_holdout_preds_local,
    selected_experts_local,
    config,
    final_cfg,
):
    cv = get_cv(config, final_stage=True)
    backbone_name = final_cfg["backbone_name"]
    seed_offsets = final_cfg["seed_offsets"]

    B_dev, _ = build_backbone_predictions_locked(
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds_local,
        backbone_name=backbone_name,
    )

    E_dev = np.column_stack(
        [clip_probs(final_dev_preds_local[m]) for m in selected_experts_local]
    ).astype(np.float32)

    meta_dev_df, _ = build_trf_final_meta_features(
        base_dev_meta_df=base_dev_meta_df,
        base_holdout_meta_df=holdout_meta_df,
        final_dev_preds=final_dev_preds_local,
        final_holdout_preds=final_holdout_preds_local,
        expert_list=selected_experts_local,
        backbone_name=backbone_name,
    )

    reference_columns = meta_dev_df.columns.tolist()
    X_meta_dev = meta_dev_df.values.astype(np.float32)
    std_idx = meta_dev_df.columns.get_loc("std_pred") if "std_pred" in meta_dev_df.columns else None

    scenario_meta = {}
    scenario_backbone = {}
    scenario_E = {}

    for sc_name, pred_dict in robust_expert_holdout_preds_local.items():
        meta_df = build_scenario_meta_df_aligned(
            X_hold_sc=X_HOLDOUT_SCENARIOS[sc_name],
            scenario_pred_dict=pred_dict,
            selected_experts_local=selected_experts_local,
            backbone_name=backbone_name,
            reference_columns=reference_columns,
        )

        scenario_meta[sc_name] = meta_df.values.astype(np.float32)
        scenario_backbone[sc_name] = build_backbone_from_pred_dict(pred_dict, backbone_name=backbone_name).astype(np.float32)
        scenario_E[sc_name] = np.column_stack(
            [clip_probs(pred_dict[m]) for m in selected_experts_local]
        ).astype(np.float32)

    scenario_preds = {sc: [] for sc in robust_expert_holdout_preds_local.keys()}

    for seed_offset in seed_offsets:
        print(f"seed_offset={seed_offset}")

        for fold_id, (tr_idx, va_idx) in enumerate(cv.split(X_meta_dev, y_dev), start=1):
            print(f"  fold={fold_id}")

            X_tr = X_meta_dev[tr_idx]
            X_va = X_meta_dev[va_idx]
            E_tr = E_dev[tr_idx]
            E_va = E_dev[va_idx]
            B_tr = B_dev[tr_idx]
            B_va = B_dev[va_idx]
            y_tr = y_dev[tr_idx].astype(np.float32)
            y_va = y_dev[va_idx].astype(np.float32)

            pos = float((y_tr == 1).sum())
            neg = float((y_tr == 0).sum())
            pos_weight = neg / max(pos, 1.0)

            dl_tr, dl_va = make_trf_final_loaders(
                X_meta_tr=X_tr,
                E_tr=E_tr,
                B_tr=B_tr,
                y_tr=y_tr,
                X_meta_va=X_va,
                E_va=E_va,
                B_va=B_va,
                y_va=y_va,
                batch_size=final_cfg["batch_size"],
            )

            set_global_seed(config.seed + seed_offset + fold_id)

            model = TRFFinalModel(
                meta_dim=X_meta_dev.shape[1],
                n_experts=E_dev.shape[1],
                hidden_dim=final_cfg["hidden_dim"],
                dropout=final_cfg["dropout"],
                max_delta=final_cfg["max_delta"],
                use_disagreement_scaling=final_cfg["use_disagreement_scaling"],
                std_feature_index=std_idx,
                disagreement_strength=final_cfg["disagreement_strength"],
                disagreement_bias=final_cfg["disagreement_bias"],
            )

            model = train_trf_final_model(
                model=model,
                train_loader=dl_tr,
                valid_loader=dl_va,
                device=DEVICE,
                lr=final_cfg["lr"],
                weight_decay=final_cfg["weight_decay"],
                max_epochs=final_cfg["max_epochs"],
                patience=final_cfg["patience"],
                pos_weight=pos_weight,
                delta_l2=final_cfg["delta_l2"],
                mean_delta_penalty=final_cfg["mean_delta_penalty"],
                trust_reg=final_cfg["trust_reg"],
            )

            for sc_name in scenario_preds.keys():
                p, _, _ = predict_trf_final_model(
                    model=model,
                    X_meta=scenario_meta[sc_name],
                    expert_probs=scenario_E[sc_name],
                    backbone_prob=scenario_backbone[sc_name],
                    batch_size=final_cfg["pred_batch_size"],
                    device=DEVICE,
                )
                scenario_preds[sc_name].append(p)

            del model, dl_tr, dl_va
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    final_preds = {}
    for sc_name in scenario_preds.keys():
        final_preds[sc_name] = clip_probs(
            np.mean(np.vstack(scenario_preds[sc_name]), axis=0)
        )

    return final_preds


ROBUST_PROPOSED_PREDS_ADV = get_proposed_model_preds_all_scenarios_adv(
    y_dev=y_dev,
    base_dev_meta_df=dev_meta_df,
    final_dev_preds_local=final_dev_preds_adv,
    final_holdout_preds_local=final_holdout_preds_adv,
    robust_expert_holdout_preds_local=ROBUST_EXPERT_HOLDOUT_PREDS,
    selected_experts_local=selected_experts_adv,
    config=CFG,
    final_cfg=FINAL_TRF_CONFIG,
)

for sc_name in ROBUST_HOLDOUT_MODELS.keys():
    ROBUST_HOLDOUT_MODELS[sc_name]["Proposed Model"] = ROBUST_PROPOSED_PREDS_ADV[sc_name]

print("Built adversarial Proposed Model predictions for all scenarios.")

# ============================================================
# 6. Evaluate all scenarios
# ============================================================

robustness_rows = []

for sc_name, model_dict in ROBUST_HOLDOUT_MODELS.items():
    for model_name, p in model_dict.items():
        row = evaluate_probs(y_holdout, clip_probs(p), name=f"{model_name}_{sc_name}")
        row["scenario"] = sc_name
        row["model_name"] = model_name
        robustness_rows.append(row)

robustness_df = pd.DataFrame(robustness_rows)
robustness_df.to_csv(TAB_DIR / "robustness_results_adv_proposed.csv", index=False)

display(
    robustness_df[
        ["scenario", "model_name", "PR_AUC", "ROC_AUC", "Brier", "ECE"]
    ].sort_values(["scenario", "PR_AUC"], ascending=[True, False])
)

# ============================================================
# 7. Degradation table vs clean
# ============================================================

clean_df = robustness_df[robustness_df["scenario"] == "clean"].copy()
clean_df = clean_df.set_index("model_name")

degradation_rows = []

for _, row in robustness_df.iterrows():
    model_name = row["model_name"]
    sc_name = row["scenario"]

    base = clean_df.loc[model_name]

    degradation_rows.append({
        "scenario": sc_name,
        "model_name": model_name,
        "PR_AUC": row["PR_AUC"],
        "Brier": row["Brier"],
        "ECE": row["ECE"],
        "delta_PR_AUC": row["PR_AUC"] - base["PR_AUC"],
        "delta_Brier": row["Brier"] - base["Brier"],
        "delta_ECE": row["ECE"] - base["ECE"],
    })

degradation_df = pd.DataFrame(degradation_rows)
degradation_df.to_csv(TAB_DIR / "robustness_degradation_adv_proposed.csv", index=False)

display(
    degradation_df.sort_values(["scenario", "delta_PR_AUC"], ascending=[True, False])
)

# ============================================================
# 8. Plot degradation curves
# ============================================================

plot_order = ["clean", "noise_0.25", "noise_0.50", "mask_0.10", "mask_0.20", "corrupt_top3"]
plot_models = ["CatBoost", "HGB", "WeightedEnsemble", "Stacking", "Proposed Model"]

plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_df[degradation_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["PR_AUC"], marker="o", label=model_name)

plt.ylabel("PR-AUC")
plt.xlabel("Scenario")
plt.title("Robustness: PR-AUC under Perturbations")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_pr_auc_adv_proposed.png", dpi=300)
plt.show()


plt.figure(figsize=(7, 5))
for model_name in plot_models:
    dfp = degradation_df[degradation_df["model_name"] == model_name].copy()
    dfp["scenario"] = pd.Categorical(dfp["scenario"], categories=plot_order, ordered=True)
    dfp = dfp.sort_values("scenario")
    plt.plot(dfp["scenario"].astype(str), dfp["Brier"], marker="o", label=model_name)

plt.ylabel("Brier Score")
plt.xlabel("Scenario")
plt.title("Robustness: Brier under Perturbations")
plt.xticks(rotation=30, ha="right")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "robustness_brier_adv_proposed.png", dpi=300)
plt.show()
```
